# R Operators

## Operators

Operators are used to perform operations on variables and values.

In the example below, we use the + operator to add together two values:

```r
# Example
10 + 5
```

## R divides the operators in the following groups:

- Arithmetic operators

- Assignment operators

- Comparison operators

- Logical operators

- Miscellaneous operators

# R Assignment Operators

Assignment operators are used to assign values to variables:

```r
# Example
my_var <- 3

my_var <<- 3

3 -> my_var

3 ->> my_var

my_var # print my_var
```

Note: <<- is a global assigner.

It is also possible to turn the direction of the assignment operator.

x <- 3 is equal to 3 -> x

```r
x <- 10

my_function <- function() {
    x <<- 20
}

my_function()
x
```

# R Logical Operators

Logical operators are used to combine conditional statements:

Operator	                        Description

&	                                Element-wise Logical AND operator. Returns TRUE if both elements are TRUE

&&	                                Logical AND operator - Returns TRUE if both statements are TRUE

|	                                Elementwise- Logical OR operator. Returns TRUE if one of the statements is TRUE

||	                                Logical OR operator. Returns TRUE if one of the statements is TRUE

!	                                Logical NOT - Returns FALSE if statement is TRUE

# THE & OPERATOR

```r
x <- c(TRUE, FALSE)
y <- c(TRUE, TRUE)

x & y
```

```r
scores <- c(45, 67, 80, 90, 50)

scores > 60 & scores < 85
```

```r
x <- c("apple", "banana", "mango")
y <- c("apple", "grape", "mango")

x==y

(x==y) & (x== "mango")
```

```r
names <- c("John", "Mary", "James", "Peter")

(names == "John") & (names != "Peter")
```

# THE && OPERATOR

```r
TRUE && FALSE
```

```r
age <- 25
income <- 20000

if (age > 18 && income > 30000) {
    print("Approved")
}
```

```r
x <- 10

if (x > 5 && x < 20) {
    print("Valid range")
}
```

# THE | OPERATOR

```r

sales <- c(100, 250, 400, 150)

sales < 150 | sales > 300
```

```r
names <- c("John", "Mary", "Peter", "Jane")

(names == "John") | (names == "Jane")
```

```r
data <- data.frame(
    name = c("John", "Mary", "Peter", "Jane"),
    city = c("Lagos", "Abuja", "Lagos", "Kano")
)

data$name == "John" | data$city == "Lagos"
```

# THE || OPERATOR

### Simple Logical Values

```r
TRUE || FALSE
```

```r
FALSE || FALSE
```

### Using Vectors

```r
x <- c(FALSE, TRUE)
y <- c(TRUE, FALSE)

x[1] || y[1]
```

### Inside an if Statement

```r
name <- "John"
city <- "Lagos"

if (name == "Peter" || city == "Lagos") {
  print("Condition met")
}

# Summary of the above code
# FALSE || TRUE => TRUE
```

### Short-Circuit Behavior

```r
TRUE || print("Hello")
```

Notice:

It does NOT print "Hello".

Why?

Because:

- First value is TRUE

- R stops immediately

- No need to check second condition

`   This is called short-circuit evaluation.    `

### Numeric Conditions

```r
x <- 5

if (x < 3 || x > 4) {
  print("Valid")
}
```

### Strings with || (Control Logic)

```r
fruit <- "apple"

if (fruit == "banana" || fruit == "apple") {
  print("We have this fruit")
}
```
