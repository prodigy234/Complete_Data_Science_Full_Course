# R Assignment Operators

Assignment operators are used to assign values to variables:

```r
# Example
my_var <- 3

my_var <<- 3

3 -> my_var

3 ->> my_var

my_var # print my_var
```

Note: <<- is a global assigner.

It is also possible to turn the direction of the assignment operator.

x <- 3 is equal to 3 -> x

```r
x <- 10

my_function <- function() {
    x <<- 20
}

my_function()
x
```

# R Logical Operators

Logical operators are used to combine conditional statements:

Operator	                        Description

&	                                Element-wise Logical AND operator. Returns TRUE if both elements are TRUE

&&	                                Logical AND operator - Returns TRUE if both statements are TRUE

|	                                Elementwise- Logical OR operator. Returns TRUE if one of the statements is TRUE

||	                                Logical OR operator. Returns TRUE if one of the statements is TRUE

!	                                Logical NOT - Returns FALSE if statement is TRUE

# THE & OPERATOR

```r
x <- c(TRUE, FALSE)
y <- c(TRUE, TRUE)

x & y
```

```r
scores <- c(45, 67, 80, 90, 50)

scores > 60 & scores < 85
```

```r
x <- c("apple", "banana", "mango")
y <- c("apple", "grape", "mango")

x==y

(x==y) & (x== "mango")
```

```r
names <- c("John", "Mary", "James", "Peter")

(names == "John") & (names != "Peter")
```

# THE && OPERATOR

```r
TRUE && FALSE
```

```r
age <- 25
income <- 50000

if (age > 18 && income > 30000) {
    print("Approved")
}
```

```r
x <- 10

if (x > 5 && x < 20) {
    print("Valid range")
}
```

# THE | OPERATOR

```r

sales <- c(100, 250, 400, 150)

sales < 150 | sales > 300
```

```r
names <- c("John", "Mary", "Peter", "Jane")

(names == "John") | (names == "Jane")
```

```r
data <- data.frame(
    name = c("John", "Mary", "Peter", "Jane"),
    city = c("Lagos", "Abuja", "Lagos", "Kano")
)

data$name == "John" | data$city == "Lagos"
```

# THE || OPERATOR

### Simple Logical Values

```r
TRUE || FALSE
```

```r
FALSE || FALSE
```

### Using Vectors

```r
x <- c(FALSE, TRUE)
y <- c(TRUE, FALSE)

x[1] || y[1]
```

### Inside an if Statement

```r
name <- "John"
city <- "Lagos"

if (name == "Peter" || city == "Lagos") {
  print("Condition met")
}

# Summary of the above code
# FALSE || TRUE => TRUE
```

### Short-Circuit Behavior

```r
TRUE || print("Hello")
```

Notice:

It does NOT print "Hello".

Why?

Because:

- First value is TRUE

- R stops immediately

- No need to check second condition

`   This is called short-circuit evaluation.    `

### Numeric Conditions

```r
x <- 5

if (x < 3 || x > 4) {
  print("Valid")
}
```

### Strings with || (Control Logic)

```r
fruit <- "apple"

if (fruit == "banana" || fruit == "apple") {
  print("We have this fruit")
}
```

# %in% Operator (Membership Test)

The %in% operator is used for membership testing.

It checks:

“Is this value inside that vector?”

Checks if elements of one vector exist in another.

Example 1: Basic membership

```r
x <- c(1,2,3,4)
2 %in% x
```

Step-by-step Explanation

(1) x <- c(1,2,3,4)

Creates a vector:

Output => 1 2 3 4

(2) 2 %in% x

This asks:

“Is 2 inside the vector x?”

Since 2 is in (1,2,3,4), the result is:

TRUE

### What `%in%` Returns

It always returns TRUE or FALSE.

TRUE → Value exists

FALSE → Value does not exist

Example 2: Multiple values

```r
c(2,5) %in% c(1,2,3,4)
```

What This Means

Now we are checking two values at once:

- Is 2 in (1,2,3,4)?

- Is 5 in (1,2,3,4)?

R checks each value separately.

Example 3: With characters

```r
names <- c("John","Mary","Peter")
"Mary" %in% names
```

Example 4: Filtering data

```r
x <- c(10,20,30,40)
x[x %in% c(20,40)]
```

Step 1: Create the vector

x <- c(10, 20, 30, 40)

output = 10 20 30 40

Step 2: Understand This Part

x %in% c(20, 40)

This checks each value in x:

| Value | Is it in (20, 40)? |
| ----- | ------------------ |
| 10    | FALSE              |
| 20    | TRUE               |
| 30    | FALSE              |
| 40    | TRUE               |

So it returns:

FALSE TRUE FALSE TRUE

Step 3: Subsetting

x[ FALSE TRUE FALSE TRUE ]


R keeps only values where condition is TRUE.

So the output is:

20 40

Example 5: Checking missing categories

```r
cities <- c("Lagos","Abuja","Kano")
"Lagos" %in% cities
```

# %*% Operator (Matrix Multiplication)

Used for proper linear algebra multiplication.

Example 1: 2×2 matrices

```r
A <- matrix(c(1,2,3,4), nrow=2)
B <- matrix(c(5,6,7,8), nrow=2)

# 1 3
# 2 4

# 5 7
# 6 8

# Matrix Multiplication

# Multiply rows of A by columns of B

# C|i,j| = ∑(Rowi * Columnj)

# (1 * 5) + (3 * 6) = 5 + 18 = 23 (Row1 * Col1)

# (1 * 7) + (3 * 8) = 7 + 24 = 31 (Row1 * Col2)

# (2 * 5) + (4 * 6) = 10 + 24 = 34 (Row2 * Col1)

# (2 * 7) + (4 * 8) = 14 + 32 = 46 (Row2 * Col2)

A %*% B
```

```r
A <- matrix(c(1,2,3,4), nrow=2, byrow = TRUE)
B <- matrix(c(5,6,7,8), nrow=2, byrow = TRUE)

# 1 2
# 3 4

A %*% B
```

Example 2: Matrix × Vector

```r
A <- matrix(c(1,2,3,4), nrow=2)
v <- c(1,2)

# 1 3
# 2 4

# 1
# 2

# Matrix multiplication is only valid and possible when the column of the first matrix is equal to the row of the second matrix

# Note: The result of a matrix multiplication will always be  a dimension of the number of rows in the first matrix and the number of column in the second matrix

# Row 1 : (1 * 1) + (3 * 2) = 1 + 6 = 7

# Row 2 : (2 * 1) + (4 * 2) = 2 + 8 = 10

A %*% v
```

Example 3: 3×3 matrix multiplication

```r
A <- matrix(1:9, nrow=3)
B <- matrix(9:1, nrow=3)

A %*% B
```

Example 4: Transpose multiplication

```r
A <- matrix(1:6, nrow=2)
t(A) %*% A
```

Example 5: Identity matrix multiplication

```r
I <- diag(2)
A <- matrix(c(1,2,3,4), nrow=2)

A %*% I
```
