In R, package installation normally happens inside R itself (console, script, or notebook cell), not from the OS terminal.

R does it a bit differently than Python's pip.

In Python, you install from the terminal, outside your script:

` pip install pandas `

then import pandas inside the script.

In R, package installation normally happens inside R itself (console, script, or notebook cell), not from the OS terminal:


` install.packages("ggplot2") `

This is why in the notebook I wrote install.packages(...) as a regular code cell rather than a terminal command — running that cell downloads and installs the package for you.

A few ways to trigger it:

| **Where you are**                     | **How to install**                       |
| ------------------------------------- | ---------------------------------------- |
| Inside R / RStudio / a Jupyter R cell | `install.packages("ggplot2")`            |
| Terminal, one-off                     | `R -e 'install.packages("ggplot2")'`     |
| Terminal, if you use conda            | `conda install -c conda-forge r-ggplot2` |

Things worth telling R Programmers, since they'll trip over these coming from pip:

- Case-sensitive, exact names — install.packages("ggplot2"), not GGPlot2.

- You only need to run install.packages() once per machine. After that, every script/notebook just needs library(ggplot2) to load it — this is the R equivalent of Python's import. Running install.packages() every time (like some people re-run pip install out of habit) just wastes time re-downloading.

- No built-in virtual environments — base R installs packages globally by default. If you want isolated environments per project (like venv/conda env), the standard tool is the renv package (install.packages("renv"), then renv::init()).

- First install can take a minute — some R packages compile C/Fortran code locally, so install.packages() can be slower than pip install, especially for plotly or GGally.

`So the pattern for the notebook is: run each install.packages(...) cell once, then re-runs of the notebook only need the library(...) lines.`

The big one first: tidyverse

Most of "data analytics" in R lives in one meta-package:


` install.packages("tidyverse") `

**This single install pulls in ~8 packages at once: dplyr (data manipulation), tidyr (reshaping), ggplot2 (plotting), readr (reading CSVs), purrr (functional programming), tibble, stringr (text), forcats (factors/categorical data). If you only install one thing, install this.**

# By category

## Data import/export


`install.packages(c("readxl", "haven", "jsonlite", "DBI", "RSQLite"))`

- Excel files, SPSS/SAS/Stata files, JSON, and database connections.

## Statistics

`install.packages(c("psych", "car", "lme4", "survival", "broom"))`

- psych — descriptive stats, reliability analysis

- car — regression diagnostics, ANOVA

- lme4 — mixed-effects models

- survival — survival analysis

- broom — turns model output into tidy data frames (very useful, use constantly)

## Visualization (beyond ggplot2)

`install.packages(c("plotly", "GGally", "patchwork", "ggthemes", "corrplot"))`

- patchwork — combine multiple ggplots into one layout

- corrplot — quick correlation matrix visuals

- ggthemes — extra themes (Economist, FiveThirtyEight styles, etc.)

## Machine learning

`install.packages(c("caret", "tidymodels", "randomForest", "xgboost", "glmnet", "e1071", "rpart", "rpart.plot"))`

- tidymodels — the modern, tidyverse-style ML framework (like scikit-learn's philosophy) — this is the one to learn if starting fresh

- caret — the older but still widely-used ML framework, good to know since lots of tutorials use it
randomForest, xgboost, glmnet (regularized regression), e1071 (SVM, Naive Bayes), rpart (decision trees)

## Time series

`install.packages(c("forecast", "tsibble", "fable"))`

## Reporting / notebooks

`install.packages(c("rmarkdown", "knitr", "shiny"))`

- rmarkdown — R's answer to Jupyter notebooks/reports (mix code, output, and prose into HTML/PDF/Word)

- shiny — build interactive web apps/dashboards from R

## One-shot install

If you'd rather just run it all at once:

```
install.packages(c(
  "tidyverse", "readxl", "haven", "jsonlite", "DBI", "RSQLite",
  "psych", "car", "lme4", "survival", "broom",
  "plotly", "GGally", "patchwork", "ggthemes", "corrplot",
  "caret", "tidymodels", "randomForest", "xgboost", "glmnet", "e1071", "rpart", "rpart.plot",
  "forecast", "tsibble", "fable",
  "rmarkdown", "knitr", "shiny"
))
```

Fair warning: this will take a while (some of these, like xgboost, compile from source and can take several minutes), so run it once, go grab a coffee, and let it finish. You'll hit the CRAN mirror, pick 0-Cloud as the option to be selected.
