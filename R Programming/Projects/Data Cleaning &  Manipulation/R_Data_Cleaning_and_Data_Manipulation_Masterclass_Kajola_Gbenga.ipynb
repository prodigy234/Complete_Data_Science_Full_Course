# PART 1 - Foundations of Data Cleaning

## 1. What is data cleaning?

Data cleaning is the process of identifying, correcting, standardizing, documenting, and validating data-quality problems.

Typical problems include:

- missing values
- duplicate records
- inconsistent categories
- invalid dates
- numbers stored as text
- currency symbols embedded in numeric fields
- inconsistent capitalization
- leading and trailing spaces
- invalid email addresses
- malformed phone numbers
- impossible business values
- outliers that require investigation
- inconsistent identifiers

## 2. What is data manipulation?

Data manipulation is the controlled transformation of data into a structure that supports analysis, reporting, modeling, or operational use.

Examples include:

- selecting columns
- filtering rows
- creating variables
- joining tables
- grouping and aggregating
- reshaping data

## 3. Cleaning versus manipulation

Cleaning focuses on data quality. Manipulation focuses on changing the structure or representation of data for a purpose. In real projects, the two activities overlap.

## 4. Raw, clean, tidy, transformed, and analysis-ready data

**Raw data** is the source as received.

**Clean data** has known quality issues addressed according to documented rules.

**Tidy data** follows the principle that each variable is a column, each observation is a row, and each observational unit forms a table.

**Transformed data** has been deliberately changed for analysis.

**Analysis-ready data** has passed the relevant quality and validation checks for its intended use.

## 5. Reproducibility

Never rely on manually editing the Excel file and then forgetting what changed. A reproducible R script or notebook should make the transformation repeatable.

## Type-conversion warning

Never use `as.numeric()` blindly on messy currency or percentage text.

For example:

```r
as.numeric("₦1,250,000")
```

does not perform currency-aware parsing.

A safer approach is often:

```r
readr::parse_number("₦1,250,000")
```

Similarly, percentages need a deliberate rule. Decide whether `"15%"` should become `15` or `0.15` before converting.

Do not delete duplicates merely because `duplicated()` returns `TRUE`.

Ask:

1. What is one row supposed to represent?
2. Is the repeated ID legitimate?

# PART 20 - Outlier Detection

An outlier is not automatically an error.

Investigate whether it is:

- a genuine unusual observation
- a data-entry error
- a unit mismatch
- a duplicate
- a legitimate business event

Approaches:

1. domain rules
2. IQR
3. z-score
4. standard deviation
5. percentiles
6. visual inspection
7. robust statistics


````r
# IQR method for a numeric vector
x <- cleaned %>% 
  select(where(is.numeric)) %>% 
  pull(1)

q1 <- quantile(x, 0.25, na.rm = TRUE)
q3 <- quantile(x, 0.75, na.rm = TRUE)
iqr_value <- IQR(x, na.rm = TRUE)

lower <- q1 - 1.5 * iqr_value
upper <- q3 + 1.5 * iqr_value

outlier_flag <- x < lower | x > upper
table(outlier_flag, useNA = "ifany")

# Visual investigation
# boxplot(x)
````

# PART 26 - Joins and Data Integrity

Before joining, identify:

- primary keys
- foreign keys
- table grain
- expected relationship

Relationships can be:

- one-to-one
- one-to-many
- many-to-many

Join functions:

- `left_join()`
- `right_join()`
- `inner_join()`
- `full_join()`
- `semi_join()`
- `anti_join()`

A major professional risk is unexpected row multiplication caused by duplicate join keys.


````r
# Example tables
customers <- tibble(
  customer_id = c(1, 2, 3),
  customer_name = c("Ada", "Bola", "Chidi")
)

transactions <- tibble(
  transaction_id = c(101, 102, 103, 104),
  customer_id = c(1, 1, 2, 4),
  amount = c(100, 150, 200, 50)
)

left_join(customers, transactions, by = "customer_id")
right_join(customers, transactions, by = "customer_id")
inner_join(customers, transactions, by = "customer_id")
full_join(customers, transactions, by = "customer_id")

# Find transactions without a matching customer
anti_join(transactions, customers, by = "customer_id")
````

## Join diagnostics

Before joining:

```r
transactions %>%
  count(customer_id) %>%
  filter(n > 1)
```

Ask whether repeated keys are expected.

After joining, compare:

```r
nrow(left_table)
nrow(joined_table)
```

If row counts change unexpectedly, investigate the relationship instead of assuming the join is correct.

# PART 39 - Base R versus tidyverse

Both approaches are valuable.

| Task | Base R | Tidyverse |
|---|---|---|
| Filter rows | `df[df$x > 10, ]` | `filter(df, x > 10)` |
| Select columns | `df[c("x", "y")]` | `select(df, x, y)` |
| Create column | `df$new <- ...` | `mutate(df, new = ...)` |
| Summarize | `aggregate()` | `summarise()` |
| Combine rows | `rbind()` | `bind_rows()` |
| Rename | `names()` | `rename()` |

Use the approach that produces clear, reliable, maintainable code. The tidyverse is particularly expressive for data-wrangling pipelines, while base R remains fundamental and widely used.
