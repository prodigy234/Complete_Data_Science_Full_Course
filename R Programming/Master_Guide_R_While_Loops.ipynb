## 1. Introduction to Loops in R

Loops allow repeated execution of a block of code.

R provides three looping constructs:
- `for`
- `while`
- `repeat`

`while` loops are used when the number of iterations is NOT known in advance.


## 2. What is a While Loop?

A `while` loop continues running as long as a condition remains TRUE.

It checks the condition BEFORE each iteration.


## 3. Syntax and Structure

```r
while (condition) {
  # code block
}
```

Key Components:
- `condition`: Logical expression evaluated before each iteration
- Loop body: Code executed if condition is TRUE
- Must update condition inside loop


## 4. Flow of Execution
1. Evaluate condition
2. If TRUE → execute block
3. Return to condition
4. Stop when FALSE


## 5. Basic Example

```r
i <- 1

while (i <= 5) {
  print(i)
  i <- i + 1
}
```

Important: Always modify the control variable inside the loop.


## 6. Controlled Iteration Example

Factorial using while loop:

```r
n <- 5
result <- 1

while (n > 0) {
  result <- result * n
  n <- n - 1
}

result
```


## 7. Infinite Loops

Example of infinite loop:

```r
i <- 1
while (i > 0) {
  print(i)
}
```

Prevention:
- Always update control variable
- Use break condition


## 8. break and next

### break Example
```r
i <- 1
while (TRUE) {
  if (i > 5) break
  print(i)
  i <- i + 1
}
```

### next Example
```r
i <- 0
while (i < 5) {
  i <- i + 1
  if (i == 3) next
  print(i)
}
```


## 9. While Loop with Vectors

```r
x <- c(2, 4, 6, 8)
i <- 1

while (i <= length(x)) {
  print(x[i] * 2)
  i <- i + 1
}
```


## 10. User Input Simulation

```r
answer <- ''

while (answer != 'quit') {
  answer <- readline('Type quit to exit: ')
}
```


## 11. Simulation Example – Dice Roll Until 6 Appears

```r
set.seed(123)
roll <- 0
count <- 0

while (roll != 6) {
  roll <- sample(1:6, 1)
  count <- count + 1
}

count
```


## 12. Financial Application – Compound Interest

```r
principal <- 1000
rate <- 0.05
target <- 2000
years <- 0

while (principal < target) {
  principal <- principal * (1 + rate)
  years <- years + 1
}

years
```


## 13. Monte Carlo Simulation Example

Estimate probability that sum of two dice > 8

```r
set.seed(123)
success <- 0
trials <- 0

while (trials < 10000) {
  dice1 <- sample(1:6, 1)
  dice2 <- sample(1:6, 1)
  if ((dice1 + dice2) > 8) {
    success <- success + 1
  }
  trials <- trials + 1
}

success / trials
```


## 14. Nested While Loops

```r
i <- 1

while (i <= 3) {
  j <- 1
  while (j <= 2) {
    print(paste('i =', i, 'j =', j))
    j <- j + 1
  }
  i <- i + 1
}
```


## 15. Performance & Best Practices

- Avoid unnecessary computations inside loop
- Use counters properly
- Add safety break conditions
- Prefer vectorization when possible
- Profile heavy loops using system.time()


## 16. Comparison: for vs while vs repeat

| Feature | for | while | repeat |
|---------|------|--------|--------|
| Known iterations | Yes | No | No |
| Condition at start | No | Yes | No |
| Requires break | No | No | Yes |

Use while when stopping condition is data-dependent.


## 17. Debugging While Loops

- Print intermediate values
- Use browser()
- Use cat() for tracking
- Check logical conditions carefully


## 18. Common Mistakes

- Forgetting to update variable
- Wrong logical condition
- Using assignment (=) instead of comparison (==)
- Not initializing variables


## 19. Advanced Control Pattern – Convergence Algorithm

Newton-Raphson style approximation:

```r
x <- 1
tolerance <- 1e-6
diff <- 1

while (diff > tolerance) {
  new_x <- 0.5 * (x + 2/x)
  diff <- abs(new_x - x)
  x <- new_x
}

x
```
