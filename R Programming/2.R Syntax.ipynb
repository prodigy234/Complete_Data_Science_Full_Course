# R Syntax

To output text in R, use single or double quotes:

```r
"Hello World!"
```

## To output numbers, just type the number (without quotes):

```r
# Example
5
10
25
```

## To do simple calculations, add numbers together:

```r
# Example
5 + 5
```

## Print

Unlike many other programming languages, you can output code in R without using a print function:

```r
# Example

"Hello World!"
```

However, R does have a print() function available if you want to use it. This might be useful if you are familiar with other programming languages, such as Python, which often uses the print() function to output code.

```r
# Example
print("Hello World!")
```

NOTE => "[1] simply means: “This is the first item."

[1] tells us the position of the value in R’s data structure, reminding us that R treats even a single value as a vector.

You can say it like this:

“R is telling us that the text Hello World! is stored as a list of items, and this one is the first item in that list.”

Simple real-life analogy

Imagine writing a shopping list with only one item:

1. Bread

The number 1 doesn’t mean anything is wrong, it just means it’s item number one.

[1] → position of the element

"Hello World!" → the value at that position

## And there are times you must use the print() function to output code, for example when working with for loops.

```r
# Example
for (x in 1:10) {
  print(x)
}
```

## Conclusion: It is up to you whether you want to use the print() function to output code. However, when your code is inside an R expression (e.g. inside curly braces {} like in the example above), use the print() function to output the result.
