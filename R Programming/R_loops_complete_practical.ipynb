## 1. Introduction to Loops in R

Loops allow you to repeat a block of code multiple times.

The main types of loops in R are:

- `for` loop
- `while` loop
- `repeat` loop
- `break` and `next` statements

Loops are useful for:
- Iterating through vectors
- Automating calculations
- Processing datasets
- Running simulations


## 2. The `for` Loop

```r

for (i in 1:5) {
  print(i)
}
```

### Loop Through a Vector

```r

names <- c("John", "Mary", "Paul")

for (name in names) {
  print(paste("Hello", name))
}
```


## 3. Practical Example 1 – Calculate Squares


```r

numbers <- 1:10

for (n in numbers) {
  print(n^2)
}
```

## 4. The `while` Loop

```r

count <- 1

while (count <= 5) {
  print(count)
  count <- count + 1
}
```


## 5. Practical Example 2 – Savings Growth Simulation


```r

balance <- 1000
year <- 1

while (balance < 2000) {
  balance <- balance * 1.10
  print(paste("Year", year, "Balance:", balance))
  year <- year + 1
}
```

```r
# Note: The result will stop printing the value above 2000
balance <- 1000
year <- 1

while (TRUE) {
    balance <- balance * 1.10

    if (balance >= 2000) {
        break
    }

    print(paste("Year", year, "Balance:", balance))
    year <- year + 1
}
```

## 6. The `repeat` Loop

```r

x <- 1

repeat {
  print(x)
  x <- x + 1
  
  if (x > 5) {
    break
  }
}
```

## 7. `break` and `next` Statements

```r

for (i in 1:10) {
  if (i == 5) {
    next   # Skip 5
  }
  if (i == 8) {
    break  # Stop at 8
  }
  print(i)
}
```


## 8. Practical Example 3 – Loop Through Data Frame


```r

students <- data.frame(
  Name = c("John", "Mary", "Paul"),
  Score = c(45, 78, 88)
)

for (i in 1:nrow(students)) {
  if (students$Score[i] >= 50) {
    print(paste(students$Name[i], "Pass"))
  } else {
    print(paste(students$Name[i], "Fail"))
  }
}
```

## 9. Nested Loops

```r

for (i in 1:3) {
  for (j in 1:2) {
    print(paste("i =", i, "j =", j))
  }
}
```


## 10. Important Note on Performance

In R, loops are sometimes slower than vectorized operations.
Whenever possible, prefer vectorized functions like:

- `apply()`
- `lapply()`
- `sapply()`
- `ifelse()`

But loops are still very important for:
- Complex logic
- Simulations
- Iterative algorithms



## 11. Practice Exercises

1. Write a loop that prints all even numbers from 1 to 20.
2. Create a multiplication table (1–5).
3. Simulate compound interest for 10 years using a for loop.


## 12. Sample Solutions

```r

# Even numbers
for (i in 1:20) {
  if (i %% 2 == 0) {
    print(i)
  }
}

# Compound Interest
principal <- 1000
rate <- 0.05

for (year in 1:10) {
  principal <- principal * (1 + rate)
  print(paste("Year", year, "Amount:", principal))
}
```
