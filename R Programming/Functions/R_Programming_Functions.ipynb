## Introduction to R Programming Functions

In programming, functions are fundamental building blocks. They allow us to group code into reusable modules, making our programs cleaner, easier to understand, and easier to maintain.

A function is a named block of code designed to perform a specific task. Once a function is defined, you can call (or execute) it whenever you need it, rather than rewriting the same code over and over.

R is a language built around functions - almost everything you do in R, from `print()` to `mean()` to `plot()`, is itself a function call. Understanding how to write your own functions is one of the most important skills you can develop as an R programmer.

### Why Functions?

* **Reusability** - Write once, use many times.
* **Organization** - Break a program into manageable sections.
* **Readability** - Code is easier to read and understand.
* **Maintainability** - Bugs are easier to isolate and fix.
* **Avoid Repetition** - Reduce redundancy (the "Don't Repeat Yourself" principle).

Think of a function like a machine: you provide some input, it processes it, and then it produces output.

### Why Master Functions?

Functions are at the heart of writing good R code. They allow you to:

* Structure your programs cleanly.
* Solve problems piece by piece.
* Build bigger projects from smaller, understandable units.
* Reuse and share code more effectively.

"Functions make your code like Lego blocks: strong, flexible, and ready to build something greater."

## 1. What is a Function?

A function is a block of code which only runs when it is called.

You can pass data, known as **parameters** (or **arguments**), into a function.

A function can return data as a result using `return()`, or simply by evaluating an expression as the last line of the function body.

## 2. Creating a Function

In R, a function is created using the `function` keyword and assigned to a name using the assignment operator `<-`.

## 3. Defining a Function

In R, you define a function by assigning `function(...)` to a variable name. That variable name becomes the name you use to call the function afterwards.

### Syntax:

```r
function_name <- function(parameters) {

  # Optional comment explaining what the function does

  # block of code

}
```

This is a very well detailed explanation on the different components in the syntax above when creating a function:

* **function_name** - the name of your function (follow standard naming conventions, e.g. `snake_case` or `camelCase`).
* **function** - the keyword R uses to tell the interpreter you are creating a function.
* **parameters** - inputs (optional) the function needs to work with.
* **Comment / roxygen note** - (optional) a brief description of what the function does. R does not have a built-in docstring like Python, but by convention, comments (or `#'` roxygen2 comments) are placed above or inside the function to document it.
* **Curly braces `{ }`** - wrap the block of code the function executes. (For a single-line function body, the braces can be omitted.)

````r
# e.g

my_function <- function() {
  print("Hello from a function")
}

my_function()
````

## 4. Calling a Function

After defining it, you call a function by writing its name followed by parentheses `()`.

In other words, to successfully call a function, use the function name followed by parentheses:

````r
# e.g

my_function <- function() {
  # I created this function to say HI from the function.
  print("Hello from a function")
}

my_function()
````

````r
# another example

greet <- function() {
  print("Hello, welcome to R!")
}

greet()  # Function call
````

````r
greet_user <- function(name) {
  cat(sprintf("Hello, %s! Welcome to R....\n", name))
}

greet_user("Jennifer")
````

## 5. Parameters and Arguments

**Parameter**: A variable listed inside the parentheses in the function definition.

**Argument**: A value sent to the function when it is called.

### Arguments

Information can be passed into functions as arguments.

Arguments are specified after the function name, inside the parentheses. You can add as many arguments as you want, just separate them with a comma.

The example below has a function with one argument (`first_name`). When the function is called, we pass along a first name, which is used inside the function to print a greeting:

````r
my_function <- function(first_name) {
  cat(" Hi ", first_name, "\n")
}

my_function("Comfort")
my_function("Tata")
my_function("Ajanlekoko")
````

````r
my_function("Gbenga")
````

````r
greet_user <- function(name) {  # 'name' is the parameter
  cat(sprintf("Thank you %s for investing with us!\n", name))
}

greet_user("Comfort")  # 'Comfort' is the argument
````

You can have multiple parameters, separated by commas.

````r
multiply_numbers <- function(a, b) {  # Creating a function called multiply_numbers which accepts two parameters
  return(a * b)                       # The output should be the multiplication of the two parameters
}

# Now to use the function
result <- multiply_numbers(15, 30)    # Enters the two required arguments to be passed into the function
print(result)
````

### Number of Arguments

By default, a function must be called with the correct number of arguments. Meaning that if your function expects 2 arguments, you have to call the function with 2 arguments, not more, and not less (unless default values are used - see Section 8 below).

````r
# In the example below, this function expects 2 arguments, and gets 2 arguments:

my_function <- function(fname, lname) {
  cat(fname, lname, "\n")
}

my_function("Comfort", "Ajanlekoko")
````

**NOTE** => if you try to call the function created above with 1 or 3 arguments (and no defaults are set), you will get an error as shown in the example below.

This is because while creating the function, two parameters were declared for it.

So for an error not to be raised while calling the function, the number of arguments supplied when it is called must match the number of parameters that do not have default values.

````r
# This function expects 2 arguments, but gets only 1:

my_function <- function(fname, lname) {
  cat(fname, lname, "\n")
}

my_function("Emil")

# Running this produces:
# Error in my_function("Emil") : argument "lname" is missing, with no default
````

## 6. Bar Charts Without Using Functions

Before wrapping repeated plotting logic inside a function, let's first see how it looks when we do it the long way, writing the plotting code directly, without a function.

````r
# BAR CHARTS WITHOUT USING FUNCTIONS

# My data
students <- c("Ade", "Bola", "Chidi", "Zainab")
scores <- c(78, 85, 69, 90)

# My plot
barplot(scores,
        names.arg = students,
        main = "Student Scores",
        xlab = "Students",
        ylab = "Scores",
        col = "skyblue")
````

## 7. Bar Charts Using a Function

Now, let's wrap that same plotting logic inside a reusable function, exactly the benefit that functions give us: write the logic once, then reuse it with different data, titles, and labels.

````r
# My data
students <- c("Ade", "Bola", "Chidi", "Zainab")
scores <- c(78, 85, 69, 90)

Gbefun <- function(students, scores, title, xlabel, ylabel) {
  barplot(scores,
          names.arg = students,
          main = title,
          xlab = xlabel,
          ylab = ylabel,
          col = "skyblue")
}

Gbefun(students, scores, "Student Scores", "Students", "Scores")
````

## 8. Default Parameter Values

R lets you give a parameter a default value. If the caller does not supply that argument, the default is used instead, and no error is raised.

````r
greet_user <- function(name, greeting = "Hello") {
  cat(greeting, ",", name, "\n")
}

greet_user("Gbenga")                 # uses the default greeting
greet_user("Gbenga", "Good morning")  # overrides the default
````

## 9. Named Arguments

Unlike Python, R matches arguments by position by default, but you can also pass arguments by name, in any order. This is one of R's most useful features once your functions have several parameters.

````r
describe_student <- function(name, score, grade) {
  cat(sprintf("%s scored %d and got grade %s\n", name, score, grade))
}

# Positional matching
describe_student("Zainab", 90, "A")

# Named matching - order no longer matters
describe_student(grade = "A", name = "Zainab", score = 90)
````

## 10. The `...` (Ellipsis) Argument

R allows a function to accept an arbitrary number of extra arguments using `...`. This is similar in spirit to `*args`/`**kwargs` in Python.

````r
sum_all <- function(...) {
  values <- c(...)
  return(sum(values))
}

sum_all(1, 2, 3)
sum_all(10, 20, 30, 40)
````

## 11. Return Values

A function in R automatically returns the value of the last expression evaluated in its body, so `return()` is technically optional. However, using `return()` explicitly is often clearer, especially when returning early.

````r
# Implicit return (last evaluated expression is returned)
square_implicit <- function(x) {
  x^2
}

# Explicit return
square_explicit <- function(x) {
  return(x^2)
}

square_implicit(5)
square_explicit(5)
````

````r
# Early return example
check_pass <- function(score) {
  if (score < 50) {
    return("Fail")
  }
  return("Pass")
}

check_pass(40)
check_pass(75)
````

## 12. Variable Scope

A variable created inside a function belongs to the **local scope** of that function and cannot be accessed outside it. A variable created outside any function belongs to the **global scope** and can be read from inside a function, but not modified with a normal `<-` (a local copy is created instead).

````r
x <- 10  # global variable

show_value <- function() {
  x <- 5  # this creates a NEW local variable, it does not touch the global x
  print(x)
}

show_value()  # prints 5 (local)
print(x)      # prints 10 (global, unchanged)
````

````r
# To modify a global variable from inside a function, use the <<- (super-assignment) operator

counter <- 0

increment <- function() {
  counter <<- counter + 1
}

increment()
increment()
print(counter)  # prints 2
````

## 13. Recursive Functions

A function that calls itself is called a **recursive function**. Recursion is useful for problems that can be broken down into smaller versions of the same problem, such as calculating a factorial.

````r
factorial <- function(n) {
  if (n <= 1) {
    return(1)
  }
  return(n * factorial(n - 1))
}

factorial(5)   # 5 * 4 * 3 * 2 * 1 = 120
````

## 14. Anonymous (Lambda) Functions

R supports anonymous functions, functions without a name, useful for quick, throwaway operations. Since R 4.1, there is a short-hand syntax using `\(x)` in addition to the classic `function(x)` form.

````r
# Classic anonymous function
square <- (function(x) x^2)(6)
print(square)

# Shorthand syntax (R >= 4.1)
cube <- (\(x) x^3)(3)
print(cube)
````

## 15. The `apply` Family: Applying Functions Over Data

One of R's greatest strengths is applying a function to every element of a vector, list, or data frame without writing an explicit loop. The most common tools for this are `sapply()`, `lapply()`, and `mapply()`.

````r
scores <- c(78, 85, 69, 90)

# sapply: apply a function to every element, and simplify the result to a vector
grade_bonus <- sapply(scores, function(s) s + 5)
print(grade_bonus)
````

````r
# lapply: like sapply, but always returns a list
squared_list <- lapply(scores, function(s) s^2)
print(squared_list)
````

````r
# mapply: apply a function over multiple vectors in parallel
students <- c("Ade", "Bola", "Chidi", "Zainab")

report <- mapply(function(name, score) {
  sprintf("%s scored %d", name, score)
}, students, scores)

print(report)
````

## 16. Nested Functions

Functions can be defined inside other functions. The inner function has access to the variables of the outer function, this is called a **closure**.

````r
make_multiplier <- function(factor) {
  multiplier <- function(x) {
    x * factor
  }
  return(multiplier)
}

double <- make_multiplier(2)
triple <- make_multiplier(3)

double(10)  # 20
triple(10)  # 30
````

## 17. Best Practices for Writing Functions in R

* Give your function a clear, descriptive name that reflects what it does (prefer verbs, e.g. `calculate_mean`).
* Keep each function focused on a single task.
* Document your function with comments (or `#'` roxygen2 tags if you plan to build a package).
* Prefer explicit `return()` statements when a function has multiple exit points.
* Give sensible default values to optional parameters.
* Avoid modifying global variables from inside a function unless absolutely necessary; prefer returning values instead of using `<<-`.
* Test your function with a few different inputs, including edge cases, before relying on it elsewhere in your code.
