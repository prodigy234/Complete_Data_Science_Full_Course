## 1. Introduction to Conditional Statements in R

Conditional statements allow your program to make decisions.

In R, the main conditional structures are:

- `if`
- `if...else`
- `if...else if...else`
- `ifelse()` (vectorized version)

They are used to:
- Make decisions
- Validate inputs
- Categorize data
- Build business logic


## 2. Basic `if` Statement

```r

x <- 10

if (x > 5) {
  print("x is greater than 5")
}
```

## 3. `if...else` Statement

```r

score <- 45

if (score >= 50) {
  print("Pass")
} else {
  print("Fail")
}
```

## 4. `if...else if...else` Ladder

```r

score <- 75

if (score >= 90) {
  print("Grade A")
} else if (score >= 70) {
  print("Grade B")
} else if (score >= 50) {
  print("Grade C")
} else {
  print("Fail")
}
```


## 5. Practical Example 1 – Loan Approval System

We simulate a basic loan approval decision system.


```r

income <- 500000
credit_score <- 720

if (income > 400000 & credit_score > 700) {
  print("Loan Approved")
} else {
  print("Loan Rejected")
}
```


## 6. Practical Example 2 – Even or Odd Checker


```r

number <- 8

if (number %% 2 == 0) {
  print("Even number")
} else {
  print("Odd number")
}
```

## 7. Nested If Statements

```r

age <- 20
has_id <- TRUE

if (age >= 18) {
  if (has_id == TRUE) {
    print("Access Granted")
  } else {
    print("ID Required")
  }
} else {
  print("Access Denied - Underage")
}
```

## 8. Vectorized `ifelse()` Function

```r

marks <- c(45, 67, 89, 32, 76)

result <- ifelse(marks >= 50, "Pass", "Fail")
        # ifelse(condition, value_if_true, value_if_false)
result
```


## 9. Practical Example 3 – Working with Data Frames


```r

students <- data.frame(
  Name = c("John", "Mary", "Paul", "Grace"),
  Score = c(55, 82, 47, 91)
)

students$Grade <- ifelse(students$Score >= 50, "Pass", "Fail")

students
```


## 10. Logical Operators Used in If Statements

- `&`  (AND)
- `|`  (OR)
- `!`  (NOT)


```r
# If Statement with the AND Operator
temperature <- 30
humidity <- 80

if (temperature > 25 & humidity > 70) {
  print("Hot and Humid")
}
```

```r
# If Statement with the OR Operator
temperature <- 10
humidity <- 80

if (temperature > 25 | humidity > 70) {
  print("Weather condition met")
}
```

```r
# If Statement with the NOT Operator
temperature <- 10

if (!(temperature < 25)) {
  print("Not Hot!")                 
} else {
    print("Hot!")
}
```

## 13. Sample Solutions

```r
# 1. Positive, Negative, Zero
num <- -3

if (num > 0) {
  print("Positive")
} else if (num < 0) {
  print("Negative")
} else {
  print("Zero")
}
```

```r
# Asking the user to input a number
num <- as.numeric(readline(prompt = "Enter a number: "))

# Check for the condition
if (num > 0) {
    print("The number is Positive")
} else if (num < 0) {
    print("The number is Negative")
} else {
   print("The number is Zero")
}
```

```r
# Grading system with A, B, C, D, F

#. Ask for student's score
score <- as.numeric(readline(prompt = "Enter student's score: "))

if (score >= 70) {
  grade <- "A"
} else if (score >= 60) {
  grade <- "B"
} else if (score >= 50) {
  grade <- "C"
} else if (score >= 45) {
  grade <- "D"
} else {
  grade <- "F"
}

print(paste("The student's grade is:", grade))
```

```r

# 3. Discount Calculator
amount <- 150000

if (amount > 100000) {
  discount <- amount * 0.10
  print(paste("Discount:", discount))
} else {
  print("No Discount")
}
```

```r
#3. Ask user for purchase amount
amount <- as.numeric(readline(prompt = "Enter purchase amount"))

if (amount > 100000) {
    discount <- amount * 0.10
    final_price <- amount - discount

    print(paste("Discount Applied:", discount))
    print(paste("Final price:", final_price))
} else {
    print("No Discount Applied")
    print(paste("Total amount to pay:", amount))
}
```
