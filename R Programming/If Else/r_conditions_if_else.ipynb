## 1. Comparison Operators

R supports the usual logical conditions from mathematics, just like most other programming languages:

- Equal to: `a == b`
- Not equal to: `a != b`
- Less than: `a < b`
- Less than or equal to: `a <= b`
- Greater than: `a > b`
- Greater than or equal to: `a >= b`

These conditions return a `logical` value, either `TRUE` or `FALSE`. They are most commonly used inside `if` statements and loops to control the flow of a program.

An `if` statement in R is written using the `if` keyword, followed by the condition in parentheses `()`, and the code to run wrapped in curly braces `{}`.

```r
# If statement
a <- 33
b <- 200

if (b > a) {
  print("b is greater than a")
}
```

**NOTE** => In the example above we used two variables, `a` and `b`, as part of the if statement to test whether `b` is greater than `a`. Since `a` is 33 and `b` is 200, we know that 200 is greater than 33, so the program prints `"b is greater than a"` to the screen.

## 2. Code Blocks: Curly Braces, Not Indentation

Unlike Python, R does not rely on indentation to define scope. R uses curly braces `{ }` to mark the start and end of a block of code that belongs to an `if` statement, loop, or function.

Indentation in R is still considered best practice for readability, but it is not required by the language itself. This means the code below runs perfectly fine even though it is not indented, because the braces define the block:

```r
# If statement, without indentation (this still works fine in R)
a <- 33
b <- 200

if (b > a) {
print("b is greater than a, and indentation was not required")
}
```

However, if you forget the curly braces on a multi-line `if` statement, R can throw an error or behave unexpectedly, especially when run as a script rather than line by line in the console. It is always safest to use braces, even for single-line if statements, to avoid this pitfall:

```r
# Missing braces can cause problems in scripts (uncomment to test in a script file)
# a <- 33
# b <- 200
#
# if (b > a)
# print("This line may not be treated as part of the if block in a script")
#
# Always prefer using { } to wrap the body of your if statement
```

## 3. else if

In R, the keyword is written as two separate words, `else if` (not `elif` like in Python). It means "if the previous condition was not true, then try this condition instead".

```r
a <- 33
b <- 33

if (b > a) {
  print("b is greater than a")
} else if (a == b) {
  print("a and b are equal")
}
```

**NOTE** => In the example above, `a` is equal to `b`, so the first condition is not true, but the `else if` condition is true, so the program prints `"a and b are equal"`.

**Important R syntax note:** the closing brace `}` of the `if` block and the `else if` keyword must be on the same line (or the `else` must not start a fresh statement), otherwise R's parser will throw an error in script mode. Writing it as shown above, with `} else if (...) {`, is the safe and standard style.

## 4. else

The `else` keyword catches anything that is not caught by the preceding conditions.

```r
a <- 200
b <- 33

if (b > a) {
  print("b is greater than a")
} else if (a == b) {
  print("a and b are equal")
} else {
  print("a is greater than b")
}
```

In the example above, `a` is greater than `b`, so the first condition is not true, and the `else if` condition is also not true, so the program falls through to the `else` block and prints `"a is greater than b"`.

You can also have an `else` without an `else if`:

```r
a <- 200
b <- 33

if (b > a) {
  print("b is greater than a")
} else {
  print("b is not greater than a")
}
```

## 5. Short Hand If (Single Line)

If you have only one statement to execute, you can keep the whole `if` statement on a single line. The curly braces are optional in this case, though many style guides still recommend keeping them for clarity.

```r
# One line if statement
a <- 200
b <- 33

if (a > b) print("a is greater than b")
```

## 6. Short Hand If ... Else (The `ifelse()` Function)

R does not have a special one-line ternary symbol like Python's `x if condition else y`. Instead, R gives you the built-in `ifelse()` function, which works beautifully for both single values and entire vectors at once. This is one of R's most powerful features and is used constantly when working with data.

Syntax: `ifelse(condition, value_if_true, value_if_false)`

```r
a <- 2
b <- 330

result <- ifelse(a > b, "A", "B")
print(result)
```

You can also nest `ifelse()` calls to check multiple conditions on the same line, similar to chaining `elif` on one line in Python:

```r
# One line if/else if/else, with 3 outcomes
a <- 330
b <- 330

result <- ifelse(a > b, "A", ifelse(a == b, "=", "B"))
print(result)
```

### `ifelse()` is Vectorized

Unlike a normal `if` statement (which only accepts a single `TRUE`/`FALSE` value), `ifelse()` can operate on an entire vector at once. This is extremely useful in data analysis, for example when creating a new column in a data frame based on a condition.

```r
scores <- c(45, 88, 67, 90, 32, 71)

result <- ifelse(scores >= 50, "Pass", "Fail")
print(result)
```

## 7. The `switch()` Statement

When you have many possible values to check against a single variable, chaining lots of `else if` statements can get messy. R provides a `switch()` function for this situation, similar in spirit to `switch` statements in other languages.

```r
day <- "Wednesday"

result <- switch(
  day,
  "Monday" = "Start of the work week",
  "Wednesday" = "Midweek check-in",
  "Friday" = "Almost the weekend",
  "Not a specially handled day"  # default / fallback value
)

print(result)
```

## 8. `and` in R: `&&` versus `&`

R's logical AND operator combines conditional statements. There are two versions, and knowing the difference is very important:

- `&&` : compares a single pair of values only. This is what you should use inside `if()` statements.
- `&` : compares element by element across entire vectors. This is what you should use when working with vectors or data frame columns.

**Key Note**

`and = all conditions must be true.`

**If even one condition is false, the entire expression evaluates to FALSE.**

A useful way to remember it: `and` means BOTH (or ALL) conditions must be satisfied.

```r
# Test if a is greater than b, AND if c is greater than a:
a <- 200
b <- 33
c <- 500

if (a > b && c > a) {   # && -> TRUE only when ALL conditions are TRUE
  print("Both conditions are True")
}
```

```r
print(TRUE && TRUE)    # TRUE
print(TRUE && FALSE)   # FALSE
print(FALSE && TRUE)   # FALSE
print(FALSE && FALSE)  # FALSE
```

#### Truth Table for `and`

| Condition A | Condition B | A and B |
| ----------- | ----------- | ------- |
| TRUE        | TRUE        | TRUE    |
| TRUE        | FALSE       | FALSE   |
| FALSE       | TRUE        | FALSE   |
| FALSE       | FALSE       | FALSE   |

## 9. `or` in R: `||` versus `|`

Just like `and`, `or` also comes in two flavors:

- `||` : compares a single pair of values only. Use this inside `if()` statements.
- `|` : compares element by element across entire vectors.

`The or operator is used when at least one condition must be true.`

```r
# Test if a is greater than b, OR if a is greater than c:
a <- 200
b <- 33
c <- 500

if (a > b || a > c) {
  print("At least one of the conditions is True")
}
```

```r
print(TRUE || TRUE)    # TRUE
print(TRUE || FALSE)   # TRUE
print(FALSE || TRUE)   # TRUE
print(FALSE || FALSE)  # FALSE
```

#### Truth Table for `or`

| Condition A | Condition B | A or B |
| ----------- | ----------- | ------ |
| TRUE        | TRUE        | TRUE   |
| TRUE        | FALSE       | TRUE   |
| FALSE       | TRUE        | TRUE   |
| FALSE       | FALSE       | FALSE  |

#### Comparison: `and` vs `or`

| Operator | Meaning                             | Result                                   |
| -------- | ----------------------------------- | ----------------------------------------- |
| `&&`     | All conditions must be true         | `TRUE` only if every condition is `TRUE`  |
| `\|\|`     | At least one condition must be true | `TRUE` if any condition is `TRUE`         |

#### Final Note

- `&&` -> BOTH conditions must be TRUE, and it is used for single (scalar) logical values inside `if()`.
- `||` -> EITHER condition can be TRUE, and it is also used for single (scalar) logical values inside `if()`.
- `&` and `|` -> Use these instead when comparing whole vectors element by element, for example inside `ifelse()` or when filtering a data frame.

```r
# Vectorized & versus scalar && 
x <- c(TRUE, FALSE, TRUE)
y <- c(TRUE, TRUE, FALSE)

print(x & y)   # element-by-element AND, returns a vector
# print(x && y) # this would raise an error or warning in modern R, since && expects single values
```

## 10. `not`: The `!` Operator

The `!` operator is a logical operator used to reverse, or negate, the result of a conditional statement.

```r
# Test if a is NOT greater than b:
a <- 33
b <- 200

if (!(a > b)) {
  print("a is NOT greater than b")
}
```

## 11. Nested If

You can have `if` statements inside other `if` statements. This is called a nested `if` statement, and it lets you check a more specific condition only after a broader condition has already been met.

```r
x <- 41

if (x > 10) {
  print("Above ten,")
  if (x > 20) {
    print("and also above 20!")
  } else {
    print("but not above 20.")
  }
} else {
  print("below 10.")
}
```

## 12. Empty Blocks: The Equivalent of `pass`

Python requires the `pass` statement when an `if` block would otherwise be empty. R does not have this restriction, because empty curly braces `{}` are already valid and simply do nothing. You can also explicitly return `NULL` if you want to be extra clear that no action is intended.

```r
a <- 33
b <- 200

if (b > a) {
  # having an empty block like this is perfectly legal in R
}

print("Program continued without errors")
```

## 13. Putting It All Together: A Practical Example

Here is a realistic example using a named `list` (R's equivalent of a Python dictionary) to check a user's role before granting access.

```r
user <- list(username = "admin", role = "admin")

if (user$role == "admin") {
  print("Access Granted!!!")
} else {
  print("Access Denied!!!")
}
```

## 14. Quick Reference: Python vs R

| Concept                | Python                          | R                                    |
| ----------------------- | -------------------------------- | -------------------------------------- |
| Block delimiter          | Indentation                     | Curly braces `{ }`                    |
| Equality check           | `==`                             | `==`                                  |
| Not equal                | `!=`                             | `!=`                                  |
| else-if keyword          | `elif`                           | `else if`                             |
| Negation                 | `not`                            | `!`                                   |
| Scalar AND / OR          | `and` / `or`                     | `&&` / `\|\|`                          |
| Vectorized AND / OR      | not built in the same way        | `&` / `\|`                             |
| Ternary / one-line if-else | `x if cond else y`             | `ifelse(cond, x, y)`                  |
| Multi-branch dispatch     | `if/elif` chain, or `match`     | `switch()`                            |
| Empty block               | requires `pass`                 | empty `{}` is valid, no keyword needed |
| Boolean literals          | `True` / `False`                | `TRUE` / `FALSE` (or `T` / `F`)       |
