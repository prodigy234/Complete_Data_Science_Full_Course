## Strings

### What is a String?

A string (called a `character` value in R) is a sequence of characters enclosed inside quotation marks. The characters can include:

- Letters
- Numbers
- Spaces
- Symbols
- Emojis
- Special characters

In R, strings (character values) are one of the most frequently used data types, because almost every program works with text at some point.

Strings in R are surrounded by either single quotation marks or double quotation marks, exactly like Python.

`'hello'` is the same as `"hello"`.

You can display a string literal with the `print()` function, or simply by typing it directly in the R console:

```r
# Example

print("Hello")
print('Hello')
```

## Quotes Inside Quotes

You can use quotes inside a string, as long as they don't match the quotes surrounding the string:

```r
# Example

print("It's alright")
print("He is called 'Johnny'")
print('He is called "Johnny"')
```

**NOTE** => When R prints a character value back to you at the console (rather than through `print()` with `cat()`), it will show escaped quotes with a backslash, like `\"Johnny\"`, because it is showing you the literal representation of the string. If you want the clean, human-readable version without escape characters, use `cat()` instead of `print()`:

```r
cat("He is called \"Johnny\"", "\n")
```

# Assign a String to a Variable

Assigning a string to a variable in R is done with the variable name, followed by the assignment operator `<-` (the idiomatic R style), and then the string. The `=` sign also works for assignment in R, but `<-` is the convention you will see in almost all R code.

```r
# Example

a <- "Hello"
print(a)
```

# Multiline Strings

R does not have a special triple-quote syntax like Python's `"""` or `'''`. Instead, you can simply write a newline directly inside a normal pair of quotes, and R will preserve it as part of the string:

```r
# Example

# You can use double quotes with an actual line break inside them:

a <- "Lorem ipsum dolor sit amet,
consectetur adipiscing elit,
sed do eiusmod tempor incididunt
ut labore et dolore magna aliqua."
cat(a)
```

### Or single quotes:

```r
# Example

a <- 'Lorem ipsum dolor sit amet,
consectetur adipiscing elit,
sed do eiusmod tempor incididunt
ut labore et dolore magna aliqua.'
cat(a)
```

### Using `\n` Explicitly

You can also build a multiline string by inserting the newline escape character `\n` explicitly, which is often clearer in scripts than a literal line break:

```r
a <- "Line one\nLine two\nLine three"
cat(a)
```

## Escaping Quotes

Just like Python, R lets you escape a quote character with a backslash `\` so that it doesn't end the string early:

```r
a <- 'Segun\'s car'
b <- 'It\'s raining'
cat(a, "\n")
cat(b, "\n")
```

- Note: when using `cat()`, the line breaks inside a string are printed at the exact same position as they appear in the code.

## Strings Are NOT Automatically Arrays (Important R Difference!)

This is one of the biggest differences between Python and R.

In Python, a string behaves like an array of characters, and you can grab a single character using square brackets like `a[7]`.

**In R, a single character string is treated as one single value of length 1, not as a vector of characters.** You cannot use square brackets directly on a string to pull out a character. Instead, R gives you the `substr()` function to extract characters by position.

```r
# Example

# Get the character at position 8 (R uses 1-based indexing, so the first character is position 1):

a <- "Hello, World!"
b <- substr(a, 8, 8)
print(b)
```

**NOTE** => R counts positions starting from 1, not from 0 like Python. So in `"Hello, World!"`, position 1 is `"H"`, and position 8 is `"W"`. This is different from the Python version of this notebook, where `a[7]` (position 7, counting from 0) also returned `"W"`. The character is the same, but the index used to reach it is different because of the different counting convention.

### Turning a String Into an Actual Vector of Characters

If you truly want a string broken up into a vector of individual characters (closer to how Python treats strings by default), use `strsplit()` with an empty separator `""`. This returns a list, so `[[1]]` is used to pull out the character vector inside it:

```r
a <- "Hello, World!"
chars <- strsplit(a, "")[[1]]
print(chars)
```

## Looping Through a String

Since R doesn't treat a string as a vector of characters by default, you first split it into one using `strsplit()`, and then loop through the resulting character vector with a `for` loop.

`In R, a vector is the most basic data structure used to store an ordered collection of items, where every element must be of the same data type - this fills a very similar role to what Python calls an array.`

```r
# Example

# Loop through the letters in the word "banana":

for (x in strsplit("banana", "")[[1]]) {
  print(x)
}
```

## String Length

To get the length of a string, use the `nchar()` function (the R equivalent of Python's `len()`).

```r
# Example

# The nchar() function returns the number of characters in a string:

a <- "Hello, World!"
print(nchar(a))
```

**Careful:** R also has a function called `length()`, but `length()` on a single string returns `1`, because it counts how many elements are in the vector (and a single string is just one element), not how many characters are inside it. Always use `nchar()` when you want the number of characters.

```r
a <- "Hello, World!"
print(length(a))  # returns 1, this is NOT the character count!
print(nchar(a))   # returns 13, this IS the character count
```

## Check String

To check if a certain phrase or character is present in a string, R provides the `grepl()` function (pattern detection), which returns `TRUE` or `FALSE`.

```r
# Example

# Check if "life" is present in the following text:

txt <- "The best things in life are free!"
print(grepl("life", txt))
```

## Use it in an if statement:

```r
# Example
# Print only if "free" is present:

txt <- "The best things in life are free!"
if (grepl("free", txt)) {
  print("Yes, 'free' is present.")
}
```

## Check if NOT

To check if a certain phrase or character is NOT present in a string, simply negate `grepl()` with `!`.

```r
# Example

# Check if "expensive" is NOT present in the following text:

txt <- "The best things in life are free!"
print(!grepl("expensive", txt))
```

## Use it in an if statement:

```r
# Example

# print only if "expensive" is NOT present:

txt <- "The best things in life are free!"
if (!grepl("expensive", txt)) {
  print("No, 'expensive' is NOT present.")
}
```

## R - Slicing Strings (`substr()`)

### Slicing

R does not have Python's `[start:end]` slice syntax for strings. Instead, you use the `substr()` function, which takes the string, a start position, and a stop position - and unlike Python, **both the start and stop positions are included** in the result.

Syntax: `substr(x, start, stop)`

```r
# Example

# Get the characters from position 3 to position 5 (both included in R):

b <- "Hello, World!"
print(substr(b, 3, 5))
```

**NOTE** => In the Python version, `b[2:5]` returns `"llo"` using 0-based, end-exclusive indexing (positions 2, 3, 4). In R, `substr(b, 3, 5)` returns the same three characters `"llo"`, but using 1-based, end-inclusive indexing (positions 3, 4, 5). The result is identical, but you must shift your thinking about the numbers by one, and remember that R always includes the final position.

## Slice From the Start

To slice from the beginning of the string, simply set the start position to `1`:

```r
# Example

# Get the characters from the start up to and including position 5:

b <- "Hello, World!"
print(substr(b, 1, 5))
```

## Slice To the End

To slice all the way to the end of the string, set the stop position using `nchar()` to find out how long the string is:

```r
# Example

# Get the characters from position 3, all the way to the end:

b <- "Hello, World!"
print(substr(b, 3, nchar(b)))
```

## Negative Indexing (Simulated)

R does not support negative indexes on strings the way Python does. To slice relative to the end of the string, you calculate the position yourself using `nchar()`.

```r
# Example

# Get the characters:
# From: "o" in "World!" (5th position counting from the end)
# To: "r" in "World!" (3rd position counting from the end)

b <- "Hello, World!"
start <- nchar(b) - 4
stop <- nchar(b) - 2
print(substr(b, start, stop))
```

## R - Modify Strings

R has a set of built-in functions that you can use on strings, playing the same role as Python's string methods (`.upper()`, `.lower()`, and so on).

### Upper Case

```r
# Example

# The toupper() function returns the string in upper case:

a <- "Hello, World!"
print(toupper(a))
```

### Lower Case

```r
# Example

# The tolower() function returns the string in lower case:

a <- "Hello, World!"
print(tolower(a))
```

### Remove Whitespace

Whitespace is the space before and/or after the actual text, and very often you want to remove this space. In R, the `trimws()` function does this job.

```r
# Example

# The trimws() function removes any whitespace from the beginning and/or end:

a <- " Hello, World! "
b <- trimws(a)
cat(a, "\n")  # prints " Hello, World! " with the surrounding whitespace
cat(b, "\n")  # prints "Hello, World!" with the whitespace removed
```

By default `trimws()` removes whitespace from both sides. You can control this with the `which` argument, using `"left"`, `"right"`, or `"both"` (the default).

```r
a <- " Hello, World! "
cat("[", trimws(a, which = "left"), "]\n", sep = "")
cat("[", trimws(a, which = "right"), "]\n", sep = "")
```

## Replace String

To replace one substring with another, R provides `sub()` (replaces only the first match) and `gsub()` (replaces every match, standing for "global sub").

```r
# Example

# The gsub() function replaces every occurrence of a substring with another string:

a <- "Hello, World!"
print(gsub("H", "J", a))
```

### Split String

The `strsplit()` function returns a list where the text between the specified separator becomes the list items - this fills the same role as Python's `.split()` method.

```r
# Example

# The strsplit() function splits the string into substrings wherever it finds the separator:

a <- "Hello, World!"
print(strsplit(a, ","))  # returns a list containing c("Hello", " World!")
```

```r
full_name <- "James Brown"
print(strsplit(full_name, " ")[[1]])
```

**NOTE** => `strsplit()` always returns a `list`, even when you only passed in a single string. That is why `[[1]]` is used above, to reach inside the list and pull out the actual character vector of results. This trips up a lot of people coming from Python, where `.split()` returns the list of pieces directly.

## R - String Concatenation

### String Concatenation

R does not use the `+` operator to combine strings the way Python does (in fact, `+` on strings in R will throw an error). Instead, R gives you the `paste()` and `paste0()` functions.

```r
# Example

# Merge variable a with variable b into variable c, using paste0() (no separator added):

a <- "Hello"
b <- "World"
c <- paste0(a, b)
print(c)
```

`paste()` behaves the same way, except that by default it inserts a single space between the pieces it joins - which is exactly the effect we want when adding a space between two words:

```r
# Example

# To add a space between them, use paste() instead of paste0():

a <- "Hello"
b <- "World"
c <- paste(a, b)
print(c)
```

You can also control the separator explicitly with the `sep` argument, which works with both `paste()` and `paste0()`:

```r
a <- "Hello"
b <- "World"
c <- paste(a, b, sep = " ")
print(c)
```

## R - Format Strings

### String Format

Just like in Python, you cannot directly combine strings and numbers in R using `paste0()`'s cousin, the `+` operator - `+` simply does not work on strings at all in R:

```r
# Example

age <- 36
txt <- "My name is John, I am " + age
print(txt)
```

**But this is not actually a problem in R**, because `paste0()` and `paste()` already handle combining strings and numbers automatically, without needing any special f-string syntax:

```r
age <- 36
txt <- paste0("My name is John, I am ", age)
print(txt)
```

## sprintf(): R's Answer to F-Strings

While `paste0()` works fine for simple joins, R's `sprintf()` function is the closer equivalent to Python's f-strings or `.format()` method, especially when you need precise control over number formatting. It uses `%s` as a placeholder for strings, `%d` for whole numbers, and `%f` for decimal numbers.

```r
# Example

# Create a formatted string using sprintf():

age <- 36
txt <- sprintf("My name is John, I am %d", age)
print(txt)
```

## Placeholders and Modifiers

A placeholder in `sprintf()` can format a value in very specific ways, similar to Python's f-string modifiers.

```r
# Example

# Add a placeholder for the price variable:

price <- 59
txt <- sprintf("The price is %d dollars", price)
print(txt)
```

### A placeholder can include a modifier to format the value.

For decimal precision, `sprintf()` uses `%.2f`, which means a fixed-point number with 2 decimal places - the exact same modifier syntax as Python:

```r
# Example

# Display the price with 2 decimals:

price <- 72.65978
txt <- sprintf("The price is %.2f dollars", price)
print(txt)
```

### A placeholder can also contain the result of an R expression, like a math operation, by computing it first:

```r
# Example

# Perform a math operation, and place the result into the string:

txt <- sprintf("The price is %d dollars", 20 * 59)
print(txt)
```

## Task 1: Using a formatted string, produce this output: `Jack is 28 years old and he has an available balance of 100,000 naira.`

To insert comma separators into a large number in R, use the `format()` function with `big.mark = ","`.

```r
name <- "Jack"
age <- 28
available_balance <- 100000

formatted_balance <- format(available_balance, big.mark = ",", scientific = FALSE)
cat(sprintf("%s is %d years old and he has an available balance of %s naira.\n", name, age, formatted_balance))
```

## Task 2: Using a formatted string, produce this output: `Jack is 28 years old and he has an available balance of ₦100,000.`

```r
name <- "Jack"
age <- 28
available_balance <- 100000

formatted_balance <- format(available_balance, big.mark = ",", scientific = FALSE)
cat(sprintf("%s is %d years old and he has an available balance of \u20a6%s.\n", name, age, formatted_balance))
```

## Bonus: The `glue` Package (An Even Closer Match to F-Strings)

Base R's `sprintf()` is powerful, but many modern R users prefer the `glue` package for string formatting, because its syntax looks almost identical to Python's f-strings - you write `{variable}` directly inside the string instead of using `%d`/`%s` placeholders. You will need to install it once with `install.packages("glue")` before using it.

```r
# install.packages("glue")  # run this once if you don't already have the package
library(glue)

name <- "Jack"
age <- 28
available_balance <- 100000

formatted_balance <- format(available_balance, big.mark = ",", scientific = FALSE)
txt <- glue("{name} is {age} years old and he has an available balance of {formatted_balance} naira.")
print(txt)
```

## Quick Reference: Python vs R String Operations

| Concept                      | Python                          | R                                          |
| ------------------------------ | ---------------------------------- | --------------------------------------------- |
| Print a string                 | `print("Hello")`                 | `print("Hello")` or `cat("Hello")`         |
| Assign to variable             | `a = "Hello"`                    | `a <- "Hello"`                              |
| Multiline string                | `"""text"""`                    | `"text"` with a literal line break, or `\n`  |
| Escape a quote                  | `\'` or `\"`                     | `\'` or `\"`                                |
| Character at position           | `a[7]` (0-indexed)                | `substr(a, 8, 8)` (1-indexed)                |
| String as a vector of characters | automatic                        | `strsplit(a, "")[[1]]`                      |
| Length of a string              | `len(a)`                          | `nchar(a)`                                   |
| Check substring present         | `"x" in a`                       | `grepl("x", a)`                             |
| Check substring NOT present     | `"x" not in a`                   | `!grepl("x", a)`                            |
| Slice                           | `a[2:5]` (end excluded)          | `substr(a, 3, 5)` (end included)             |
| Slice from start                | `a[:5]`                           | `substr(a, 1, 5)`                            |
| Slice to end                    | `a[2:]`                           | `substr(a, 3, nchar(a))`                     |
| Negative indexing               | `a[-5:-2]`                        | not supported, compute with `nchar()`        |
| Upper case                      | `a.upper()`                       | `toupper(a)`                                 |
| Lower case                      | `a.lower()`                       | `tolower(a)`                                 |
| Strip whitespace                | `a.strip()`                       | `trimws(a)`                                  |
| Replace substring               | `a.replace("x", "y")`            | `gsub("x", "y", a)`                         |
| Split string                    | `a.split(",")`                    | `strsplit(a, ",")[[1]]`                     |
| Concatenation                    | `a + b`                           | `paste0(a, b)` or `paste(a, b)`              |
| Formatted string / f-string      | `f"{a} is {b}"`                   | `sprintf("%s is %s", a, b)` or `glue("{a} is {b}")` |
| Thousands separator              | `f"{x:,}"`                        | `format(x, big.mark = ",")`                 |
