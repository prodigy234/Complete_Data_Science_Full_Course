## 1. What Is a String in R?

A **string** is a sequence of characters such as a name, sentence, email address, product description, or transaction label.

In R, strings are usually created using either double quotes or single quotes:

```r
name <- "Gbenga"
city <- 'Lagos'
```

Strings are particularly important in data analytics because many real-world datasets contain text that needs to be cleaned, transformed, searched, or extracted.

## 18. Clean Customer Names

### Practical Question

Given `customers <- c("JOHN", "mary", "PETER", "susan")`, convert the names so that the first letter is uppercase and the remaining letters are lowercase.

### Answer

```r
customers <- c("JOHN", "mary", "PETER", "susan")

customers <- paste0(
  toupper(substr(customers, 1, 1)),
  tolower(substr(customers, 2, nchar(customers)))
)

print(customers)
```


**Explanation:** `substr()` extracts the first and remaining characters, while `toupper()` and `tolower()` standardize the case. `paste0()` joins the pieces without adding a space.

````r
customers <- c("JOHN", "mary", "PETER", "susan")

customers <- paste0(
  toupper(substr(customers, 1, 1)),
  tolower(substr(customers, 2, nchar(customers)))
)

print(customers)
````

**Expected output:**

```text
[1] "John"  "Mary"  "Peter" "Susan"
```

# 24. Challenge: Customer Data Cleaning

### Practical Question

You receive the following customer names:

```r
customers <- c(
  "  JOHN DOE",
  "mary jane  ",
  "PETER SMITH",
  "  Susan Williams "
)
```

Perform the following tasks:

1. Remove unnecessary leading and trailing spaces.
2. Convert all names to lowercase.
3. Convert the first character of each name to uppercase.
4. Display the cleaned names.

### Answer


````r
customers <- c(
  "  JOHN DOE",
  "mary jane  ",
  "PETER SMITH",
  "  Susan Williams "
)

# Step 1: Remove leading and trailing spaces
customers <- trimws(customers)

# Step 2: Convert all names to lowercase
customers <- tolower(customers)

# Step 3: Convert the first character of each name to uppercase
customers <- paste0(
  toupper(substr(customers, 1, 1)),
  substr(customers, 2, nchar(customers))
)

# Step 4: Display the cleaned names
print(customers)
````

**Expected output:**

```text
[1] "John doe"
[2] "Mary jane"
[3] "Peter smith"
[4] "Susan williams"
```

### Note

This solution capitalizes only the first character of the complete string. If you want **every word** in a person's name to be title-cased (for example, `John Doe` rather than `John doe`), you would use a different approach or a string-processing package such as `stringr`.


# 25. String Functions Quick Reference

| Function | Purpose | Example |
|---|---|---|
| `nchar()` | Count characters | `nchar("Hello")` |
| `toupper()` | Convert to uppercase | `toupper("hello")` |
| `tolower()` | Convert to lowercase | `tolower("HELLO")` |
| `substr()` | Extract characters | `substr("Hello", 1, 2)` |
| `substring()` | Extract characters | `substring("Hello", 1, 2)` |
| `paste()` | Combine strings with a separator | `paste("Data", "Science")` |
| `paste0()` | Combine strings without a separator | `paste0("Data", "Science")` |
| `sub()` | Replace first occurrence | `sub("R", "Python", text)` |
| `gsub()` | Replace all occurrences | `gsub("R", "Python", text)` |
| `strsplit()` | Split a string | `strsplit("A B C", " ")` |
| `trimws()` | Remove leading/trailing whitespace | `trimws(text)` |
| `grepl()` | Test whether a pattern exists | `grepl("R", text)` |
| `grep()` | Find matching vector elements | `grep("R", text)` |
| `regexpr()` | Find character position of a pattern | `regexpr("R", text)` |
| `length()` | Count vector elements | `length(words)` |
| `sapply()` | Apply a function over elements | `sapply(x, function)` |

## Key Distinctions

### `sub()` vs `gsub()`
- `sub()` → first matching occurrence
- `gsub()` → every matching occurrence

### `grep()` vs `grepl()`
- `grep()` → returns matching positions
- `grepl()` → returns `TRUE`/`FALSE`

### `paste()` vs `paste0()`
- `paste()` → inserts a separator (space by default)
- `paste0()` → no separator by default
