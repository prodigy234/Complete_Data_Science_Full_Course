## 1. Why Use the Apply Family?


In R, vectorized operations are often faster than traditional loops.

Instead of:

for (i in 1:n) { ... }

We use higher-order functions like:
- apply()
- lapply()
- sapply()

These follow functional programming principles:
- Functions as first-class objects
- Passing functions as arguments


## 2. apply() – For Matrices

```r

# Create a matrix
m <- matrix(1:12, nrow=3)

# Row sums
apply(m, 1, sum)

# Column means
apply(m, 2, mean)
```

## 3. lapply() – Returns List

```r

numbers <- list(a=1:5, b=6:10)

lapply(numbers, sum)
```

## 4. sapply() – Simplified Output

```r

numbers <- list(a=1:5, b=6:10)

sapply(numbers, mean)
```

## 5. vapply() – Safer Version

```r

numbers <- list(a=1:5, b=6:10)

vapply(numbers, mean, numeric(1))
```

## 6. tapply() – Grouped Operations

```r

scores <- c(80, 75, 90, 60, 85)
groups <- c("A", "A", "B", "B", "B")

tapply(scores, groups, mean)
```

## 7. mapply() – Multiple Arguments

```r

x <- 1:5
y <- 6:10

mapply(function(a, b) a + b, x, y)
```

## 8. Custom Functions & Functional Programming

```r

# Custom function
square <- function(x) {
  return(x^2)
}

sapply(1:10, square)
```

## 9. Anonymous Functions

```r

sapply(1:10, function(x) x^3)
```

## 10. Real-World Example: Financial Returns

```r

returns <- list(
  stockA = c(0.05, 0.02, -0.01, 0.04),
  stockB = c(0.03, 0.01, 0.06, -0.02)
)

# Calculate average return per stock
sapply(returns, mean)
```
