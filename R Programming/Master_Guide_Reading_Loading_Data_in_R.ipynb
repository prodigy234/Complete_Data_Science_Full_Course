## 1. Introduction to Data Import in R
R can import data from local files, Excel, databases, APIs, web URLs, statistical software, and cloud platforms.
Data import is the foundation of data analysis, data science, and financial modeling workflows.


## 2. Working Directory & File Paths
```r
getwd()
setwd('C:/Users/YourName/Documents')
list.files()
```
Understand absolute vs relative paths. Use file.choose() for interactive selection.


## 3. Reading CSV Files (Base R)
```r
data <- read.csv('data.csv', header=TRUE, sep=',', na.strings=c('', 'NA'), fileEncoding='UTF-8')
```
Key arguments: header, sep, na.strings, stringsAsFactors, fileEncoding.


## 4. Reading CSV (readr - tidyverse)
```r
library(readr)
data <- read_csv('data.csv')
```
Faster, better parsing, improved diagnostics.


## 5. Reading Excel Files
```r
library(readxl)
data <- read_excel('data.xlsx')
```
Supports .xls and .xlsx formats.


## 6. Reading Multiple Excel Sheets
```r
sheets <- excel_sheets('data.xlsx')
all_data <- lapply(sheets, function(s) read_excel('data.xlsx', sheet=s))
```


## 7. Reading Text Files
```r
read.table('data.txt', header=TRUE)
read.delim('data.txt')
```


## 8. R Native Formats
```r
save(data, file='data.RData')
load('data.RData')
readRDS('file.rds')
```


## 9. Reading JSON
```r
library(jsonlite)
data <- fromJSON('data.json')
```


## 10. Reading XML
```r
library(XML)
data <- xmlParse('data.xml')
```


## 11. Reading from Web
```r
read.csv('https://example.com/data.csv')
```


## 12. REST APIs
```r
library(httr)
res <- GET('https://api.example.com/data')
content(res)
```


## 13. Databases
```r
library(DBI)
con <- dbConnect(RSQLite::SQLite(), 'db.sqlite')
data <- dbGetQuery(con, 'SELECT * FROM table')
dbDisconnect(con)
```


## 14. Large Files
```r
library(data.table)
data <- fread('large.csv')
```


## 15. Encoding Issues
```r
read.csv('data.csv', fileEncoding='latin1')
```


## 16. Missing Values
Specify na.strings during import.


## 17. Column Types
```r
read_csv('data.csv', col_types = cols(id=col_integer(), salary=col_double()))
```


## 18. Compressed Files
```r
read.csv(gzfile('data.csv.gz'))
unzip('file.zip')
```


## 19. Google Sheets
```r
library(googlesheets4)
read_sheet('SheetURL')
```


## 20. SPSS, Stata, SAS
```r
library(haven)
read_spss('file.sav')
read_dta('file.dta')
read_sas('file.sas7bdat')
```
