## 1. Introduction to Loops in R

A loop is a programming construct that repeats a block of code multiple times.

In R, there are three main types of loops:
- `for` loop
- `while` loop
- `repeat` loop

The `for` loop is used when the number of iterations is known in advance.


## 2. Structure of a For Loop

Basic syntax:

```r
for (variable in sequence) {
  # code to execute
}
```

Components:
- `variable`: temporary variable that takes each value in the sequence
- `sequence`: vector or list to iterate over
- `{}`: block of code executed for each iteration


## 3. Basic Examples

### Example 1: Printing numbers 1 to 5

```r
for (i in 1:5) {
  print(i)
}
```

### Example 2: Using seq()

```r
for (i in seq(1, 10, by = 2)) {
  print(i)
}
```


## 4. Looping Over Vectors

```r
numbers <- c(10, 20, 30, 40)

for (num in numbers) {
  print(num * 2)
}
```

Looping using index:

```r
for (i in seq_along(numbers)) {
  print(numbers[i])
}
```


## 5. Looping Over Lists

```r
my_list <- list(a = 1, b = 2, c = 3)

for (item in my_list) {
  print(item)
}
```


## 6. Looping Over Matrices and Data Frames

### Matrix Example
```r
mat <- matrix(1:9, nrow = 3)

for (i in 1:nrow(mat)) {
  for (j in 1:ncol(mat)) {
    print(mat[i, j])
  }
}
```

### Data Frame Example
```r
df <- data.frame(name = c('A', 'B'), score = c(90, 85))

for (i in 1:nrow(df)) {
  print(paste(df$name[i], df$score[i]))
}
```


## 7. Nested For Loops

Nested loops are loops inside loops.

```r
for (i in 1:3) {
  for (j in 1:2) {
    print(paste('i =', i, 'j =', j))
  }
}
```


## 8. break and next Statements

### break
Stops the loop immediately.

```r
for (i in 1:10) {
  if (i == 5) break
  print(i)
}
```

### next
Skips current iteration.

```r
for (i in 1:5) {
  if (i == 3) next
  print(i)
}
```


## 9. Preallocating Memory (Best Practice)

Avoid growing vectors inside loops.

Bad Practice:
```r
result <- c()
for (i in 1:5) {
  result <- c(result, i^2)
}
```

Good Practice:
```r
result <- numeric(5)
for (i in 1:5) {
  result[i] <- i^2
}
```


## 10. Performance Considerations

- For loops are slower than vectorized operations
- Prefer vectorization when possible
- Use apply family for cleaner code

Vectorized alternative:
```r
x <- 1:5
x^2
```


## 11. For Loops vs apply Family

Example using lapply:

```r
lapply(1:5, function(x) x^2)
```

Apply functions are often more efficient and readable.

## 14. Advanced Patterns

### Dynamic Column Creation

```r
df <- data.frame(a = 1:5)
for (i in 1:3) {
  df[paste0('col_', i)] <- df$a * i
}
```

### Simulation Example

```r
set.seed(123)
means <- numeric(1000)

for (i in 1:1000) {
  sample_data <- rnorm(30)
  means[i] <- mean(sample_data)
}
hist(means)
```
