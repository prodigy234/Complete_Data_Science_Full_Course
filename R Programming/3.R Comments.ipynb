# R Comments

## Comments

Comments in R programming provide extra details about various aspects of a program. They are usually included at the start of a program to indicate its purpose, author, and the date on which it was written. They also provide clarification about complicated statements in the program. On executing the program, R will not consider it a program statement and will ignore it.

Comments can be used to explain R code, and to make it more readable. It can also be used to prevent execution when testing alternative code.

Comments starts with a #. 

When executing code, R will ignore anything that starts with #.

The # (hash) is used in R for single-line comments.

## This example uses a comment before a line of code:

```r
# This is a comment
"Hello World!"
```

## This example uses a comment at the end of a line of code:

```r
# Example
"Hello World!" # This is a comment
```

## Comments does not have to be text to explain the code, it can also be used to prevent R from executing the code:

```r
# "Good morning!"
"Good night!"
```

## Multiline Comments

Unlike other programming languages, such as Java, there are no syntax in R for multiline comments. However, we can just insert a # for each line to create multiline comments:

```r
# Example
# This is a comment
# written in
# more than just one line
"Hello World!"
```
