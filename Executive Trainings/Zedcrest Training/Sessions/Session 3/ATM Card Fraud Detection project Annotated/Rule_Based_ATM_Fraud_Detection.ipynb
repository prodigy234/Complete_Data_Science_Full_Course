## 📂 Section 0 - Loading the CSV File

The first thing any data analyst does when given a new dataset is **load it and look at it**.

In Python, the `csv` module is a built-in tool that reads CSV files — the same files you open in Excel every day. No installation needed.

### What is a CSV file?
CSV stands for **Comma-Separated Values**. It is a plain text file where:
- The **first row** is the column headers
- Every other row is one record
- Values are separated by commas

Our file `atm_transactions.csv` contains **2,229 bank transactions** across **80 customers** from January to March 2026.


```python
# ─────────────────────────────────────────────────────────────────────────────
# STEP 1 - Import the csv module (built into Python, no installation needed)
# ─────────────────────────────────────────────────────────────────────────────

import csv    # reads and writes CSV files

# ─────────────────────────────────────────────────────────────────────────────
# STEP 2 - Load the CSV file into a list of dictionaries
#
# csv.DictReader reads each row as a dictionary:
#   {"transaction_id": "TXN000001", "customer_name": "Emeka Okafor", ...}
#
# This means we access values by column name, not by position — much clearer!
# ─────────────────────────────────────────────────────────────────────────────

transactions = []   # empty list — will hold all 2,229 rows

with open("atm_transactions.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)          # DictReader reads each row as a dict
    for row in reader:                     # loop through every row
        transactions.append(dict(row))     # add it to our list

# ─────────────────────────────────────────────────────────────────────────────
# STEP 3 - Confirm how many rows were loaded
# ─────────────────────────────────────────────────────────────────────────────

print(f"✅ File loaded successfully!")
print(f"   Total transactions loaded : {len(transactions)}")
print(f"   Columns per row           : {len(transactions[0])}")
print()
print("Column names:")
```

> Numbers like `250000.00` are read as the text `"250000.00"`, not the number `250000.00`.  
> We must convert them using `float()` or `int()` before doing any maths.


```python
# ─────────────────────────────────────────────────────────────────────────────
# STEP 5 - Convert numeric columns from strings to numbers
#
# CSV files store everything as text. We need actual numbers for comparisons.
# float("250000.00") converts the text "250000.00" to the number 250000.0
# int("50000")       converts the text "50000" to the number 50000
# ─────────────────────────────────────────────────────────────────────────────

for txn in transactions:
    txn["amount_naira"]       = float(txn["amount_naira"])
    txn["balance_before"]     = float(txn["balance_before"])
    txn["balance_after"]      = float(txn["balance_after"])
    txn["typical_max_amount"] = float(txn["typical_max_amount"])
    txn["daily_limit"]        = float(txn["daily_limit"])
    txn["daily_total_so_far"] = float(txn["daily_total_so_far"])

print("✅ Numeric columns converted from text to numbers.")
print()

# Verify - check the type of amount_naira
sample = transactions[0]
print(f"amount_naira value : {sample['amount_naira']}")
print(f"amount_naira type  : {type(sample['amount_naira'])}")
print()

# Now we can do maths on it
print(f"Example calculation:")
print(f"  amount * 2 = ₦{sample['amount_naira'] * 2:,.2f}")
print(f"  (This would fail if it were still a string)")
```

## 📐 Section 2 - Comparison Operators Applied to Fraud Detection

Now we apply the comparison operators we learned in Session 2 — but this time on **real banking data**.

### The Four Fraud Rules We Will Apply

| Rule | Condition | Why It Matters |
|------|-----------|----------------|
| **Rule 1** | Amount > 3× customer's typical max | Unusual spending spike |
| **Rule 2** | Transaction hour between 0 and 4 (midnight–4am) | High-risk time window |
| **Rule 3** | ATM location is far from home state | Card may have been cloned |
| **Rule 4** | Daily total exceeds customer's daily limit | Possible card takeover |


```python
# ─────────────────────────────────────────────────────────────────────────────
# SECTION 2- Applying comparison operators to a single transaction
#
# Let's pick ONE transaction and apply all four fraud rules manually.
# We will see clearly what each comparison returns.
# ─────────────────────────────────────────────────────────────────────────────

# Find a suspicious transaction to examine
suspicious_txn = None
for txn in transactions:
    if txn["is_fraud_flag"] == "SUSPICIOUS":
        suspicious_txn = txn
        break

t = suspicious_txn

print("Examining this transaction:")
print("-" * 55)
print(f"  Transaction ID   : {t['transaction_id']}")
print(f"  Customer         : {t['customer_name']} ({t['customer_id']})")
print(f"  Date & Time      : {t['date']} at {t['time']}")
print(f"  Amount           : ₦{t['amount_naira']:,.2f}")
print(f"  Typical Max      : ₦{t['typical_max_amount']:,.2f}")
print(f"  ATM Location     : {t['atm_location']}")
print(f"  Home State       : {t['home_state']}")
print(f"  Daily Total      : ₦{t['daily_total_so_far']:,.2f}")
print(f"  Daily Limit      : ₦{t['daily_limit']:,.2f}")
print(f"  Ground Truth     : {t['is_fraud_flag']}")
print()

# ── RULE 1: Unusually large amount ───────────────────────────────────────────
# If the amount is more than 3× the customer's typical maximum, it is suspicious.
# We extract the hour from the time string "HH:MM:SS" using string slicing.

rule1 = t["amount_naira"] > (t["typical_max_amount"] * 3)
print(f"Rule 1 — Amount > 3× typical max?")
print(f"  ₦{t['amount_naira']:,.2f} > ₦{t['typical_max_amount']*3:,.2f}  →  {rule1}")
print()

# ── RULE 2: Unusual hour (midnight to 4am) ───────────────────────────────────
# The time column is a string: "23:47:12". We take the first 2 characters to
# get the hour, then convert to integer: int("23") = 23
hour = int(t["time"][:2])    # "23:47:12" → "23" → 23
rule2 = (hour >= 0) and (hour <= 4)
print(f"Rule 2 — Transaction between midnight and 4am?")
print(f"  Hour = {hour}  →  (0 <= {hour} <= 4)  →  {rule2}")
print()

# ── RULE 3: Transaction in a different state from home ───────────────────────
# The home_state tells us where the customer normally transacts.
# The atm_location column may contain a different state if the card was used elsewhere.
home = t["home_state"]
atm_loc = t["atm_location"]
rule3 = (home not in atm_loc) and ("HOME STATE" in atm_loc)
print(f"Rule 3 — Card used far from home state?")
print(f"  Home state: {home}")
print(f"  ATM:        {atm_loc[:50]}...")
print(f"  Result:     {rule3}")
print()

# ── RULE 4: Daily limit exceeded ─────────────────────────────────────────────
rule4 = t["daily_total_so_far"] > t["daily_limit"]
print(f"Rule 4 — Daily limit exceeded?")
print(f"  Daily total ₦{t['daily_total_so_far']:,.2f} > Limit ₦{t['daily_limit']:,.2f}  →  {rule4}")
print()

# ── FINAL VERDICT using OR: suspicious if ANY rule triggers ──────────────────
is_suspicious = rule1 or rule2 or rule3 or rule4
print(f"VERDICT (Rule1 OR Rule2 OR Rule3 OR Rule4): {is_suspicious}")
```

## 🔀 Section 3 — if / elif / else — Classifying Each Transaction

Now we write the classification logic. Each transaction will be assigned a **risk level**:

| Level | Meaning | Action |
|-------|---------|--------|
| `🔴 HIGH RISK` | Multiple fraud rules triggered | Block card immediately, call customer |
| `🟠 MEDIUM RISK` | One serious rule triggered | Flag for analyst review within 1 hour |
| `🟡 LOW RISK` | Minor suspicious pattern | Log and monitor |
| `🟢 NORMAL` | No rules triggered | No action needed |


```python
# ─────────────────────────────────────────────────────────────────────────────
# SECTION 3A — Classify a single transaction using if/elif/else
# ─────────────────────────────────────────────────────────────────────────────

def get_hour(time_str):
    """Extract hour as integer from time string 'HH:MM:SS'."""
    return int(time_str[:2])

def check_rules(txn):
    """
    Apply all 4 fraud rules to one transaction.
    Returns a dictionary showing which rules were triggered.
    """
    hour = get_hour(txn["time"])
    
    rule1 = txn["amount_naira"] > (txn["typical_max_amount"] * 3)
    rule2 = (hour >= 0) and (hour <= 4)
    rule3 = (txn["home_state"] not in txn["atm_location"]) and ("HOME STATE" in txn["atm_location"])
    rule4 = txn["daily_total_so_far"] > txn["daily_limit"]
    
    return {
        "rule1_large_amount":  rule1,
        "rule2_unusual_hour":  rule2,
        "rule3_wrong_location":rule3,
        "rule4_limit_exceeded":rule4,
        "rules_triggered":     sum([rule1, rule2, rule3, rule4])
    }


def classify_transaction(txn):
    """
    Classify a transaction into a risk level based on how many rules triggered.
    
    Returns:
        dict: {risk_level, emoji, action, rules}
    """
    rules = check_rules(txn)
    n     = rules["rules_triggered"]   # number of rules that fired
    
    # Multiple rules triggered = very high confidence of fraud
    if n >= 3:
        risk_level = "HIGH RISK"
        emoji      = "🔴"
        action     = "BLOCK CARD IMMEDIATELY. Call customer. Escalate to fraud team."
    
    elif n == 2:
        risk_level = "HIGH RISK"
        emoji      = "🔴"
        action     = "Flag for immediate review. Temporary card hold recommended."
    
    elif n == 1:
        # Which specific rule triggered determines the severity
        if rules["rule1_large_amount"]:
            risk_level = "MEDIUM RISK"
            emoji      = "🟠"
            action     = "Send SMS alert to customer. Analyst review within 1 hour."
        
        elif rules["rule2_unusual_hour"]:
            risk_level = "MEDIUM RISK"
            emoji      = "🟠"
            action     = "Late-night transaction flagged. Monitor for follow-up activity."
        
        elif rules["rule3_wrong_location"]:
            risk_level = "MEDIUM RISK"
            emoji      = "🟠"
            action     = "Out-of-state transaction. Verify with customer via SMS."
        
        elif rules["rule4_limit_exceeded"]:
            risk_level = "LOW RISK"
            emoji      = "🟡"
            action     = "Daily limit exceeded. Flag and log for review."
        
        else:
            risk_level = "LOW RISK"
            emoji      = "🟡"
            action     = "Minor pattern detected. Monitor."
    
    else:
        risk_level = "NORMAL"
        emoji      = "🟢"
        action     = "No action required."
    
    return {
        "risk_level": risk_level,
        "emoji":      emoji,
        "action":     action,
        "rules":      rules
    }


# ── Test on our suspicious transaction ───────────────────────────────────────
result = classify_transaction(suspicious_txn)

print(f"Transaction : {suspicious_txn['transaction_id']}")
print(f"Customer    : {suspicious_txn['customer_name']}")
print(f"Amount      : ₦{suspicious_txn['amount_naira']:,.2f}")
print()
print(f"Rules triggered:")
for rule, fired in result['rules'].items():
    if rule != "rules_triggered":
        marker = "  ✅ YES" if fired else "  ❌ No"
        print(f"{marker}  —  {rule}")
print()
print(f"Risk Level  : {result['emoji']} {result['risk_level']}")
print(f"Action      : {result['action']}")
print()
print(f"Ground truth: {suspicious_txn['is_fraud_flag']}")
print(f"Reason      : {suspicious_txn['fraud_reason']}")
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# SECTION 3B — Test on multiple transactions — see different outcomes
# ─────────────────────────────────────────────────────────────────────────────

# Pick 6 transactions: 3 suspicious, 3 normal
suspicious_sample = [t for t in transactions if t["is_fraud_flag"] == "SUSPICIOUS"][:3]
normal_sample     = [t for t in transactions if t["is_fraud_flag"] == "NORMAL"][:3]
test_sample       = suspicious_sample + normal_sample

print("=" * 80)
print("  CLASSIFICATION TEST — 6 Transactions")
print("=" * 80)

for txn in test_sample:
    result = classify_transaction(txn)
    hour   = get_hour(txn["time"])
    print(f"\n  {result['emoji']} [{result['risk_level']:<12}] {txn['transaction_id']}")
    print(f"     Customer : {txn['customer_name']:<25}  Amount: ₦{txn['amount_naira']:>12,.2f}")
    print(f"     Time     : {txn['date']} {txn['time']}  (Hour: {hour})")
    print(f"     Location : {txn['atm_location'][:55]}")
    print(f"     Rules hit: {result['rules']['rules_triggered']}")
    print(f"     Action   : {result['action']}")
    print(f"     Truth    : {txn['is_fraud_flag']}")
```

## 🔁 Section 4 — for Loops — Scanning All 2,229 Transactions Automatically

We just classified 6 transactions manually. Now we scale that to all **2,229 transactions** with one loop.

This is the power of loops: the exact same logic runs on every single row.


```python
# ─────────────────────────────────────────────────────────────────────────────
# SECTION 4A — Loop over ALL 2,229 transactions and classify each one
# ─────────────────────────────────────────────────────────────────────────────

flagged_transactions = []   # will hold all HIGH and MEDIUM risk transactions
all_results          = []   # will hold the result for every transaction

for txn in transactions:                    # loop over all 2,229 rows
    result = classify_transaction(txn)      # classify this transaction
    
    # Attach the result to the transaction for easy access later
    txn["detected_risk"]   = result["risk_level"]
    txn["detected_emoji"]  = result["emoji"]
    txn["detected_action"] = result["action"]
    txn["rules_triggered"] = result["rules"]["rules_triggered"]
    
    all_results.append(result)
    
    # Collect flagged ones for our fraud report
    if result["risk_level"] in ("HIGH RISK", "MEDIUM RISK"):
        flagged_transactions.append(txn)

# ── Summary counts ────────────────────────────────────────────────────────────
high_risk   = sum(1 for r in all_results if r["risk_level"] == "HIGH RISK")
medium_risk = sum(1 for r in all_results if r["risk_level"] == "MEDIUM RISK")
low_risk    = sum(1 for r in all_results if r["risk_level"] == "LOW RISK")
normal      = sum(1 for r in all_results if r["risk_level"] == "NORMAL")

print(f"✅ Scan complete — {len(transactions):,} transactions processed.")
print()
print("=" * 55)
print("  SCAN RESULTS")
print("=" * 55)
print(f"  🔴 HIGH RISK   : {high_risk:>5} transactions")
print(f"  🟠 MEDIUM RISK : {medium_risk:>5} transactions")
print(f"  🟡 LOW RISK    : {low_risk:>5} transactions")
print(f"  🟢 NORMAL      : {normal:>5} transactions")
print(f"  {'─'*35}")
print(f"  Total flagged  : {high_risk + medium_risk:>5} ({(high_risk+medium_risk)/len(transactions)*100:.1f}% of all transactions)")
print("=" * 55)
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# SECTION 4B — Summary by bank
# Which bank has the most flagged transactions?
# ─────────────────────────────────────────────────────────────────────────────

banks = ["Access Bank","GTBank","Zenith Bank","First Bank","UBA","Fidelity Bank","Stanbic IBTC"]

print("Flagged Transactions by Bank:")
print("=" * 55)
print(f"  {'Bank':<20}  {'Total Txns':>10}  {'Flagged':>8}  {'Rate':>6}")
print("=" * 55)

for bank in banks:
    bank_txns    = [t for t in transactions if t["bank"] == bank]
    bank_flagged = [t for t in bank_txns if t["detected_risk"] in ("HIGH RISK","MEDIUM RISK")]
    rate = len(bank_flagged)/len(bank_txns)*100 if bank_txns else 0
    print(f"  {bank:<20}  {len(bank_txns):>10,}  {len(bank_flagged):>8}  {rate:>5.1f}%")

print("=" * 55)
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# SECTION 4C — Summary by transaction type
# Which type of transaction is most often flagged?
# ─────────────────────────────────────────────────────────────────────────────

txn_types = ["ATM Withdrawal","POS Purchase","Online Transfer","ATM Balance Enquiry","ATM Mini Statement"]

print("Flagged Transactions by Transaction Type:")
print("=" * 65)
print(f"  {'Type':<25}  {'Total':>8}  {'Flagged':>8}  {'Rate':>6}")
print("=" * 65)

for ttype in txn_types:
    type_txns    = [t for t in transactions if t["transaction_type"] == ttype]
    type_flagged = [t for t in type_txns if t["detected_risk"] in ("HIGH RISK","MEDIUM RISK")]
    rate = len(type_flagged)/len(type_txns)*100 if type_txns else 0
    print(f"  {ttype:<25}  {len(type_txns):>8,}  {len(type_flagged):>8}  {rate:>5.1f}%")

print("=" * 65)
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# SECTION 4D — Hourly distribution of suspicious transactions
# At what time of day is fraud most common?
# ─────────────────────────────────────────────────────────────────────────────

print("Suspicious Transactions by Hour of Day:")
print("=" * 55)
print(f"  {'Hour':<8}  {'Total Txns':>10}  {'Suspicious':>10}  {'Rate':>6}  Bar Chart")
print("=" * 55)

for hour_num in range(24):
    hour_txns  = [t for t in transactions if get_hour(t["time"]) == hour_num]
    hour_susp  = [t for t in hour_txns if t["detected_risk"] in ("HIGH RISK","MEDIUM RISK")]
    rate       = len(hour_susp)/len(hour_txns)*100 if hour_txns else 0
    bar        = "█" * int(rate / 2)   # 1 block per 2%
    time_label = f"{hour_num:02d}:00"
    print(f"  {time_label:<8}  {len(hour_txns):>10,}  {len(hour_susp):>10}  {rate:>5.1f}%  {bar}")

print("=" * 55)
```

## ⚙️ Section 5 — Functions — Building a Reusable Fraud Detection Engine

Now we build the full, professional fraud detection system as a set of clean, reusable functions. Each function does ONE job. They work together like an assembly line.


```python
# ─────────────────────────────────────────────────────────────────────────────
# SECTION 5A — A complete, professional fraud detection function
# ─────────────────────────────────────────────────────────────────────────────

def detect_fraud(txn, large_amount_multiplier=3.0, max_hour=4):
    """
    Full fraud detection pipeline for a single bank transaction.
    
    Applies 4 fraud detection rules and returns a complete analysis.
    
    Args:
        txn                    (dict) : One transaction row from the dataset
        large_amount_multiplier (float): How many times typical_max counts as suspicious (default 3.0)
        max_hour               (int)  : Latest 'suspicious' hour after midnight (default 4 = 4am)
    
    Returns:
        dict: {
            'transaction_id': str,
            'customer':       str,
            'amount':         float,
            'risk_level':     str,
            'emoji':          str,
            'rules_hit':      list,
            'action':         str,
            'confidence':     str
        }
    """
    
    hour       = get_hour(txn["time"])
    rules_hit  = []
    
    # ── Rule 1: Unusually large amount ────────────────────────────────────────
    threshold_amount = txn["typical_max_amount"] * large_amount_multiplier
    if txn["amount_naira"] > threshold_amount:
        ratio = txn["amount_naira"] / txn["typical_max_amount"]
        rules_hit.append(f"LARGE AMOUNT ({ratio:.1f}x typical max of ₦{txn['typical_max_amount']:,.0f})")
    
    # ── Rule 2: Unusual hour ──────────────────────────────────────────────────
    if 0 <= hour <= max_hour:
        rules_hit.append(f"UNUSUAL HOUR ({txn['time']} — between midnight and {max_hour}am)")
    
    # ── Rule 3: Out-of-state transaction ─────────────────────────────────────
    if txn["home_state"] not in txn["atm_location"] and "HOME STATE" in txn["atm_location"]:
        rules_hit.append(f"WRONG LOCATION (card from {txn['home_state']} used at: {txn['atm_location'][:40]}...)")
    
    # ── Rule 4: Daily limit exceeded ──────────────────────────────────────────
    if txn["daily_total_so_far"] > txn["daily_limit"]:
        excess = txn["daily_total_so_far"] - txn["daily_limit"]
        rules_hit.append(f"LIMIT EXCEEDED (₦{txn['daily_total_so_far']:,.0f} vs limit ₦{txn['daily_limit']:,.0f}, excess: ₦{excess:,.0f})")
    
    # ── Classification ────────────────────────────────────────────────────────
    n = len(rules_hit)
    
    if n >= 2:
        risk_level  = "HIGH RISK"
        emoji       = "🔴"
        action      = "BLOCK CARD. Call customer immediately. Escalate to fraud team."
        confidence  = "HIGH"
    elif n == 1:
        # Large amount and wrong location are more serious than others
        if "LARGE AMOUNT" in rules_hit[0] or "WRONG LOCATION" in rules_hit[0]:
            risk_level = "MEDIUM RISK"
            emoji      = "🟠"
            action     = "Send SMS to customer. Analyst review within 1 hour."
            confidence = "MEDIUM"
        else:
            risk_level = "LOW RISK"
            emoji      = "🟡"
            action     = "Log and monitor. No immediate action."
            confidence = "LOW"
    else:
        risk_level = "NORMAL"
        emoji      = "🟢"
        action     = "No action required."
        confidence = "N/A"
    
    return {
        "transaction_id": txn["transaction_id"],
        "customer":       txn["customer_name"],
        "date":           txn["date"],
        "time":           txn["time"],
        "amount":         txn["amount_naira"],
        "bank":           txn["bank"],
        "risk_level":     risk_level,
        "emoji":          emoji,
        "rules_hit":      rules_hit,
        "action":         action,
        "confidence":     confidence,
        "ground_truth":   txn["is_fraud_flag"],
    }


# ── Test on one transaction ───────────────────────────────────────────────────
result = detect_fraud(suspicious_txn)

print(f"FRAUD DETECTION RESULT")
print("=" * 60)
print(f"  Transaction : {result['transaction_id']}")
print(f"  Customer    : {result['customer']}")
print(f"  Date/Time   : {result['date']} at {result['time']}")
print(f"  Amount      : ₦{result['amount']:,.2f}")
print(f"  Risk Level  : {result['emoji']} {result['risk_level']}")
print(f"  Confidence  : {result['confidence']}")
print(f"  Action      : {result['action']}")
print()
print(f"  Rules triggered ({len(result['rules_hit'])}):")
for rule in result["rules_hit"]:
    print(f"    ▶ {rule}")
print()
print(f"  Ground truth: {result['ground_truth']}")
```

```python
# ── ACCURACY EVALUATION ──────────────────────────────────────────────────────
#
# Our dataset has a ground truth column: is_fraud_flag (NORMAL or SUSPICIOUS)
# We planted known fraud in the data. Now we measure how well we detected it.
#
# True Positive  (TP): We flagged it AND it was real fraud    ← CORRECT
# False Positive (FP): We flagged it BUT it was normal        ← FALSE ALARM
# False Negative (FN): We missed it AND it was real fraud     ← DANGEROUS MISS
# True Negative  (TN): We cleared it AND it was normal        ← CORRECT

TP = sum(1 for r in all_detections
         if r["risk_level"] in ("HIGH RISK","MEDIUM RISK")
         and r["ground_truth"] == "SUSPICIOUS")

FP = sum(1 for r in all_detections
         if r["risk_level"] in ("HIGH RISK","MEDIUM RISK")
         and r["ground_truth"] == "NORMAL")

FN = sum(1 for r in all_detections
         if r["risk_level"] == "NORMAL"
         and r["ground_truth"] == "SUSPICIOUS")

TN = sum(1 for r in all_detections
         if r["risk_level"] in ("NORMAL","LOW RISK")
         and r["ground_truth"] == "NORMAL")

total_real_fraud = sum(1 for t in transactions if t["is_fraud_flag"] == "SUSPICIOUS")

precision = (TP / (TP + FP) * 100) if (TP + FP) > 0 else 0
recall    = (TP / (TP + FN) * 100) if (TP + FN) > 0 else 0

print()
print("=" * 60)
print("     DETECTOR ACCURACY REPORT")
print("=" * 60)
print(f"  Total real fraud in dataset    : {total_real_fraud}")
print(f"  Total transactions we flagged  : {len(flagged)}")
print()
print(f"  ✅ True Positives  (TP) : {TP:>5}  — real fraud we caught")
print(f"  ❌ False Positives (FP) : {FP:>5}  — normal txns we wrongly flagged")
print(f"  ⚠️  False Negatives (FN) : {FN:>5}  — real fraud we MISSED")
print(f"  ✅ True Negatives  (TN) : {TN:>5}  — normal txns correctly cleared")
print()
print(f"  Precision : {precision:.1f}%  (of all our alerts, how many were real fraud?)")
print(f"  Recall    : {recall:.1f}%  (of all real fraud, how much did we catch?)")
print("=" * 60)
print()
if recall >= 75:
    print("  ✅ Good recall — catching most real fraud cases.")
else:
    print("  ⚠️  Low recall — consider lowering thresholds to catch more.")
if precision >= 60:
    print("  ✅ Acceptable precision — most alerts are genuine.")
else:
    print("  ⚠️  Many false alarms — consider raising the large amount multiplier.")
```
