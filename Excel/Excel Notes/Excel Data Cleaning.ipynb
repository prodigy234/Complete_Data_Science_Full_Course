# Excel Data Cleaning

## What is Data Cleaning?

Data cleaning is the process of finding, correcting, removing, or transforming inaccurate, incomplete, inconsistent, duplicate, or incorrectly formatted data so that the data becomes reliable and ready for analysis.

In simple terms:

`Data cleaning means taking messy data and turning it into clean, accurate, consistent, and usable data.`

Simple Example

Imagine you have this customer data:

| Customer Name | Age | Gender | City  |   Sales |
| ------------- | --: | ------ | ----- | ------: |
| John Smith    |  25 | Male   | Lagos |   50000 |
| mary johnson  |  30 | female | lagos | ₦75,000 |
| John Smith    |  25 | Male   | Lagos |   50000 |
| Peter Adams   |     | M      | Abuja |         |
| DAVID BROWN   |  42 | MALE   | ABUJA |   90000 |

There are several problems:

- mary johnson has unnecessary spaces.

- female, Male, MALE, and M are inconsistent.

- lagos and Lagos are inconsistent.

- ₦75,000 may be stored as text rather than a number.

- Peter's age is missing.

- Peter's sales value is missing.

- John Smith appears twice.

After cleaning:

| Customer Name | Age | Gender | City  | Sales |
| ------------- | --: | ------ | ----- | ----: |
| John Smith    |  25 | Male   | Lagos | 50000 |
| Mary Johnson  |  30 | Female | Lagos | 75000 |
| Peter Adams   |     | Male   | Abuja |       |
| David Brown   |  42 | Male   | Abuja | 90000 |

`Now the data is much more suitable for analysis`

## Common Data Cleaning Activities

| Cleaning Activity      | Example                                |
| ---------------------- | -------------------------------------- |
| Remove duplicates      | Remove repeated customer records       |
| Handle missing values  | Deal with blank Age or Sales           |
| Correct spelling       | `Lagos` instead of `Lagso`             |
| Standardize text       | `MALE`, `male`, `M` → `Male`           |
| Remove spaces          | `"  John Smith "` → `"John Smith"`     |
| Fix data types         | Text `"50000"` → Number `50000`        |
| Fix dates              | Standardize different date formats     |
| Remove errors          | Correct invalid values                 |
| Handle outliers        | Investigate unusually large values     |
| Standardize categories | `Lagos`, `lagos`, `LAGOS` → `Lagos`    |
| Validate data          | Ensure values follow expected rules    |
| Remove irrelevant data | Delete columns not needed for analysis |

In one sentence:

`Data cleaning is the process of improving the quality, accuracy, consistency, completeness, and usability of raw data before using it for analysis, reporting, decision-making, or machine learning.`

## Functions of Data Cleaning

Data cleaning performs several important functions in the data lifecycle.

| #  | Data Cleaning Function             | What It Does                                     | Example                                |
| -- | ---------------------------------- | ------------------------------------------------ | -------------------------------------- |
| 1  | **Removing duplicates**            | Removes repeated records                         | Same customer appears twice            |
| 2  | **Handling missing values**        | Identifies and treats blank/null values          | Missing customer age                   |
| 3  | **Correcting errors**              | Fixes incorrect data                             | `Lagos` entered as `Lagso`             |
| 4  | **Standardizing data**             | Makes values follow one format                   | `Male`, `M`, `male` → `Male`           |
| 5  | **Removing unnecessary spaces**    | Removes leading/trailing/unwanted spaces         | `" John "` → `"John"`                  |
| 6  | **Correcting data types**          | Converts data into appropriate types             | `"50000"` → `50000`                    |
| 7  | **Date standardization**           | Makes dates consistent                           | `01/08/2026` and `2026-08-01`          |
| 8  | **Text cleaning**                  | Removes unwanted characters and formatting       | Removing special characters            |
| 9  | **Handling outliers**              | Identifies unusual values for investigation      | Age = 250                              |
| 10 | **Handling invalid values**        | Finds values that violate rules                  | Quantity = -50                         |
| 11 | **Data validation**                | Checks whether data meets predefined rules       | Email must contain `@`                 |
| 12 | **Data integration**               | Makes data from different sources compatible     | CRM + Excel sales data                 |
| 13 | **Data restructuring**             | Changes data into an analysis-friendly structure | Unpivoting monthly columns             |
| 14 | **Categorization**                 | Groups values into meaningful categories         | Age → Teenager/Adult/Senior            |
| 15 | **Deduplication**                  | Identifies records representing the same entity  | Same customer with different spellings |
| 16 | **Standardizing units**            | Converts measurements to a common unit           | KG → Grams                             |
| 17 | **Formatting correction**          | Corrects inconsistent formats                    | Phone numbers                          |
| 18 | **Referential integrity checking** | Ensures relationships between datasets are valid | Every Order_ID has a valid Customer_ID |
| 19 | **Anomaly detection**              | Identifies suspicious or unusual records         | ₦50 million transaction                |
| 20 | **Data enrichment**                | Adds useful information from another source      | Adding customer region                 |
| 21 | **Consistency checking**           | Ensures related fields agree                     | Age and date of birth                  |
| 22 | **Range checking**                 | Ensures values fall within acceptable limits     | Exam score must be 0-100               |
| 23 | **Completeness checking**          | Determines whether required fields are populated | Every transaction needs Transaction_ID |
| 24 | **Uniqueness checking**            | Ensures fields that should be unique are unique  | Employee ID                            |
| 25 | **Quality assurance**              | Confirms the final dataset is fit for use        | Final validation before dashboarding   |


## Business Scenarios Where Data Cleaning Is Used

Data cleaning is not restricted to Excel or data analysts. It happens in almost every business function that collects or uses data.

### 1. Banking

#### Scenario

A bank receives customer and transaction data from:

- Mobile banking

- ATM transactions

- POS

- Internet banking

- Branch transactions

#### Data cleaning activities

- Remove duplicate transactions

- Standardize customer names

- Fix transaction dates

- Handle missing customer information

- Standardize account numbers

- Identify invalid transactions

- Detect unusual values

- Standardize currencies

#### Where it is carried out

`Core Banking → Data Warehouse → SQL/Python → Power BI/Tableau`
