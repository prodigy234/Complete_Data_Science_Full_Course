## Chapter 1: Why Learn SPSS Syntax

SPSS is usually taught through its point-and-click menus, and that is a reasonable place to start. But every menu click that SPSS performs is, underneath, a line of syntax. Once you can write that syntax directly, you gain four things the menus cannot give you: reproducibility (anyone can re-run your exact analysis), speed (one paragraph of syntax can replace forty clicks), auditability (a syntax file is a written record of every decision you made about the data), and automation (syntax can be looped, macro-ised, and scheduled).

In a professional or research setting, a saved .sps syntax file is your proof of work. If a supervisor, examiner, or auditor asks 'how did you get this number', the honest answer with syntax is 'run this file and you will get it again'. That is not true of a project built entirely by clicking through menus.

This book teaches SPSS the way a programmer would teach a language: starting from the syntax window itself, through data definition, data management, every major statistical procedure, and finally macros and automation. Every example runs against one practice dataset so that the syntax you write in Chapter 4 still works when you reach Chapter 34.


> **TIP:** You can always recover the syntax behind any menu action: run the analysis once through the menus, then click 'Paste' instead of 'OK'. SPSS writes the exact syntax for what you just configured into a new Syntax Editor window.

## Chapter 2: The Syntax Editor and How Commands Are Structured

The Syntax Editor is a plain text window (File > New > Syntax) where you type SPSS commands. Every SPSS command follows the same skeleton: a COMMAND KEYWORD, one or more optional /SUBCOMMANDS, and a closing period. The period is not decorative - it is how SPSS knows a command has ended, exactly like a semicolon in SQL or a full stop in a sentence.

Command keywords are conventionally written in upper case and variable names in the case you defined them, though SPSS itself is not case sensitive for either. Indentation and line breaks are free-form: a command can span many lines, and SPSS only cares where the period is.

Comments start with an asterisk (*) or the word COMMENT, and always end with a period. You can also use C-style block comments with /* text */ inside a command. Get in the habit of commenting your syntax files as if someone else will read them in a year, because that someone is usually you.


**Anatomy of a command**

```spss
* This is a full-line comment describing what the command below does.
FREQUENCIES VARIABLES=Gender Region
 /ORDER=ANALYSIS.

* Everything before the period, across as many lines as needed, is ONE command.
* A period with nothing after it ends the command; the next line starts a new one.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `* This is a full-line comment describing what the command below does.` | A full-line comment. It starts with an asterisk and ends with a period; SPSS ignores everything between them. |
| 2 | `FREQUENCIES VARIABLES=Gender Region` | The command keyword FREQUENCIES begins, naming Gender and Region as the variables to analyse. The command is not finished yet because there is no period. |
| 3 | ` /ORDER=ANALYSIS.` | A subcommand starting with a forward slash, still part of the same FREQUENCIES command; it tells SPSS to order the output categories by the order it discovers them in the data. The period here ends the whole command. |
| 4 | *(blank line)* | A blank line. SPSS ignores blank lines completely; they exist only to make the file easier for a human to read. |
| 5 | `* Everything before the period, across as many lines as needed, is ONE command.` | Another full-line comment explaining that a command can span several physical lines and still counts as one command, up to its closing period. |
| 6 | `* A period with nothing after it ends the command; the next line starts a new one.` | A final comment clarifying that the period is the true end-of-command marker, not the line break. |

**Running syntax**

```spss
* Highlight one or more complete commands and press Ctrl+R (or the Run > Selection
* menu / green play-triangle icon) to execute just the highlighted syntax.
* Run > All Syntax executes the entire file from top to bottom.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `* Highlight one or more complete commands and press Ctrl+R (or the Run > Selection` | Comment explaining that highlighting one or more complete commands and pressing Ctrl+R runs only what is selected. |
| 2 | `* menu / green play-triangle icon) to execute just the highlighted syntax.` | Continuation of the same comment, naming the alternative ways to trigger the same action (the Run menu or the green play-triangle toolbar icon). |
| 3 | `* Run > All Syntax executes the entire file from top to bottom.` | Comment explaining the Run > All Syntax menu option, which executes every command in the file from the first line to the last. |

> **TIP:** A very common beginner error is a missing period at the end of a command - SPSS will then try to interpret the next line as part of the same command and throw a confusing error. If a command 'does not run', check the period first.

## Chapter 3: Session Setup: SET, SHOW, and Working Directories

Before touching data it helps to configure the session. SET controls global options such as the width of printed output, the decimal symbol, and the random number seed (important for reproducible simulations or bootstrapping). SHOW displays the current value of any of those settings. CD changes SPSS's current working directory, which is useful so that later GET DATA and SAVE commands can use short relative file names instead of full paths.


**Typical session header**

```spss
SET SEED=234.
SET DECIMAL=DOT.
CD 'C:\CarePulse\SPSS_Course'.
SHOW SEED.
SHOW DIRECTORY.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `SET SEED=234.` | SET SEED fixes the starting point of SPSS's random number generator to the value 234, so any random process (sampling, bootstrapping) gives identical results every time this file is run. |
| 2 | `SET DECIMAL=DOT.` | SET DECIMAL tells SPSS to expect and display a full stop as the decimal separator (DOT), rather than a comma, regardless of the Windows regional settings on the machine. |
| 3 | `CD 'C:\CarePulse\SPSS_Course'.` | CD changes SPSS's current working directory to the folder path given in quotes, so later file references can use short file names instead of the full path. |
| 4 | `SHOW SEED.` | SHOW SEED prints the current random seed value to the Output Viewer, confirming the SET SEED command above took effect. |
| 5 | `SHOW DIRECTORY.` | SHOW DIRECTORY prints the current working directory to the Output Viewer, confirming the CD command above took effect. |

> **TIP:** Put a SET SEED command at the very top of any syntax file that involves random sampling (SAMPLE, bootstrapping, multiple imputation) so your results are identical every time the file is re-run.

## Chapter 4: GET DATA: Importing CSV and Excel Files

GET DATA is the modern, general-purpose command for importing external files. It always follows the same pattern: /TYPE names the file format, /FILE gives the path, and format-specific subcommands control how rows and columns are read. This chapter imports the practice dataset directly from CSV and, separately, from Excel - you only need to run one of the two, but both are shown because you will meet each format in real work.


**Import the CSV file**

```spss
GET DATA /TYPE=TXT
 /FILE='carepulse_patient_survey.csv'
 /DELCASE=LINE
 /DELIMITERS=','
 /QUALIFIER='"'
 /ARRANGEMENT=DELIMITED
 /FIRSTCASE=2
 /IMPORTCASE=ALL
 /VARIABLES=
 PatientID A6
 Region A15
 Gender A6
 Age F3
 InsuranceType A13
 ConsultationChannel A14
 YearsOnPlatform F5.1
 MonthlyConsultationSpend F8
 NumConsultationsMonth F3
 Q1_DoctorResponsiveness F1
 Q2_AppEaseOfUse F1
 Q3_ConsultationQuality F1
 Q4_WaitTimeSatisfaction F1
 Q5_OverallSatisfaction F1
 ComplaintFiled A3
 EngagementIndex F3
 MissedFollowUp A3
 ChurnRisk A6.
CACHE.
EXECUTE.
DATASET NAME CarePulse WINDOW=FRONT.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `GET DATA /TYPE=TXT` | Begins the GET DATA command and tells SPSS the source is a plain text file (/TYPE=TXT), which covers CSV files. |
| 2 | ` /FILE='carepulse_patient_survey.csv'` | Names the exact file to read, in single quotes, relative to the working directory set by CD in Chapter 3. |
| 3 | ` /DELCASE=LINE` | DELCASE=LINE tells SPSS that every new line in the file is a new case (row) of data. |
| 4 | ` /DELIMITERS=','` | DELIMITERS names the character that separates one column's value from the next; a comma, matching a standard CSV file. |
| 5 | ` /QUALIFIER='"'` | QUALIFIER names the character used to wrap text that itself might contain the delimiter, here a double quote mark, exactly as CSV files typically use. |
| 6 | ` /ARRANGEMENT=DELIMITED` | ARRANGEMENT=DELIMITED confirms the file is column-delimited (as opposed to fixed-width columns). |
| 7 | ` /FIRSTCASE=2` | FIRSTCASE=2 tells SPSS that row 1 is a header row of column names, so actual data starts on row 2. |
| 8 | ` /IMPORTCASE=ALL` | IMPORTCASE=ALL tells SPSS to read every case in the file rather than stopping after a sample or a fixed count. |
| 9 | ` /VARIABLES=` | Opens the /VARIABLES subcommand, which is where every column's name and data type will be listed. |
| 10 | ` PatientID A6` | Defines PatientID as a string (A) variable up to 6 characters wide. |
| 11 | ` Region A15` | Defines Region as a string variable up to 15 characters wide. |
| 12 | ` Gender A6` | Defines Gender as a string variable up to 6 characters wide. |
| 13 | ` Age F3` | Defines Age as a numeric (F, meaning standard numeric) variable up to 3 digits wide. |
| 14 | ` InsuranceType A13` | Defines InsuranceType as a string variable up to 13 characters wide. |
| 15 | ` ConsultationChannel A14` | Defines ConsultationChannel as a string variable up to 14 characters wide. |
| 16 | ` YearsOnPlatform F5.1` | Defines YearsOnPlatform as numeric with 5 total digits and 1 decimal place (F5.1). |
| 17 | ` MonthlyConsultationSpend F8` | Defines MonthlyConsultationSpend as numeric with up to 8 digits and no decimal places. |
| 18 | ` NumConsultationsMonth F3` | Defines NumConsultationsMonth as numeric with up to 3 digits. |
| 19 | ` Q1_DoctorResponsiveness F1` | Defines Q1_DoctorResponsiveness as a single-digit numeric variable (the 1-5 Likert code). |
| 20 | ` Q2_AppEaseOfUse F1` | Defines Q2_AppEaseOfUse the same way, as a single-digit numeric variable. |
| 21 | ` Q3_ConsultationQuality F1` | Defines Q3_ConsultationQuality the same way, as a single-digit numeric variable. |
| 22 | ` Q4_WaitTimeSatisfaction F1` | Defines Q4_WaitTimeSatisfaction the same way, as a single-digit numeric variable. |
| 23 | ` Q5_OverallSatisfaction F1` | Defines Q5_OverallSatisfaction the same way, as a single-digit numeric variable. This is the last Likert item in the list. |
| 24 | ` ComplaintFiled A3` | Defines ComplaintFiled as a string variable up to 3 characters wide (holds 'Yes' or 'No'). |
| 25 | ` EngagementIndex F3` | Defines EngagementIndex as numeric with up to 3 digits. |
| 26 | ` MissedFollowUp A3` | Defines MissedFollowUp as a string variable up to 3 characters wide (holds 'Yes' or 'No'). |
| 27 | ` ChurnRisk A6.` | Defines ChurnRisk as a string variable up to 6 characters wide, and the period here closes the entire GET DATA command. |
| 28 | `CACHE.` | CACHE tells SPSS to read the whole file into memory immediately rather than only when it is next needed. |
| 29 | `EXECUTE.` | EXECUTE forces SPSS to actually carry out the pending GET DATA and CACHE instructions right now. |
| 30 | `DATASET NAME CarePulse WINDOW=FRONT.` | DATASET NAME gives the newly opened data the short handle CarePulse, and WINDOW=FRONT brings its Data Editor window to the front of the screen. |

**Import the Excel file instead**

```spss
GET DATA /TYPE=XLSX
 /FILE='carepulse_patient_survey.xlsx'
 /SHEET=name 'CarePulse_Survey'
 /CELLRANGE=full
 /READNAMES=on
 /DATATYPEMIN PERCENTAGE=95.0.
EXECUTE.
DATASET NAME CarePulse WINDOW=FRONT.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `GET DATA /TYPE=XLSX` | Begins the GET DATA command, this time naming the source type as XLSX (a modern Excel workbook). |
| 2 | ` /FILE='carepulse_patient_survey.xlsx'` | Names the exact Excel file to open, in single quotes. |
| 3 | ` /SHEET=name 'CarePulse_Survey'` | SHEET=name tells SPSS which worksheet inside the workbook to read, here the one named CarePulse_Survey. |
| 4 | ` /CELLRANGE=full` | CELLRANGE=full tells SPSS to read every populated cell in that worksheet rather than a manually specified range. |
| 5 | ` /READNAMES=on` | READNAMES=on tells SPSS that the first row of the sheet contains column headers to use as variable names. |
| 6 | ` /DATATYPEMIN PERCENTAGE=95.0.` | DATATYPEMIN PERCENTAGE=95.0 tells SPSS how confident it must be (95%) about a column's data type before assigning it automatically; the period ends the command. |
| 7 | `EXECUTE.` | EXECUTE forces SPSS to actually read the Excel file now. |
| 8 | `DATASET NAME CarePulse WINDOW=FRONT.` | DATASET NAME gives the data the handle CarePulse and brings its window to the front, exactly as in the CSV version above. |

> **TIP:** CACHE. followed by EXECUTE. forces SPSS to read the whole file into memory immediately rather than lazily, which avoids surprises later if the source file is moved or deleted.

> **TIP:** DATASET NAME gives the open data a short handle so you can refer to it later when working with more than one dataset open at once (see MATCH FILES in Chapter 14).

## Chapter 5: DATA LIST: Defining Data Inline or From Fixed-Width Files

Before GET DATA existed, DATA LIST was (and still is) the workhorse command for defining a dataset's structure by hand, either reading an external fixed-width or delimited file, or reading data typed directly inside the syntax file using BEGIN DATA / END DATA. This is invaluable for small teaching datasets, quick simulations, or when you need full manual control of column positions.


**Inline data with BEGIN DATA / END DATA**

```spss
DATA LIST FREE / ClinicCode (A4) StaffCount (F3) AvgWaitMinutes (F4.1).
BEGIN DATA
LAG1 42 6.2
ABJ1 27 4.8
PHC1 19 7.1
KAN1 15 5.5
IBD1 22 6.0
END DATA.
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `DATA LIST FREE / ClinicCode (A4) StaffCount (F3) AvgWaitMinutes (F4.1).` | DATA LIST FREE declares three new variables read in free (whitespace-separated) format: ClinicCode as up to 4 characters of text, StaffCount as a 3-digit number, and AvgWaitMinutes as a number with 4 total digits and 1 decimal place. |
| 2 | `BEGIN DATA` | BEGIN DATA marks the start of the actual data rows typed directly into the syntax file. |
| 3 | `LAG1 42 6.2` | The first data row: ClinicCode='LAG1', StaffCount=42, AvgWaitMinutes=6.2. |
| 4 | `ABJ1 27 4.8` | The second data row: ClinicCode='ABJ1', StaffCount=27, AvgWaitMinutes=4.8. |
| 5 | `PHC1 19 7.1` | The third data row: ClinicCode='PHC1', StaffCount=19, AvgWaitMinutes=7.1. |
| 6 | `KAN1 15 5.5` | The fourth data row: ClinicCode='KAN1', StaffCount=15, AvgWaitMinutes=5.5. |
| 7 | `IBD1 22 6.0` | The fifth data row: ClinicCode='IBD1', StaffCount=22, AvgWaitMinutes=6.0. |
| 8 | `END DATA.` | END DATA marks the end of the inline data rows, matching the BEGIN DATA above. |
| 9 | `EXECUTE.` | EXECUTE forces SPSS to actually build this small dataset in memory now. |

**Fixed-column format from an external file**

```spss
DATA LIST FILE='clinic_master.txt' FIXED RECORDS=1
 / ClinicCode 1-4 (A) StaffCount 6-8 AvgWaitMinutes 10-13 (1).
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `DATA LIST FILE='clinic_master.txt' FIXED RECORDS=1` | DATA LIST FILE names an external fixed-width text file to read, and FIXED RECORDS=1 states that each case occupies exactly one physical line of that file. |
| 2 | ` / ClinicCode 1-4 (A) StaffCount 6-8 AvgWaitMinutes 10-13 (1).` | Defines the column positions within each line: ClinicCode occupies characters 1 to 4 and is text (A); StaffCount occupies characters 6 to 8; AvgWaitMinutes occupies characters 10 to 13 with 1 implied decimal place. |
| 3 | `EXECUTE.` | EXECUTE forces SPSS to actually read the fixed-width file now. |

> **TIP:** FREE format reads values separated by any whitespace and is quickest for small inline tables; FIXED format reads exact column positions and is used for legacy government or mainframe extract files.

## Chapter 6: Labelling the Data: VARIABLE LABELS, VALUE LABELS, MISSING VALUES

Raw column names and numeric codes are rarely self-explanatory to a reader of your output. VARIABLE LABELS attaches a descriptive sentence to a variable name; VALUE LABELS attaches a text meaning to specific numeric or string codes (essential for Likert items and Yes/No flags); MISSING VALUES declares which codes should be treated as missing rather than as real data (e.g., 99 for 'refused to answer').


**Variable labels**

```spss
VARIABLE LABELS
 Q1_DoctorResponsiveness 'Doctors respond promptly to my consultation requests'
 Q2_AppEaseOfUse 'The CarePulse app is easy to use'
 Q3_ConsultationQuality 'The quality of medical consultations meets my expectations'
 Q4_WaitTimeSatisfaction 'I am satisfied with the waiting time before a consultation'
 Q5_OverallSatisfaction 'Overall, I am satisfied with CarePulse Health Tech'
 MonthlyConsultationSpend 'Monthly consultation spend in Naira (NGN)'.
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `VARIABLE LABELS` | Begins the VARIABLE LABELS command, which will attach a descriptive label to one or more variables. |
| 2 | ` Q1_DoctorResponsiveness 'Doctors respond promptly to my consultation requests'` | Attaches the label 'Doctors respond promptly to my consultation requests' to the variable Q1_DoctorResponsiveness. |
| 3 | ` Q2_AppEaseOfUse 'The CarePulse app is easy to use'` | Attaches the label 'The CarePulse app is easy to use' to Q2_AppEaseOfUse. |
| 4 | ` Q3_ConsultationQuality 'The quality of medical consultations meets my expectations'` | Attaches the label 'The quality of medical consultations meets my expectations' to Q3_ConsultationQuality. |
| 5 | ` Q4_WaitTimeSatisfaction 'I am satisfied with the waiting time before a consultation'` | Attaches the label 'I am satisfied with the waiting time before a consultation' to Q4_WaitTimeSatisfaction. |
| 6 | ` Q5_OverallSatisfaction 'Overall, I am satisfied with CarePulse Health Tech'` | Attaches the label 'Overall, I am satisfied with CarePulse Health Tech' to Q5_OverallSatisfaction. |
| 7 | ` MonthlyConsultationSpend 'Monthly consultation spend in Naira (NGN)'.` | Attaches the label 'Monthly consultation spend in Naira (NGN)' to MonthlyConsultationSpend; the period ends the whole command. |
| 8 | `EXECUTE.` | EXECUTE forces SPSS to apply all six labels now. |

**Value labels for the Likert items and flags**

```spss
VALUE LABELS Q1_DoctorResponsiveness Q2_AppEaseOfUse Q3_ConsultationQuality
 Q4_WaitTimeSatisfaction Q5_OverallSatisfaction
 1 'Strongly Disagree' 2 'Disagree' 3 'Neutral' 4 'Agree' 5 'Strongly Agree'.

VALUE LABELS ComplaintFiled MissedFollowUp 'No' 'No' 'Yes' 'Yes'.
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `VALUE LABELS Q1_DoctorResponsiveness Q2_AppEaseOfUse Q3_ConsultationQuality` | Begins a VALUE LABELS command that will apply the same set of numeric-to-text mappings to all five Likert variables listed here. |
| 2 | ` Q4_WaitTimeSatisfaction Q5_OverallSatisfaction` | Continues the list of variables that will share the labels defined below (still part of the same command as line 0). |
| 3 | ` 1 'Strongly Disagree' 2 'Disagree' 3 'Neutral' 4 'Agree' 5 'Strongly Agree'.` | Defines the five shared value labels: 1 means 'Strongly Disagree', 2 means 'Disagree', 3 means 'Neutral', 4 means 'Agree', 5 means 'Strongly Agree'; the period ends this command. |
| 4 | *(blank line)* | A blank line separating the two independent VALUE LABELS commands for readability. |
| 5 | `VALUE LABELS ComplaintFiled MissedFollowUp 'No' 'No' 'Yes' 'Yes'.` | A second, separate VALUE LABELS command: for both ComplaintFiled and MissedFollowUp, the string 'No' is labelled 'No' and 'Yes' is labelled 'Yes' (this mainly documents intent since the codes and labels are identical strings). |
| 6 | `EXECUTE.` | EXECUTE applies both VALUE LABELS commands now. |

**Declaring missing codes**

```spss
MISSING VALUES MonthlyConsultationSpend (999999) EngagementIndex (0).
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `MISSING VALUES MonthlyConsultationSpend (999999) EngagementIndex (0).` | MISSING VALUES declares that, for MonthlyConsultationSpend, the code 999999 should be treated as missing, and for EngagementIndex, the code 0 should be treated as missing, rather than as genuine data values. |
| 2 | `EXECUTE.` | EXECUTE applies this missing-value declaration now. |

> **TIP:** Labelling every categorical and Likert variable at the start of a project takes ten minutes and saves hours later, because every FREQUENCIES, CROSSTABS, and chart you run afterwards will automatically print the readable labels instead of raw codes.

## Chapter 7: VARIABLE LEVEL, FORMATS, and Display Properties

SPSS distinguishes three measurement levels - Nominal, Ordinal, Scale - which control which chart types and statistics some menus offer you (syntax itself does not enforce this, but many procedures assume you set it correctly). VARIABLE LEVEL sets this explicitly. FORMATS controls how numbers print (decimal places, width), and VARIABLE WIDTH / VARIABLE ALIGNMENT control column display in Data View.


**Setting measurement level and display format**

```spss
VARIABLE LEVEL Region Gender InsuranceType ConsultationChannel ComplaintFiled
 MissedFollowUp ChurnRisk (NOMINAL).
VARIABLE LEVEL Q1_DoctorResponsiveness TO Q5_OverallSatisfaction (ORDINAL).
VARIABLE LEVEL Age YearsOnPlatform MonthlyConsultationSpend NumConsultationsMonth
 EngagementIndex (SCALE).

FORMATS MonthlyConsultationSpend EngagementIndex (F8.0) YearsOnPlatform (F5.1).
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `VARIABLE LEVEL Region Gender InsuranceType ConsultationChannel ComplaintFiled` | Begins a VARIABLE LEVEL command listing several categorical variables. |
| 2 | ` MissedFollowUp ChurnRisk (NOMINAL).` | Continues the variable list from the line above and sets all of them, as a group, to the NOMINAL measurement level (categories with no inherent order). |
| 3 | `VARIABLE LEVEL Q1_DoctorResponsiveness TO Q5_OverallSatisfaction (ORDINAL).` | A second VARIABLE LEVEL command sets every variable from Q1_DoctorResponsiveness through Q5_OverallSatisfaction (using the TO keyword) to ORDINAL, since Likert categories have a meaningful order. |
| 4 | `VARIABLE LEVEL Age YearsOnPlatform MonthlyConsultationSpend NumConsultationsMonth` | Begins a third VARIABLE LEVEL command listing four scale (continuous numeric) variables. |
| 5 | ` EngagementIndex (SCALE).` | Continues that list and sets all five named variables (including EngagementIndex) to SCALE, the measurement level for continuous numeric data. |
| 6 | *(blank line)* | A blank line separating the VARIABLE LEVEL commands from the FORMATS command below. |
| 7 | `FORMATS MonthlyConsultationSpend EngagementIndex (F8.0) YearsOnPlatform (F5.1).` | FORMATS sets the printed display format: MonthlyConsultationSpend and EngagementIndex each show up to 8 digits with 0 decimal places, while YearsOnPlatform shows 5 total digits with 1 decimal place. |
| 8 | `EXECUTE.` | EXECUTE applies all of the level and format changes now. |

> **TIP:** The TO keyword (Q1_DoctorResponsiveness TO Q5_OverallSatisfaction) refers to every variable between two names in file order - a fast way to address a block of consecutively positioned variables without typing each name.

## Chapter 8: COMPUTE: Creating New Variables

COMPUTE creates or overwrites a variable using an arithmetic expression. SPSS supports the usual operators (+ - * / **), a large function library (SUM, MEAN, SQRT, LN, LG10, RND, TRUNC, ABS, and many more), and can reference any existing numeric variable. Every COMPUTE should be followed by EXECUTE (or will run automatically at the next procedure) so you can see the result immediately.


**Building a composite satisfaction score**

```spss
COMPUTE SatisfactionIndex = MEAN(Q1_DoctorResponsiveness, Q2_AppEaseOfUse,
 Q3_ConsultationQuality, Q4_WaitTimeSatisfaction, Q5_OverallSatisfaction).
VARIABLE LABELS SatisfactionIndex 'Mean of 5 satisfaction items'.
EXECUTE.

COMPUTE AnnualConsultationSpend = MonthlyConsultationSpend * 12.
COMPUTE SpendPerConsultation = MonthlyConsultationSpend / NumConsultationsMonth.
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `COMPUTE SatisfactionIndex = MEAN(Q1_DoctorResponsiveness, Q2_AppEaseOfUse,` | Begins a COMPUTE command creating SatisfactionIndex as the MEAN() of several listed variables; the function call continues onto the next line. |
| 2 | ` Q3_ConsultationQuality, Q4_WaitTimeSatisfaction, Q5_OverallSatisfaction).` | Completes the list of variables being averaged and closes the MEAN() function and the COMPUTE command with a period. |
| 3 | `VARIABLE LABELS SatisfactionIndex 'Mean of 5 satisfaction items'.` | Attaches a descriptive label, 'Mean of 5 satisfaction items', to the newly created SatisfactionIndex variable. |
| 4 | `EXECUTE.` | EXECUTE actually calculates SatisfactionIndex for every case now. |
| 5 | *(blank line)* | A blank line separating this pair of computations from the next. |
| 6 | `COMPUTE AnnualConsultationSpend = MonthlyConsultationSpend * 12.` | A second COMPUTE creates AnnualConsultationSpend by multiplying MonthlyConsultationSpend by 12. |
| 7 | `COMPUTE SpendPerConsultation = MonthlyConsultationSpend / NumConsultationsMonth.` | A third COMPUTE creates SpendPerConsultation by dividing MonthlyConsultationSpend by NumConsultationsMonth. |
| 8 | `EXECUTE.` | EXECUTE actually calculates both AnnualConsultationSpend and SpendPerConsultation now. |

> **TIP:** MEAN(), SUM(), and SD() ignore missing values by default unless you request a minimum number of valid values, e.g. MEAN.3(Q1,Q2,Q3,Q4,Q5) requires at least 3 of the 5 to be non-missing before it computes a result.

## Chapter 9: IF and DO IF / ELSE IF / END IF: Conditional Logic

The standalone IF command computes a value only for cases meeting a condition, leaving other cases untouched. DO IF ... ELSE IF ... END IF is the block form, letting you route each case through one of several regions - closer to a traditional if/elif/else block in a programming language, and the right tool when you need to assign several variables per region.


**Standalone IF**

```spss
IF (MonthlyConsultationSpend >= 20000) HighSpendFlag = 1.
IF (MonthlyConsultationSpend < 20000) HighSpendFlag = 0.
VARIABLE LABELS HighSpendFlag 'Monthly consultation spend at or above N20,000'.
VALUE LABELS HighSpendFlag 0 'No' 1 'Yes'.
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `IF (MonthlyConsultationSpend >= 20000) HighSpendFlag = 1.` | A standalone IF: for any case where MonthlyConsultationSpend is 20000 or more, it sets HighSpendFlag to 1; cases that fail the condition are left untouched by this line. |
| 2 | `IF (MonthlyConsultationSpend < 20000) HighSpendFlag = 0.` | A second standalone IF: for any case where MonthlyConsultationSpend is below 20000, it sets HighSpendFlag to 0, covering the cases the first IF did not touch. |
| 3 | `VARIABLE LABELS HighSpendFlag 'Monthly consultation spend at or above N20,000'.` | Attaches the descriptive label 'Monthly consultation spend at or above N20,000' to HighSpendFlag. |
| 4 | `VALUE LABELS HighSpendFlag 0 'No' 1 'Yes'.` | Attaches value labels to HighSpendFlag so that 0 displays as 'No' and 1 displays as 'Yes'. |
| 5 | `EXECUTE.` | EXECUTE actually evaluates both IF conditions and applies the labels now. |

**DO IF / ELSE IF block for a multi-category recode**

```spss
DO IF (Age < 25).
 COMPUTE AgeBand = 1.
ELSE IF (Age >= 25 AND Age <= 40).
 COMPUTE AgeBand = 2.
ELSE IF (Age >= 41 AND Age <= 55).
 COMPUTE AgeBand = 3.
ELSE.
 COMPUTE AgeBand = 4.
END IF.
VALUE LABELS AgeBand 1 'Under 25' 2 '25-40' 3 '41-55' 4 '56 and above'.
VARIABLE LABELS AgeBand 'Age band'.
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `DO IF (Age < 25).` | Opens a DO IF block: cases where Age is under 25 will run the next line only. |
| 2 | ` COMPUTE AgeBand = 1.` | For cases entering the region above, sets AgeBand to 1. |
| 3 | `ELSE IF (Age >= 25 AND Age <= 40).` | ELSE IF adds a second condition, tested only for cases that failed the first: Age between 25 and 40 inclusive. |
| 4 | ` COMPUTE AgeBand = 2.` | For cases entering this second region, sets AgeBand to 2. |
| 5 | `ELSE IF (Age >= 41 AND Age <= 55).` | ELSE IF adds a third condition, tested only for cases that failed the first two: Age between 41 and 55 inclusive. |
| 6 | ` COMPUTE AgeBand = 3.` | For cases entering this third region, sets AgeBand to 3. |
| 7 | `ELSE.` | ELSE catches every remaining case that failed all three conditions above (Age 56 and over). |
| 8 | ` COMPUTE AgeBand = 4.` | For cases entering this final catch-all region, sets AgeBand to 4. |
| 9 | `END IF.` | END IF closes the DO IF block that began on the first line. |
| 10 | `VALUE LABELS AgeBand 1 'Under 25' 2 '25-40' 3 '41-55' 4 '56 and above'.` | Attaches value labels to AgeBand: 1='Under 25', 2='25-40', 3='41-55', 4='56 and above'. |
| 11 | `VARIABLE LABELS AgeBand 'Age band'.` | Attaches the descriptive label 'Age band' to AgeBand. |
| 12 | `EXECUTE.` | EXECUTE actually runs the whole DO IF block and applies the labels now. |

> **TIP:** A DO IF block with no matching ELSE region leaves the target variable as system-missing for any case that fails every condition - always add a final ELSE. to catch the remainder unless that is genuinely what you want.

## Chapter 10: RECODE and AUTORECODE

RECODE changes existing values into new ones, either overwriting the same variable or, with INTO, writing to a new one (the safer habit, since it preserves the original). AUTORECODE takes any string or scattered-numeric variable and assigns clean sequential integer codes 1, 2, 3... in alphabetical/ascending order - extremely useful before running procedures that require pure numeric grouping variables.


**RECODE into a new variable**

```spss
RECODE Age (18 THRU 25=1) (26 THRU 40=2) (41 THRU 55=3) (56 THRU HIGHEST=4)
 INTO AgeBandRecode.
VARIABLE LABELS AgeBandRecode 'Age band (via RECODE)'.
VALUE LABELS AgeBandRecode 1 'Under 25' 2 '25-40' 3 '41-55' 4 '56 and above'.
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `RECODE Age (18 THRU 25=1) (26 THRU 40=2) (41 THRU 55=3) (56 THRU HIGHEST=4)` | Begins a RECODE command on Age: values 18 through 25 become 1, 26 through 40 become 2, 41 through 55 become 3, and 56 through the highest value present become 4; the command continues on the next line. |
| 2 | ` INTO AgeBandRecode.` | INTO AgeBandRecode tells SPSS to write these new codes into a brand-new variable of that name, leaving the original Age variable completely unchanged. |
| 3 | `VARIABLE LABELS AgeBandRecode 'Age band (via RECODE)'.` | Attaches the descriptive label 'Age band (via RECODE)' to AgeBandRecode. |
| 4 | `VALUE LABELS AgeBandRecode 1 'Under 25' 2 '25-40' 3 '41-55' 4 '56 and above'.` | Attaches value labels to AgeBandRecode matching the four bands defined above. |
| 5 | `EXECUTE.` | EXECUTE actually performs the recode and applies the labels now. |

**AUTORECODE for a clean numeric grouping variable**

```spss
AUTORECODE VARIABLES=Region /INTO RegionNumeric /PRINT.
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `AUTORECODE VARIABLES=Region /INTO RegionNumeric /PRINT.` | AUTORECODE takes the string variable Region, assigns each distinct value a sequential integer code in alphabetical order, stores those codes in a new variable called RegionNumeric, and PRINT requests a table showing which original value maps to which new code. |
| 2 | `EXECUTE.` | EXECUTE actually performs the autorecode and prints the mapping table now. |

> **TIP:** THRU and the keywords LOWEST / HIGHEST let you write open-ended ranges without hard-coding the minimum or maximum value in the data, so the RECODE still works correctly if new, more extreme cases are added later.

## Chapter 11: SELECT IF, FILTER, and TEMPORARY

SELECT IF permanently discards cases that fail a condition for the rest of the session (or until you reload the data) - use it when you genuinely want to work only with a subset from this point on. FILTER BY is non-destructive: it hides cases from later procedures without deleting them, and can be switched off with FILTER OFF. TEMPORARY makes the very next transformation apply only to the next procedure, after which the full dataset returns.


**Permanently keep only Lagos patients**

```spss
SELECT IF (Region = 'Lagos').
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `SELECT IF (Region = 'Lagos').` | SELECT IF permanently discards, for the rest of the session, every case where Region is not equal to 'Lagos'. |
| 2 | `EXECUTE.` | EXECUTE actually removes the non-Lagos cases from memory now. |

**Non-destructive filter (preferred for exploratory work)**

```spss
COMPUTE FilterHighSpend = (MonthlyConsultationSpend >= 20000).
FILTER BY FilterHighSpend.
EXECUTE.

* ... run any procedure here and it only sees high-income patients ...

FILTER OFF.
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `COMPUTE FilterHighSpend = (MonthlyConsultationSpend >= 20000).` | COMPUTE creates a new 0/1 flag variable, FilterHighSpend, that equals 1 (true) wherever MonthlyConsultationSpend is 20000 or higher and 0 (false) otherwise. |
| 2 | `FILTER BY FilterHighSpend.` | FILTER BY tells every subsequent procedure to only look at cases where FilterHighSpend is 1 (non-zero), without deleting the other cases. |
| 3 | `EXECUTE.` | EXECUTE actually computes the flag and switches the filter on now. |
| 4 | *(blank line)* | A blank line for readability, separating the filter setup from the analysis that would run while it is active. |
| 5 | `* ... run any procedure here and it only sees high-income patients ...` | A comment reminding the reader that any procedure placed here would only see the high-income patients. |
| 6 | *(blank line)* | A blank line for readability before turning the filter back off. |
| 7 | `FILTER OFF.` | FILTER OFF cancels the filter, so every subsequent procedure sees the full dataset again. |
| 8 | `EXECUTE.` | EXECUTE actually applies the FILTER OFF instruction now. |

**TEMPORARY for a one-off selection**

```spss
TEMPORARY.
SELECT IF (Gender = 'Female').
FREQUENCIES VARIABLES=ChurnRisk.
* The next procedure after this one sees the FULL dataset again.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `TEMPORARY.` | TEMPORARY marks that the very next transformation command should only apply to the next procedure, after which the full, unfiltered dataset returns automatically. |
| 2 | `SELECT IF (Gender = 'Female').` | SELECT IF, made temporary by the line above, keeps only female patients for the single procedure that follows. |
| 3 | `FREQUENCIES VARIABLES=ChurnRisk.` | FREQUENCIES runs against only the temporarily filtered (female) subset defined above. |
| 4 | `* The next procedure after this one sees the FULL dataset again.` | A comment noting that any procedure after this FREQUENCIES command would see the complete dataset again, because the TEMPORARY effect has already expired. |

> **TIP:** Prefer FILTER BY over SELECT IF while you are still exploring data, since a mistaken SELECT IF can silently and permanently remove cases you needed later in the same session.

## Chapter 12: SORT CASES and SPLIT FILE

SORT CASES reorders the rows in memory by one or more variables, ascending or descending - required before some sequential commands (like some LAG functions or AGGREGATE with BREAK) which assume sorted input. SPLIT FILE repeats every subsequent procedure separately for each level of a grouping variable, printing a separate results block per group, which is the fastest way to produce region-by-region or gender-by-gender tables without writing a loop.


**Sort by region, then by spend descending**

```spss
SORT CASES BY Region (A) MonthlyConsultationSpend (D).
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `SORT CASES BY Region (A) MonthlyConsultationSpend (D).` | SORT CASES reorders every row in memory: first by Region in ascending (A) order, then within each region by MonthlyConsultationSpend in descending (D) order. |
| 2 | `EXECUTE.` | EXECUTE actually performs the sort now. |

**Split the output by region**

```spss
SORT CASES BY Region.
SPLIT FILE SEPARATE BY Region.
DESCRIPTIVES VARIABLES=MonthlyConsultationSpend EngagementIndex.

* Always turn splitting back off once you are done:
SPLIT FILE OFF.
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `SORT CASES BY Region.` | SORT CASES BY Region puts the data in region order, which SPLIT FILE below requires. |
| 2 | `SPLIT FILE SEPARATE BY Region.` | SPLIT FILE SEPARATE BY Region tells every following procedure to run once per region, printing a separate results block for each one. |
| 3 | `DESCRIPTIVES VARIABLES=MonthlyConsultationSpend EngagementIndex.` | DESCRIPTIVES now runs separately for each region because the split is active, producing five small tables instead of one combined table. |
| 4 | *(blank line)* | A blank line for readability before switching the split back off. |
| 5 | `* Always turn splitting back off once you are done:` | A comment reminding the reader to always turn SPLIT FILE off once the region-by-region output is no longer needed. |
| 6 | `SPLIT FILE OFF.` | SPLIT FILE OFF cancels the split, so later procedures again treat the dataset as one combined group. |
| 7 | `EXECUTE.` | EXECUTE actually applies the SPLIT FILE OFF instruction now. |

> **TIP:** SPLIT FILE requires the data to already be sorted by the same grouping variable, which is why SORT CASES almost always appears on the line immediately above it.

## Chapter 13: AGGREGATE: Building Summary Tables

AGGREGATE collapses many rows into one row per group, computing statistics such as means, sums, and counts along the way. Written with /OUTFILE=* it replaces the current dataset; written with a real file name it creates a brand-new region-level (or any other group-level) dataset while leaving the original case-level data untouched.


**Region-level summary as a new dataset**

```spss
DATASET DECLARE RegionSummary.
AGGREGATE
 /OUTFILE='RegionSummary'
 /BREAK=Region
 /MeanIncome_Region = MEAN(MonthlyConsultationSpend)
 /MeanSatisfaction_Region = MEAN(Q5_OverallSatisfaction)
 /PatientCount = N.

DATASET ACTIVATE RegionSummary.
LIST.
DATASET ACTIVATE CarePulse.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `DATASET DECLARE RegionSummary.` | DATASET DECLARE reserves the name RegionSummary for a new, currently-empty dataset that AGGREGATE below will fill. |
| 2 | `AGGREGATE` | Begins the AGGREGATE command. |
| 3 | ` /OUTFILE='RegionSummary'` | OUTFILE names RegionSummary as the destination for the new, summarised dataset, leaving the original case-level CarePulse data untouched. |
| 4 | ` /BREAK=Region` | BREAK=Region tells SPSS to produce one output row per distinct value of Region. |
| 5 | ` /MeanIncome_Region = MEAN(MonthlyConsultationSpend)` | Defines a new summary variable, MeanIncome_Region, equal to the average MonthlyConsultationSpend within each region. |
| 6 | ` /MeanSatisfaction_Region = MEAN(Q5_OverallSatisfaction)` | Defines a second summary variable, MeanSatisfaction_Region, equal to the average Q5_OverallSatisfaction within each region. |
| 7 | ` /PatientCount = N.` | Defines a third summary variable, PatientCount, equal to the number of cases (N) in each region; the period ends the AGGREGATE command. |
| 8 | *(blank line)* | A blank line for readability before switching to the new summarised dataset. |
| 9 | `DATASET ACTIVATE RegionSummary.` | DATASET ACTIVATE makes RegionSummary the active dataset so later commands operate on it instead of the case-level data. |
| 10 | `LIST.` | LIST prints every row of the now-active RegionSummary dataset to the Output Viewer. |
| 11 | `DATASET ACTIVATE CarePulse.` | DATASET ACTIVATE switches back to the original CarePulse case-level dataset, ready for further work. |

> **TIP:** Always give /OUTFILE a real dataset name (as above) rather than '*' the first time you use AGGREGATE - '*' overwrites your case-level data in memory with the summary table, which is rarely what a beginner intends.

## Chapter 14: MATCH FILES and ADD FILES: Combining Datasets

MATCH FILES joins two datasets side-by-side on a common key (like a SQL JOIN), adding new columns to matching cases. ADD FILES stacks two datasets with the same variables on top of each other (like a SQL UNION), adding new rows. Both require the input datasets to be sorted by the BY key first, and both operate on datasets you have already opened and named with DATASET NAME.


**MATCH FILES: bring region-level metrics onto every patient row**

```spss
DATASET ACTIVATE CarePulse.
SORT CASES BY Region.
DATASET ACTIVATE RegionSummary.
SORT CASES BY Region.

MATCH FILES /FILE=CarePulse /TABLE=RegionSummary /BY Region.
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `DATASET ACTIVATE CarePulse.` | DATASET ACTIVATE makes CarePulse the active dataset before sorting it. |
| 2 | `SORT CASES BY Region.` | SORT CASES BY Region puts the case-level data in region order, which MATCH FILES requires. |
| 3 | `DATASET ACTIVATE RegionSummary.` | DATASET ACTIVATE switches to the RegionSummary dataset built in Chapter 13 so it can be sorted too. |
| 4 | `SORT CASES BY Region.` | SORT CASES BY Region puts the region-level summary data in the same region order. |
| 5 | *(blank line)* | A blank line for readability before the actual join. |
| 6 | `MATCH FILES /FILE=CarePulse /TABLE=RegionSummary /BY Region.` | MATCH FILES joins the two datasets: FILE=CarePulse is the main, case-level file that keeps one row per patient; TABLE=RegionSummary is the smaller lookup file; BY Region is the shared key the two are matched on. |
| 7 | `EXECUTE.` | EXECUTE actually performs the join now, adding the region-level columns onto every patient row. |

**ADD FILES: stack two survey waves**

```spss
ADD FILES /FILE=Wave1Survey /FILE=Wave2Survey /IN=SurveyWave.
VALUE LABELS SurveyWave 0 'Wave 1' 1 'Wave 2'.
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `ADD FILES /FILE=Wave1Survey /FILE=Wave2Survey /IN=SurveyWave.` | ADD FILES stacks two already-open datasets, Wave1Survey and Wave2Survey, on top of each other, and IN=SurveyWave creates a new flag variable recording which of the two source files each resulting row came from (0 for the first, 1 for the second). |
| 2 | `VALUE LABELS SurveyWave 0 'Wave 1' 1 'Wave 2'.` | Attaches value labels to SurveyWave so that 0 displays as 'Wave 1' and 1 displays as 'Wave 2'. |
| 3 | `EXECUTE.` | EXECUTE actually performs the stacking now. |

> **TIP:** /TABLE= in MATCH FILES marks the smaller 'lookup' dataset (here, one row per region); the main /FILE= dataset keeps one row per original case, exactly like the left table in a LEFT JOIN.

## Chapter 15: String Functions and Date/Time Functions

SPSS syntax includes a full string-handling library - CONCAT, SUBSTR, LENGTH, UPCASE/LOWER, LTRIM/RTRIM, INDEX/RINDEX, REPLACE - for cleaning text fields, and a matching date library - DATE.DMY, XDATE.YEAR/MONTH/MDAY, CTIME.DAYS, DATEDIFF - for anything involving calendar dates, such as calculating a patient's tenure or age directly from a date-of-birth field.


**String cleaning example**

```spss
STRING RegionUpper (A15).
COMPUTE RegionUpper = UPCASE(Region).
COMPUTE RegionInitial = SUBSTR(Region,1,3).
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `STRING RegionUpper (A15).` | STRING declares a brand-new text variable, RegionUpper, up to 15 characters wide, before it can be filled with COMPUTE. |
| 2 | `COMPUTE RegionUpper = UPCASE(Region).` | COMPUTE fills RegionUpper with the upper-case version of Region, using the UPCASE() function. |
| 3 | `COMPUTE RegionInitial = SUBSTR(Region,1,3).` | COMPUTE creates RegionInitial as the first 3 characters of Region, using the SUBSTR() function (starting at character 1, taking a length of 3). |
| 4 | `EXECUTE.` | EXECUTE actually performs both computations now. |

**Working with dates**

```spss
* Suppose JoinDate is a real SPSS date variable (already formatted as DATE11).
COMPUTE TodayDate = $SYSDATE.
COMPUTE TenureDays = DATEDIFF(TodayDate, JoinDate, 'days').
COMPUTE TenureYears = TenureDays / 365.25.
FORMATS TenureYears (F5.1).
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `* Suppose JoinDate is a real SPSS date variable (already formatted as DATE11).` | A comment noting the assumption that JoinDate already exists in the data as a genuine SPSS date variable. |
| 2 | `COMPUTE TodayDate = $SYSDATE.` | COMPUTE creates TodayDate equal to $SYSDATE, the built-in system variable holding the current date at the moment the syntax runs. |
| 3 | `COMPUTE TenureDays = DATEDIFF(TodayDate, JoinDate, 'days').` | COMPUTE creates TenureDays as the difference, in days, between TodayDate and JoinDate, using the DATEDIFF() function. |
| 4 | `COMPUTE TenureYears = TenureDays / 365.25.` | COMPUTE creates TenureYears by dividing TenureDays by 365.25 (an average year length that accounts for leap years). |
| 5 | `FORMATS TenureYears (F5.1).` | FORMATS sets TenureYears to display with 5 total digits and 1 decimal place. |
| 6 | `EXECUTE.` | EXECUTE actually performs all three computations and applies the display format now. |

> **TIP:** $SYSDATE is a built-in system variable holding today's date at the moment the syntax runs - useful for age or tenure calculations that should always be current.

## Chapter 16: WEIGHT BY: Weighted Analysis

WEIGHT BY tells every subsequent procedure to treat each case as if it appeared as many times as the value of the weight variable - used to correct for unequal sampling probabilities, or to make a convenience sample match known population proportions (e.g., region size). Weighting stays active until you run WEIGHT OFF.


**Weighting to correct region representation**

```spss
COMPUTE RegionWeight = 1.
IF (Region = 'Lagos') RegionWeight = 0.9.
IF (Region = 'Kano') RegionWeight = 1.3.
EXECUTE.

WEIGHT BY RegionWeight.
FREQUENCIES VARIABLES=Region.
WEIGHT OFF.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `COMPUTE RegionWeight = 1.` | COMPUTE creates RegionWeight and sets every case's value to 1 as a default starting point. |
| 2 | `IF (Region = 'Lagos') RegionWeight = 0.9.` | IF overrides that default: cases from the Lagos region get a weight of 0.9 (slightly down-weighted). |
| 3 | `IF (Region = 'Kano') RegionWeight = 1.3.` | IF overrides the default again for a different subset: cases from the Kano region get a weight of 1.3 (up-weighted). |
| 4 | `EXECUTE.` | EXECUTE actually computes the weight variable for every case now. |
| 5 | *(blank line)* | A blank line for readability before turning weighting on. |
| 6 | `WEIGHT BY RegionWeight.` | WEIGHT BY switches on case weighting using RegionWeight, so every procedure that follows treats each case as if it appeared that many times. |
| 7 | `FREQUENCIES VARIABLES=Region.` | FREQUENCIES now runs on the weighted data, so region counts reflect the adjusted weights rather than the raw case counts. |
| 8 | `WEIGHT OFF.` | WEIGHT OFF switches weighting back off so later procedures return to treating every case equally. |

> **TIP:** Always run WEIGHT OFF as soon as the weighted analysis is finished - a weight left silently active is one of the most common causes of 'my sample size looks wrong' errors later in a session.

## Chapter 17: FREQUENCIES

FREQUENCIES is the default first step for any categorical or ordinal variable: it prints a count and percentage for every distinct value, plus optional summary statistics and charts. It is also the fastest way to eyeball a scale variable for outliers or data-entry errors before deciding on the right procedure.


**Categorical frequencies with a bar chart**

```spss
FREQUENCIES VARIABLES=Region Gender InsuranceType ConsultationChannel ChurnRisk
 /ORDER=ANALYSIS
 /BARCHART PERCENT.
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `FREQUENCIES VARIABLES=Region Gender InsuranceType ConsultationChannel ChurnRisk` | Begins FREQUENCIES on five categorical variables. |
| 2 | ` /ORDER=ANALYSIS` | ORDER=ANALYSIS lists each variable's categories in the order they are first encountered in the data, rather than sorted alphabetically. |
| 3 | ` /BARCHART PERCENT.` | BARCHART PERCENT requests a bar chart for each variable, with bar heights shown as percentages; the period ends the command. |
| 4 | `EXECUTE.` | EXECUTE actually runs the frequency tables and charts now. |

**Scale variable frequencies with summary statistics**

```spss
FREQUENCIES VARIABLES=Age MonthlyConsultationSpend EngagementIndex
 /FORMAT=NOTABLE
 /STATISTICS=MEAN MEDIAN STDDEV MINIMUM MAXIMUM SKEWNESS
 /HISTOGRAM NORMAL.
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `FREQUENCIES VARIABLES=Age MonthlyConsultationSpend EngagementIndex` | Begins FREQUENCIES on three scale (continuous) variables. |
| 2 | ` /FORMAT=NOTABLE` | FORMAT=NOTABLE suppresses the long value-by-value frequency table, since these variables have too many distinct values for such a table to be useful. |
| 3 | ` /STATISTICS=MEAN MEDIAN STDDEV MINIMUM MAXIMUM SKEWNESS` | STATISTICS requests the mean, median, standard deviation, minimum, maximum, and skewness for each variable. |
| 4 | ` /HISTOGRAM NORMAL.` | HISTOGRAM NORMAL requests a histogram for each variable with a normal distribution curve overlaid for visual comparison; the period ends the command. |
| 5 | `EXECUTE.` | EXECUTE actually runs the statistics and histograms now. |

> **TIP:** /FORMAT=NOTABLE suppresses the long value-by-value table for a continuous variable while still printing the requested summary statistics and histogram - use it whenever a scale variable has too many distinct values for a frequency table to be useful.

## Chapter 18: DESCRIPTIVES

DESCRIPTIVES is the leaner counterpart to FREQUENCIES, purpose-built for scale variables: one compact table with N, mean, standard deviation, minimum, and maximum for as many variables as you list. Its /SAVE subcommand can also save standardised (z-score) versions of each variable directly into new columns.


**Standard descriptive summary**

```spss
DESCRIPTIVES VARIABLES=Age YearsOnPlatform MonthlyConsultationSpend
 NumConsultationsMonth EngagementIndex
 /STATISTICS=MEAN STDDEV MIN MAX SKEWNESS KURTOSIS.
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `DESCRIPTIVES VARIABLES=Age YearsOnPlatform MonthlyConsultationSpend` | Begins DESCRIPTIVES on the first three scale variables. |
| 2 | ` NumConsultationsMonth EngagementIndex` | Continues the variable list with two more scale variables. |
| 3 | ` /STATISTICS=MEAN STDDEV MIN MAX SKEWNESS KURTOSIS.` | STATISTICS requests mean, standard deviation, minimum, maximum, skewness, and kurtosis for every listed variable; the period ends the command. |
| 4 | `EXECUTE.` | EXECUTE actually calculates all the requested statistics now. |

**Saving z-scores for later use**

```spss
DESCRIPTIVES VARIABLES=MonthlyConsultationSpend EngagementIndex /SAVE.
EXECUTE.
* This creates two new variables, ZMonthlyConsultationSpend and ZEngagementIndex.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|

| 2 | `EXECUTE.` | EXECUTE actually runs the command and creates the new z-score variables now. |

## Chapter 19: EXAMINE: Deeper Distributional Diagnostics

EXAMINE goes further than DESCRIPTIVES or FREQUENCIES: it produces normality tests (Shapiro-Wilk, Kolmogorov-Smirnov), confidence intervals for the mean, stem-and-leaf plots, boxplots, and can split all of that by a grouping variable in one command - the standard first check before choosing a parametric or non-parametric test.


**Full distributional check split by gender**

```spss
EXAMINE VARIABLES=MonthlyConsultationSpend EngagementIndex BY Gender
 /PLOT BOXPLOT STEMLEAF
 /STATISTICS DESCRIPTIVES EXTREME
 /CINTERVAL 95
 /MISSING LISTWISE.
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `EXAMINE VARIABLES=MonthlyConsultationSpend EngagementIndex BY Gender` | Begins EXAMINE on MonthlyConsultationSpend and EngagementIndex, split separately for each level of Gender. |
| 2 | ` /PLOT BOXPLOT STEMLEAF` | PLOT requests a boxplot and a stem-and-leaf plot for each variable, within each Gender group. |
| 3 | ` /STATISTICS DESCRIPTIVES EXTREME` | STATISTICS requests descriptive statistics and a table of the most extreme (highest and lowest) individual values. |
| 4 | ` /CINTERVAL 95` | CINTERVAL 95 requests a 95 percent confidence interval around each mean. |
| 5 | ` /MISSING LISTWISE.` | MISSING LISTWISE excludes any case with a missing value on either variable from the entire analysis; the period ends the command. |
| 6 | `EXECUTE.` | EXECUTE actually runs the full distributional check now. |

> **TIP:** Shapiro-Wilk (printed automatically for groups under 50 cases) and Kolmogorov-Smirnov (for larger groups) are the two normality tests EXAMINE reports - a significant result (p < .05) suggests the variable deviates from a normal distribution, which is one of the signals used in Chapter 31 to decide between a parametric and a non-parametric test.

## Chapter 20: MEANS: Group Comparisons of Averages

MEANS prints the mean, count, and standard deviation of one or more dependent variables broken down by one or more categorical variables, and can layer several grouping variables to build a crossed table of averages - a lightweight alternative to full ANOVA when you only need the descriptive picture, not a significance test.


**Average satisfaction by region and consultation channel**

```spss
MEANS TABLES=Q5_OverallSatisfaction MonthlyConsultationSpend BY Region BY ConsultationChannel
 /CELLS=MEAN COUNT STDDEV
 /STATISTICS ANOVA.
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `MEANS TABLES=Q5_OverallSatisfaction MonthlyConsultationSpend BY Region BY ConsultationChannel` | Begins MEANS, requesting average Q5_OverallSatisfaction and MonthlyConsultationSpend, broken down first by Region and then, within each region, by ConsultationChannel. |
| 2 | ` /CELLS=MEAN COUNT STDDEV` | CELLS requests that each cell of the resulting table show the mean, the case count, and the standard deviation. |
| 3 | ` /STATISTICS ANOVA.` | STATISTICS ANOVA additionally layers a one-way ANOVA significance test onto the table for each grouping variable; the period ends the command. |
| 4 | `EXECUTE.` | EXECUTE actually runs the grouped averages and ANOVA tests now. |

> **TIP:** Adding /STATISTICS ANOVA to a MEANS command layers a one-way ANOVA significance test onto the descriptive table for each grouping variable - a convenient shortcut before running the full ONEWAY procedure in Chapter 24.

## Chapter 21: CROSSTABS and the Chi-Square Test

CROSSTABS builds a contingency table between two (or more) categorical variables and, with /STATISTICS=CHISQ, tests whether the variables are independent. /CELLS controls which percentages print inside each cell - row, column, and total percentages are the three you will use most often.


**Is complaint filing associated with consultation channel?**

```spss
CROSSTABS
 /TABLES=ConsultationChannel BY ComplaintFiled
 /CELLS=COUNT ROW COLUMN
 /STATISTICS=CHISQ PHI.
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `CROSSTABS` | Begins the CROSSTABS command. |
| 2 | ` /TABLES=ConsultationChannel BY ComplaintFiled` | TABLES defines the contingency table: ConsultationChannel as rows, ComplaintFiled as columns. |
| 3 | ` /CELLS=COUNT ROW COLUMN` | CELLS requests the raw count plus row and column percentages inside every cell of the table. |
| 4 | ` /STATISTICS=CHISQ PHI.` | STATISTICS requests the chi-square test of independence and the Phi association measure; the period ends the command. |
| 5 | `EXECUTE.` | EXECUTE actually builds the table and runs the tests now. |

> **TIP:** Chi-square requires every expected cell count to be reasonably large (a common rule of thumb is at least 5); CROSSTABS prints a footnote telling you exactly how many cells fell below that threshold, and PHI/Cramer's V (add CC to /STATISTICS for a table larger than 2x2) gives you the effect size alongside the p-value.

## Chapter 22: T-TEST: One-Sample, Independent-Samples, and Paired-Samples

The T-TEST command has three distinct forms sharing one keyword. A one-sample t-test checks whether a variable's mean differs from a fixed value. An independent-samples t-test compares the means of two separate groups (e.g., male vs. female income). A paired-samples t-test compares two related measurements taken on the same cases (e.g., a before/after satisfaction score).


**One-sample: is average engagement index different from 650?**

```spss
T-TEST /TESTVAL=650
 /VARIABLES=EngagementIndex
 /CRITERIA=CI(.95).
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `T-TEST /TESTVAL=650` | Begins a one-sample T-TEST, comparing the data against a fixed test value of 650. |
| 2 | ` /VARIABLES=EngagementIndex` | VARIABLES names EngagementIndex as the variable being tested against that fixed value. |
| 3 | ` /CRITERIA=CI(.95).` | CRITERIA=CI(.95) requests a 95 percent confidence interval for the difference; the period ends the command. |

**Independent-samples: consultation spend by gender**

```spss
T-TEST GROUPS=Gender('Male' 'Female')
 /VARIABLES=MonthlyConsultationSpend
 /CRITERIA=CI(.95).
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `T-TEST GROUPS=Gender('Male' 'Female')` | Begins an independent-samples T-TEST, comparing the two named categories of Gender ('Male' and 'Female') against each other. |
| 2 | ` /VARIABLES=MonthlyConsultationSpend` | VARIABLES names MonthlyConsultationSpend as the outcome being compared between the two groups. |
| 3 | ` /CRITERIA=CI(.95).` | CRITERIA=CI(.95) requests a 95 percent confidence interval for the difference between the two group means; the period ends the command. |

**Paired-samples: two related satisfaction items**

```spss
T-TEST PAIRS=Q1_DoctorResponsiveness WITH Q4_WaitTimeSatisfaction (PAIRED)
 /CRITERIA=CI(.95).
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `T-TEST PAIRS=Q1_DoctorResponsiveness WITH Q4_WaitTimeSatisfaction (PAIRED)` | Begins a paired-samples T-TEST, comparing Q1_DoctorResponsiveness WITH Q4_WaitTimeSatisfaction as one matched pair measured on the same cases (PAIRED). |
| 2 | ` /CRITERIA=CI(.95).` | CRITERIA=CI(.95) requests a 95 percent confidence interval for the mean difference between the paired scores; the period ends the command. |

> **TIP:** For the independent-samples form, always read the Levene's Test for Equality of Variances row first: if it is significant (p < .05), report the 'Equal variances not assumed' row of the t-test instead of the default row.

## Chapter 23: ONEWAY ANOVA and Post-Hoc Tests

ONEWAY compares the means of a scale variable across three or more groups of a single categorical factor. On its own it only tells you 'at least one group differs' - the /POSTHOC subcommand runs pairwise comparisons (Tukey, Bonferroni, Scheffe are the most common) so you can see exactly which groups differ from which.


**Satisfaction across five regions, with post-hoc tests**

```spss
ONEWAY Q5_OverallSatisfaction BY Region
 /STATISTICS DESCRIPTIVES HOMOGENEITY
 /POSTHOC=TUKEY BONFERRONI ALPHA(0.05)
 /PLOT MEANS.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `ONEWAY Q5_OverallSatisfaction BY Region` | Begins ONEWAY, testing whether the mean of Q5_OverallSatisfaction differs across the categories of Region. |
| 2 | ` /STATISTICS DESCRIPTIVES HOMOGENEITY` | STATISTICS requests group descriptive statistics plus Levene's test for homogeneity (equality) of variances across regions. |
| 3 | ` /POSTHOC=TUKEY BONFERRONI ALPHA(0.05)` | POSTHOC requests two different pairwise comparison methods, Tukey and Bonferroni, both using a significance threshold (ALPHA) of 0.05, to show exactly which region pairs differ. |
| 4 | ` /PLOT MEANS.` | PLOT MEANS requests a simple line/point chart of the region means for a quick visual comparison; the period ends the command. |

> **TIP:** HOMOGENEITY requests Levene's test of equal variances across groups, the ANOVA equivalent of the t-test's Levene check - if it is significant, consider the Welch or Brown-Forsythe robust F available under /STATISTICS WELCH.

## Chapter 24: UNIANOVA and GLM: Factorial and Covariate-Adjusted ANOVA

ONEWAY only handles a single factor. UNIANOVA (and the more general GLM command) handles multiple categorical factors at once, their interactions, and continuous covariates - the syntax route to a two-way ANOVA or ANCOVA. Interaction terms are written with an asterisk between factor names.


**Two-way ANOVA: Region x Gender on satisfaction, with an interaction**

```spss
UNIANOVA Q5_OverallSatisfaction BY Region Gender
 /METHOD=SSTYPE(3)
 /INTERCEPT=INCLUDE
 /PLOT=PROFILE(Region*Gender)
 /EMMEANS=TABLES(Region*Gender)
 /PRINT=DESCRIPTIVE ETASQ HOMOGENEITY
 /DESIGN=Region Gender Region*Gender.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `UNIANOVA Q5_OverallSatisfaction BY Region Gender` | Begins UNIANOVA, modelling Q5_OverallSatisfaction as a function of the two categorical factors Region and Gender. |
| 2 | ` /METHOD=SSTYPE(3)` | METHOD=SSTYPE(3) selects Type III sums of squares, the standard default for a balanced or unbalanced factorial ANOVA. |
| 3 | ` /INTERCEPT=INCLUDE` | INTERCEPT=INCLUDE keeps the model's intercept term (the overall baseline level) in the analysis. |
| 4 | ` /PLOT=PROFILE(Region*Gender)` | PLOT requests an interaction (profile) plot showing how the Region by Gender combination affects the mean outcome. |
| 5 | ` /EMMEANS=TABLES(Region*Gender)` | EMMEANS requests estimated marginal means for every combination of Region and Gender. |
| 6 | ` /PRINT=DESCRIPTIVE ETASQ HOMOGENEITY` | PRINT requests descriptive statistics, partial eta-squared effect sizes, and the homogeneity-of-variance test. |
| 7 | ` /DESIGN=Region Gender Region*Gender.` | DESIGN explicitly lists the model terms to fit: the main effect of Region, the main effect of Gender, and their interaction (Region*Gender); the period ends the command. |

**ANCOVA: adjusting for a continuous covariate (Age)**

```spss
UNIANOVA MonthlyConsultationSpend BY Region WITH Age
 /PRINT=DESCRIPTIVE PARAMETER
 /DESIGN=Age Region.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `UNIANOVA MonthlyConsultationSpend BY Region WITH Age` | Begins UNIANOVA, modelling MonthlyConsultationSpend as a function of the categorical factor Region while statistically adjusting for the continuous covariate Age (introduced with WITH). |
| 2 | ` /PRINT=DESCRIPTIVE PARAMETER` | PRINT requests descriptive statistics and the model's parameter (coefficient) estimates. |
| 3 | ` /DESIGN=Age Region.` | DESIGN fits Age first, then Region, so the Region effect is reported after adjusting for Age; the period ends the command. |

> **TIP:** ETASQ in /PRINT adds partial eta-squared effect sizes next to the F-tests - always report an effect size alongside significance for a factorial design.

## Chapter 25: CORRELATIONS and NONPAR CORR

CORRELATIONS computes Pearson's r between every pair of listed scale variables (a full correlation matrix if you list more than two). NONPAR CORR computes the non-parametric equivalents - Spearman's rho and Kendall's tau-b - for ordinal data or when the normality assumption behind Pearson's r is doubtful.


**Pearson correlation matrix**

```spss
CORRELATIONS
 /VARIABLES=Age YearsOnPlatform MonthlyConsultationSpend NumConsultationsMonth EngagementIndex
 /PRINT=TWOTAIL NOSIG
 /MISSING=PAIRWISE.
```

| Line | Code | What this line does |
|---|---|---|
| 1 | `CORRELATIONS` | Begins the CORRELATIONS command. |
| 2 | ` /VARIABLES=Age YearsOnPlatform MonthlyConsultationSpend NumConsultationsMonth EngagementIndex` | VARIABLES lists the five scale variables that will be correlated with each other, producing a full correlation matrix. |
| 3 | ` /PRINT=TWOTAIL NOSIG` | PRINT requests two-tailed significance tests and turns off (NOSIG) the asterisk-flagging of non-significant correlations, for a cleaner-looking matrix. |
| 4 | ` /MISSING=PAIRWISE.` | MISSING=PAIRWISE tells SPSS to use all available data for each pair of variables individually, rather than dropping a case entirely if it is missing on any one variable; the period ends the command. |

**Spearman correlation for the Likert items**

```spss
NONPAR CORR
 /VARIABLES=Q1_DoctorResponsiveness Q2_AppEaseOfUse Q3_ConsultationQuality
 Q4_WaitTimeSatisfaction Q5_OverallSatisfaction
 /PRINT=SPEARMAN TWOTAIL NOSIG
 /MISSING=PAIRWISE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `NONPAR CORR` | Begins the NONPAR CORR command, the non-parametric counterpart to CORRELATIONS. |
| 2 | ` /VARIABLES=Q1_DoctorResponsiveness Q2_AppEaseOfUse Q3_ConsultationQuality` | VARIABLES begins listing the five Likert satisfaction items to be correlated. |
| 3 | ` Q4_WaitTimeSatisfaction Q5_OverallSatisfaction` | Continues the variable list from the line above. |

> **TIP:** NOSIG suppresses the asterisk-flagging of non-significant correlations so the matrix reads cleanly; remove it if you want every cell's significance flagged automatically.

## Chapter 26: REGRESSION: Linear Regression

REGRESSION fits an ordinary least squares model predicting one scale dependent variable from one or more independent variables. /DEPENDENT names the outcome, /METHOD=ENTER forces every listed predictor into the model in one step (other methods like STEPWISE add them conditionally), and /STATISTICS controls which diagnostic tables print alongside the coefficients.


**Predicting monthly consultation spend**

```spss
REGRESSION
 /MISSING LISTWISE
 /STATISTICS COEFF OUTS R ANOVA COLLIN TOL
 /CRITERIA=PIN(.05) POUT(.10)
 /NOORIGIN
 /DEPENDENT MonthlyConsultationSpend
 /METHOD=ENTER Age YearsOnPlatform NumConsultationsMonth
 /SCATTERPLOT (*ZRESID ,*ZPRED)
 /RESIDUALS HISTOGRAM(ZRESID).
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `REGRESSION` | Begins the REGRESSION command. |
| 2 | ` /MISSING LISTWISE` | MISSING LISTWISE drops any case with a missing value on any variable used anywhere in the model. |
| 3 | ` /STATISTICS COEFF OUTS R ANOVA COLLIN TOL` | STATISTICS requests coefficients, excluded-variable statistics, the multiple R, an ANOVA table for the overall model, and collinearity diagnostics (COLLIN) and tolerance values (TOL). |
| 4 | ` /CRITERIA=PIN(.05) POUT(.10)` | CRITERIA sets the probability thresholds SPSS uses to decide whether to enter (PIN) or remove (POUT) a predictor, relevant mainly for stepwise-style methods. |
| 5 | ` /NOORIGIN` | NOORIGIN tells SPSS to include the normal intercept term rather than forcing the regression line through the origin. |
| 6 | ` /DEPENDENT MonthlyConsultationSpend` | DEPENDENT names MonthlyConsultationSpend as the outcome variable being predicted. |
| 7 | ` /METHOD=ENTER Age YearsOnPlatform NumConsultationsMonth` | METHOD=ENTER forces all three named predictors, Age, YearsOnPlatform, and NumConsultationsMonth, into the model in a single step. |
| 8 | ` /SCATTERPLOT (*ZRESID ,*ZPRED)` | SCATTERPLOT requests a plot of the standardised residuals (*ZRESID) against the standardised predicted values (*ZPRED), used to check for patterns that would indicate a poor model fit. |
| 9 | ` /RESIDUALS HISTOGRAM(ZRESID).` | RESIDUALS HISTOGRAM(ZRESID) requests a histogram of the standardised residuals, used to check they are roughly normally distributed; the period ends the command. |

> **TIP:** COLLIN TOL requests collinearity diagnostics (tolerance and VIF) - always check these when a model has more than one predictor, since severely correlated predictors make individual coefficients unreliable even when the overall model fits well.

## Chapter 27: LOGISTIC REGRESSION

When the outcome is binary (Yes/No, 0/1) rather than continuous, ordinary regression is the wrong tool - LOGISTIC REGRESSION models the log-odds of the outcome instead, and prints odds ratios (Exp(B)) that are directly interpretable as 'this predictor multiplies the odds of the outcome by X'.


**Predicting missed follow-ups from engagement index and spend**

```spss
LOGISTIC REGRESSION VARIABLES MissedFollowUp
 /METHOD=ENTER EngagementIndex MonthlyConsultationSpend Age
 /CONTRAST (Gender)=Indicator
 /PRINT=GOODFIT CI(95)
 /CRITERIA=PIN(0.05) POUT(0.10) ITERATE(20) CUT(0.5).
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `LOGISTIC REGRESSION VARIABLES MissedFollowUp` | Begins LOGISTIC REGRESSION VARIABLES, naming MissedFollowUp as the binary (0/1) outcome being predicted. |
| 2 | ` /METHOD=ENTER EngagementIndex MonthlyConsultationSpend Age` | METHOD=ENTER forces the three named predictors, EngagementIndex, MonthlyConsultationSpend, and Age, into the model in a single step. |
| 3 | ` /CONTRAST (Gender)=Indicator` | CONTRAST tells SPSS to treat Gender as a categorical predictor using indicator (dummy) coding, in case it were added to the model. |
| 4 | ` /PRINT=GOODFIT CI(95)` | PRINT requests a goodness-of-fit test (GOODFIT) and 95 percent confidence intervals (CI(95)) for the odds ratios. |
| 5 | ` /CRITERIA=PIN(0.05) POUT(0.10) ITERATE(20) CUT(0.5).` | CRITERIA sets the entry/removal probability thresholds, the maximum number of iterations (20) the model may take to converge, and the probability cutoff (0.5) used to classify a predicted case as an event or non-event; the period ends the command. |

> **TIP:** The dependent variable in LOGISTIC REGRESSION must be coded numerically as 0/1 (the higher code is treated as the 'event'). Recode a Yes/No string variable with RECODE ... INTO before running this command if it is not already numeric.

## Chapter 29: FACTOR: Exploratory Factor Analysis / PCA

FACTOR reduces a set of correlated variables down to a smaller number of underlying dimensions (factors or components). It is commonly used on a health tech platform of Likert items to check whether they really measure one construct or several, and to produce factor scores for later use as predictors.


**Exploratory factor analysis on the satisfaction items**

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `FACTOR` | Begins the FACTOR command. |
| 2 | ` /VARIABLES=Q1_DoctorResponsiveness Q2_AppEaseOfUse Q3_ConsultationQuality` | VARIABLES begins listing the items to be included in the factor analysis, continuing onto the next line. |
| 3 | ` Q4_WaitTimeSatisfaction Q5_OverallSatisfaction` | Continues the variable list, completing all five satisfaction items. |
| 4 | ` /MISSING LISTWISE` | MISSING LISTWISE drops any case with a missing value on any of the listed items. |
| 5 | ` /ANALYSIS=Q1_DoctorResponsiveness Q2_AppEaseOfUse Q3_ConsultationQuality` | ANALYSIS repeats the same variable list, confirming which variables are actually analysed (this can differ from VARIABLES in more advanced setups), continuing onto the next line. |

## Chapter 30: NPAR TESTS: Non-Parametric Alternatives

**Kruskal-Wallis H: satisfaction across regions**

**Wilcoxon Signed-Rank: two paired Likert items**

## Chapter 31: Exporting Output: OUTPUT SAVE, OMS, and Direct Export

Once your analysis works, you often need the tables and charts outside SPSS entirely. OUTPUT EXPORT writes the whole Viewer to a Word, PDF, HTML, or Excel file in one command. The Output Management System (OMS) is more surgical: it intercepts specific tables as they are produced and routes them straight into a chosen file format, which is what you want inside an automated or looped job.


**Export the entire Viewer contents to Word**

```spss
OUTPUT EXPORT
 /CONTENTS EXPORT=ALL
 /DOCX DOCUMENTFILE='CarePulse_Results.docx'
 DOCUMENTTYPE=DOCUMENT.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `OUTPUT EXPORT` | Begins the OUTPUT EXPORT command. |
| 2 | ` /CONTENTS EXPORT=ALL` | CONTENTS EXPORT=ALL tells SPSS to export every table and chart currently in the Output Viewer, not a selected subset. |
| 3 | ` /DOCX DOCUMENTFILE='CarePulse_Results.docx'` | DOCX DOCUMENTFILE names the destination Word file to create. |
| 4 | ` DOCUMENTTYPE=DOCUMENT.` | DOCUMENTTYPE=DOCUMENT confirms a standard Word document (rather than, for example, a template) should be produced; the period ends the command. |

**OMS: capture only the Descriptives table to Excel**

```spss
OMS
 /SELECT TABLES
 /IF COMMANDS=['Descriptives'] SUBTYPES=['Descriptive Statistics']
 /DESTINATION FORMAT=XLSX OUTFILE='Descriptives_Only.xlsx'
 /TAG='DescCapture'.

DESCRIPTIVES VARIABLES=MonthlyConsultationSpend EngagementIndex.

OMSEND TAG='DescCapture'.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `OMS` | Begins the OMS (Output Management System) command, which will intercept specific output tables as they are produced. |
| 2 | ` /SELECT TABLES` | SELECT TABLES tells OMS to only capture statistical tables (not charts, logs, or other output types). |
| 3 | ` /IF COMMANDS=['Descriptives'] SUBTYPES=['Descriptive Statistics']` | IF filters which tables to capture: only ones produced by the Descriptives command, and specifically the 'Descriptive Statistics' subtype of table. |
| 4 | ` /DESTINATION FORMAT=XLSX OUTFILE='Descriptives_Only.xlsx'` | DESTINATION sends every captured table straight into a new Excel file named Descriptives_Only.xlsx. |
| 5 | ` /TAG='DescCapture'.` | TAG gives this OMS interception a name, DescCapture, so it can be matched to a specific OMSEND later; the period ends the command. |
| 6 | *(blank line)* | A blank line for readability before the procedure whose output will be captured. |
| 7 | `DESCRIPTIVES VARIABLES=MonthlyConsultationSpend EngagementIndex.` | DESCRIPTIVES runs normally, but because the OMS block above is active, its Descriptive Statistics table is silently also written to the Excel file. |
| 8 | *(blank line)* | A blank line for readability before switching the interception off. |
| 9 | `OMSEND TAG='DescCapture'.` | OMSEND, matched by TAG to the OMS command above, stops the interception so no further tables are captured after this point. |

> **TIP:** Always pair an OMS command with a matching OMSEND - an OMS block left open keeps intercepting every future table of that type for the rest of the session, which silently pollutes files you did not mean to write to.

## Chapter 32: Charts via Syntax: GGRAPH and GRAPH

GGRAPH generates modern, highly customisable charts using the Graphics Production Language (GPL) that SPSS builds internally whenever you use the Chart Builder. The legacy GRAPH command is shorter to type for simple bar and line charts. Pasting a chart you built visually in Chart Builder is the fastest way to learn GPL syntax for a specific chart type.


**Simple legacy bar chart**

```spss
GRAPH
 /BAR(SIMPLE)=MEAN(MonthlyConsultationSpend) BY Region
 /TITLE='Average Monthly Income by Region'.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `GRAPH` | Begins the legacy GRAPH command. |
| 2 | ` /BAR(SIMPLE)=MEAN(MonthlyConsultationSpend) BY Region` | BAR(SIMPLE) requests a simple bar chart showing the mean of MonthlyConsultationSpend, one bar per category of Region. |
| 3 | ` /TITLE='Average Monthly Income by Region'.` | TITLE sets the chart's on-screen title text; the period ends the command. |

**GGRAPH: clustered bar of satisfaction by region and gender**

```spss
GGRAPH
 /GRAPHDATASET NAME='graphdataset' VARIABLES=Region Gender
 MEAN(Q5_OverallSatisfaction)[name='MeanSat']
 MISSING=LISTWISE REPORTMISSING=NO
 /GRAPHSPEC SOURCE=INLINE.
BEGIN GPL
 SOURCE: s=userSource(id('graphdataset'))
 DATA: Region=col(source(s), name('Region'), unit.category())
 DATA: Gender=col(source(s), name('Gender'), unit.category())
 DATA: MeanSat=col(source(s), name('MeanSat'))
 GUIDE: axis(dim(1), label('Region'))
 GUIDE: axis(dim(2), label('Mean Overall Satisfaction'))
 GUIDE: legend(aesthetic(aesthetic.color.interior), label('Gender'))
 ELEMENT: interval(position(Region*MeanSat), color.interior(Gender),
 shape.interior(shape.square))
END GPL.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `GGRAPH` | Begins the GGRAPH command, which builds a chart using the modern Graphics Production Language (GPL). |
| 2 | ` /GRAPHDATASET NAME='graphdataset' VARIABLES=Region Gender` | GRAPHDATASET names an internal dataset called 'graphdataset' and specifies that it should contain Region, Gender, and the mean of Q5_OverallSatisfaction (aliased as MeanSat) for each combination. |
| 3 | ` MEAN(Q5_OverallSatisfaction)[name='MeanSat']` | Continues defining the summary measure MeanSat, naming it so GPL can refer to it by that alias later. |
| 4 | ` MISSING=LISTWISE REPORTMISSING=NO` | MISSING=LISTWISE REPORTMISSING=NO drops cases with missing values from the chart data and suppresses any 'missing data' note on the chart. |
| 5 | ` /GRAPHSPEC SOURCE=INLINE.` | GRAPHSPEC SOURCE=INLINE tells SPSS that the actual chart specification, in GPL, is written directly in the lines that follow. |
| 6 | `BEGIN GPL` | BEGIN GPL marks the start of the embedded Graphics Production Language block. |
| 7 | ` SOURCE: s=userSource(id('graphdataset'))` | SOURCE names s as a reference to the graphdataset defined above, so the GPL statements below can pull data from it. |
| 8 | ` DATA: Region=col(source(s), name('Region'), unit.category())` | DATA declares that the Region column from the source will be treated as a categorical (unit.category) dimension called Region. |
| 9 | ` DATA: Gender=col(source(s), name('Gender'), unit.category())` | DATA declares that the Gender column will likewise be treated as a categorical dimension called Gender. |
| 10 | ` DATA: MeanSat=col(source(s), name('MeanSat'))` | DATA declares that the MeanSat column (the pre-computed mean satisfaction) will be used as a plain numeric measure. |
| 11 | ` GUIDE: axis(dim(1), label('Region'))` | GUIDE sets the first chart axis's label to 'Region'. |
| 12 | ` GUIDE: axis(dim(2), label('Mean Overall Satisfaction'))` | GUIDE sets the second chart axis's label to 'Mean Overall Satisfaction'. |
| 13 | ` GUIDE: legend(aesthetic(aesthetic.color.interior), label('Gender'))` | GUIDE adds a colour legend labelled 'Gender', mapped to the interior colour aesthetic used below. |
| 14 | ` ELEMENT: interval(position(Region*MeanSat), color.interior(Gender),` | ELEMENT draws the actual bars: position them by Region and MeanSat, colour them by Gender, and use square-shaped bar interiors; the element definition continues on the next line. |
| 15 | ` shape.interior(shape.square))` | Completes the ELEMENT statement's shape specification from the line above. |
| 16 | `END GPL.` | END GPL closes the embedded GPL block, and the following period ends the whole GGRAPH command. |

> **TIP:** You rarely write GPL from a blank page: build the chart once in Graphs > Chart Builder, click Paste, and edit the generated GGRAPH block - this is far faster than memorising GPL syntax.

## Chapter 33: Macros: DEFINE ... !ENDDEFINE

A macro is a reusable block of syntax with parameters, similar to a function in a programming language. Once defined with DEFINE ... !ENDDEFINE, you call it anywhere later in the session simply by writing its name and arguments, and SPSS expands it into the full syntax before running it. Macros remove repetition when you need the same analysis run many times with only one detail changing.


**A macro that runs the same descriptive block for any variable**

```spss
DEFINE !QuickDescribe (!POS !TOKENS(1)).
 DESCRIPTIVES VARIABLES=!1
 /STATISTICS=MEAN STDDEV MIN MAX.
 FREQUENCIES VARIABLES=!1 /HISTOGRAM NORMAL /FORMAT=NOTABLE.
!ENDDEFINE.

!QuickDescribe MonthlyConsultationSpend.
!QuickDescribe EngagementIndex.
!QuickDescribe Age.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `DEFINE !QuickDescribe (!POS !TOKENS(1)).` | DEFINE begins a new macro named !QuickDescribe, which accepts one positional token argument (!POS !TOKENS(1)), meaning it takes exactly one plain value when called. |
| 2 | ` DESCRIPTIVES VARIABLES=!1` | Inside the macro body, DESCRIPTIVES runs on whichever variable name is substituted for !1 (the macro's first argument) when it is called. |
| 3 | ` /STATISTICS=MEAN STDDEV MIN MAX.` | STATISTICS requests mean, standard deviation, minimum, and maximum for that same variable. |
| 4 | ` FREQUENCIES VARIABLES=!1 /HISTOGRAM NORMAL /FORMAT=NOTABLE.` | FREQUENCIES also runs on !1, requesting a normal-curve histogram while suppressing the full frequency table with NOTABLE. |
| 5 | `!ENDDEFINE.` | !ENDDEFINE closes the macro definition that began with DEFINE above. |
| 6 | *(blank line)* | A blank line for readability before the macro is actually used. |
| 7 | `!QuickDescribe MonthlyConsultationSpend.` | Calls the macro with MonthlyConsultationSpend as the argument, so !1 is replaced by MonthlyConsultationSpend and both statements inside the macro run against it. |
| 8 | `!QuickDescribe EngagementIndex.` | Calls the same macro again with EngagementIndex as the argument. |
| 9 | `!QuickDescribe Age.` | Calls the same macro a third time with Age as the argument. |

> **TIP:** Macro names conventionally start with an exclamation mark (!) so they are visually distinct from ordinary command keywords and variable names in the syntax file.

## Chapter 34: Looping: LOOP / END LOOP and DO REPEAT

LOOP ... END LOOP repeats a block of transformation commands a set number of times or while a condition holds, case by case - useful for iterative calculations like running totals. DO REPEAT ... END REPEAT is different: it applies the same transformation to a whole list of variables in parallel, one substitution at a time, which is the everyday tool for recoding, renaming, or recalculating many similarly-structured variables at once.


**DO REPEAT: rescale every Likert item to a 0-100 scale at once**

```spss
DO REPEAT src=Q1_DoctorResponsiveness Q2_AppEaseOfUse Q3_ConsultationQuality
 Q4_WaitTimeSatisfaction Q5_OverallSatisfaction
 /dst=Q1_100 Q2_100 Q3_100 Q4_100 Q5_100.
 COMPUTE dst = ((src - 1) / 4) * 100.
END REPEAT.
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `DO REPEAT src=Q1_DoctorResponsiveness Q2_AppEaseOfUse Q3_ConsultationQuality` | DO REPEAT begins a parallel substitution block: src will stand in turn for each of the five source Likert variables listed here, continuing onto the next line. |
| 2 | ` Q4_WaitTimeSatisfaction Q5_OverallSatisfaction` | Completes the list of source variables that src will represent, one at a time. |
| 3 | ` /dst=Q1_100 Q2_100 Q3_100 Q4_100 Q5_100.` | /dst lists five destination variable names, one to be created for each source variable, in the same left-to-right order. |
| 4 | ` COMPUTE dst = ((src - 1) / 4) * 100.` | COMPUTE rescales whichever src is currently active from its original 1-5 range onto a 0-100 range, storing the result in the matching dst variable. |
| 5 | `END REPEAT.` | END REPEAT closes the DO REPEAT block, having by now run the COMPUTE line once for each of the five src/dst pairs. |
| 6 | `EXECUTE.` | EXECUTE actually performs all five rescaling computations now. |

**LOOP: a simple case-by-case running index**

```spss
COMPUTE RunningFlagCount = 0.
LOOP #i = 1 TO 5.
 DO IF (#i = 1 AND Q1_DoctorResponsiveness >= 4).
 COMPUTE RunningFlagCount = RunningFlagCount + 1.
 END IF.
END LOOP.
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `COMPUTE RunningFlagCount = 0.` | COMPUTE initialises RunningFlagCount to 0 for every case, before the loop below adjusts it. |
| 2 | `LOOP #i = 1 TO 5.` | LOOP begins a repeat block that runs its contents for #i equal to 1, then 2, then 3, up through 5, for every case. |
| 3 | ` DO IF (#i = 1 AND Q1_DoctorResponsiveness >= 4).` | DO IF restricts the action inside this iteration to the very first pass (#i=1) where Q1_DoctorResponsiveness is also 4 or higher. |
| 4 | ` COMPUTE RunningFlagCount = RunningFlagCount + 1.` | For cases meeting that condition, adds 1 to RunningFlagCount. |
| 5 | ` END IF.` | END IF closes the DO IF block opened two lines above. |
| 6 | `END LOOP.` | END LOOP closes the LOOP block opened above, having run through all 5 iterations of #i for every case. |
| 7 | `EXECUTE.` | EXECUTE actually runs the whole loop now. |

> **TIP:** DO REPEAT is worth learning before LOOP: in real teaching and analysis work it is used far more often, because most repetitive tasks are 'do the same thing to many variables', not 'do the same thing many times to one case'.

## Chapter 35: Extending SPSS with Python: BEGIN PROGRAM / END PROGRAM

This is an advanced, optional bridge rather than a requirement for everyday syntax work.

## Chapter 36: Writing Clean, Maintainable Syntax

A syntax file is code, and the same habits that make code maintainable apply here. Keep one command per logical block with a comment above it stating its purpose. Prefer INTO when recoding, so the original variable is never destroyed. ## Chapter 37: Common Errors and How to Read Them

Most beginner syntax errors fall into a handful of repeating patterns: a missing period at the end of a command (SPSS reports an error on the line after the one actually at fault); referencing a variable name that does not exist yet, usually from a typo or because an earlier COMPUTE/RECODE was never actually run; a string value compared without quotes (Region = Lagos instead of Region = 'Lagos'); and forgetting EXECUTE after a transformation, so nothing appears to have happened even though the command 'ran' without error.

When SPSS reports an error, read the error number and the quoted fragment of syntax carefully - it almost always points at the exact token that confused the parser, and the fix is usually smaller than the error message makes it look.


**Common mistake: unquoted string comparison**

```spss
* Wrong - SPSS will treat Lagos as an undefined variable name:
* SELECT IF (Region = Lagos).

* Correct - string values must always be in quotes:
SELECT IF (Region = 'Lagos').
EXECUTE.
```

*Line-by-line explanation:*

| Line | Code | What this line does |
|---|---|---|
| 1 | `* Wrong - SPSS will treat Lagos as an undefined variable name:` | A comment labelling the line below as an example of what NOT to do. |
| 2 | `* SELECT IF (Region = Lagos).` | A commented-out line showing the mistake: comparing Region to Lagos without quotes, which SPSS would try to read as an undefined variable name rather than the text 'Lagos'. |
| 3 | *(blank line)* | A blank line separating the wrong example from the corrected one. |
| 4 | `* Correct - string values must always be in quotes:` | A comment labelling the line below as the corrected version. |
| 5 | `SELECT IF (Region = 'Lagos').` | The corrected SELECT IF: 'Lagos' is wrapped in single quotes, so SPSS correctly compares Region against the literal text value. |
| 6 | `EXECUTE.` | EXECUTE actually applies the corrected selection now. |

> **TIP:** If a command silently 'does nothing', check for a missing EXECUTE first - SPSS queues transformation commands and only actually processes the data when it hits EXECUTE or the next procedure command (FREQUENCIES, DESCRIPTIVES, etc.).

## Chapter 38: Appendix A: Full Worked Syntax Script

The script below chains together one representative command from most chapters in this book into a single runnable file against the practice dataset, from raw import through to a final exported results document. Use it as a template to copy and adapt for your own projects, and as a quick self-test: if every line below runs without an error message on your machine, your SPSS installation and syntax fundamentals are both in good shape.


**End-to-end pipeline**

```spss
SET SEED=234.

GET DATA /TYPE=TXT
 /FILE='carepulse_patient_survey.csv'
 /DELCASE=LINE /DELIMITERS=',' /QUALIFIER='"'
 /ARRANGEMENT=DELIMITED /FIRSTCASE=2 /IMPORTCASE=ALL
 /VARIABLES=
 PatientID A6 Region A15 Gender A6 Age F3 InsuranceType A13
 ConsultationChannel A14 YearsOnPlatform F5.1 MonthlyConsultationSpend F8
 NumConsultationsMonth F3 Q1_DoctorResponsiveness F1 Q2_AppEaseOfUse F1
 Q3_ConsultationQuality F1 Q4_WaitTimeSatisfaction F1 Q5_OverallSatisfaction F1
 ComplaintFiled A3 EngagementIndex F3 MissedFollowUp A3 ChurnRisk A6.
CACHE.
EXECUTE.
DATASET NAME CarePulse WINDOW=FRONT.

VARIABLE LEVEL Q1_DoctorResponsiveness TO Q5_OverallSatisfaction (ORDINAL).
VALUE LABELS Q1_DoctorResponsiveness Q2_AppEaseOfUse Q3_ConsultationQuality
 Q4_WaitTimeSatisfaction Q5_OverallSatisfaction
 1 'Strongly Disagree' 2 'Disagree' 3 'Neutral' 4 'Agree' 5 'Strongly Agree'.

COMPUTE SatisfactionIndex = MEAN(Q1_DoctorResponsiveness, Q2_AppEaseOfUse,
 Q3_ConsultationQuality, Q4_WaitTimeSatisfaction, Q5_OverallSatisfaction).
RECODE Age (18 THRU 25=1)(26 THRU 40=2)(41 THRU 55=3)(56 THRU HIGHEST=4)
 INTO AgeBand.
EXECUTE.

FREQUENCIES VARIABLES=Region Gender ChurnRisk.
DESCRIPTIVES VARIABLES=MonthlyConsultationSpend EngagementIndex SatisfactionIndex.
CROSSTABS /TABLES=ConsultationChannel BY ComplaintFiled /STATISTICS=CHISQ.
T-TEST GROUPS=Gender('Male' 'Female') /VARIABLES=MonthlyConsultationSpend.
ONEWAY Q5_OverallSatisfaction BY Region /POSTHOC=TUKEY.
CORRELATIONS /VARIABLES=Age MonthlyConsultationSpend EngagementIndex.
REGRESSION /DEPENDENT MonthlyConsultationSpend /METHOD=ENTER Age YearsOnPlatform.
RELIABILITY /VARIABLES=Q1_DoctorResponsiveness TO Q5_OverallSatisfaction
 /MODEL=ALPHA.

OUTPUT EXPORT /CONTENTS EXPORT=ALL
 /DOCX DOCUMENTFILE='CarePulse_FullResults.docx' DOCUMENTTYPE=DOCUMENT.
```
