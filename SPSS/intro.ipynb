# SPSS

## INTRODUCTION TO SPSS (Statistical Package for the Social Sciences)

### 1. What is SPSS?

SPSS (Statistical Package for the Social Sciences) is a software application used for:

- Data management

- Statistical analysis

- Data visualization

- Reporting

It is widely used in:

- Academic research

- Market research

- Healthcare analytics

- Government and policy analysis

 Originally developed in 1968, now owned by IBM, hence often called IBM SPSS Statistics.

### 2. Why SPSS? 

| Tool   | Strength                                     |
| ------ | -------------------------------------------- |
| Python | Automation, ML, flexibility                  |
| R      | Advanced statistics, visualization           |
| SPSS   | GUI-based, fast analysis, no coding required |

SPSS is best for:

Non-programmers
Quick statistical testing
Structured survey data
Academic projects

### 3. SPSS Interface Overview

SPSS has 3 major windows:

#### A. Data Editor Window (Main Workspace)

This has two tabs:

🔹 Data View
Rows = Observations (records)
Columns = Variables (features)

 Equivalent:

Python → Pandas DataFrame

R → Data frame

🔹 Variable View

This is where you define your dataset structure.

| Column   | Meaning                   |
| -------- | ------------------------- |
| Name     | Variable name (no spaces) |
| Type     | Numeric, String, Date     |
| Width    | Number of characters      |
| Decimals | Decimal places            |
| Label    | Full description          |
| Values   | Categorical encoding      |
| Missing  | Missing value handling    |
| Measure  | Scale type                |


#### B. Output Viewer

- Displays results of analysis

- Tables, charts, statistical outputs

- Can export to Word, PDF, Excel

#### C. Syntax Editor

- Where you write SPSS commands (optional)

- Helps automation and reproducibility

Equivalent:

- Python → script (.py)

- R → script (.R)

### 4. Types of Variables in SPSS


🔹 1. Numeric Variables

- Numbers (e.g., age, salary)

🔹 2. String Variables

- Text (e.g., names, gender labels)

Measurement Levels (VERY IMPORTANT)

| Type    | Meaning                  | Example         |
| ------- | ------------------------ | --------------- |
| Nominal | Categories, no order     | Gender          |
| Ordinal | Ordered categories       | Education level |
| Scale   | Numeric (interval/ratio) | Income, Age     |
