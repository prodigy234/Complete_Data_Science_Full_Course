## Lesson 1.1 — Introduction to SPSS and Its Role in Data Science

### What is SPSS?

**SPSS** stands for **Statistical Package for the Social Sciences**. It was originally developed at the University of Chicago in 1968 and is now owned by IBM as **IBM SPSS Statistics**. 

SPSS is a point-and-click statistical software used widely in:
- **Business and HR analytics** — analysing employee data, performance, compensation
- **Healthcare research** — clinical trials, patient outcome studies
- **Social science and survey research** — government surveys, academic studies
- **Banking and finance** — risk analysis, customer segmentation
- **Marketing research** — consumer behaviour, brand studies

### SPSS vs Other Tools

| Tool | Strengths | When to use |
|---|---|---|
| **SPSS** | Menu-driven, output tables, statistical depth | Research, surveys, HR analytics, academia |
| **Excel** | Familiar, fast calculations | Simple summaries, small datasets |
| **Python/R** | Flexible, automation, machine learning | Large-scale, custom analysis |
| **Power BI** | Visualisation, dashboards | Reporting and monitoring |

**Key message for students:** SPSS is not a replacement for Python or R — it is a *complementary* tool that is faster for standard statistical tests and produces publication-ready output tables without writing code.

### The SPSS Workflow

```
Raw Data (CSV/Excel)
      ↓
  Import into SPSS
      ↓
  Define Variables (Variable View)
      ↓
  Clean the Data (Transform menu)
      ↓
  Analyse (Analyze menu)
      ↓
  Interpret Output (Output Viewer)
      ↓
  Report (Export to Word/PDF)
```

### SPSS File Types

| Extension | What it is |
|---|---|
| `.sav` | SPSS data file — your main working file |
| `.spv` | SPSS output viewer file |
| `.sps` | SPSS syntax (code) file |
| `.csv` | Comma-separated values — imported from Excel or other sources |

## Lesson 1.3 — Variable View: Defining Variables Correctly

Variable View is where you tell SPSS *what each column means*. Getting this right is critical — incorrect variable definitions cause wrong analyses.

### The Variable View Columns Explained

| Column | What it means | Example |
|---|---|---|
| **Name** | Variable name — no spaces, max 64 chars | `Monthly_Salary_NGN` |
| **Type** | Data format | Numeric, String, Date |
| **Width** | Number of characters stored | 8 (default for numbers) |
| **Decimals** | Decimal places displayed | 0 for salary, 1 for scores |
| **Label** | Full descriptive label shown in output | "Monthly Salary (NGN)" |
| **Values** | Labels for coded numbers | 0 = "Not Promoted", 1 = "Promoted" |
| **Missing** | Which values mean "missing" | 99 = missing for age |
| **Columns** | Display width in Data View | |
| **Align** | Left/right/centre alignment | |
| **Measure** | Measurement level — CRITICAL | Nominal, Ordinal, Scale |
| **Role** | Input/Target/Both for modelling | |

---

### Measurement Levels — The Most Important Setting

SPSS uses the measurement level to decide which analyses are appropriate. Setting this wrongly produces incorrect results.

| Level | What it means | Examples from our dataset | SPSS icon |
|---|---|---|---|
| **Nominal** | Categories with no order | Gender, Department, Region, Promoted | 🔵 |
| **Ordinal** | Categories with a meaningful order | Education_Level, Job_Grade, Performance_Rating | 🟡 |
| **Scale** | Continuous numbers — mean and std dev make sense | Age, Salary, Score, Training_Hours | 📏 |

**Common mistake students make:** They import a CSV and SPSS assigns everything as Scale by default. Students must manually set Nominal and Ordinal variables.

---

### Value Labels — Why They Matter

When you have a binary variable like `Promoted` (coded 0 and 1), you should define value labels:

**Steps:**
1. In Variable View, click the cell in the **Values** column for `Promoted`
2. Click the small blue button (...)
3. In the dialog: Value = `0`, Label = `Not Promoted` → click **Add**
4. Value = `1`, Label = `Promoted` → click **Add**
5. Click **OK**

Now SPSS output will show "Not Promoted / Promoted" instead of "0 / 1".

Do the same for `Gender`: 1 = Male, 2 = Female (if you re-code to numeric).

### SPSS Syntax: Defining Variable Labels and Value Labels

```spss
* ============================================================
* DEFINE VARIABLE LABELS
* This makes output tables more readable
* ============================================================

VARIABLE LABELS
  Employee_ID       'Unique Employee Identifier'
  Department        'Department Name'
  Gender            'Employee Gender'
  Region            'Work Region'
  Age               'Age in Years'
  Education_Level   'Highest Educational Qualification'
  Years_Experience  'Total Years of Work Experience'
  Job_Grade         'Job Grade (1=Lowest, 5=Highest)'
  Monthly_Salary_NGN 'Monthly Gross Salary (Nigerian Naira)'
  Training_Hours    'Annual Training Hours Attended'
  Performance_Score 'Performance Score (20-100)'
  Performance_Rating 'Performance Rating Category'
  Job_Satisfaction  'Job Satisfaction Score (1.0-5.0)'
  Absenteeism_Days  'Number of Absence Days in the Year'
  Promoted          'Was Employee Promoted? (0=No, 1=Yes)'.

* ============================================================
* DEFINE VALUE LABELS FOR CATEGORICAL VARIABLES
* ============================================================

VALUE LABELS Promoted
  0 'Not Promoted'
  1 'Promoted'.

* ============================================================
* SET MEASUREMENT LEVELS CORRECTLY
* ============================================================

VARIABLE LEVEL
  Department Gender Region Education_Level Job_Grade Performance_Rating (NOMINAL)
  Promoted (NOMINAL)
  Age Years_Experience Monthly_Salary_NGN Training_Hours 
  Performance_Score Job_Satisfaction Absenteeism_Days (SCALE).

EXECUTE.
```

## Lesson 2.1 — Identifying and Handling Missing Data

### Why Missing Data Matters

Missing data is one of the most common real-world problems in analytics. If not handled properly:
- Statistical tests may produce biased results
- SPSS procedures may silently exclude cases, leading to misleading sample sizes
- Reports may overstate or understate findings

### Types of Missing Data in SPSS

| Type | What it means | How to handle |
|---|---|---|
| **System Missing** | Cell is completely blank — the dot (.) in SPSS | SPSS excludes automatically |
| **User Missing** | A numeric value that means "not applicable" — e.g., 99 or -1 | You declare it as missing in Variable View |

Our dataset has **system missing** values in `Training_Hours` (15 cases) and `Job_Satisfaction` (14 cases).

---

### Step 1: Check for Missing Values — Frequencies Method

**Menu path:** `Analyze → Descriptive Statistics → Frequencies`

1. Move `Training_Hours` and `Job_Satisfaction` into the Variables box
2. Make sure **Display frequency tables** is checked
3. Click **OK**

Look at the output. The **Valid** count and **Missing** count at the top of each table tell you exactly how many missing values exist.

---

### Step 2: Check All Variables at Once — Missing Values Analysis

**Menu path:** `Analyze → Missing Value Analysis`

This dialog gives a full summary across all variables.

---

### Step 3: Handling Missing Values — Options

**Option A: Listwise deletion (SPSS default)**  
SPSS automatically excludes any case that has a missing value on *any* variable involved in the analysis. This is fine when missingness is low (< 5%).

**Option B: Replace with Mean (Mean Substitution)**  
Replace missing values with the variable mean. Use for scale variables when data is missing at random.

**Menu path:** `Transform → Replace Missing Values`

**Option C: Exclude from specific analyses**  
Use SPSS syntax `MISSING VALUES` to mark values as user-missing without deleting them.

---

### SPSS Syntax: Missing Value Analysis and Mean Substitution

```spss
* ============================================================
* CHECK MISSING VALUES — FREQUENCIES APPROACH
* ============================================================

FREQUENCIES VARIABLES=Training_Hours Job_Satisfaction
  /ORDER=ANALYSIS.

* ============================================================
* REPLACE MISSING VALUES WITH VARIABLE MEAN
* This creates a NEW variable — never overwrite the original
* ============================================================

RMV
  /Training_Hours_clean=SMEAN(Training_Hours)
  /Job_Satisfaction_clean=SMEAN(Job_Satisfaction).

EXECUTE.

* ============================================================
* VERIFY — COMPARE ORIGINAL VS CLEANED VARIABLE
* ============================================================

DESCRIPTIVES VARIABLES=Training_Hours Training_Hours_clean
  /STATISTICS=MEAN STDDEV MIN MAX N.
```

---

### Teaching Note on Missing Data Strategy

For this course, we will use **mean substitution** for the two variables with missing values. In a real-world setting, you would investigate *why* data is missing before deciding how to treat it. The three main types of missingness are:
- **MCAR** — Missing Completely At Random (safe to substitute)
- **MAR** — Missing At Random (requires more care)
- **MNAR** — Missing Not At Random (requires specialist methods)

This is an important concept to introduce even at beginner level.

## Lesson 2.2 — Recoding Variables

### What is Recoding?

Recoding means transforming the values of an existing variable — grouping them into categories, reversing a scale, or collapsing many levels into fewer.

**Use cases:**
- Group `Age` into age bands: 20–29, 30–39, 40–49, 50+
- Collapse `Performance_Rating` into 2 groups: Below Average vs Average and Above
- Reverse-score a satisfaction item (so 1 becomes 5 and 5 becomes 1)

### Recode Into Different Variables (Best Practice)

Always recode **Into Different Variables** — never overwrite your original data.

**Menu path:** `Transform → Recode Into Different Variables`

1. Move `Age` into the Input Variable box
2. In **Output Variable Name**, type `Age_Group`
3. Click **Change**
4. Click **Old and New Values**
5. Use **Range** to define: 22 through 29 → 1, 30 through 39 → 2, 40 through 49 → 3, 50 through 99 → 4
6. Click **Add** after each pair
7. Click **Continue → OK**
8. Then in Variable View, add value labels to `Age_Group`

---

### SPSS Syntax: Recode Age into Age Groups

```spss
* ============================================================
* RECODE AGE INTO AGE BANDS
* Creates a new variable: Age_Group
* ============================================================

RECODE Age
  (22 THRU 29=1)
  (30 THRU 39=2)
  (40 THRU 49=3)
  (50 THRU 58=4)
  INTO Age_Group.

VARIABLE LABELS Age_Group 'Age Group (Banded)'.

VALUE LABELS Age_Group
  1 '22 to 29'
  2 '30 to 39'
  3 '40 to 49'
  4 '50 and above'.

VARIABLE LEVEL Age_Group (ORDINAL).

EXECUTE.

* ============================================================
* RECODE PERFORMANCE RATING INTO BINARY (Below vs At/Above)
* ============================================================

RECODE Performance_Rating
  ('Poor'='Below Average')(ELSE=COPY)
  INTO Perf_Binary.

* Alternative: create numeric binary version
RECODE Performance_Score
  (LO THRU 54.99=0)
  (55 THRU HI=1)
  INTO Perf_Meets_Target.

VARIABLE LABELS Perf_Meets_Target 'Meets Performance Target (0=No, 1=Yes)'.
VALUE LABELS Perf_Meets_Target 0 'Does Not Meet Target' 1 'Meets Target'.
VARIABLE LEVEL Perf_Meets_Target (NOMINAL).

EXECUTE.
```

---

### SPSS Syntax: Recode Education Level to Numeric Ordinal

```spss
* ============================================================
* RECODE EDUCATION LEVEL (String) TO NUMERIC ORDINAL
* Needed for correlation and regression
* ============================================================

RECODE Education_Level
  ('Secondary School'=1)
  ('OND/NCE'=2)
  ('HND/BSc'=3)
  ('MSc/MBA'=4)
  ('PhD'=5)
  INTO Education_Numeric.

VARIABLE LABELS Education_Numeric 'Education Level (Numeric)'.
VALUE LABELS Education_Numeric
  1 'Secondary School'
  2 'OND/NCE'
  3 'HND/BSc'
  4 'MSc/MBA'
  5 'PhD'.
VARIABLE LEVEL Education_Numeric (ORDINAL).

EXECUTE.
```

## Lesson 2.3 — Computing New Variables

### What is Compute Variable?

The `Compute Variable` function creates a new variable by applying a formula to existing variables. This is equivalent to creating a new column in Excel with a formula — except SPSS applies it to all rows at once.

**Menu path:** `Transform → Compute Variable`

---

### Examples We Will Build

1. **Annual Salary** — multiply monthly salary × 12
2. **Salary per Year of Experience** — salary ÷ experience (productivity ratio)
3. **Standardised Performance Score** — re-scale score to 0–1

---

### SPSS Syntax: Compute New Variables

```spss
* ============================================================
* COMPUTE ANNUAL SALARY
* ============================================================

COMPUTE Annual_Salary_NGN = Monthly_Salary_NGN * 12.
VARIABLE LABELS Annual_Salary_NGN 'Annual Gross Salary (NGN)'.
VARIABLE LEVEL Annual_Salary_NGN (SCALE).
EXECUTE.

* ============================================================
* COMPUTE SALARY PER YEAR OF EXPERIENCE
* Avoid divide-by-zero: only compute for cases with experience > 0
* ============================================================

IF (Years_Experience > 0) Salary_Per_Exp_Year = Monthly_Salary_NGN / Years_Experience.
VARIABLE LABELS Salary_Per_Exp_Year 'Monthly Salary per Year of Experience'.
VARIABLE LEVEL Salary_Per_Exp_Year (SCALE).
EXECUTE.

* ============================================================
* COMPUTE NORMALISED PERFORMANCE SCORE (0 to 1 scale)
* Formula: (x - min) / (max - min)
* ============================================================

COMPUTE Performance_Norm = (Performance_Score - 20) / (100 - 20).
VARIABLE LABELS Performance_Norm 'Normalised Performance Score (0-1)'.
VARIABLE LEVEL Performance_Norm (SCALE).
EXECUTE.

* ============================================================
* COMPUTE TOTAL COMPENSATION PROXY
* Salary + Training value (assuming NGN 5,000 per training hour)
* ============================================================

COMPUTE Total_Compensation = Monthly_Salary_NGN + (Training_Hours * 5000).
VARIABLE LABELS Total_Compensation 'Total Compensation Proxy (NGN)'.
VARIABLE LEVEL Total_Compensation (SCALE).
EXECUTE.

* ============================================================
* CONDITIONAL COMPUTE: High Performer Flag
* 1 if Performance_Score >= 80, else 0
* ============================================================

COMPUTE High_Performer = 0.
IF (Performance_Score >= 80) High_Performer = 1.
VARIABLE LABELS High_Performer 'High Performer Flag (Score >= 80)'.
VALUE LABELS High_Performer 0 'Standard' 1 'High Performer'.
VARIABLE LEVEL High_Performer (NOMINAL).
EXECUTE.
```

| Function | What it does | Example |

## Lesson 3.1 — Frequencies Procedure

### What Does Frequencies Do?

The Frequencies procedure is your first look at any dataset. For categorical variables it counts how many cases fall in each category. For scale variables it produces summary statistics and optional histograms.

**Menu path:** `Analyze → Descriptive Statistics → Frequencies`

---

### Running Frequencies on Categorical Variables

Move these into the Variables box:
- `Department`, `Gender`, `Region`, `Education_Level`, `Job_Grade`, `Performance_Rating`, `Promoted`

In **Statistics**, check: None needed for categorical  
In **Charts**, select: **Bar charts** → check **Show chart in viewer**

---

### Reading the Frequency Table Output

The output table has these columns:

| Column | Meaning |
|---|---|
| **Frequency** | Raw count of cases in that category |
| **Percent** | Percentage of ALL cases (including missing) |
| **Valid Percent** | Percentage of non-missing cases — use this one |
| **Cumulative Percent** | Running total percentage (useful for ordinal) |

**Teaching point:** Always use **Valid Percent**, not Percent, when there are missing values.

---

### SPSS Syntax: Frequencies on Categorical Variables

```spss
* ============================================================
* FREQUENCIES — CATEGORICAL VARIABLES
* ============================================================

FREQUENCIES VARIABLES=Department Gender Region Education_Level 
    Job_Grade Performance_Rating Promoted
  /BARCHART FREQ
  /ORDER=ANALYSIS.
```

---

### Running Frequencies on Numeric Variables

For scale variables, add summary statistics:

```spss
* ============================================================
* FREQUENCIES — SCALE VARIABLES WITH STATISTICS
* ============================================================

FREQUENCIES VARIABLES=Age Years_Experience Monthly_Salary_NGN 
    Training_Hours Performance_Score Job_Satisfaction Absenteeism_Days
  /STATISTICS=MEAN MEDIAN MODE STDDEV VARIANCE RANGE 
               MINIMUM MAXIMUM SKEWNESS KURTOSIS
  /HISTOGRAM NORMAL
  /ORDER=ANALYSIS.
```

---

### Interpreting Skewness and Kurtosis

| Statistic | Value | What it means |
|---|---|---|
| **Skewness** | Close to 0 | Roughly symmetric distribution |
| **Skewness** | Positive (> 1) | Right-skewed — tail on the right (e.g., salary data) |
| **Skewness** | Negative (< -1) | Left-skewed — tail on the left |
| **Kurtosis** | Close to 0 | Normal-shaped peak |
| **Kurtosis** | Positive | Peaked (leptokurtic) |
| **Kurtosis** | Negative | Flat (platykurtic) |

**Rule of thumb for normality from skewness/kurtosis:**  
If skewness ÷ its standard error is between -2 and +2, the variable is approximately normal.

**Expected findings in our dataset:**
- `Monthly_Salary_NGN` — likely right-skewed (a few very high earners)
- `Absenteeism_Days` — right-skewed (most people absent few days, a few many days)
- `Performance_Score` — roughly symmetric around 60–65

## Lesson 4.2 — Box Plots for Distribution Comparison

### What is a Box Plot?

A box plot (box-and-whisker plot) shows the distribution of a variable in a compact form. It is especially powerful for **comparing groups**.

**Parts of a box plot:**

```
          ─── Whisker (extends to 1.5× IQR or max, whichever is smaller)
    ┌─────┤
    │     │   ← Upper quartile (Q3, 75th percentile)
    │─────│   ← Median (Q2, 50th percentile) — the line in the box
    │     │   ← Lower quartile (Q1, 25th percentile)
    └─────┤
          ─── Whisker (extends to 1.5× IQR or min, whichever is larger)
       ○      Outlier (beyond 1.5× IQR)
       *      Extreme outlier (beyond 3× IQR)
```

**IQR (Interquartile Range)** = Q3 − Q1. The box represents the middle 50% of the data.

### How to Read Box Plots in Your Lessons

When comparing box plots across groups, teach students to look at:
1. **Median position** — is one group's median higher or lower?
2. **Box size (IQR)** — which group has more spread in the middle 50%?
3. **Whisker length** — which group has more overall range?
4. **Outliers (circles)** — are there unusually high or low values in any group?
5. **Overlap** — do the boxes overlap substantially? If not, groups are likely significantly different.

## Lesson 5.1 — Foundations of Hypothesis Testing

### The Core Logic

Hypothesis testing is a framework for making decisions from data when there is uncertainty. The question is always: **"Could this result have happened by chance alone?"**

### The Two Hypotheses

| Hypothesis | Symbol | What it says |
|---|---|---|
| **Null Hypothesis** | H₀ | No effect, no difference, no relationship — the boring result |
| **Alternative Hypothesis** | H₁ or Ha | There IS an effect, difference, or relationship |

**Example for our dataset:**  
H₀: The mean salary of males and females is equal  
H₁: The mean salary of males and females is NOT equal

---

### The p-value Explained

The **p-value** is the probability of getting a result as extreme as yours (or more extreme) *if the null hypothesis were true*.

**Simple way to explain to students:**

> "The p-value is how surprised you should be if you assumed there was no difference. A very small p-value means the result would be very surprising if H₀ were true — so we reject H₀."

| p-value | Interpretation |
|---|---|
| p > .05 | Not significant — do NOT reject H₀ |
| p ≤ .05 | Significant at the 5% level — reject H₀ |
| p ≤ .01 | Significant at the 1% level — strong evidence |
| p ≤ .001 | Significant at the 0.1% level — very strong evidence |

**The significance level (α)** is typically set at **0.05** (5%). This means you accept a 5% chance of wrongly rejecting H₀ (Type I error).

---

### Type I and Type II Errors

| | H₀ is True | H₀ is False |
|---|---|---|
| **Reject H₀** | Type I Error (False Positive) — probability = α | Correct decision ✓ |
| **Fail to Reject H₀** | Correct decision ✓ | Type II Error (False Negative) — probability = β |

**Memorable analogy:** A pregnancy test.
- Type I Error: Test says positive when you are NOT pregnant (false alarm)
- Type II Error: Test says negative when you ARE pregnant (missed detection)

---

### Parametric vs Non-Parametric Tests

| Type | Assumption | Examples |
|---|---|---|
| **Parametric** | Data is normally distributed | t-test, ANOVA, Pearson r, regression |
| **Non-Parametric** | No normality assumption | Mann-Whitney U, Kruskal-Wallis, Spearman r |

**Rule of thumb:** With n > 30 per group, parametric tests are generally robust to non-normality due to the Central Limit Theorem. With small samples, check normality with Shapiro-Wilk first.


---
## Lesson 5.2 — T-Tests in SPSS

### Three Types of T-Tests

| Test | Question it answers |
|---|---|
| **One-Sample T-Test** | Is the sample mean different from a known/hypothesised value? |
| **Independent Samples T-Test** | Do two independent groups have different means? |
| **Paired Samples T-Test** | Did the same group change between two time points? |

---

### One-Sample T-Test

**Question:** Is the mean monthly salary in our company significantly different from the national average of NGN 200,000?

**Menu path:** `Analyze → Compare Means → One-Sample T Test`

```spss
* ============================================================
* ONE-SAMPLE T-TEST
* H0: Mean salary = 200,000 NGN
* H1: Mean salary ≠ 200,000 NGN
* ============================================================

T-TEST
  /TESTVAL=200000
  /MISSING=ANALYSIS
  /VARIABLES=Monthly_Salary_NGN
  /CRITERIA=CI(.95).
```

**How to read the output:**
1. **Descriptives table:** Check the sample mean and std dev
2. **One-Sample Test table:** Look at **Sig. (2-tailed)**
   - If Sig. < .05 → Reject H₀ — salary is significantly different from 200,000
   - If Sig. > .05 → Fail to reject H₀
3. **t statistic:** The test statistic — how many standard errors the mean is from the test value
4. **95% CI of the difference:** Does the confidence interval include 0?

---

### Independent Samples T-Test

**Question:** Do male and female employees earn significantly different salaries?

**Menu path:** `Analyze → Compare Means → Independent-Samples T Test`

```spss
* ============================================================
* INDEPENDENT SAMPLES T-TEST
* H0: Mean salary (Male) = Mean salary (Female)
* H1: Mean salary (Male) ≠ Mean salary (Female)
* ============================================================

T-TEST GROUPS=Gender('Male' 'Female')
  /MISSING=ANALYSIS
  /VARIABLES=Monthly_Salary_NGN Performance_Score Training_Hours_clean
  /CRITERIA=CI(.95).
```

**How to read the output — TWO tables:**

**Table 1: Group Statistics**
- Shows mean, std dev, std error for each group
- Compare means visually here

**Table 2: Independent Samples Test**
- **Levene's Test for Equality of Variances:**
  - If Sig. > .05 → variances are equal → read the TOP row ("Equal variances assumed")
  - If Sig. < .05 → variances are NOT equal → read the BOTTOM row ("Equal variances not assumed")
- **t, df, Sig. (2-tailed):** The t-test results
- **Mean Difference:** Actual difference between group means
- **95% CI:** If this interval does NOT include 0, the difference is significant

---

### Paired Samples T-Test

**Question:** Did employee performance scores change significantly between Year 1 and Year 2?  
(Not applicable to our cross-sectional dataset, but important to teach conceptually)

```spss
* ============================================================
* PAIRED SAMPLES T-TEST — CONCEPTUAL TEMPLATE
* Use when you have two measurements from the SAME subjects
* ============================================================

T-TEST PAIRS=Performance_Score_Year1 WITH Performance_Score_Year2
  (PAIRED)
  /CRITERIA=CI(.95)
  /MISSING=ANALYSIS.
```

---

### Effect Size — Cohen's d

A significant result does not mean a large or important result. Always report effect size.

**Cohen's d** = (Mean₁ − Mean₂) / Pooled Standard Deviation

| d value | Interpretation |
|---|---|
| 0.2 | Small effect |
| 0.5 | Medium effect |
| 0.8 | Large effect |

SPSS does not compute Cohen's d automatically in the T-Test dialog. Calculate manually:  
`d = Mean Difference / Pooled Std Dev`

Or use the syntax below to compute it from the output values.

## Lesson 5.3 — One-Way ANOVA

### What is ANOVA?

**ANOVA** (Analysis of Variance) compares the means of **three or more groups** simultaneously. If you ran multiple t-tests instead (A vs B, A vs C, B vs C), you would inflate your Type I error rate. ANOVA controls for this.

**Question for our dataset:** Do the five departments have significantly different mean performance scores?

H₀: μ_Sales = μ_Finance = μ_Operations = μ_HR = μ_IT = μ_Marketing = μ_CustomerService  
H₁: At least one department mean is significantly different from the others

**Menu path:** `Analyze → Compare Means → One-Way ANOVA`

---

### SPSS Syntax: One-Way ANOVA with Post-Hoc Tests

```spss
* ============================================================
* ONE-WAY ANOVA
* DV: Performance Score
* Factor: Department (7 groups)
* ============================================================

ONEWAY Performance_Score BY Department
  /STATISTICS DESCRIPTIVES HOMOGENEITY
  /PLOT MEANS
  /MISSING ANALYSIS
  /POSTHOC=TUKEY BONFERRONI ALPHA(0.05).

* ============================================================
* ONE-WAY ANOVA: SALARY BY EDUCATION LEVEL
* ============================================================

ONEWAY Monthly_Salary_NGN BY Education_Numeric
  /STATISTICS DESCRIPTIVES HOMOGENEITY
  /PLOT MEANS
  /MISSING ANALYSIS
  /POSTHOC=TUKEY ALPHA(0.05).

* ============================================================
* ONE-WAY ANOVA: TRAINING HOURS BY JOB GRADE
* ============================================================

ONEWAY Training_Hours_clean BY Job_Grade
  /STATISTICS DESCRIPTIVES HOMOGENEITY
  /MISSING ANALYSIS
  /POSTHOC=TUKEY ALPHA(0.05).
```

---

### Reading the ANOVA Output — Step by Step

**Step 1: Descriptives Table**
- Check sample sizes, means, and std devs per group
- Look for patterns in means visually before testing

**Step 2: Test of Homogeneity of Variances (Levene's Test)**
- If Sig. > .05 → variances are equal across groups → proceed with standard ANOVA
- If Sig. < .05 → variances are NOT equal → use Welch's F (separate procedure) or consider Kruskal-Wallis

**Step 3: ANOVA Table**

| Source | Sum of Squares | df | Mean Square | F | Sig. |
|---|---|---|---|---|---|
| Between Groups | SSB | k-1 | MSB | MSB/MSW | p-value |
| Within Groups | SSW | N-k | MSW | | |
| Total | SST | N-1 | | | |

- **F statistic** = Mean Square Between / Mean Square Within
- If **Sig. (p) < .05** → Reject H₀ → at least one group mean is different

**Step 4: Post-Hoc Tests (Multiple Comparisons Table)**

The ANOVA only tells you *that* at least one group differs — not *which* groups differ. Post-hoc tests tell you the specific pairings.

- **Tukey HSD** — most commonly used, appropriate when group sizes are roughly equal
- **Bonferroni** — more conservative, better when making fewer comparisons
- Look at the **Sig.** column in the Multiple Comparisons table. Pairs with Sig. < .05 are significantly different.

---

### Effect Size for ANOVA — Eta Squared (η²)

η² = SS_Between / SS_Total

| η² | Interpretation |
|---|---|
| .01 | Small effect |
| .06 | Medium effect |
| .14 | Large effect |

Compute manually from the ANOVA table:  
η² = Between Groups SS / Total SS

## Lesson 5.4 — Chi-Square Test of Independence

### What is the Chi-Square Test?

The chi-square (χ²) test of independence tests whether **two categorical variables are related** or whether they are independent of each other.

**Question for our dataset:** Is there a statistically significant association between `Gender` and `Promoted`?

H₀: Gender and Promoted are independent (no association)  
H₁: Gender and Promoted are NOT independent (there is an association)

**Menu path:** `Analyze → Descriptive Statistics → Crosstabs`  
Then click **Statistics** → check **Chi-square** and **Phi and Cramer's V**  
Then click **Cells** → check **Observed**, **Expected**, **Row %**

---

### SPSS Syntax: Chi-Square Test

```spss
* ============================================================
* CHI-SQUARE TEST: GENDER × PROMOTED
* ============================================================

CROSSTABS
  /TABLES=Gender BY Promoted
  /FORMAT=AVALUE TABLES
  /STATISTICS=CHISQ PHI
  /CELLS=COUNT EXPECTED ROW
  /COUNT ROUND CELL.

* ============================================================
* CHI-SQUARE TEST: DEPARTMENT × PROMOTED
* ============================================================

CROSSTABS
  /TABLES=Department BY Promoted
  /FORMAT=AVALUE TABLES
  /STATISTICS=CHISQ CRAMER
  /CELLS=COUNT EXPECTED ROW
  /COUNT ROUND CELL.

* ============================================================
* CHI-SQUARE TEST: EDUCATION LEVEL × PERFORMANCE RATING
* ============================================================

CROSSTABS
  /TABLES=Education_Level BY Performance_Rating
  /FORMAT=AVALUE TABLES
  /STATISTICS=CHISQ CRAMER LAMBDA
  /CELLS=COUNT EXPECTED ROW COLUMN
  /COUNT ROUND CELL.
```

---

### Reading the Chi-Square Output

**Table 1: Crosstabulation**
- Check observed counts and row percentages
- Do the row percentages differ substantially across groups?

**Table 2: Chi-Square Tests**

| Row | What it is |
|---|---|
| **Pearson Chi-Square** | The main test — use this one |
| **Continuity Correction** | Used only for 2×2 tables |
| **Likelihood Ratio** | Alternative — use for small samples |
| **N of Valid Cases** | Sample size used |

- Look at **Asymptotic Significance (2-sided)** for the Pearson Chi-Square row
- If Sig. < .05 → the variables are significantly associated → reject H₀
- **Important:** Check the footnote. If more than 20% of cells have expected count < 5, the chi-square result is unreliable. Consider combining categories.

**Table 3: Symmetric Measures**
- **Phi (φ):** Effect size for 2×2 tables (like Pearson r for categories)
- **Cramer's V:** Effect size for larger tables

| Cramer's V | Effect Size |
|---|---|
| .10 | Small |
| .30 | Medium |
| .50 | Large |

## Lesson 6.1 — Correlation Analysis

### What is Correlation?

Correlation measures the **strength and direction of the linear relationship** between two variables. It does NOT imply causation.

**The correlation coefficient (r):**

| r value | Interpretation |
|---|---|
| +1.0 | Perfect positive relationship |
| +0.7 to +0.9 | Strong positive |
| +0.4 to +0.6 | Moderate positive |
| +0.1 to +0.3 | Weak positive |
| 0 | No linear relationship |
| −0.1 to −0.3 | Weak negative |
| −0.4 to −0.6 | Moderate negative |
| −0.7 to −0.9 | Strong negative |
| −1.0 | Perfect negative relationship |

---

### Pearson vs Spearman

| Coefficient | Use when | Assumption |
|---|---|---|
| **Pearson r** | Both variables are scale and approximately normal | Linear relationship |
| **Spearman ρ** | Ordinal variables, or scale variables that are not normal | Monotonic relationship |

For our dataset:
- Pearson r: `Monthly_Salary_NGN` vs `Years_Experience`
- Spearman ρ: `Education_Numeric` vs `Performance_Score`

## Lesson 6.2 — Simple Linear Regression

### What is Simple Linear Regression?

Simple linear regression models the relationship between **one predictor (X)** and **one outcome (Y)** using a straight line:

```
Y = b₀ + b₁X + ε

Where:
  Y  = Dependent variable (what we are predicting)
  X  = Independent variable (the predictor)
  b₀ = Intercept (value of Y when X = 0)
  b₁ = Slope (how much Y changes for each 1-unit increase in X)
  ε  = Error term (residual — what the model cannot explain)
```

**Our first model:**  
Can we predict monthly salary from years of experience?

Y = Monthly_Salary_NGN  
X = Years_Experience

**Menu path:** `Analyze → Regression → Linear`

---

### SPSS Syntax: Simple Linear Regression

```spss
* ============================================================
* SIMPLE LINEAR REGRESSION
* DV: Monthly Salary
* IV: Years of Experience
* ============================================================

REGRESSION
  /MISSING LISTWISE
  /STATISTICS COEFF OUTS R ANOVA COLLIN TOL CHANGE
  /CRITERIA=PIN(.05) POUT(.10)
  /NOORIGIN
  /DEPENDENT Monthly_Salary_NGN
  /METHOD=ENTER Years_Experience
  /SCATTERPLOT=(*ZRESID ,*ZPRED)
  /RESIDUALS HISTOGRAM(ZRESID) NORMPROB(ZRESID)
  /CASEWISE PLOT(ZRESID) THRESHOLD(3).
```

---

### Reading the Regression Output — Four Key Tables

**Table 1: Variables Entered/Removed**
- Confirms which variables were entered into the model
- Method: Enter means all variables were forced in simultaneously

---

**Table 2: Model Summary**

| Column | What it means |
|---|---|
| **R** | Multiple correlation coefficient — correlation between predicted and actual Y |
| **R²** | Coefficient of determination — % of variance in Y explained by X |
| **Adjusted R²** | R² penalised for number of predictors — use this for model comparison |
| **Std. Error of the Estimate** | Average prediction error in Y's units |

**Example interpretation:**  
R² = 0.45 → "Years of experience explains 45% of the variance in monthly salary."  
Adjusted R² is always lower than R² and more conservative.

---

**Table 3: ANOVA Table**

Tests whether the model as a whole is statistically significant.

| Row | Meaning |
|---|---|
| Regression | Variation in Y explained by X |
| Residual | Variation in Y NOT explained by X |
| Total | Total variation in Y |

- Look at **F** and **Sig.**
- If Sig. < .05 → the model is statistically significant — X is a useful predictor

---

**Table 4: Coefficients Table**

| Column | What it means |
|---|---|
| **B (Unstandardised)** | Raw slope — in original units |
| **Std. Error** | Uncertainty in B |
| **Beta (Standardised)** | Slope in standard deviation units — use for comparing predictors |
| **t** | Test statistic for this coefficient |
| **Sig.** | Is this predictor significant? (< .05 = yes) |
| **95% CI for B** | Range within which the true slope likely falls |

**Writing up the equation:**  
If B₀ = 130,000 and B₁ = 5,200:

`Monthly Salary = 130,000 + 5,200 × Years_Experience`

**Interpretation:** Each additional year of experience is associated with an increase of NGN 5,200 in monthly salary, holding all other factors constant.

---

### Checking Regression Assumptions

Before trusting your regression results, verify these four assumptions:

| Assumption | How to check in SPSS |
|---|---|
| **1. Linearity** | Scatter plot of X vs Y — should look roughly linear |
| **2. Normality of residuals** | Histogram of residuals + Normal P-P plot |
| **3. Homoscedasticity** | Scatter plot of *ZPRED (x-axis) vs *ZRESID (y-axis) — should be random scatter |
| **4. Independence of errors** | Durbin-Watson statistic — should be close to 2.0 |

**Reading the residual plots:**
- **Histogram of residuals:** Should look bell-shaped. If skewed, normality assumption is violated.
- **Normal P-P Plot:** Points should follow the diagonal line closely.
- **Scatter of *ZPRED vs *ZRESID:** Should show a random cloud with no pattern. A funnel shape = heteroscedasticity.

```spss
* ============================================================
* CHECK DURBIN-WATSON FOR AUTOCORRELATION
* ============================================================

REGRESSION
  /MISSING LISTWISE
  /STATISTICS COEFF OUTS R ANOVA DWATSON
  /DEPENDENT Monthly_Salary_NGN
  /METHOD=ENTER Years_Experience.

* Durbin-Watson:
*   < 1.5 = positive autocorrelation (concern)
*   1.5 to 2.5 = acceptable range
*   > 2.5 = negative autocorrelation (concern)
```

## Lesson 6.3 — Multiple Linear Regression

### What is Multiple Linear Regression?

Multiple regression extends simple regression to include **more than one predictor variable**:

```
Y = b₀ + b₁X₁ + b₂X₂ + b₃X₃ + ... + bₖXₖ + ε
```

**Our model:** Predict `Monthly_Salary_NGN` from:
- `Years_Experience` (X₁)
- `Education_Numeric` (X₂)
- `Training_Hours_clean` (X₃)
- `Performance_Score` (X₄)
- `Absenteeism_Days` (X₅)

---

### SPSS Syntax: Multiple Linear Regression

```spss
* ============================================================
* MULTIPLE LINEAR REGRESSION — ENTER ALL PREDICTORS
* DV: Monthly Salary (NGN)
* IVs: Experience, Education, Training, Performance, Absenteeism
* ============================================================

REGRESSION
  /MISSING LISTWISE
  /STATISTICS COEFF OUTS R ANOVA COLLIN TOL CHANGE ZPP
  /CRITERIA=PIN(.05) POUT(.10)
  /NOORIGIN
  /DEPENDENT Monthly_Salary_NGN
  /METHOD=ENTER Years_Experience Education_Numeric 
    Training_Hours_clean Performance_Score Absenteeism_Days
  /SCATTERPLOT=(*ZRESID ,*ZPRED)
  /RESIDUALS HISTOGRAM(ZRESID) NORMPROB(ZRESID).

* ============================================================
* HIERARCHICAL REGRESSION — ADD PREDICTORS IN BLOCKS
* Block 1: Demographics (Age, Experience, Education)
* Block 2: Performance (Training, Score, Satisfaction)
* ============================================================

REGRESSION
  /MISSING LISTWISE
  /STATISTICS COEFF OUTS R ANOVA CHANGE
  /CRITERIA=PIN(.05) POUT(.10)
  /NOORIGIN
  /DEPENDENT Monthly_Salary_NGN
  /METHOD=ENTER Years_Experience Education_Numeric Age
  /METHOD=ENTER Training_Hours_clean Performance_Score Job_Satisfaction_clean.
```

---

### New Concepts in Multiple Regression Output

**Adjusted R² and R² Change:**  
In hierarchical regression, look at the **R² Change** column to see how much additional variance each new block of predictors explains. If Block 2 adds ΔR² = 0.08, the performance variables explain an extra 8% of salary variation beyond demographics alone.

---

### Checking for Multicollinearity

**Multicollinearity** occurs when two or more predictors are very highly correlated with each other. This inflates standard errors and makes coefficients unstable and hard to interpret.

**How to detect it:**

| Statistic | Where to find it | What it means |
|---|---|---|
| **Tolerance** | Coefficients table | < 0.10 = serious multicollinearity |
| **VIF (Variance Inflation Factor)** | Coefficients table | > 10 = serious (> 5 = moderate concern) |
| **Condition Index** | Collinearity Diagnostics table | > 30 = serious multicollinearity |

**SPSS Syntax to request collinearity diagnostics:**
```spss
REGRESSION
  /MISSING LISTWISE
  /STATISTICS COEFF OUTS R ANOVA COLLIN TOL
  /DEPENDENT Monthly_Salary_NGN
  /METHOD=ENTER Years_Experience Education_Numeric 
    Training_Hours_clean Performance_Score Absenteeism_Days.
```

---

### Interpreting the Multiple Regression Coefficients Table

For each predictor:
- **B (Unstandardised coefficient):** Change in Y for a 1-unit increase in this predictor, holding all other predictors constant
- **Beta (Standardised coefficient):** Which predictor has the strongest effect? The predictor with the largest absolute Beta is the most important predictor
- **Sig.:** Is this predictor making a statistically significant contribution to the model?

**Example write-up:**  
*"A multiple regression was conducted to predict monthly salary from years of experience, education level, training hours, performance score, and absenteeism days. The model was statistically significant, F(5, 274) = 47.3, p < .001, accounting for 46% of the variance in monthly salary (Adjusted R² = .46). Years of experience was the strongest predictor (β = .42, p < .001), followed by education level (β = .29, p < .001). Absenteeism days did not significantly predict salary (β = −.04, p = .38)."*

## Lesson 6.4 — Binary Logistic Regression

### What is Logistic Regression?

Logistic regression predicts the **probability of a binary outcome** (0 or 1) based on one or more predictor variables.

Unlike linear regression (which predicts a continuous number), logistic regression predicts:
- Was the employee promoted? (0 = No, 1 = Yes)
- Will a customer churn? (0 = No, 1 = Yes)
- Does a patient have a disease? (0 = No, 1 = Yes)

**The logistic model:**
```
log(p / 1-p) = b₀ + b₁X₁ + b₂X₂ + ... + bₖXₖ

Where:
  p         = Probability of the outcome (Y=1)
  p / (1-p) = Odds of the outcome
  log(p/1-p)= Log-odds (logit)
```

The model predicts log-odds, which we convert to an **odds ratio** by exponentiating: **Exp(B)**.

---

### Our Model

**Outcome:** `Promoted` (0 = Not promoted, 1 = Promoted)  
**Predictors:** `Performance_Score`, `Years_Experience`, `Education_Numeric`, `Training_Hours_clean`, `Job_Satisfaction_clean`

**Menu path:** `Analyze → Regression → Binary Logistic`

---

### SPSS Syntax: Binary Logistic Regression

```spss
* ============================================================
* BINARY LOGISTIC REGRESSION
* DV: Promoted (0=No, 1=Yes)
* IVs: Performance, Experience, Education, Training, Satisfaction
* ============================================================

LOGISTIC REGRESSION VARIABLES Promoted
  /METHOD=ENTER Performance_Score Years_Experience 
    Education_Numeric Training_Hours_clean Job_Satisfaction_clean
  /CONTRAST (Promoted)=Indicator(1)
  /SAVE=PRED COOK LEVER ZRESID DFBETA
  /CLASSPLOT
  /CASEWISE OUTLIER(2)
  /PRINT=GOODFIT ITER(1) CI(95)
  /CRITERIA=PIN(.05) POUT(.10) ITERATE(20) CUT(.5).
```

---

### Reading the Logistic Regression Output — Table by Table

**Table 1: Case Processing Summary**
- Confirms sample size and the coding of the dependent variable (which category is "event")

**Table 2: Block 0 — Beginning Block**
- **Classification Table:** How well does the model classify cases using NO predictors? If 70% of employees were not promoted, always predicting "Not Promoted" gives 70% accuracy — this is the baseline.

**Table 3: Omnibus Tests of Model Coefficients**
- Chi-square test: Is the model with predictors significantly better than the null (no predictor) model?
- If Sig. < .05 → the overall model is significant

**Table 4: Model Summary**

| Statistic | Meaning |
|---|---|
| **-2 Log Likelihood** | Lower = better fit |
| **Cox & Snell R²** | Pseudo R² — approximate variance explained |
| **Nagelkerke R²** | Adjusted pseudo R² — ranges 0 to 1, preferred |

**Table 5: Hosmer and Lemeshow Test**
- Tests whether predicted probabilities match observed frequencies
- **You want Sig. > .05** here (this is the one test where non-significance is desirable)
- If Sig. < .05 → the model does NOT fit the data well

**Table 6: Variables in the Equation — THE KEY TABLE**

| Column | Meaning |
|---|---|
| **B** | Log-odds coefficient — difficult to interpret directly |
| **S.E.** | Standard error of B |
| **Wald** | Test statistic (z-squared) |
| **Sig.** | Is this predictor significant? |
| **Exp(B)** | **Odds Ratio — the key statistic to interpret** |
| **95% CI for Exp(B)** | Confidence interval for the odds ratio |

---

### Interpreting Odds Ratios — Exp(B)

The **odds ratio** tells you how the odds of promotion change for a 1-unit increase in the predictor.

| Exp(B) | Interpretation |
|---|---|
| > 1 | Predictor increases odds of promotion |
| = 1 | Predictor has no effect on odds |
| < 1 | Predictor decreases odds of promotion |

**Example interpretations:**
- Exp(B) for `Performance_Score` = 1.08 → "Each 1-point increase in performance score increases the odds of promotion by 8%"
- Exp(B) for `Years_Experience` = 1.12 → "Each additional year of experience increases the odds of promotion by 12%"
- Exp(B) for `Absenteeism_Days` = 0.90 → "Each additional absence day decreases the odds of promotion by 10%"

**95% Confidence Interval:** If the CI for Exp(B) **includes 1.0**, the predictor is NOT statistically significant.

---

### Classification Table

| | Predicted: No | Predicted: Yes | % Correct |
|---|---|---|---|
| **Observed: No** | True Negatives (TN) | False Positives (FP) | Specificity |
| **Observed: Yes** | False Negatives (FN) | True Positives (TP) | Sensitivity |

- **Overall % Correct:** What % of all cases did the model classify correctly?
- **Cut value = 0.5:** SPSS predicts "Promoted" if predicted probability ≥ 0.50

---

### SPSS Syntax: Save Predicted Probabilities and Classify

```spss
* ============================================================
* SAVE PREDICTED PROBABILITIES TO DATA FILE
* After logistic regression, examine who had highest promotion probability
* ============================================================

LOGISTIC REGRESSION VARIABLES Promoted
  /METHOD=ENTER Performance_Score Years_Experience 
    Education_Numeric Training_Hours_clean Job_Satisfaction_clean
  /SAVE=PRED(Prob_Promoted).

* SORT BY PREDICTED PROBABILITY TO SEE TOP CANDIDATES
SORT CASES BY Prob_Promoted (DESCENDING).

* EXAMINE TOP 20 MOST LIKELY TO BE PROMOTED
FREQUENCIES VARIABLES=Department Gender Education_Level
  /ORDER=ANALYSIS.
```
