## SECTION 02 - Missing Values (Slides 6-9)

```python
# Which rows have any null?
rows_with_nulls = df[df.isnull().any(axis=1)]
print(f'Rows with at least one missing value: {len(rows_with_nulls)}')
rows_with_nulls.head(10)
```

```python
# Note: empty strings in CSV are loaded as NaN by pandas automatically
# But let's also replace any literal empty string that slipped through
df_clean = df_clean.replace('', np.nan)
print('After replacing empty strings with NaN:')
print(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])
```

```python
# Strategy 1 - Fill payment_mode with constant 'Unknown'
df_clean['payment_mode'] = df_clean['payment_mode'].fillna('Unknown')

# Strategy 2 - Fill category with mode (most frequent category)
mode_cat = df_clean['category'].mode()[0]
df_clean['category'] = df_clean['category'].fillna(mode_cat)

print(f'payment_mode nulls: {df_clean["payment_mode"].isnull().sum()}')
print(f'category nulls:     {df_clean["category"].isnull().sum()}')
```

```python
# Strategy 3 - Fill gender missing values with mode
# But first we need to normalise M/F -> Male/Female (see Section 05)
# For now fill missing with 'Unknown'
df_clean['gender'] = df_clean['gender'].fillna('Unknown')
print(f'gender nulls: {df_clean["gender"].isnull().sum()}')
```

```python
# Strategy 4 - Group-wise median fill for price (BEST PRACTICE)
# Fill missing price with the median price for that category
df_clean['price'] = df_clean.groupby('category')['price'].transform(
    lambda x: x.fillna(x.median())
)
print(f'price nulls after group-wise fill: {df_clean["price"].isnull().sum()}')
```

```python
# Strategy 5 - Group-wise median fill for qty
df_clean['qty']      = df_clean.groupby('product')['qty'].transform(
    lambda x: x.fillna(x.median())
)
df_clean['quantity'] = df_clean['qty'].copy()
print(f'qty nulls after fill: {df_clean["qty"].isnull().sum()}')
```

```python
# Strategy 6 - Fill rating with median rating (global)
# But first remove the outlier values (99, 0, -1, 150) from median calculation
valid_ratings = df_clean['rating'][df_clean['rating'].between(1, 5)]
median_rating = valid_ratings.median()
df_clean['rating'] = df_clean['rating'].fillna(median_rating)
print(f'rating nulls after fill: {df_clean["rating"].isnull().sum()}')
```

```python
# Verify - all target columns now have no missing values
print('Remaining nulls per column:')
remaining = df_clean.isnull().sum()
print(remaining[remaining > 0])
print(f'Total nulls remaining: {df_clean.isnull().sum().sum()}')
```

## SECTION 03 - Duplicate Rows (Slides 10-11)

```python
# Count duplicates before
print(f'Before: {len(df_clean):,} rows')
print(f'Exact duplicate rows:    {df_clean.duplicated().sum()}')
print(f'Duplicate order_ids:     {df_clean.duplicated(subset=["order_id"]).sum()}')

# View the duplicated rows
df_clean[df_clean.duplicated(keep=False)].sort_values('order_id').head(10)
```

```python
# Remove duplicates - keep first occurrence
df_clean = df_clean.drop_duplicates()
df_clean = df_clean.reset_index(drop=True)

print(f'After:  {len(df_clean):,} rows')
print(f'Removed: {4248 - len(df_clean)} rows')
print(f'Duplicates remaining: {df_clean.duplicated().sum()}')
```

## SECTION 04 - Data Types (Slides 12-13)

```python
# Check current dtypes
print(df_clean.dtypes)
print()
# Problems:
# - price is object because of 'NGN 45,000' strings - needs cleaning first
# - date is object - needs to_datetime()
# - qty/quantity might still be object
```

```python
# Fix price column
# Step 1: view the messy prices
messy_prices = df_clean[df_clean['price'].astype(str).str.contains('NGN', na=False)]
print(f'Rows with NGN in price: {len(messy_prices)}')
print(messy_prices['price'].head(5).tolist())
```

```python
# Step 2: clean the price string - remove NGN, commas, spaces
df_clean['price'] = (
    df_clean['price']
    .astype(str)
    .str.replace('NGN', '', regex=False)
    .str.replace(',',   '', regex=False)
    .str.strip()
)

# Step 3: safe conversion - errors='coerce' turns unparseable values into NaN
df_clean['price'] = pd.to_numeric(df_clean['price'], errors='coerce')

print(f'price dtype:  {df_clean["price"].dtype}')
print(f'price nulls after conversion: {df_clean["price"].isnull().sum()}')
```

```python
# Safe convert other numeric columns
df_clean['qty']        = pd.to_numeric(df_clean['qty'],        errors='coerce')
df_clean['quantity']   = pd.to_numeric(df_clean['quantity'],   errors='coerce')
df_clean['rating']     = pd.to_numeric(df_clean['rating'],     errors='coerce')
df_clean['margin_pct'] = pd.to_numeric(df_clean['margin_pct'], errors='coerce')
df_clean['age']        = pd.to_numeric(df_clean['age'],        errors='coerce')
df_clean['SALES_AMT']  = pd.to_numeric(df_clean['SALES_AMT'],  errors='coerce')
df_clean['sales_amount']= pd.to_numeric(df_clean['sales_amount'],errors='coerce')
df_clean['unit_cost']  = pd.to_numeric(df_clean['unit_cost'],  errors='coerce')
df_clean['profit']     = pd.to_numeric(df_clean['profit'],     errors='coerce')

print('Numeric dtypes confirmed:')
print(df_clean[['price','qty','rating','age','margin_pct']].dtypes)
```

```python
# Convert date column - errors='coerce' turns invalid dates into NaT
print('Invalid dates before conversion:')
bad_dates = df_clean[~df_clean['date'].str.match(r'^\d{4}-\d{2}-\d{2}$', na=False)]
print(bad_dates['date'].unique())

df_clean['date'] = pd.to_datetime(df_clean['date'], errors='coerce')

print(f'\ndate dtype: {df_clean["date"].dtype}')
print(f'NaT (failed conversion): {df_clean["date"].isnull().sum()}')
```

```python
# Extract useful date parts
df_clean['year']        = df_clean['date'].dt.year
df_clean['month']       = df_clean['date'].dt.month
df_clean['month_name']  = df_clean['date'].dt.month_name()
df_clean['day_of_week'] = df_clean['date'].dt.day_name()
df_clean['quarter']     = df_clean['date'].dt.quarter

df_clean[['date','year','month','month_name','day_of_week','quarter']].head(5)
```

```python
# Convert low-cardinality text to 'category' dtype (saves memory)
df_clean['region']       = df_clean['region'].astype('object')   # will fix case first
df_clean['DOB']          = pd.to_datetime(df_clean['DOB'], errors='coerce')
df_clean['Reg?']         = df_clean['Reg?'].map({'Yes': True, 'No': False})

print('Types after conversion:')
print(df_clean[['region','Reg?']].dtypes)
```

## SECTION 05 - String Cleaning (Slides 14-15)

```python
# See the dirty region values
print('All unique region values before cleaning:')
print(df_clean['region'].value_counts().to_string())
```

```python
# Fix region: strip whitespace then title case
df_clean['region'] = df_clean['region'].str.strip().str.title()

print('After cleaning:')
print(df_clean['region'].value_counts())
```

```python
# Fix category
df_clean['category'] = df_clean['category'].str.strip().str.title()
print('Categories:', df_clean['category'].unique())
```

```python
# Fix gender - map M/F -> Male/Female
print('Gender before:')
print(df_clean['gender'].value_counts())

gender_map = {
    'M':      'Male',
    'F':      'Female',
    'Male':   'Male',
    'Female': 'Female',
    'Unknown':'Unknown'
}
df_clean['gender'] = df_clean['gender'].map(gender_map).fillna('Unknown')

print('\nGender after:')
print(df_clean['gender'].value_counts())
```

```python
# Fix product column
df_clean['product'] = df_clean['product'].str.strip().str.title()

# Fix channel
df_clean['channel'] = df_clean['channel'].str.strip().str.title()

print('String columns cleaned successfully.')
```

## SECTION 07 - Filtering Invalid Rows (Slides 18-19)

```python
print(f'Before filtering: {len(df_clean):,} rows')

# See negative prices
print(f'Negative prices: {(df_clean["price"] < 0).sum()}')
print(f'Zero qty rows:   {(df_clean["qty"] == 0).sum()}')
```

## SECTION 08 - Outlier Detection (Slides 20-21)

```python
# IQR method on price
Q1  = df_clean['price'].quantile(0.25)
Q3  = df_clean['price'].quantile(0.75)
IQR = Q3 - Q1

lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

print(f'Q1:           {Q1:,.0f}')
print(f'Q3:           {Q3:,.0f}')
print(f'IQR:          {IQR:,.0f}')
print(f'Lower fence:  {lower_fence:,.0f}')
print(f'Upper fence:  {upper_fence:,.0f}')

outliers = df_clean[(df_clean['price'] < lower_fence) | (df_clean['price'] > upper_fence)]
print(f'\nOutliers detected: {len(outliers)}')
print(outliers[['order_id','product','price']].head(10))
```

```python
# Option 1: Remove outliers
df_no_outliers = df_clean[
    (df_clean['price'] >= lower_fence) &
    (df_clean['price'] <= upper_fence)
]
print(f'After removing outliers: {len(df_no_outliers):,} rows')
```

```python
# Option 2: Cap outliers (winsorize) - safer approach
df_clean['price'] = df_clean['price'].clip(
    lower=lower_fence,
    upper=upper_fence
)
print('Price after capping:')
print(df_clean['price'].describe())
```

```python
# Z-score method on margin_pct
from scipy import stats

z_scores = np.abs(stats.zscore(df_clean['margin_pct'].dropna()))
outlier_mask = z_scores > 3
print(f'Margin_pct outliers by Z-score: {outlier_mask.sum()}')
```

## SECTION 09 - groupby and agg() (Slides 22-23)

```python
# Recompute sales_amount with clean price and qty
df_clean['sales_amount'] = df_clean['price'] * df_clean['qty']

# Named aggregation - best practice
region_summary = df_clean.groupby('region').agg(
    total_revenue  = ('sales_amount', 'sum'),
    avg_price      = ('price',        'mean'),
    order_count    = ('order_id',     'count'),
    avg_rating     = ('rating',       'mean'),
    unique_products= ('product',      'nunique')
).reset_index()

region_summary['total_revenue'] = region_summary['total_revenue'].round(0)
region_summary['avg_price']     = region_summary['avg_price'].round(0)
region_summary['avg_rating']    = region_summary['avg_rating'].round(2)
region_summary.sort_values('total_revenue', ascending=False)
```

```python
# Multi-key groupby
cat_region = df_clean.groupby(['region', 'category']).agg(
    total_sales = ('sales_amount', 'sum'),
    orders      = ('order_id',     'count')
).reset_index()
cat_region.head(10)
```

```python
# Add group mean column using transform (keeps original shape)
df_clean['region_avg_price'] = df_clean.groupby('region')['price'].transform('mean')
print('Added region_avg_price column:')
df_clean[['order_id','region','price','region_avg_price']].head(8)
```

```python
# Pivot table - cross-tabulate region vs category
pivot = pd.pivot_table(
    df_clean,
    index='region',
    columns='category',
    values='sales_amount',
    aggfunc='sum',
    fill_value=0
)
pivot.round(0)
```

## SECTION 11 - apply(), map(), and applymap() (Slides 26-27)

```python
# map() - classify gender using a dictionary (already done above)
# map() on channel column
channel_map = {
    'In-Store':  'Offline',
    'Wholesale': 'Offline',
    'Online':    'Online',
    'B2B':       'Online'
}
df_clean['channel_type'] = df_clean['channel'].map(channel_map)
print(df_clean['channel_type'].value_counts())
```

```python
# apply() with a custom function to classify margin
def classify_margin(margin):
    if pd.isna(margin):  return 'Unknown'
    if margin >= 30:     return 'High'
    elif margin >= 15:   return 'Medium'
    else:                return 'Low'

df_clean['margin_class'] = df_clean['margin_pct'].apply(classify_margin)
print(df_clean['margin_class'].value_counts())
```

```python
# apply() with axis=1 (across rows) - create full name column
def full_name(row):
    first = str(row['first_name']).strip()
    last  = str(row['last_name']).strip()
    return first + ' ' + last

df_clean['full_name'] = df_clean.apply(full_name, axis=1)
print(df_clean['full_name'].head(5).tolist())
```

```python
# np.where() - fast if/else
df_clean['status'] = np.where(df_clean['sales_amount'] > 500_000, 'High Value', 'Standard')
print(df_clean['status'].value_counts())
```

```python
# pd.cut() - bin age into groups
df_clean['age_group'] = pd.cut(
    df_clean['age'],
    bins=[0, 25, 35, 50, 100],
    labels=['18-25', '26-35', '36-50', '51+']
)
print(df_clean['age_group'].value_counts().sort_index())
```

## SECTION 12 - Advanced Techniques (Slides 28-29)

```python
# explode - the tags column has comma-separated values
print('Tags before explode:')
print(df_clean['tags'].head(5).tolist())

df_tags = df_clean[['order_id','tags']].copy()
df_tags['tags'] = df_tags['tags'].str.split(',')
df_tags = df_tags.explode('tags')
df_tags['tags'] = df_tags['tags'].str.strip()

print(f'\nAfter explode: {len(df_tags):,} rows')
print('Most common tags:')
print(df_tags['tags'].value_counts().head(8))
```

```python
# melt - wide to long format (quarterly data example)
# Create a wide pivot first
quarterly = df_clean.groupby(['region','quarter'])['sales_amount'].sum().reset_index()
wide = quarterly.pivot(index='region', columns='quarter', values='sales_amount').reset_index()
wide.columns = ['region','Q1','Q2','Q3','Q4']
wide = wide.fillna(0)
print('Wide format:')
print(wide)

# Now melt back to long format
long = pd.melt(
    wide,
    id_vars=['region'],
    value_vars=['Q1','Q2','Q3','Q4'],
    var_name='quarter',
    value_name='sales'
)
print('\nLong format (first 10):')
long.head(10)
```

```python
# str.extract with regex - extract year from order_id if it had one,
# or extract numeric part from customer_id
df_clean['cust_num'] = df_clean['customer_id'].str.extract(r'(\d+)')
print(df_clean[['customer_id','cust_num']].head(5))
```

```python
# pd.get_dummies - one-hot encode gender (for ML)
gender_dummies = pd.get_dummies(df_clean['gender'], prefix='gender')
print('One-hot encoded gender:')
print(gender_dummies.head(5))
print(f'Columns: {gender_dummies.columns.tolist()}')
```
