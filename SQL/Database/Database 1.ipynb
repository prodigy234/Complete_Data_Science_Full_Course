1. Create the Database
CREATE DATABASE business_analytics_db;

Use the database:

USE business_analytics_db;

2. Create Customers Table

Stores information about customers.

CREATE TABLE customers (
    customer_id INT AUTO_INCREMENT PRIMARY KEY,
    first_name VARCHAR(50) NOT NULL,
    last_name VARCHAR(50) NOT NULL,
    email VARCHAR(100) UNIQUE,
    phone VARCHAR(20),
    city VARCHAR(50),
    country VARCHAR(50),
    registration_date DATE
);

3. Create Categories Table

Groups products into categories.

CREATE TABLE categories (
    category_id INT AUTO_INCREMENT PRIMARY KEY,
    category_name VARCHAR(100) NOT NULL,
    description TEXT
);

4. Create Products Table

Each product belongs to a category.

CREATE TABLE products (
    product_id INT AUTO_INCREMENT PRIMARY KEY,
    product_name VARCHAR(100) NOT NULL,
    category_id INT,
    price DECIMAL(10,2),
    stock_quantity INT,
    created_at DATETIME DEFAULT CURRENT_TIMESTAMP,
    FOREIGN KEY (category_id) REFERENCES categories(category_id)
);

5. Create Employees Table

Stores employee details.

CREATE TABLE employees (
    employee_id INT AUTO_INCREMENT PRIMARY KEY,
    first_name VARCHAR(50),
    last_name VARCHAR(50),
    email VARCHAR(100),
    job_title VARCHAR(100),
    hire_date DATE,
    salary DECIMAL(10,2)
);

6. Create Orders Table

Each order is linked to a customer and an employee.

CREATE TABLE orders (
    order_id INT AUTO_INCREMENT PRIMARY KEY,
    customer_id INT,
    employee_id INT,
    order_date DATETIME DEFAULT CURRENT_TIMESTAMP,
    order_status VARCHAR(50),    
    FOREIGN KEY (customer_id) REFERENCES customers(customer_id),    
    FOREIGN KEY (employee_id) REFERENCES employees(employee_id)
);

7. Create Order Details Table

This is a junction table connecting orders and products.

CREATE TABLE order_details (
    order_detail_id INT AUTO_INCREMENT PRIMARY KEY,
    order_id INT,
    product_id INT,
    quantity INT,
    unit_price DECIMAL(10,2),
    discount DECIMAL(5,2) DEFAULT 0,    
    FOREIGN KEY (order_id) REFERENCES orders(order_id),    
    FOREIGN KEY (product_id) REFERENCES products(product_id)
);

8. Create Payments Table
CREATE TABLE payments (
    payment_id INT AUTO_INCREMENT PRIMARY KEY,
    order_id INT,
    payment_date DATETIME,
    payment_method VARCHAR(50),
    amount DECIMAL(10,2),    
    FOREIGN KEY (order_id) REFERENCES orders(order_id)
);

9. Create Suppliers Table
CREATE TABLE suppliers (
    supplier_id INT AUTO_INCREMENT PRIMARY KEY,
    supplier_name VARCHAR(100),
    contact_name VARCHAR(100),
    phone VARCHAR(20),
    city VARCHAR(50),
    country VARCHAR(50)
);

10. Create Product Suppliers Table

Many-to-many relationship.

CREATE TABLE product_suppliers (
    product_supplier_id INT AUTO_INCREMENT PRIMARY KEY,
    product_id INT,
    supplier_id INT,    
    FOREIGN KEY (product_id) REFERENCES products(product_id),
    FOREIGN KEY (supplier_id) REFERENCES suppliers(supplier_id)
);

1. INNER JOIN Questions

Question 1

Display all customers and the orders they placed.

Answer

SELECT c.first_name, c.last_name, o.order_id, o.order_date
FROM customers c
INNER JOIN orders o
ON c.customer_id = o.customer_id;

Question 2

Show the products that were ordered along with their quantities.

Answer

SELECT p.product_name, od.quantity
FROM products p
INNER JOIN order_details od
ON p.product_id = od.product_id;

Question 3

Show all orders with the employee who handled them.

Answer

SELECT o.order_id, e.first_name, e.last_name
FROM orders o
INNER JOIN employees e
ON o.employee_id = e.employee_id;

2. LEFT JOIN Questions

Question 4

Show all customers and their orders (including customers who have not placed orders).

Answer

SELECT c.first_name, c.last_name, o.order_id
FROM customers c
LEFT JOIN orders o
ON c.customer_id = o.customer_id;

Question 5

List all products and the quantities sold (including products never sold).

Answer

SELECT p.product_name, od.quantity
FROM products p
LEFT JOIN order_details od
ON p.product_id = od.product_id;

3. RIGHT JOIN Questions

Question 6

Show all orders and the customers who placed them.

Answer

SELECT c.first_name, c.last_name, o.order_id
FROM customers c
RIGHT JOIN orders o
ON c.customer_id = o.customer_id;

4. MULTIPLE JOIN Questions

Question 7

Display the customer's name, product name, and quantity purchased.

Answer

SELECT c.first_name, c.last_name, p.product_name, od.quantity
FROM customers c
JOIN orders o
ON c.customer_id = o.customer_id
JOIN order_details od
ON o.order_id = od.order_id
JOIN products p
ON od.product_id = p.product_id;

Question 8

Display the category of products ordered by customers.

Answer

SELECT c.first_name, p.product_name, cat.category_name
FROM customers c
JOIN orders o
ON c.customer_id = o.customer_id
JOIN order_details od
ON o.order_id = od.order_id
JOIN products p
ON od.product_id = p.product_id
JOIN categories cat
ON p.category_id = cat.category_id;

5. JOIN with Aggregation

Question 9

Find the total quantity sold for each product.

Answer

SELECT p.product_name, SUM(od.quantity) AS total_quantity
FROM products p
JOIN order_details od
ON p.product_id = od.product_id
GROUP BY p.product_name;

Question 10

Find the total amount spent by each customer.

Answer

SELECT c.first_name, c.last_name,
SUM(od.quantity * od.unit_price) AS total_spent
FROM customers c
JOIN orders o
ON c.customer_id = o.customer_id
JOIN order_details od
ON o.order_id = od.order_id
GROUP BY c.customer_id;

1. Second Best Performing Product

Question

Find the second highest selling product by quantity.

Answer (Using LIMIT & OFFSET)

SELECT p.product_name, SUM(od.quantity) AS total_sold
FROM products p
JOIN order_details od ON p.product_id = od.product_id
GROUP BY p.product_id
ORDER BY total_sold DESC
LIMIT 1 OFFSET 1;

Alternative (Using Subquery – Interview Style)

SELECT product_name, total_sold
FROM (
    SELECT p.product_name, SUM(od.quantity) AS total_sold
    FROM products p
    JOIN order_details od ON p.product_id = od.product_id
    GROUP BY p.product_id
) ranked_products
ORDER BY total_sold DESC
LIMIT 1 OFFSET 1;

19. Running Total Revenue (Advanced 🔥)

SUM(revenue) OVER (ORDER BY order_day) AS running_total
FROM (
    SELECT DATE(o.order_date) AS order_day,

    FROM orders o
    JOIN order_details od ON o.order_id = od.order_id
    GROUP BY order_day
) t;

20. Rank Products by Sales (Window Function)

SELECT product_name, total_sold,
RANK() OVER (ORDER BY total_sold DESC) AS ranking
FROM (
    SELECT p.product_name,
    SUM(od.quantity) AS total_sold
    FROM products p
    JOIN order_details od ON p.product_id = od.product_id
    GROUP BY p.product_id
) t;

Basic Query (Today’s Total Sales)

SELECT SUM(od.quantity * od.unit_price) AS today_sales
FROM orders o
JOIN order_details od ON o.order_id = od.order_id
WHERE DATE(o.order_date) = CURDATE();

Explanation

CURDATE() → returns today’s date

DATE(o.order_date) → removes time (so it matches today correctly)

SUM(quantity * unit_price) → calculates total revenue

If You Want More Details (Breakdown by Product)

SELECT p.product_name,
SUM(od.quantity * od.unit_price) AS today_sales
FROM orders o
JOIN order_details od ON o.order_id = od.order_id
JOIN products p ON od.product_id = p.product_id
WHERE DATE(o.order_date) = CURDATE()
GROUP BY p.product_name
ORDER BY today_sales DESC;

If You Want Number of Orders Today

SELECT COUNT(*) AS total_orders_today
FROM orders
WHERE DATE(order_date) = CURDATE();

If You Want Today’s Sales + Payment Info

SELECT p.payment_method,
SUM(p.amount) AS total_sales
FROM payments p
WHERE DATE(p.payment_date) = CURDATE()
