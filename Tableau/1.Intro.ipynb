# TABLEAU

## INTRODUCTION TO TABLEAU

### 1. What is Tableau?

Tableau is a powerful data visualization and business intelligence (BI) tool used to:

- Analyze data

- Create interactive dashboards

- Share insights visually

**It enables users to turn raw data into actionable insights without heavy coding.**

### 2. Why Tableau? (Position it vs Excel & Power BI)

| Tool     | Strength                                  | Limitation                      |
| -------- | ----------------------------------------- | ------------------------------- |
| Excel    | Easy, flexible                            | Weak for large-scale dashboards |
| Power BI | Strong modeling, DAX                      | Slight learning curve           |
| Tableau  | Best-in-class visualization & exploration | Weaker data modeling            |

Key takeaway:

- Excel = Data handling

- Power BI = Data modeling + reporting

- Tableau = Data storytelling + exploration

### 4. Key Concepts in Tableau

4.1 Data Visualization Philosophy

Tableau is built on:

- Visual analytics

- Drag-and-drop interface

- Real-time exploration

Unlike Power BI:

- Tableau encourages exploration first

- Power BI encourages modeling first

4.2 Types of Data in Tableau

Dimensions vs Measures

| Type      | Description               | Examples                |
| --------- | ------------------------- | ----------------------- |
| Dimension | Qualitative (categorical) | Name, Country, Product  |
| Measure   | Quantitative (numeric)    | Sales, Profit, Quantity |

Rule:

- Dimensions = slice data

- Measures = aggregate data

4.3 Discrete vs Continuous

| Type       | Appearance     | Example  |
| ---------- | -------------- | -------- |
| Discrete   | Blue (headers) | Category |
| Continuous | Green (axis)   | Sales    |

4.4 Aggregation

Tableau automatically aggregates data:

- SUM

- AVG

- COUNT

- MIN / MAX

Example:

SUM(Sales)

AVG(Profit)

### 7. Data Handling in Tableau

7.1 Live vs Extract

| Type    | Meaning              |
| ------- | -------------------- |
| Live    | Real-time connection |
| Extract | Snapshot of data     |

Tip:

- Use Extract for performance

- Use Live for real-time dashboards

7.2 Joins & Relationships

Tableau supports:

- Inner Join

- Left Join

- Right Join

- Full Outer Join

Difference from Power BI:

Tableau uses Relationships (logical layer) more flexibly

This controls how data appears:

| Feature | Function          |
| ------- | ----------------- |

| Label   | Shows values      |
| Detail  | Adds granularity  |
| Tooltip | Hover info        |

### 10. Filters in Tableau

Types:

- Extract Filter

- Data Source Filter

- Context Filter

- Dimension Filter

- Measure Filter

### 11. Calculated Fields

Like DAX in Power BI but simpler.

Example:

Profit Ratio = SUM(Profit) / SUM(Sales)

Types:

- Basic calculations

- String functions

- Date functions

- Logical statements

### 12. Table Calculations

Used for:

- Running totals

- Percent of total

- Moving averages

Example:

RUNNING_SUM(SUM(Sales))

### 13. Level of Detail (LOD Expressions)

Very important advanced concept.

Types:

- FIXED

- INCLUDE

- EXCLUDE

Example:

{FIXED Region : SUM(Sales)}

Difference from Power BI:

- Equivalent to controlling aggregation context

### 14. Dashboards in Tableau

### 17. Best Practices

Design:

- Avoid clutter

- Use consistent colors

- Highlight key insights

Performance:

- Optimize calculations

Analytics:

- Ask business questions first

Don’t just build charts
