# DAX

## WHAT IS DAX?

 Definition:

DAX (Data Analysis Expressions) is the formula language in Power BI that allows us to create calculations like total sales, profit, and growth. It works dynamically based on filters in the report.

DAX is a formula language used in Power BI, Excel Power Pivot, and Analysis Services to:

-  Perform calculations

- Create measures

- Create calculated columns

- Build logic in your data model

Simple Explanation:

`“DAX is the language Power BI uses to calculate and analyze data.”`

### WHERE DAX IS USED

DAX is used in:

- Power BI

- Excel (Power Pivot)

- SQL Server Analysis Services (SSAS)

### WHAT DAX IS USED FOR

🔵 1. Creating Measures (MOST IMPORTANT)

- Measures are dynamic calculations.

Example:

`Total Sales = SUM(Sales[Sales Amount])`

This calculates total sales based on filters in your report.

 Why Measures Matter:

If you select:

- Region = Lagos

- Category = Accesdories

The measure updates automatically

🟢 2. Creating Calculated Columns

Columns are static calculations stored in the table.

Example:

`Profit = Sales[Sales Amount] - Sales[Cost]`

This creates a new column for each row

🟡 3. Creating Calculated Tables

You can create new tables using DAX.

Example:

`TopCustomers = FILTER(Customers, Customers[TotalSales] > 10000)`

🔴 4. Time Intelligence Calculations

Used for date-based analysis:

Examples:

`Sales YTD = TOTALYTD([Total Sales], DateTable[Date])`

`Sales Last Year = CALCULATE([Total Sales], SAMEPERIODLASTYEAR(DateTable[Date]))`

🔷 5. Filtering and Context Control

DAX controls how data is filtered.

Example:

`Sales in Lagos = CALCULATE([Total Sales], Customers[Region] = "Lagos")`

🔷 KEY CONCEPT: CONTEXT (VERY IMPORTANT)

DAX works based on:

🔹 1. Filter Context

Filters applied from:

- Slicers

- Charts

- Tables

 Example:

If user selects:

Region = Abuja

DAX recalculates automatically

🔹 2. Row Context

Used in calculated columns

Works row by row

| --------------------- | -------- | -------------- |

- AVERAGE()

- COUNT()

🔹 Logical

- IF()

- SWITCH()

🔹 Filtering

- CALCULATE() ⭐ (Most important)

- FILTER()

🔹 Time Intelligence

- TOTALYTD()

- SAMEPERIODLASTYEAR()

🔷 REAL EXAMPLES FROM YOUR MODEL

- Total Sales

`Total Sales = SUM(Sales[Sales Amount])`

- Profit Margin

`Profit Margin = DIVIDE(SUM(Sales[Profit]), SUM(Sales[Sales Amount]))`

- Sales by Region (Using Filter)

`Sales Lagos = CALCULATE([Total Sales], Customers[Region] = "Lagos")`

🔷 WHY DAX IS POWERFUL

DAX allows you to:

- Analyze millions of rows instantly

- Build interactive dashboards

- Perform business calculations

- Create KPIs

- Handle complex logic

🔷 COMMON MISTAKES WITH DAX

- Using calculated columns instead of measures

- Ignoring filter context

- Not using CALCULATE properly

- No date table for time intelligence

🔷 FINAL SUMMARY

 DAX is:

- A formula language

- Used in Power BI

- For calculations and analysis

 It is used to:

- Create measures

- Perform time analysis

- Control filter behaviour
