## 1. Introduction

Power BI is often introduced to beginners as "a tool for making charts and dashboards." While that is true on the surface, a professional Power BI solution is really the visible tip of a much larger data pipeline. Underneath every dashboard sits a chain of decisions about where data comes from, how it is cleaned, where it is transformed, how it is modeled, and how it is refreshed.

Two of the most important architectural decisions in that pipeline are **ETL** and **ELT**. These are not just abstract data engineering terms. They directly affect:

- How fast a Power BI report can be built
- How much data volume the solution can handle
- How flexible the solution is for future analytics and AI projects
- How much control an organization has over data quality before it reaches the model

This notebook builds the full picture, from the overall workflow down to specific business scenarios, so that students can reason about ETL and ELT the way a working data professional would.


<a id="section2"></a>
## 2. The Overall Power BI Workflow

A typical end to end Power BI data workflow looks like this:

```
Data Sources -> Connect -> Extract -> Transform -> Load -> Model -> DAX -> Visualize -> Publish -> Share
```

Expanded into a more detailed pipeline:

```
Excel / SQL Server / APIs / CSV
        |
        v
   Power Query
        |
        v
Clean and Transform Data
        |
        v
   Data Model
        |
        v
  Relationships
        |
        v
  DAX Measures
        |
        v
Reports and Dashboards
        |
        v
 Power BI Service
        |
        v
 Refresh and Sharing
```

### Why This Matters for Students

Notice that "making a chart" is actually step 8 or 9 in a much longer chain. Most of the real engineering work, and most of the value delivered to a business, happens *before* a single visual is created. This is why data professionals spend far more time in Power Query and data modeling than in the visualization pane.


<a id="section3"></a>
## 3. What is ETL?

**ETL** stands for **Extract, Transform, Load**.

The data is:

1. **Extracted** from the source system
2. **Transformed** and cleaned
3. **Loaded** into the destination system (data warehouse or data model)

### Worked Example

Suppose you have sales data in Excel:

| Customer | Product | Sales   | Date       |
|----------|---------|---------|------------|
| John     | Laptop  | 500000  | 01/08/2026 |
| Mary     | Phone   | 250000  | 02/08/2026 |
| John     | Laptop  | 500000  | 02/08/2026 |

**Step 1: Extract**
Power BI connects to the Excel file.

**Step 2: Transform**
Using Power Query, you would:

- Remove duplicates
- Change data types
- Remove unnecessary columns
- Handle missing values
- Split columns
- Merge tables
- Append tables
- Create calculated columns

**Step 3: Load**
The transformed, clean data is loaded into the Power BI data model.

```
Excel -> Power Query -> Data Model
```

This entire chain is an ETL workflow: the data is cleaned *before* it settles into its final home.


<a id="section4"></a>
## 4. What is ELT?

**ELT** stands for **Extract, Load, Transform**.

The key difference from ETL is *when* transformation happens. Instead of cleaning the data before loading it, you:

1. **Extract** the data
2. **Load** it into a data warehouse or lakehouse in raw (or near raw) form
3. **Transform** it after it has landed

### Example Flow

```
SQL Server -> Data Warehouse -> Transformation -> Power BI
```

Or in a modern Microsoft environment:

```
Source Systems -> Microsoft Fabric Lakehouse/Warehouse -> Transform -> Power BI
```

Here, the heavy transformation work is performed by the warehouse or lakehouse engine (which can be far more powerful than a laptop running Power Query), rather than primarily inside Power BI itself.

### Why ELT Has Become Popular

Modern cloud platforms such as Microsoft Fabric, Snowflake, Databricks, BigQuery, and Azure Synapse can process enormous volumes of data very quickly. This makes it practical to load raw data first and decide how to shape it later, sometimes in several different ways for several different purposes.


<a id="section5"></a>
## 5. ETL vs ELT: Side by Side Comparison

| ETL | ELT |
|-----|-----|
| Extract, then Transform, then Load | Extract, then Load, then Transform |
| Transformation happens before loading | Transformation happens after loading |
| Power Query performs the transformations | The data warehouse or lakehouse performs the transformations |
| Suitable for smaller to moderate workloads | Excellent for large scale data |
| Often used directly inside Power BI | Often used with Fabric, Synapse, Snowflake, BigQuery, and similar platforms |
| Data is cleaned before entering the model | Data can initially be loaded in raw form |
| Easier to guarantee data quality up front | Easier to preserve raw data for future, unplanned analysis |
| Slower to adapt to new transformation needs | Faster to adapt because raw data is still available |

### The One Sentence Version

- **ETL:** Clean the data *before* it goes into storage.
- **ELT:** Put the data into storage *first*, then decide how to clean and shape it.


<a id="section6"></a>
## 6. Where Power Query Fits In

Power Query is the main data preparation engine inside Power BI. You can think of it as the **data preparation layer** that sits between raw data and the data model.

```
Raw Data
   |
   v
Power Query
   - Remove duplicates
   - Filter rows
   - Replace values
   - Change data types
   - Merge queries
   - Append queries
   - Pivot / Unpivot
   - Group By
   |
   v
Clean Data
   |
   v
Power BI Data Model
```

Power Query uses a functional language called **M** behind the scenes. Every click you make in the Power Query Editor (Remove Duplicates, Change Type, Merge Queries, and so on) is automatically translated into M code, which you can view and edit directly in the Advanced Editor.


<a id="section7"></a>
## 7. The Complete Power BI Architecture (5 Layers)

A professional Power BI solution can be broken down into five distinct layers.

### Layer 1: Data Sources

Data can come from many places, including:

- Excel and CSV files
- SQL Server, MySQL, Oracle
- SharePoint
- Web APIs
- Azure services
- Salesforce
- Other databases and data warehouses
- Microsoft Fabric

### Layer 2: Data Extraction

Power BI connects to the sources above and retrieves the raw data.

### Layer 3: Data Transformation

Usually performed using Power Query:

```
Raw Sales Data
      |
      v
Remove duplicates
      |
      v
Handle missing values
      |
      v
Change data types
      |
      v
Clean customer names
      |
      v
Merge customer table
      |
      v
Create Date table
```

### Layer 4: Data Loading

The transformed data is loaded into the Power BI data model.

### Layer 5: Data Modeling

This is where you create relationships between tables.

```
             Customers
                 |
                 v
              Sales
              /    \
             /      \
            v        v
        Products    Date
```

A common and highly recommended approach is the **Star Schema**:

```
                Date
                 |
                 |
Products --- Sales --- Customers
                 |
                 |
             Geography
```

- **Sales** is the fact table (it holds transactional, numeric data such as sales amount, quantity, and profit)
- **Customers, Products, Date, Geography** are dimension tables (they describe the "who, what, when, where" of each transaction)

A star schema keeps the model simple, fast, and easy for DAX to query, which is why it is the standard recommended pattern in Power BI.


<a id="section8"></a>
## 8. The DAX Layer

After the model is built, **DAX (Data Analysis Expressions)** is used to create business calculations.

### Example Measures

```DAX
Total Sales = SUM(Sales[SalesAmount])
```

```DAX
Total Profit = SUM(Sales[Profit])
```

```DAX
Profit Margin = DIVIDE([Total Profit], [Total Sales])
```

### Power Query vs DAX

| Power Query | DAX |
|-------------|-----|
| Prepares the data | Analyzes the data |
| Runs before data is loaded into the model | Runs after data is in the model |
| Produces tables and columns | Produces measures, calculated columns, and calculated tables |
| Uses the M language | Uses the DAX language |

A simple way to remember the difference:

> **Power Query prepares the data. DAX analyzes the data.**

<a id="section10"></a>
## 10. Publishing to Power BI Service

After building the report in Power BI Desktop, you publish it to the Power BI Service (the cloud platform).

```
Power BI Desktop
        |
        v
     Publish
        |
        v
 Power BI Service
        |
        v
     Workspace
        |
        v
 Dashboard / Report
        |
        v
       Users
```

Workspaces in the Power BI Service are where teams collaborate, apply row level security, schedule refreshes, and package content into apps for wider distribution.


<a id="section11"></a>
## 11. Data Refresh and Gateways

A very important part of any real world Power BI workflow is refresh. Suppose a company adds new sales records every day. The Power BI report needs to receive that new data automatically.

```
Database
   |
   v
Power BI
   |
   v
 Refresh
   |
   v
Updated Data Model
   |
   v
Updated Dashboard
```

For **on premises** databases (a SQL Server sitting inside a company's own building, for example), Power BI Service usually needs an **On premises Data Gateway** installed to safely bridge the cloud service and the local database.

For **cloud based** data sources (such as data already in Azure or Microsoft Fabric), a gateway is often not required because the Power BI Service can connect directly.

## 12. Practical Business Scenarios: ETL vs ELT

The easiest way to truly understand ETL vs ELT is to look at what a real business is trying to accomplish. Below are seven realistic scenarios, each showing both an ETL and an ELT approach where relevant.


### 12.1 Banking: Customer and Transaction Reporting

Imagine a bank receives transaction data from:

- ATM systems
- Mobile banking
- POS terminals
- Internet banking
- Branch transactions

Raw data might look like this:

| Transaction ID | Customer | Amount  | Date     | Channel |
|-----------------|----------|---------|----------|---------|
| T001             | C001     | 50,000  | 18/08/26 | ATM     |
| T002             | C002     | 120,000 | 18/08/26 | Mobile  |
| T003             | C003     | 75,000  | 18/08/26 | POS     |

The bank does not want dirty data flowing directly into its reporting database, so it uses an **ETL** approach:

```
ATM -----------+
Mobile ---------|
POS ------------|--- EXTRACT --> TRANSFORM --> LOAD --> Data Warehouse --> Power BI
Internet -------|
Branch ---------+

TRANSFORM step includes:
  - Remove duplicates
  - Fix data types
  - Handle missing values
  - Standardize currencies
  - Validate transactions
  - Remove invalid records
```

**Why ETL here?** Data quality is critical in banking, transformation rules are well established (regulatory reporting rules rarely change week to week), and the destination database should contain only clean, validated data.


### 12.2 E-commerce: Jumia Style Analytics

An e-commerce company receives millions of records every day:

- Customer clicks
- Product views
- Orders
- Payments
- Cancellations
- Deliveries
- Customer reviews
- Website activity

Instead of transforming everything up front, the company loads raw data first, using an **ELT** approach:

```
Website --------+
Mobile App -------|
Orders ------------|--- EXTRACT --> LOAD --> Cloud Data Lake/Warehouse --> TRANSFORM --> Power BI
Payments ----------|
Logistics ---------|
Customer Data -----+

TRANSFORM step produces:
  - Sales dataset
  - Customer dataset
  - Product dataset
  - Marketing dataset
  - Logistics dataset
```

**Why ELT here?** The data volume is huge, and modern cloud platforms can process it efficiently. Keeping the raw data also lets the company build several different transformed datasets from the same source, for different teams and different questions.


### 12.3 Oil and Gas: Production and Sensor Data

An oil and gas company collects:

- Production data
- Well data
- Equipment sensor data
- Maintenance data
- Financial data
- Employee data
- Procurement data

**ETL approach** (for controlled, standardized reporting):

```
Production Systems -> Extract -> Transform -> Load -> Data Warehouse -> Power BI
```

Example transformation: convert barrels per day to standardized units, remove invalid sensor readings, aggregate production, then load into the warehouse.

**ELT approach** (for billions of IoT sensor readings):

```
IoT Sensors -> Extract -> Load RAW DATA -> Cloud Data Lake -> Transform -> Production Analytics
            -> Predictive Maintenance -> Power BI / AI Models
```

**Why ELT here?** The company can retain raw sensor data and later reuse it for predictive maintenance, production forecasting, anomaly detection, and equipment failure prediction, none of which may have been anticipated at the time the data was first collected.


### 12.4 Retail: Multi Branch Supermarket

Suppose a supermarket chain has 500 branches, each sending:

- Daily sales
- Inventory
- Customer information
- Supplier information
- Product information

**ETL:** If management mainly needs standardized reporting:

```
Raw sales -> Clean -> Validate -> Aggregate -> Data Warehouse
```

**ELT:** If the company wants to preserve every transaction and customer interaction for future analytics:

```
                 RAW DATA
                    |
            Cloud Data Warehouse
                    |
       +------------+------------+
       |            |            |
       v            v            v
 Sales Analytics  Customer AI   Inventory
       |            |            |
       v            v            v
    Power BI     ML Models    Power BI
```


### 12.5 Healthcare: Hospital Systems

A hospital manages:

- Patient records
- Laboratory results
- Pharmacy data
- Doctor consultations
- Medical imaging
- Billing data

**ETL** (for a regulatory report):

```
Hospital Systems -> Extract -> Clean and Validate -> Transform -> Data Warehouse -> Regulatory Report
```

Here the hospital wants only validated, standardized information entering the reporting database, since regulators expect consistent, auditable figures.

**ELT** (for advanced AI use cases):

```
Hospital Systems -> Extract -> Load Raw Data -> Data Lake -> Transform -> Machine Learning -> Disease Prediction
```

Raw data is preserved because future AI projects may need information that was not part of the original transformation scope, such as free text clinical notes or imaging metadata.


### 12.6 Marketing: Multi Channel Campaign Analytics

A company wants to analyze customer behavior from:

- Facebook and Instagram campaigns
- Google Ads
- Website
- Email
- CRM
- Mobile application

**ETL** (for a standardized Marketing Performance Dashboard):

```
Different Sources -> Extract -> Transform -> Standardize KPIs -> Load -> Power BI
```

Example KPIs: cost per click, conversion rate, customer acquisition cost, return on investment.

**ELT** (for deeper customer analytics):

```
All Customer Data -> Load -> Cloud Data Lake -> Transform -> Customer Segmentation
                  -> Churn Prediction -> Recommendation System
```


### 12.7 Telecommunications: Network and Customer Data (Bonus Scenario)

A telecom operator such as an MTN style business collects:

- Call detail records (CDRs)
- Data usage logs
- Network cell tower performance data
- Customer complaints and support tickets
- Billing and payment history
- SIM registration data

**ETL** (for regulatory and billing accuracy):

```
Billing Systems -> Extract -> Validate Charges -> Standardize Currency and Tariff Codes
               -> Load -> Billing Data Warehouse -> Power BI
```

Billing must be extremely accurate before it reaches any reporting layer, since incorrect figures directly affect customer invoices and regulatory filings.

**ELT** (for network optimization and churn prediction):

```
Cell Towers and CDRs -> Extract -> Load Raw Logs -> Data Lake -> Transform
                     -> Network Analytics -> Churn Prediction Models -> Power BI
```

With billions of call detail records generated daily, it is far more practical to load everything raw into a lakehouse first, then transform subsets of it as specific analytical questions arise, such as identifying dropped call hotspots or predicting which customers are likely to switch networks.


<a id="section13"></a>
## 13. When Should a Business Choose ETL?

| Business Situation | Why ETL? |
|---------------------|----------|
| Data must be highly validated | Clean the data before it ever reaches the warehouse |
| Strict regulatory reporting | Full control over exactly what enters the warehouse |
| Sensitive data involved | Remove or mask unnecessary information early |
| Relatively moderate data volume | Transformation workload stays manageable |
| Fixed, well known reporting requirements | Prepare exactly what is needed, nothing more |
| Legacy data warehouse in place | Traditional ETL architecture usually fits better |

<a id="section14"></a>
## 14. When Should a Business Choose ELT?

| Business Situation | Why ELT? |
|---------------------|----------|
| Massive data volumes | Cloud platforms can process large datasets efficiently |
| Data changes frequently | Raw data can always be retained for reprocessing |
| Multiple analytical requirements | The same raw data can be transformed differently for different teams |
| AI and machine learning projects | Raw information is preserved for future, unplanned modeling |
| Cloud data warehouse or lakehouse already in use | Processing can happen natively inside the platform |
| Exploratory analytics | Analysts can create their own transformations as questions arise |


<a id="section15"></a>
## 15. The Restaurant Analogy

A simple way to help students remember the difference between ETL and ELT is to imagine a restaurant kitchen.

### ETL: Prepare Before Storage

```
Buy ingredients -> Clean and cut and prepare -> Store prepared ingredients in the kitchen -> Cook
```

### ELT: Store Before Preparation

```
Buy ingredients -> Store them in the kitchen -> Prepare them only when needed -> Cook
```

**In one line:**

> ETL prepares before storage. ELT stores before preparation.

## 16. Complete Enterprise Architecture Diagram

As a Data Scientist, AI Engineer, or Power BI professional, it is important not to think of Power BI as only a visualization tool. A modern Power BI solution typically sits at the end of a much longer chain:

```
              DATA SOURCES
                   |
                   v
          +-----------------+
          |   ETL / ELT     |
          +-----------------+
                   |
                   v
          DATA WAREHOUSE /
             LAKEHOUSE
                   |
                   v
          +-----------------+
          | DATA TRANSFORM  |
          +-----------------+
                   |
                   v
          +-----------------+
          |  DATA MODEL     |
          |  Star Schema    |
          +-----------------+
                   |
                   v
          +-----------------+
          |      DAX        |
          +-----------------+
                   |
                   v
          +-----------------+
          | POWER BI REPORT |
          +-----------------+
                   |
                   v
          +-----------------+
          | POWER BI SERVICE|
          +-----------------+
                   |
                   v
               END USERS
```

### One Sentence Summary

> ETL/ELT gets and prepares the data, data modeling organizes it, DAX analyzes it, Power BI visualizes it, and Power BI Service distributes and refreshes it. That is the end to end Power BI workflow you should understand professionally.

Two architecture patterns follow from this:

**ETL Architecture**
```
Sources -> ETL -> Data Warehouse -> Power BI
```

**ELT Architecture**
```
Sources -> Data Lake/Warehouse -> ELT -> Data Model -> Power BI
```

## 17. Power Query (M) Quick Reference Cheat Sheet

| Task | Power Query Ribbon Action | Underlying M Function (example) |
|------|----------------------------|----------------------------------|
| Remove duplicate rows | Home > Remove Rows > Remove Duplicates | `Table.Distinct(Source)` |
| Change a column's data type | Right click column > Change Type | `Table.TransformColumnTypes(...)` |
| Remove a column | Right click column > Remove | `Table.RemoveColumns(...)` |
| Filter rows | Click the filter arrow on a column | `Table.SelectRows(...)` |
| Split a column | Transform > Split Column | `Table.SplitColumn(...)` |
| Merge two tables | Home > Merge Queries | `Table.NestedJoin(...)` |
| Append two tables | Home > Append Queries | `Table.Combine({Table1, Table2})` |
| Group rows | Transform > Group By | `Table.Group(...)` |
| Pivot a column | Transform > Pivot Column | `Table.Pivot(...)` |
| Unpivot columns | Transform > Unpivot Columns | `Table.UnpivotOtherColumns(...)` |
| Replace values | Transform > Replace Values | `Table.ReplaceValue(...)` |
| Fill down blank cells | Transform > Fill > Down | `Table.FillDown(...)` |

**Teaching tip:** Encourage students to open the Advanced Editor after every few clicks in Power Query. Seeing the M code generated automatically helps them understand that the user interface is really just writing a functional script for them, step by step.

## 18. DAX Quick Reference Cheat Sheet

| Category | Example | Purpose |
|----------|---------|---------|
| Simple aggregation | `Total Sales = SUM(Sales[SalesAmount])` | Adds up a numeric column |
| Ratio / division | `Profit Margin = DIVIDE([Total Profit], [Total Sales])` | Safely divides two measures, avoiding divide by zero errors |
| Row filtering | `High Value Sales = CALCULATE([Total Sales], Sales[SalesAmount] > 100000)` | Filters a calculation by a condition |
| Time intelligence | `YTD Sales = TOTALYTD([Total Sales], 'Date'[Date])` | Calculates a year to date running total |
| Ranking | `Product Rank = RANKX(ALL(Products[Product]), [Total Sales])` | Ranks items by a measure |
| Related table lookup | `Customer Region = RELATED(Customers[Region])` | Pulls a value from a related table |
| Conditional logic | `Sales Status = IF([Total Sales] > 1000000, "High", "Low")` | Creates simple business logic |
| Counting | `Number of Orders = COUNTROWS(Sales)` | Counts rows in a table |

**Teaching tip:** Always teach `CALCULATE` early. It is the single most important function in DAX because almost every advanced formula (time intelligence, ranking, comparisons) is built using `CALCULATE` to modify filter context.

## 19. Glossary of Key Terms

| Term | Definition |
|------|------------|
| **ETL** | Extract, Transform, Load. Data is cleaned before it is loaded into its destination. |
| **ELT** | Extract, Load, Transform. Data is loaded first, in raw form, and transformed afterward. |
| **Power Query** | The data preparation tool inside Power BI, powered by the M language. |
| **M Language** | The functional programming language used behind the scenes in Power Query. |
| **Data Model** | The set of tables and relationships that Power BI uses to answer questions. |
| **Star Schema** | A data modeling pattern with one central fact table connected to several dimension tables. |
| **Fact Table** | A table holding measurable, numeric business events, such as sales or transactions. |
| **Dimension Table** | A table describing the who, what, when, or where of a fact, such as Customers or Products. |
| **DAX** | Data Analysis Expressions. The formula language used to build measures and calculated columns in Power BI. |
| **Measure** | A dynamic calculation, written in DAX, that responds to filters in a report. |
| **Data Warehouse** | A structured, centralized repository designed for reporting and analysis, usually holding cleaned data. |
| **Data Lake / Lakehouse** | A storage system that can hold large volumes of raw, semi structured, or unstructured data, often used as the landing zone in ELT. |
| **Power BI Desktop** | The Windows application used to build reports and data models. |
| **Power BI Service** | The cloud platform where reports are published, shared, and refreshed. |
| **On Premises Data Gateway** | Software that securely connects the Power BI Service to data sources hosted inside a company's own network. |
| **Refresh** | The process of updating a Power BI data model with the latest data from its sources. |
| **Microsoft Fabric** | Microsoft's unified analytics platform combining data engineering, data warehousing, and business intelligence. |

## 21. Key Takeaways and Career Guidance

### Key Takeaways

- Power BI is the visible layer of a much larger data pipeline that begins with data sources and ends with refreshed dashboards in the hands of business users.
- ETL cleans data before loading it; ELT loads data first and transforms it afterward.
- Power Query prepares the data; DAX analyzes the data.
- A star schema, built from one fact table and several dimension tables, is the standard, recommended way to model data in Power BI.
- The right choice between ETL and ELT depends on data volume, regulatory requirements, and how many different ways the data will eventually need to be used.
- Real businesses often use both approaches at once, applying ETL to strictly regulated data and ELT to large, exploratory, or AI focused datasets.

### Career Guidance

For a Power BI or data engineering career, it is worth learning both ETL and ELT thoroughly, but placing particular emphasis on understanding ELT with modern cloud data warehouses and lakehouses.
