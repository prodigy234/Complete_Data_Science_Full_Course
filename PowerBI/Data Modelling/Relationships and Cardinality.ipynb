# 1. What is a Relationship in Power BI?

A relationship is a connection between two tables using a common column.

Example:

- Customers table → has CustomerID

- Orders table → has CustomerID

Power BI links them using that column so you can analyze data across both tables.

# 2. What is Cardinality?

Cardinality defines how rows in one table relate to rows in another table.

It answers this question:

**“How many records in Table A relate to Table B?”**

 ### TYPES OF CARDINALITY (VERY IMPORTANT)

🔹 1. One-to-Many (1 : *)

This is the MOST COMMON and BEST practice

Meaning:

One record in Table A → relates to MANY records in Table B

 Example:

- Customers → one customer

- Orders → many orders

 One customer can place multiple orders

 In Power BI:

- Customers[CustomerID] → unique

- Orders[CustomerID] → repeated

**This is the ideal relationship structure**

🔹 2. Many-to-One (* : 1)

This is just the reverse of One-to-Many

 Meaning:

Many records in Table A → relate to one record in Table B

 Same example:

Many Orders → One Customer

 **Power BI just flips the direction depending on how you connect it.**

🔹 3. One-to-One (1 : 1)

 Meaning:

One record in Table A → matches exactly one record in Table B

 Example:

- EmployeeDetails ↔ EmployeeSalary

Each employee has exactly one salary record.

 Rare in real-world analytics

🔹 4. Many-to-Many (* : *)
 
 Advanced (and risky if misunderstood)

 Meaning:

Many records in Table A → relate to many records in Table B

 Example:

Students ↔ Courses

A student takes many courses

A course has many students

# 3. Relationship Direction (VERY IMPORTANT)

Power BI has Cross Filter Direction

🔹 Single Direction (Default)

 Data flows one way only

 Cardinality = Relationship type in real life

- One-to-Many → One teacher, many students

- One-to-One → One person, one passport

- Many-to-Many → Students and courses

Example:

Customers → Orders

✔️ Safe

✔️ Fast

✔️ Recommended

🔹 Both Direction (Bi-directional)

 Data flows both ways

Example:

Customers ↔ Orders

 Can cause:

- Ambiguous relationships

- Wrong calculations

- Performance issues

✔️ Use ONLY when necessary

# 4. Active vs Inactive Relationships

🔹 Active Relationship

 - The main relationship used automatically
 
 - Only one active relationship allowed between two tables

🔹 Inactive Relationship
 
 - Exists but not used unless specified

 Used with DAX:

USERELATIONSHIP()

 Example:

- Orders[OrderDate]

- Orders[DeliveryDate]

You may have:

- Active → OrderDate

- Inactive → DeliveryDate

# 5. Practical Example (Real Business Scenario)

Imagine you are building a Sales Dashboard:

Tables:

- Customers

- Orders

- Products

- Sales

**Relationships:**

- Customers (1) → Orders (*)

- Products (1) → Sales (*)

- Orders (1) → Sales (*)

This forms a:

STAR SCHEMA

# 6. Star Schema (BEST PRACTICE)

🔹 Fact Table (Central Table)

- Contains transactions

Example: Sales

🔹 Dimension Tables

Describe the data

- Customers

- Products

- Date

Structure:
        Customers
            |
Products — Sales — Date
            |
         Orders

✔️ Clean

✔️ Fast

✔️ Scalable

# 7. Common Mistakes Data Analyst Make

🚫 Using Many-to-Many unnecessarily

🚫 Turning on Bi-directional filtering everywhere

🚫 Not having unique keys in dimension tables

🚫 Creating relationships on wrong columns

🚫 Ignoring data duplication
