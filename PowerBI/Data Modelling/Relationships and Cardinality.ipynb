# 1. What is a Relationship in Power BI?

A relationship is a connection between two tables using a common column.

Example:

- Customers table → has CustomerID

- Orders table → has CustomerID

Power BI links them using that column so you can analyze data across both tables.

# 2. What is Cardinality?

Cardinality defines how rows in one table relate to rows in another table.

It answers this question:

**“How many records in Table A relate to Table B?”**

 ### TYPES OF CARDINALITY (VERY IMPORTANT)

🔹 1. One-to-Many (1 : *)

This is the MOST COMMON and BEST practice

Meaning:

One record in Table A → relates to MANY records in Table B

 Example:

- Customers → one customer

- Orders → many orders

 One customer can place multiple orders

 In Power BI:

- Customers[CustomerID] → unique

- Orders[CustomerID] → repeated

**This is the ideal relationship structure**

🔹 2. Many-to-One (* : 1)

This is just the reverse of One-to-Many

 Meaning:

Many records in Table A → relate to one record in Table B

 Same example:

Many Orders → One Customer

 **Power BI just flips the direction depending on how you connect it.**

🔹 3. One-to-One (1 : 1)

 Meaning:

One record in Table A → matches exactly one record in Table B

 Example:

EmployeeDetails ↔ EmployeeSalary

Each employee has exactly one salary record.

 Rare in real-world analytics

🔹 4. Many-to-Many (* : *)
 Advanced (and risky if misunderstood)

 Meaning:

Many records in Table A → relate to many records in Table B

 Example:

Students ↔ Courses

A student takes many courses

A course has many students
