WHAT IS DATA MODELLING IN POWER BI?

Data Modelling is the process of:

Structuring your data into relationships so it can be analyzed correctly.

In simple terms:

Connecting tables
Defining relationships
Creating a structure for analysis

WHY DATA MODELLING IS IMPORTANT

Without modelling:

❌ Wrong totals
❌ Duplicate counts
❌ Broken dashboards

With modelling:

✅ Accurate insights
✅ Fast queries
✅ Scalable reports

TYPES OF DATA MODELS

1. Star Schema (BEST PRACTICE)
One Fact Table
Multiple Dimension Tables

Example with your dataset:

Fact Table → Sales_Data
Dimension Tables → Customers, Products

2. Snowflake Schema
Dimensions split into sub-tables
More complex

3. Flat Table
Everything merged into one table (NOT recommended)
