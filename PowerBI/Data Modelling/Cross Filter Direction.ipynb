# 1. What is Cross Filter Direction?

When you create a relationship between two tables in Power BI, filtering can flow between them.

 Cross filter direction answers:

“When I filter one table, which other table should be affected?”

# 2. SINGLE DIRECTION (→)

 Meaning

Filtering flows in one direction only

 Typically:

Dimension Table → Fact Table

Example Setup

Tables:

Customers (Dimension Table)

| CustomerID | Name |
| ---------- | ---- |
| 1          | John |
| 2          | Mary |

Orders (Fact Table)

| OrderID | CustomerID | Amount |
| ------- | ---------- | ------ |
| 101     | 1          | 500    |
| 102     | 1          | 300    |
| 103     | 2          | 200    |

Relationship:

- Customers (1) → Orders (*)

- Cross filter: SINGLE

 Practical Outputs

 Scenario 1: Filter from Customers

If you select:

 Customer = John

Output:

| OrderID | CustomerID | Amount |
| ------- | ---------- | ------ |
| 101     | 1          | 500    |
| 102     | 1          | 300    |

✔️ Orders table is filtered

✔️ Total = 800

❌ Scenario 2: Filter from Orders

If you filter:

 Orders where Amount > 300

Output:

| OrderID | CustomerID | Amount |
| ------- | ---------- | ------ |
| 101     | 1          | 500    |

Will Customers table change?

❌ NO

Customers table still shows:

| CustomerID | Name |
| ---------- | ---- |
| 1          | John |
| 2          | Mary |

Key Insight

 In SINGLE direction:

Filter flows only from Customers → Orders

NOT from Orders → Customers

When to Use SINGLE

✔️ Star Schema

✔️ Standard dashboards

✔️ Best performance

✔️ Avoids ambiguity

# 3. BOTH DIRECTION (↔)

 Meaning

Filtering flows in BOTH directions

Customers ↔ Orders

Same Example, But Now BOTH

 Practical Outputs

 Scenario 1: Filter from Customers

 Customer = John

✔️ Same as before:

Orders filtered to John's orders

Total = 800

 Scenario 2: Filter from Orders

👉 Orders where Amount > 300

Now result:

Orders:

| OrderID | CustomerID | Amount |
| ------- | ---------- | ------ |
| 101     | 1          | 500    |

Customers:

| CustomerID | Name |
| ---------- | ---- |
| 1          | John |

✔️ Customers table is now filtered automatically

 Big Difference

 | Action                    | SINGLE | BOTH |
| ------------------------- | ------ | ---- |
| Filter Customers → Orders | ✅      | ✅    |
| Filter Orders → Customers | ❌      | ✅    |
