# Analytics

Analytics is the process of collecting, organizing, analyzing, and interpreting data to support decision-making. In Data Science, analytics transforms raw data into meaningful insights that help businesses, governments, healthcare organizations, and individuals make informed decisions.


# Types of Analytics in Data Science

The four major types of analytics are:

Descriptive Analytics – What happened?

Diagnostic Analytics – Why did it happen?

Predictive Analytics – What is likely to happen?

Prescriptive Analytics – What should we do?

These four types build upon one another, moving from understanding the past to making intelligent future decisions.

## 1. Descriptive Analytics (What Happened?)

Definition

Descriptive Analytics summarizes historical data to explain what has already occurred.

It answers questions such as:

What happened?

How many?

How often?

What are the trends?

What is the current situation?

This is the foundation of data analytics because before making predictions, organizations must understand their historical performance.

Goal
To convert raw historical data into meaningful summaries.

Business Example
A supermarket owner wants to know:

Total sales this month

Best-selling products

Number of customers

Revenue generated

Monthly profit

Descriptive analytics provides these answers.

## 2. Diagnostic Analytics (Why Did It Happen?)
Definition
Diagnostic Analytics investigates data to determine the reasons behind past events.

It answers:

Why did sales drop?

Why did customers leave?

Why did production increase?

Why did profits decline?

Goal
To discover the root causes of business problems.

Business Example
Suppose sales fell by 30%.

Descriptive analytics tells you:

Sales decreased.

Diagnostic analytics explains:

Prices increased

Competitors offered discounts

Products were unavailable

Marketing campaign failed

Real-Life Examples
Banking

Possible reasons:

High inflation

Poor credit assessment

Rising unemployment

Hospital
Why did patient waiting time increase?

Reasons may include:

Fewer doctors

Equipment failure

High patient volume

Manufacturing
Why did production reduce?

Reasons:

Machine breakdown

Raw material shortage

Power outage

Common Techniques
Drill-down analysis

Correlation analysis

Root Cause Analysis (RCA)

Data mining

Variance analysis

Pareto Analysis (80/20 Rule)

Python Example

Advantages
Finds causes of problems

Improves decision-making

Identifies business weaknesses

Limitations
Requires high-quality data

Correlation does not always imply causation

Can be time-consuming

## 3. Predictive Analytics (What Will Likely Happen?)

It answers:

Which customer is likely to leave?

Which loan applicant may default?

Which machine will fail soon?

Goal
To estimate future outcomes before they occur.

Business Example

Amazon predicts products customers will buy.

Banks predict customers likely to default.

Hospitals predict disease outbreaks.

# Comparison of the Four Types of Analytics

| Analytics Type   | Main Question             | Focus                     | Typical Techniques                                        | Example                              |
| ---------------- | ------------------------- | ------------------------- | --------------------------------------------------------- | ------------------------------------ |
| **Descriptive**  | What happened?            | Summarize historical data | Aggregation, reporting, dashboards                        | Sales report showing monthly revenue |
| **Diagnostic**   | Why did it happen?        | Identify root causes      | Correlation, drill-down, root cause analysis              | Investigating why sales dropped      |
| **Predictive**   | What is likely to happen? | Forecast future outcomes  | Machine learning, regression, classification, time series | Predicting customer churn            |
| **Prescriptive** | What should we do?        | Recommend optimal actions | Optimization, simulation, reinforcement learning          | Suggesting the best pricing strategy |

# End-to-End Business Example

Imagine an online retail company:

Descriptive Analytics: Sales were ₦120 million last quarter, with laptops generating the highest revenue.

Diagnostic Analytics: Sales declined in one region because products were frequently out of stock and delivery times increased.

Predictive Analytics: A forecasting model predicts a 20% increase in laptop demand during the next holiday season.

Prescriptive Analytics: The system recommends increasing laptop inventory by 25%, hiring additional warehouse staff, and launching targeted digital marketing campaigns to maximize revenue while minimizing stockouts.

# Summary

The four types of analytics form a progression from understanding the past to optimizing future decisions:

Descriptive Analytics explains what happened by summarizing historical data.

Diagnostic Analytics explains why it happened by uncovering root causes.

Predictive Analytics estimates what is likely to happen using statistical and machine learning models.

Prescriptive Analytics recommends what should be done by evaluating possible actions and identifying the best course based on business objectives and constraints.

Together, these analytics approaches enable organizations to move from reporting and investigation to forecasting, optimization, and data-driven decision-making across industries such as finance, healthcare, manufacturing, retail, telecommunications, logistics, and government.
