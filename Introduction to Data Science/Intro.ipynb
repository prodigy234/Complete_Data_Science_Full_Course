### JSON stands for JavaScript Object Notation. 
It is a lightweight, text-based data format used to store and exchange structured data between a server and an application.

### XML stands for Extensible Markup Language. 
It is a flexible, text-based file format used to store, structure, and transport data between different computer systems.

# DATA SOURCES

- Surveys & Questionnaires

- Databases (MySQL, PostgreSQL)

- APIs

- Sensors (IoT devices)

- Web scraping

- Business systems (ERP, CRM)

An API (Application Programming Interface) is a set of rules and protocols that allows different software systems to communicate and exchange data. It acts as a secure intermediary like a messenger or a waiter taking your requests to another system, fetching the required data, and delivering it back to you.
