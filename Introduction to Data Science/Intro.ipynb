# Introduction to Data Science - Comprehensive Notes

## 1. What is Data Science?

`Data Science is an interdisciplinary field that uses scientific methods, statistical techniques, programming, machine learning, and domain knowledge to extract meaningful insights and knowledge from data for decision-making.`

In simple terms:

`Data Science is the process of collecting, cleaning, analyzing, visualizing, and interpreting data to solve real-world problems and support better decisions.`

Data Science combines elements of:

- Statistics

- Mathematics

- Computer Science

- Artificial Intelligence (AI)

- Machine Learning (ML)

- Business Knowledge

- Data Visualization

## 2. Definition of Data

`Data refers to raw facts, figures, observations, measurements, or records that can be processed into meaningful information.`

Examples of Data

| Raw Data | Meaning           |
| -------- | ----------------- |
| 25       | Customer age      |
| ₦50,000  | Monthly salary    |
| Male     | Gender            |
| Lagos    | Location          |
| 85       | Examination score |

## 3. Data vs Information vs Knowledge

### Data

`Raw facts without context.`

Example:

- 70, 80, 90

### Information

`Processed data with meaning.`

Example:

- Student scores are 70, 80, and 90.

### Knowledge

`Insights derived from information.`

Example:

- The student performs above average.

## 6.The Data Science Lifecycle

The Data Science lifecycle is a structured process for solving data problems.

### Step 1: Problem Definition

`Understand the business problem.`

Example:

- "Why are customers leaving our company?"

### Step 2: Data Collection

`Gather data from sources such as:`

- Databases

- APIs

- Surveys

- Sensors

- Excel files

### Step 3: Data Cleaning

Remove:

- Missing values

- Duplicates

- Outliers

- Errors

`Data cleaning often takes 60-80% of project time.`

### Step 4: Exploratory Data Analysis (EDA)

`Explore data patterns.`

Questions include:

- What is the average?

- Are there trends?

- Are variables related?

Techniques:

- Histograms

- Boxplots

- Scatterplots

### Step 5: Feature Engineering

`Create useful variables.`

Example:

- From Date of Birth:

Age = Current Year – Birth Year

### Step 6: Model Building

`Develop predictive models.`

Examples:

- Linear Regression

- Logistic Regression

- Decision Trees

- Random Forest

- XGBoost

### Step 7: Model Evaluation

- Assess model performance.

Metrics:

Regression
MAE
MSE
RMSE
R²
Classification
Accuracy
Precision
Recall
F1 Score
Step 8: Deployment

Make the model available for users.

Tools:

Streamlit
Flask
FastAPI
Azure
AWS
Step 9: Monitoring

Continuously evaluate model performance.

Monitor:

Accuracy
Data drift
Business impact
