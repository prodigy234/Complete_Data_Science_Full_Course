## Common Methods

- get() - Safely retrieve a value.
- keys() - Returns all keys.
- values() - Returns all values.
- items() - Returns key-value pairs.
- update() - Updates a dictionary.
- pop() - Removes a key.
- popitem() - Removes the last inserted item.
- setdefault() - Gets or inserts a default value.
- clear() - Removes all items.
- copy() - Creates a shallow copy.
- fromkeys() - Creates a dictionary from keys.

# Python Dictionary Methods - Practical Examples

## get()
**Purpose:** Safely retrieves values.

```python
student={'Name':'John','Age':22}
print(student.get('Name'))
print(student.get('Gender'))
print(student.get('Gender','Not Available'))
```

## keys()
**Purpose:** Returns all keys.

```python
employee={'Name':'James','Department':'IT','Salary':350000}
print(employee.keys())
for k in employee.keys():
    print(k)
```

## values()
**Purpose:** Returns all values.

```python
employee={'Name':'James','Department':'IT','Salary':350000}
print(employee.values())
for v in employee.values():
    print(v)
```

## items()
**Purpose:** Returns key-value pairs.

```python
employee={'Name':'James','Department':'IT','Salary':350000}
print(employee.items())
for k,v in employee.items():
    print(k,':',v)
```

## update()
**Purpose:** Updates existing keys or adds new ones.

```python
student={'Name':'Mary','Age':20}
student.update({'Age':21,'Department':'Computer Science'})
print(student)
```

## pop()
**Purpose:** Removes a specified key.

```python
student={'Name':'John','Age':22,'Course':'Python'}
removed=student.pop('Course')
print(removed)
print(student)
```

## popitem()
**Purpose:** Removes the last inserted item.

```python
student={'Name':'John','Age':22,'Course':'Python'}
print(student.popitem())
print(student)
```

## setdefault()
**Purpose:** Returns existing value or inserts default.

```python
student={'Name':'John','Age':22}
student.setdefault('Gender','Male')
print(student)
print(student.setdefault('Name','James'))
```

## clear()
**Purpose:** Removes everything.

```python
student={'Name':'John','Age':22}
student.clear()
print(student)
```

## copy()
**Purpose:** Creates a shallow copy.

```python
student={'Name':'John','Age':22}
cp=student.copy()
cp['Age']=30
print(student)
print(cp)
```

## fromkeys()
**Purpose:** Creates a new dictionary from keys.

```python
subjects=['Math','English','Physics','Chemistry']
scores=dict.fromkeys(subjects,0)
print(scores)
```

## Compiling them together

```python
student={"name":"John","age":20}

print(student.get("name"))
print(student.keys())
print(student.values())
print(student.items())

student.update({"age":21,"course":"Python"})
print(student)

print(student.pop("course"))
print(student.setdefault("gender","Male"))
print(student.copy())
print(dict.fromkeys(["A","B","C"],0))
```

## Real World Example - Employee Database

```python
employees={
1001:{'Name':'James','Department':'IT','Salary':450000},
1002:{'Name':'Mary','Department':'HR','Salary':380000},
1003:{'Name':'David','Department':'Finance','Salary':520000}
}
print(employees.get(1002))
print(employees.keys())
print(employees.values())
employees[1001]['Salary']=500000
employees[1004]={'Name':'Grace','Department':'Marketing','Salary':410000}
employees.pop(1003)
for emp_id,details in employees.items():
    print(emp_id,details)
```

## Summary

| Method | Purpose | When to Use |
| - | - | - |
| get() | Safe lookup | Avoid KeyError |
| keys() | Return keys | Iterate through keys |
| values() | Return values | Analyze values |
| items() | Return key-value pairs | Loop through both |
| update() | Update dictionary | Merge or modify |
| pop() | Remove key | Delete one item |
| popitem() | Remove last item | Stack-like removal |
| setdefault() | Insert default | Missing keys |
| clear() | Empty dictionary | Reset |
| copy() | Duplicate dictionary | Preserve original |
| fromkeys() | Create dictionary | Initialize keys |
