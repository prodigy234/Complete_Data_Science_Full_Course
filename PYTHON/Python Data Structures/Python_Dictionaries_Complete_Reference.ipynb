## 1. What is a Dictionary?

A dictionary in Python is an unordered (insertion-ordered since Python 3.7), mutable collection of key-value pairs. Keys must be unique and hashable (strings, numbers, tuples of hashables, and so on). Values can be of any type.

Dictionaries are created using curly braces `{}`, the `dict()` constructor, or a dictionary comprehension.


```python
# Different ways to create a dictionary

# 1. Using curly braces
student = {"name": "Ada", "age": 22, "course": "Data Science"}
print(student)

# 2. Using the dict() constructor with keyword arguments
teacher = dict(name="Kajola", age=30, subject="AI Engineering")
print(teacher)

# 3. Using dict() with a list of tuples
scores = dict([("Ada", 85), ("Musa", 90), ("Ife", 78)])
print(scores)

# 4. Using a dictionary comprehension
squares = {x: x ** 2 for x in range(1, 6)}
print(squares)

# 5. Empty dictionary
empty = {}
also_empty = dict()
print(empty, also_empty)
```

## 2. `dict()` - The Constructor

**What it does:** Creates a new dictionary object from various inputs: keyword arguments, an iterable of key-value pairs, another dictionary, or nothing at all (empty dictionary).

**When to use it:** Use it whenever you need to build a dictionary programmatically, especially when converting from another data structure such as a list of tuples, two parallel lists (with `zip`), or another mapping.

**Conditions necessary for its use:**
- If passing keyword arguments, the keys must be valid Python identifiers (no spaces, cannot start with a digit).
- If passing an iterable, each item must itself be an iterable of exactly two elements (key, value).
- Keys used must be hashable.


```python
# dict() from zip of two lists
keys = ["id", "name", "department"]
values = [101, "Chidi", "Data Analytics"]
staff = dict(zip(keys, values))
print(staff)

# dict() copying another dictionary (shallow copy)
staff_copy = dict(staff)
print(staff_copy)
```

## 3. `dict.fromkeys(iterable, value=None)`

**What it does:** Creates a new dictionary with keys taken from the given iterable, all mapped to the same specified value (or `None` if no value is given).

**When to use it:** Use it when you need to quickly initialize a dictionary where every key should start with the same default value, such as setting up counters or default statuses.

**Conditions necessary for its use:**
- The iterable supplied must contain only hashable items, since they become dictionary keys.
- Be careful when the default value is a mutable object (like a list); all keys will reference the *same* object, not separate copies.


```python
# Basic use: initialize attendance status for a list of students
students = ["Ada", "Musa", "Ife", "Tunde"]
attendance = dict.fromkeys(students, "Absent")
print(attendance)

# Default value of None when not specified
placeholders = dict.fromkeys(["a", "b", "c"])
print(placeholders)

# Pitfall: using a mutable default value
grouped = dict.fromkeys(["math", "science"], [])
grouped["math"].append(90)
print(grouped)  # Notice both keys got updated - they share the same list object
```

## 4. `dict.get(key, default=None)`

**What it does:** Returns the value for `key` if it exists in the dictionary; otherwise returns `default` (which itself defaults to `None`) without raising an error.

**When to use it:** Use it whenever you are not certain a key exists and you want to avoid a `KeyError`. It is the safest way to read a value from a dictionary.

**Conditions necessary for its use:**
- None strictly required. It works safely whether or not the key exists, which is exactly why it is preferred over direct indexing (`dict[key]`) in uncertain cases.


```python
inventory = {"pen": 50, "notebook": 120, "eraser": 30}

# Key exists
print(inventory.get("pen"))

# Key does not exist - returns None instead of raising KeyError
print(inventory.get("marker"))

# Key does not exist - returns a custom default
print(inventory.get("marker", 0))

# Compare with direct indexing (this would raise an error if uncommented)
# print(inventory["marker"])  # KeyError: 'marker'
```

## 5. `dict.keys()`

**What it does:** Returns a view object that displays all the keys of the dictionary. This view is dynamic - it reflects changes made to the dictionary after the view was created.

**When to use it:** Use it when you need to iterate over, check membership in, or convert the set of keys, without needing the associated values.

**Conditions necessary for its use:**
- None. Works on any dictionary, including an empty one (returns an empty view).


```python
capitals = {"Nigeria": "Abuja", "Ghana": "Accra", "Kenya": "Nairobi"}

# Get all keys
print(capitals.keys())

# Convert to a list
print(list(capitals.keys()))

# Membership check
print("Nigeria" in capitals.keys())

# Views are dynamic - they update automatically
country_keys = capitals.keys()
capitals["Egypt"] = "Cairo"
print(country_keys)  # Egypt now appears without recomputing the view
```

## 6. `dict.values()`

**What it does:** Returns a view object displaying all the values in the dictionary. Like `.keys()`, it is dynamic and reflects live changes.

**When to use it:** Use it when you only care about the values, for example when summing, averaging, or checking whether a particular value exists somewhere in the dictionary.

**Conditions necessary for its use:**
- None. Note that unlike keys, values are not required to be unique or hashable.


```python
prices = {"bread": 1200, "milk": 900, "rice": 2500}

# Get all values
print(prices.values())

# Sum all values
print(sum(prices.values()))

# Check if a value exists
print(2500 in prices.values())

# Convert to a list for further processing
print(sorted(prices.values(), reverse=True))
```

## 7. `dict.items()`

**What it does:** Returns a view object displaying each key-value pair as a tuple. This is the most common way to loop through a dictionary when you need both the key and the value.

**When to use it:** Use it whenever a loop needs access to both key and value simultaneously, such as printing formatted reports or building a new dictionary based on both.

**Conditions necessary for its use:**
- None, but unpacking in a `for` loop requires exactly two elements per pair, which is guaranteed since each item is a (key, value) tuple.


```python
employees = {"E001": "Ngozi", "E002": "Femi", "E003": "Aisha"}

for emp_id, name in employees.items():
    print(f"{emp_id} -> {name}")

# Building a new dictionary using items()
reversed_lookup = {name: emp_id for emp_id, name in employees.items()}
print(reversed_lookup)
```

## 8. `dict.update(other)`

**What it does:** Updates the dictionary with key-value pairs from `other`, overwriting existing keys and adding new ones. `other` can be another dictionary, an iterable of key-value pairs, or keyword arguments.

**When to use it:** Use it when merging data from two sources, applying a batch of changes, or refreshing a dictionary with newer values (for example, updating configuration settings).

**Conditions necessary for its use:**
- `other` must be a mapping, an iterable of two-item pairs, or valid keyword arguments.
- Existing keys will be silently overwritten, so use this carefully when you do not want to lose old data.


```python
config = {"debug": False, "timeout": 30}

# Update using another dictionary
config.update({"timeout": 60, "retries": 3})
print(config)

# Update using keyword arguments
config.update(debug=True)
print(config)

# Update using an iterable of pairs
config.update([("verbose", True)])
print(config)
```

## 9. `dict.pop(key, default)`

**What it does:** Removes the specified key from the dictionary and returns its value. If the key is missing and a `default` is provided, that default is returned instead of raising an error. If no default is provided and the key is missing, a `KeyError` is raised.

**When to use it:** Use it when you need to both retrieve and delete a value in one step, for example processing a queue-like structure or extracting one field while cleaning up a record.

**Conditions necessary for its use:**
- If you expect the key might be missing, always supply a `default` value to avoid a runtime error.


```python
cart = {"apple": 3, "banana": 5, "orange": 2}

# Remove and return the value for a key that exists
removed_qty = cart.pop("banana")
print(removed_qty, cart)

# Attempting to pop a missing key with a default avoids an error
removed_qty = cart.pop("grape", 0)
print(removed_qty, cart)

# Without a default, popping a missing key raises KeyError
# cart.pop("grape")  # KeyError: 'grape'
```

## 10. `dict.popitem()`

**What it does:** Removes and returns the last inserted key-value pair as a tuple (LIFO order), since Python 3.7 guarantees insertion order. Raises a `KeyError` if the dictionary is empty.

**When to use it:** Use it when you need stack-like behavior, removing the most recently added entry, for example undoing the last change made to a dictionary.

**Conditions necessary for its use:**
- The dictionary must not be empty, otherwise a `KeyError` is raised.
- Relies on Python 3.7+ insertion-order guarantee for predictable behavior.


```python
log = {"step1": "start", "step2": "load data", "step3": "clean data"}

last_step = log.popitem()
print(last_step)
print(log)

# Calling popitem() on an empty dictionary raises an error
empty_dict = {}
# empty_dict.popitem()  # KeyError: 'popitem(): dictionary is empty'
```

## 11. `dict.setdefault(key, default=None)`

**What it does:** Returns the value of `key` if it exists. If it does not exist, inserts `key` with the value `default` and then returns `default`.

**When to use it:** Use it when building up collections inside a dictionary, such as grouping items by category, without needing an extra `if key not in dict` check.

**Conditions necessary for its use:**
- Works on any dictionary. Most useful when `default` is a mutable object such as a list or dictionary, used as an accumulator.


```python
# Grouping words by their first letter
words = ["apple", "ant", "banana", "bee", "cat"]
groups = {}

for word in words:
    groups.setdefault(word[0], []).append(word)

print(groups)

# Simple use without a mutable default
settings = {"theme": "dark"}
font_size = settings.setdefault("font_size", 12)
print(font_size, settings)
```

## 12. `dict.clear()`

**What it does:** Removes all key-value pairs from the dictionary, leaving it empty. It modifies the dictionary in place and returns `None`.

**When to use it:** Use it when you want to reuse the same dictionary object (for example one referenced elsewhere in your program) but wipe out all its current contents.

**Conditions necessary for its use:**
- None. Works even if the dictionary is already empty.


```python
session_data = {"user": "Gbenga", "logged_in": True, "cart_items": 5}
print("Before:", session_data)

session_data.clear()
print("After:", session_data)
```

## 13. `dict.copy()`

**What it does:** Returns a shallow copy of the dictionary. A new dictionary object is created, but nested mutable objects (like lists or dictionaries inside) are still shared with the original.

**When to use it:** Use it when you want to modify a dictionary without affecting the original, as long as the values themselves are not nested mutable structures you also plan to change.

**Conditions necessary for its use:**
- If your dictionary has nested mutable values and you need a fully independent copy, use `copy.deepcopy()` from the `copy` module instead.


```python
original = {"name": "Kajola", "skills": ["Python", "SQL"]}
shallow = original.copy()

shallow["name"] = "Gbenga"  # Does not affect original, since strings are immutable
print(original["name"], shallow["name"])

shallow["skills"].append("Flutter")  # Affects both, since the list is shared
print(original["skills"])
print(shallow["skills"])

# For a fully independent copy of nested structures, use deepcopy
import copy
deep = copy.deepcopy(original)
deep["skills"].append("Tableau")
print(original["skills"])
print(deep["skills"])
```

## 14. The `del` Keyword

**What it does:** Deletes a specific key-value pair from the dictionary, or deletes the entire dictionary object itself.

**When to use it:** Use it when you know for certain the key exists and simply want to remove it without needing the removed value returned (unlike `pop()`).

**Conditions necessary for its use:**
- The key must exist, otherwise a `KeyError` is raised. Use an `if key in dict` check first if unsure, or use `pop()` with a default instead.


```python
profile = {"username": "kgbenga", "email": "kgbenga@example.com", "active": True}

del profile["active"]
print(profile)

# Deleting a missing key raises an error
# del profile["age"]  # KeyError: 'age'

# Safe pattern: check before deleting
if "email" in profile:
    del profile["email"]
print(profile)
```

## 15. Membership Testing with `in` and `not in`

**What it does:** Checks whether a given key exists (or does not exist) in the dictionary. This checks keys only, not values, by default.

**When to use it:** Use it before accessing a key directly with `dict[key]`, or whenever your logic branches based on the presence or absence of a key.

**Conditions necessary for its use:**
- None. Works on any dictionary, and runs in average O(1) time due to the underlying hash table.


```python
menu = {"jollof rice": 1500, "fried rice": 1600, "fish": 2000}

print("jollof rice" in menu)
print("pounded yam" in menu)
print("pounded yam" not in menu)

# Common pattern combining membership testing with direct access
dish = "fish"
if dish in menu:
    print(f"{dish} costs {menu[dish]} naira")
```

## 16. `len(dict)`

**What it does:** Returns the number of key-value pairs currently in the dictionary.

**When to use it:** Use it whenever you need to know the size of a dictionary, for example to check if it is empty or to report how many records were processed.

**Conditions necessary for its use:**
- None.


```python
records = {"A": 1, "B": 2, "C": 3}
print(len(records))

if len(records) == 0:
    print("No records found")
else:
    print(f"{len(records)} records found")
```

## 17. Dictionary Comprehensions

**What it does:** A concise syntax `{key_expr: value_expr for item in iterable if condition}` used to build a new dictionary from an iterable in a single readable line.

**When to use it:** Use it when transforming, filtering, or building a dictionary from another sequence, instead of writing a multi-line loop with `dict[key] = value`.

**Conditions necessary for its use:**
- The `key_expr` used must always evaluate to something hashable.


```python
# Build a dictionary of squares for even numbers only
even_squares = {n: n ** 2 for n in range(1, 11) if n % 2 == 0}
print(even_squares)

# Transform an existing dictionary's values
prices = {"pen": 100, "book": 500, "bag": 3000}
discounted = {item: price * 0.9 for item, price in prices.items()}
print(discounted)

# Swap keys and values
swapped = {value: key for key, value in prices.items()}
print(swapped)
```

## 18. Merge Operators `|` and `|=` (Python 3.9+)

**What it does:** The `|` operator merges two dictionaries into a brand new one. The `|=` operator merges a dictionary into the existing one in place, similar to `update()`.

**When to use it:** Use `|` when you want a fresh merged dictionary without touching either original. Use `|=` as a shorter alternative to `update()`.

**Conditions necessary for its use:**
- Requires Python 3.9 or later.
- When both dictionaries share a key, the value from the right-hand dictionary wins.


```python
defaults = {"theme": "light", "language": "en"}
user_prefs = {"language": "fr", "notifications": True}

# Merge into a new dictionary using the pipe operator
merged = defaults | user_prefs
print(merged)

# In-place merge using |=
defaults |= user_prefs
print(defaults)
```

## 19. Sorting a Dictionary with `sorted()`

**What it does:** `sorted()` is not a dictionary method itself, but it is commonly applied to `.items()`, `.keys()`, or `.values()` to produce an ordered list, since dictionaries are ordered by insertion, not by value or key.

**When to use it:** Use it whenever you need to display or process dictionary data in a specific order, for example ranking scores from highest to lowest.

**Conditions necessary for its use:**
- The `key` argument passed to `sorted()` must return a comparable value across all items (all numbers, or all strings, not a mix).


```python
scores = {"Ada": 85, "Musa": 92, "Ife": 78, "Tunde": 92}

# Sort by key alphabetically
by_name = dict(sorted(scores.items()))
print(by_name)

# Sort by value, descending
by_score_desc = dict(sorted(scores.items(), key=lambda pair: pair[1], reverse=True))
print(by_score_desc)
```

## 20. Nested Dictionaries

**What it does:** A dictionary can contain other dictionaries as values, allowing you to represent hierarchical or structured data such as records with sub-fields.

**When to use it:** Use it when modeling real-world entities with multiple attributes, such as a company's employee records or a JSON-like structure loaded from an API.

**Conditions necessary for its use:**
- Keys at every level must still be hashable. Access to deep values requires chaining keys, and missing intermediate keys will raise a `KeyError` unless `.get()` is used carefully at each level.


```python
company = {
    "E001": {"name": "Ngozi", "department": "Analytics", "salary": 350000},
    "E002": {"name": "Femi", "department": "Engineering", "salary": 420000},
}

# Accessing nested values
print(company["E001"]["name"])

# Safely accessing a nested value that might not exist
bonus = company["E001"].get("bonus", 0)
print(bonus)

# Looping through nested dictionaries
for emp_id, details in company.items():
    print(f"{emp_id}: {details['name']} works in {details['department']}")
```
