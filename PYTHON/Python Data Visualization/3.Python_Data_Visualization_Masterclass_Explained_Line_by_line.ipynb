## Section 3 — Anatomy of a Matplotlib Chart

````python
fig, ax = plt.subplots(figsize=(8, 4))
months = ["Jan","Feb","Mar","Apr","May","Jun"]
sales  = [120000, 145000, 132000, 178000, 165000, 210000]
ax.plot(months, sales, color="royalblue", linewidth=2.5, marker="o", markersize=7)
ax.set_title("Monthly Revenue — Anatomy of a Matplotlib Chart", fontsize=14, fontweight="bold")
ax.set_xlabel("Month (X-Axis)", fontsize=12)
ax.set_ylabel("Revenue in NGN (Y-Axis)", fontsize=12)
ax.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
print("fig = entire canvas | ax = chart area inside canvas")
````

#### LINE-BY-LINE BREAKDOWN

`fig, ax = plt.subplots(figsize=(8, 4))` — the fundamental matplotlib setup line:
- Creates a **figure** (`fig`) = the blank canvas
- Creates an **axes** (`ax`) = the actual chart drawing area placed on that canvas
- `figsize=(8, 4)` → 8 inches wide, 4 inches tall
- **Tuple unpacking**: `subplots()` returns two objects; we catch both simultaneously

`months = [...]` and `sales = [...]` — two paired Python lists. Jan corresponds to 120000, Feb to 145000, etc.

`ax.plot(months, sales, color="royalblue", linewidth=2.5, marker="o", markersize=7)`:
- `months, sales` → X and Y data
- `color="royalblue"` → named colour (also accepts hex like `"#5B8DB8"`)
- `linewidth=2.5` → line thickness in points
- `marker="o"` → circle dot at each data point. Other options: `"s"`=square, `"^"`=triangle, `"D"`=diamond
- `markersize=7` → diameter of each dot

`ax.set_title(...)`, `ax.set_xlabel(...)`, `ax.set_ylabel(...)` — text labels. Always label your axes — unlabelled charts are meaningless to anyone else.

`ax.grid(axis="y", linestyle="--", alpha=0.5)` — horizontal grid lines only, dashed style, 50% transparent.

`plt.tight_layout()` — auto-adjusts spacing so labels don't get cut off. Always call before `plt.show()`.

`plt.show()` — renders and displays the chart. Without this, the chart exists in memory but is never shown.

#### OUTPUT AND INTERPRETATION

A simple 6-point line chart showing rising sales from Jan to Jun with a dip in Mar.

The fig/ax mental model:
```
fig  (the whole canvas)
 └── ax  (the chart area)
     ├── Line  (the data)
     ├── Title, X-axis label, Y-axis label
     └── Grid lines
```
Everything you draw goes onto `ax`. This pattern repeats in every matplotlib chart you will ever write.

## Section 4 — All Chart Types Explained
### 4.1 Line Chart — Monthly Total Revenue

````python
month_order = ["January","February","March","April","May","June",
               "July","August","September","October","November","December"]

monthly_rev = (df.groupby("Month")["Revenue_NGN"]
                 .sum()
                 .reindex(month_order)
                 .dropna()
                 .reset_index())

fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(monthly_rev["Month"], monthly_rev["Revenue_NGN"],
        color="steelblue", linewidth=2.5, marker="o", markersize=8,
        markerfacecolor="white", markeredgewidth=2)

for i, row in monthly_rev.iterrows():
    ax.annotate(f"₦{row['Revenue_NGN']/1e6:.1f}M",
                xy=(row["Month"], row["Revenue_NGN"]),
                xytext=(0, 10), textcoords="offset points",
                ha="center", fontsize=9, color="steelblue")

ax.set_title("Monthly Total Revenue — Nigeria Retail Sales 2023", fontsize=14, fontweight="bold")
ax.set_xlabel("Month"); ax.set_ylabel("Total Revenue (NGN)")
ax.tick_params(axis="x", rotation=30)
plt.tight_layout()
plt.show()
````

#### LINE-BY-LINE BREAKDOWN

`month_order` — defines the correct chronological order. Without this, pandas sorts months alphabetically (April, August, December...) which destroys the time-series meaning.

**Method chaining** — each step feeds its result into the next:
- `df.groupby("Month")` → 12 month groups
- `["Revenue_NGN"].sum()` → one total revenue number per month
- `.reindex(month_order)` → reorders results Jan to Dec
- `.dropna()` → removes months with no sales data
- `.reset_index()` → converts Month from the index into a regular column

`markerfacecolor="white", markeredgewidth=2` — hollow circle: white fill inside, thick ring outside.

**Annotation loop:**
- `monthly_rev.iterrows()` → loops row by row, returning index `i` and row data `row`
- `f"₦{row['Revenue_NGN']/1e6:.1f}M"` → formats as millions. `1e6` = 1,000,000. ₦45,000,000 → "₦45.0M"
- `xytext=(0, 10)` → places label 10 pixels ABOVE the data point
- `textcoords="offset points"` → the offset is in screen pixels, not data units
- `ha="center"` → centers the label horizontally over the point

`ax.tick_params(axis="x", rotation=30)` — rotates X-axis labels 30 degrees to prevent overlap.

#### OUTPUT AND INTERPRETATION

A line tracking 12 months, each point labelled with revenue in millions. Rising trend = growing business. Peak month = strongest demand period. Dips = slow seasons or data gaps.


### 4.1 Line Chart — Multiple Lines by Channel

````python
channel_month = (df.groupby(["Month","Sales_Channel"])["Revenue_NGN"]
                   .sum().reset_index())
channel_month["Month"] = pd.Categorical(channel_month["Month"], categories=month_order, ordered=True)
channel_month = channel_month.sort_values("Month")

fig, ax = plt.subplots(figsize=(13, 5))
colors = {"Online":"royalblue","Physical Store":"coral","Wholesale":"seagreen"}

for channel, grp in channel_month.groupby("Sales_Channel"):
    ax.plot(grp["Month"], grp["Revenue_NGN"],
            label=channel, color=colors[channel], linewidth=2.5, marker="s", markersize=6)

ax.set_title("Monthly Revenue by Sales Channel", fontsize=14, fontweight="bold")
ax.set_xlabel("Month"); ax.set_ylabel("Revenue (NGN)")
ax.legend(title="Sales Channel", loc="upper left")
ax.tick_params(axis="x", rotation=30)
plt.tight_layout()
plt.show()
````

#### LINE-BY-LINE BREAKDOWN

`groupby(["Month","Sales_Channel"])` — groups by TWO columns simultaneously, creating up to 12x3=36 groups.

`pd.Categorical(..., ordered=True)` — makes Month an ordered category type so `.sort_values()` sorts chronologically (Jan to Dec), not alphabetically.

`colors = {"Online":"royalblue", ...}` — a Python **dictionary** mapping each channel name to a colour. `colors[channel]` looks up the colour for the current channel in the loop.

**The loop:** loops over 3 channel groups. `channel` = channel name (string), `grp` = that channel's rows (12 months). `marker="s"` = square markers.

`ax.legend(title="Sales Channel", loc="upper left")` — creates the legend box in the top-left corner.

#### OUTPUT AND INTERPRETATION

3 coloured lines. Lines converging = channels performing similarly. Lines diverging = one channel dominating. Lines crossing = a channel overtook another — a strategic reversal point.

### 4.2 Bar Charts — Vertical Bar by State

````python
state_rev = df.groupby("State")["Revenue_NGN"].sum().sort_values(ascending=False).reset_index()

fig, ax = plt.subplots(figsize=(11, 5))
bars = ax.bar(state_rev["State"], state_rev["Revenue_NGN"],
              color=sns.color_palette("Blues_d", len(state_rev)),
              edgecolor="white", linewidth=0.8)

for bar in bars:
    h = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2, h + 200000,
            f"₦{h/1e6:.1f}M", ha="center", va="bottom", fontsize=9, fontweight="bold")

ax.set_title("Total Revenue by State — Nigeria Retail Sales 2023", fontsize=14, fontweight="bold")
ax.set_xlabel("State"); ax.set_ylabel("Total Revenue (NGN)")
plt.tight_layout()
plt.show()
````

#### LINE-BY-LINE BREAKDOWN

`.sort_values(ascending=False)` — sorts from highest to lowest, creating a natural ranking left to right.

`bars = ax.bar(...)` — saves the return value (a list of bar objects) so we can loop over them for labelling.

`color=sns.color_palette("Blues_d", len(state_rev))` — generates 8 shades of blue. Combined with descending sort: tallest bar = darkest.

**Label loop** for each bar:
- `h = bar.get_height()` → the bar's height = its revenue total
- `bar.get_x() + bar.get_width()/2` → horizontal center of the bar
- `h + 200000` → places text slightly above the bar top
- `f"₦{h/1e6:.1f}M"` → formats as millions e.g. "₦45.2M"

#### OUTPUT AND INTERPRETATION

8 bars in descending height, each labelled with total revenue in millions. Tallest bar = highest-revenue state = best market.


#### Horizontal Bar — Profit by Category

````python
cat_profit = df.groupby("Category")["Profit_NGN"].sum().sort_values().reset_index()

fig, ax = plt.subplots(figsize=(10, 5))
bars = ax.barh(cat_profit["Category"], cat_profit["Profit_NGN"],
               color=sns.color_palette("Greens_d", len(cat_profit)), edgecolor="white")

for bar in bars:
    w = bar.get_width()
    ax.text(w + 50000, bar.get_y() + bar.get_height()/2,
            f"₦{w/1e6:.2f}M", va="center", fontsize=9, fontweight="bold")

ax.set_title("Total Profit by Product Category", fontsize=14, fontweight="bold")
ax.set_xlabel("Total Profit (NGN)"); ax.set_ylabel("Category")
plt.tight_layout()
plt.show()
````

#### LINE-BY-LINE BREAKDOWN

`.sort_values()` (ascending=True, default) — for horizontal bars, ascending sort places smallest at top and largest at bottom — the most natural reading direction.

`ax.barh()` vs `ax.bar()` — the `h` means horizontal. Axes are swapped: first argument → Y-axis (category names), second argument → X-axis (bar length = profit value).

For horizontal bar labels: `w = bar.get_width()` → `get_width()` returns the measured value. `bar.get_y() + bar.get_height()/2` → vertical center of each bar. `w + 50000` → slightly to the right of the bar end.

**Why horizontal?** Long category names like "Food & Beverage" overlap on vertical bars.

#### OUTPUT AND INTERPRETATION

6 horizontal bars sorted shortest (top) to longest (bottom). Longest bar = most profitable product line.

- `width=0.7` → each group occupies 70% of its slot

When using `.plot()` on a DataFrame, use `plt.title()` etc. instead of `ax.set_title()`.

#### Stacked Bar — Grade Distribution

````python
grade_prog = students.groupby(["Programme","Grade"]).size().unstack(fill_value=0)
grade_prog = grade_prog[["A","B","C","D","F"]]

grade_prog.plot(kind="bar", stacked=True, figsize=(12, 5),
                colormap="RdYlGn", edgecolor="white")

plt.title("Grade Distribution by Programme (Stacked)", fontsize=14, fontweight="bold")
plt.xlabel("Programme"); plt.ylabel("Number of Students")
plt.xticks(rotation=20, ha="right")
plt.legend(title="Grade", bbox_to_anchor=(1.01, 1))
plt.tight_layout()
plt.show()
````

#### LINE-BY-LINE BREAKDOWN

`.size()` — counts rows per group (how many students received each grade per programme). We want COUNTS, not `.sum()` or `.mean()`. `fill_value=0` fills missing combinations with 0.

`grade_prog[["A","B","C","D","F"]]` — forces grade columns into logical order A to F.

`stacked=True` — transforms grouped bars into stacked bars. Total bar height = total students in that programme.

`colormap="RdYlGn"` — Red-Yellow-Green. Grade A = dark green, Grade F = dark red. Colour instantly communicates quality.

`bbox_to_anchor=(1.01, 1)` — positions legend outside the chart to the right, preventing it from covering bars.

#### OUTPUT AND INTERPRETATION

One stacked bar per programme divided into coloured grade segments. Tall green = many A students. Large red = many failures. Total bar height = enrolment size.

### 4.3 Histogram — Customer Age Distribution

````python
fig, ax = plt.subplots(figsize=(10, 5))
ax.hist(df["Customer_Age"], bins=20, color="steelblue", edgecolor="white", linewidth=0.8)
ax.axvline(df["Customer_Age"].mean(), color="red", linestyle="--", linewidth=1.8,
           label=f"Mean Age = {df['Customer_Age'].mean():.1f}")
ax.axvline(df["Customer_Age"].median(), color="orange", linestyle="--", linewidth=1.8,
           label=f"Median Age = {df['Customer_Age'].median():.1f}")
ax.set_title("Distribution of Customer Age", fontsize=14, fontweight="bold")
ax.set_xlabel("Customer Age (years)"); ax.set_ylabel("Number of Customers")
ax.legend()
plt.tight_layout()
plt.show()
print(f"Mean:   {df['Customer_Age'].mean():.2f}")
print(f"Median: {df['Customer_Age'].median():.2f}")
print(f"Std:    {df['Customer_Age'].std():.2f}")
````

#### LINE-BY-LINE BREAKDOWN

`ax.hist(df["Customer_Age"], bins=20, ...)` — passes all 500 age values to the histogram.
- `bins=20` → divides the age range into 20 equal-width buckets. Python counts how many customers fall in each bucket and draws a bar. 20 bins is a good default for ~500 data points.
- `edgecolor="white"` → white border between bins for visual separation

`ax.axvline(...)` — draws a full-height vertical line at a specific X position.
- First call: red dashed line at the mean. `linestyle="--"` = dashed.
- Second call: orange dashed line at the median. Different colour to distinguish from mean.
- `label=f"Mean Age = {df['Customer_Age'].mean():.1f}"` → legend text. `:.1f` = 1 decimal place.

`df["Customer_Age"].std()` — standard deviation: how spread out the ages are. If std=15 and mean=35, most customers are aged 20 to 50 (mean plus or minus std).

#### OUTPUT AND INTERPRETATION

Histogram bars with red (mean) and orange (median) dashed vertical lines.

Distribution shape guide:
- Mean = Median → roughly symmetric
- Mean > Median → right-skewed (few very old customers pull the average up)
- Mean < Median → left-skewed (few very young customers pull the average down)

Business insight: if most customers are 25-40, marketing should target that demographic.

### 4.4 Box Plot — Revenue by Category

````python
fig, ax = plt.subplots(figsize=(12, 5))
sns.boxplot(data=df, x="Category", y="Revenue_NGN", ax=ax,
            palette="Set2", width=0.5, flierprops=dict(marker="o", color="red", alpha=0.5))
ax.set_title("Revenue Distribution by Product Category", fontsize=14, fontweight="bold")
ax.set_xlabel("Category"); ax.set_ylabel("Revenue per Transaction (NGN)")
ax.tick_params(axis="x", rotation=15)
plt.tight_layout()
plt.show()
print("Red dots = outliers. Line in box = median. Tall box = high variability.")
````

#### LINE-BY-LINE BREAKDOWN

`sns.boxplot(data=df, x="Category", y="Revenue_NGN", ...)` — seaborn calculates all 5 box plot statistics automatically:
- `x="Category"` → one box per category on the X-axis
- `y="Revenue_NGN"` → the distribution shown for each group
- `palette="Set2"` → 8 soft, distinct colours
- `width=0.5` → each box occupies 50% of its slot
- `flierprops=dict(marker="o", color="red", alpha=0.5)` → outlier dots: red circles, 50% transparent

**What seaborn calculates automatically:**

| Part | Formula |
|------|---------|
| Box bottom | Q1 = 25th percentile |
| Line in box | Median = 50th percentile |
| Box top | Q3 = 75th percentile |
| Lower whisker | Lowest value >= Q1 - 1.5 x IQR |
| Upper whisker | Highest value <= Q3 + 1.5 x IQR |
| Red dots | Values beyond the whiskers (outliers) |

#### OUTPUT AND INTERPRETATION

6 coloured box plots side by side.
- Median near bottom of box → right-skewed (few very high-value transactions)
- Tall box = high IQR → highly variable revenues (Electronics: cheap phones vs expensive laptops)
- Short box → consistent values (Books: narrow price range)
- Red dots above top whisker → unusually expensive purchases

`palette={"Male":"#5B8DB8","Female":"#E07B54"}` — dictionary mapping each gender to a specific hex colour.

#### Annotated 5-Number Summary

````python
data_sample = df[df["Category"] == "Electronics"]["Revenue_NGN"]
q1  = data_sample.quantile(0.25)
med = data_sample.median()
q3  = data_sample.quantile(0.75)
iqr = q3 - q1
lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

fig, ax = plt.subplots(figsize=(8, 5))
ax.boxplot(data_sample, vert=True, patch_artist=True,
           boxprops=dict(facecolor="lightblue", color="navy"),
           medianprops=dict(color="red", linewidth=2.5),
           whiskerprops=dict(color="navy"),
           flierprops=dict(marker="o", color="orange", markersize=6))
ax.set_title("Electronics Revenue — Annotated Box Plot", fontsize=13, fontweight="bold")
ax.set_ylabel("Revenue (NGN)"); ax.set_xticklabels(["Electronics"])

print("5-Number Summary for Electronics Revenue")
print(f"  Min (excl outliers): ₦{data_sample[data_sample >= lower_fence].min():,.0f}")
print(f"  Q1 (25th pct):       ₦{q1:,.0f}")
print(f"  Median (50th pct):   ₦{med:,.0f}")
print(f"  Q3 (75th pct):       ₦{q3:,.0f}")
print(f"  Max (excl outliers): ₦{data_sample[data_sample <= upper_fence].max():,.0f}")
print(f"  IQR (Q3 - Q1):       ₦{iqr:,.0f}")
print(f"  Upper fence:         ₦{upper_fence:,.0f}")
print(f"  Outlier count:       {(data_sample > upper_fence).sum()} transactions")
plt.tight_layout()
plt.show()
````

#### LINE-BY-LINE BREAKDOWN

`df[df["Category"] == "Electronics"]["Revenue_NGN"]` — Boolean filtering: `df["Category"] == "Electronics"` creates a True/False mask; `df[...]` keeps only rows where True; `["Revenue_NGN"]` selects just the revenue column.

Manual 5-number calculation:
- `quantile(0.25)` → Q1: 25% of revenues are below this value
- `.median()` → Q2 = middle value
- `quantile(0.75)` → Q3: 75% are below this
- `iqr = q3 - q1` → **Interquartile Range**: span containing the middle 50% of data
- `lower_fence = q1 - 1.5 * iqr` → Tukey's lower outlier threshold
- `upper_fence = q3 + 1.5 * iqr` → Tukey's upper outlier threshold

`patch_artist=True` — fills the box with colour (default is hollow).

`boxprops`, `medianprops`, `whiskerprops`, `flierprops` — styling dictionaries for each box plot element.

`:,.0f` format spec — `,` adds thousands separators (1450000 → 1,450,000), `.0f` = zero decimals.

`(data_sample > upper_fence).sum()` — True/False series summed = count of outliers (True=1, False=0).

#### OUTPUT AND INTERPRETATION

A styled box plot (light blue fill, red median line, orange outlier dots) + printed 5-number summary. The printed numbers map directly to the visual: Q1 = bottom of box, Q3 = top of box, Median = red line, upper fence = top of top whisker.

### 4.7 Pie Chart and Donut Chart

````python
channel_rev = df.groupby("Sales_Channel")["Revenue_NGN"].sum()
colors = ["#5B8DB8", "#E07B54", "#6DBE6D"]
explode = [0.04] * len(channel_rev)

fig, ax = plt.subplots(figsize=(8, 8))
wedges, texts, autotexts = ax.pie(
    channel_rev, labels=channel_rev.index, autopct="%1.1f%%",
    colors=colors, explode=explode, startangle=140,
    pctdistance=0.75, textprops={"fontsize": 12}
)
for t in autotexts:
    t.set_fontweight("bold"); t.set_fontsize(13)
ax.set_title("Revenue Share by Sales Channel", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.show()
````

#### LINE-BY-LINE BREAKDOWN

`explode = [0.04] * len(channel_rev)` — `[0.04] * 3` creates `[0.04, 0.04, 0.04]`. Each value = how far to pull that slice outward (4% of pie radius). Uniform value creates a slight gap between all slices.

`ax.pie(...)` key parameters:
- `channel_rev` → values (determines slice sizes proportionally)
- `labels=channel_rev.index` → text labels for each slice
- `autopct="%1.1f%%"` → auto-calculates percentages. `%1.1f` = 1 digit + 1 decimal. `%%` = literal % sign.
- `explode=explode` → pull-out distances per slice
- `startangle=140` → rotates the first slice 140 degrees (aesthetic choice)
- `pctdistance=0.75` → percentage labels at 75% of the way from centre to edge

`ax.pie()` returns 3 objects: wedges (slices), texts (labels), autotexts (percentages). We loop autotexts to make them bold and larger.

#### OUTPUT AND INTERPRETATION

A 3-slice pie with percentage labels. Use a bar chart instead if slices are similar sizes — human eyes cannot accurately judge which angle is larger when proportions are close.

### 4.16 Waterfall Chart

````python
labels  = ["Gross Revenue","Discounts","Returns","Net Revenue","COGS","Gross Profit"]
values  = [850000000, -62000000, -18000000, 0, -420000000, 0]
measure = ["absolute","relative","relative","total","relative","total"]

fig = go.Figure(go.Waterfall(
    name="2023 P&L", orientation="v",
    measure=measure, x=labels, y=values,
    text=[f"₦{abs(v)/1e6:.0f}M" for v in [850e6,-62e6,-18e6,770e6,-420e6,350e6]],
    textposition="outside",
    connector={"line": {"color": "gray", "dash":"dot"}},
    increasing={"marker":{"color":"#27AE60"}},
    decreasing={"marker":{"color":"#E74C3C"}},
    totals={"marker":{"color":"#2980B9"}},
))
fig.update_layout(title="2023 P&L Waterfall — Nigeria Retail", height=480)
fig.show()
````

#### LINE-BY-LINE BREAKDOWN

`measure` list controls how each bar is positioned:
- `"absolute"` → bar starts from zero (Gross Revenue: standalone bar)

- `"total"` → draws the current cumulative total starting from zero (Net Revenue, Gross Profit)

When `measure="total"`, the values entry is 0 because the total is auto-calculated from all relative bars before it.

`text=[f"₦{abs(v)/1e6:.0f}M" for v in [...]]` — list comprehension creating label text. `abs(v)` removes minus signs from negative values.

Colour coding: green = money coming in, red = money going out, blue = running subtotals.

#### OUTPUT AND INTERPRETATION

### 4.18 Radar (Spider) Chart

````python
categories = ["Quiz_Score","Midterm_Score","Final_Score","Project_Score","Attendance_Pct"]
prog_means = students.groupby("Programme")[categories].mean()

angles = np.linspace(0, 2 * np.pi, len(categories), endpoint=False).tolist()
angles += angles[:1]  # close the polygon

fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))
colors = ["#5B8DB8","#E07B54","#6DBE6D","#F5A623","#9B59B6"]

for idx, (prog, row) in enumerate(prog_means.iterrows()):
    values = row.tolist() + row.tolist()[:1]
    ax.plot(angles, values, color=colors[idx], linewidth=2, label=prog)
    ax.fill(angles, values, color=colors[idx], alpha=0.07)

ax.set_xticks(angles[:-1])
ax.set_xticklabels(["Quiz","Midterm","Final Exam","Project","Attendance"], fontsize=11)
ax.set_ylim(0, 100)
ax.set_title("Programme Performance Radar Chart", fontsize=14, fontweight="bold", pad=20)
ax.legend(loc="upper right", bbox_to_anchor=(1.3, 1.1), title="Programme")
plt.tight_layout()
plt.show()
print("Larger polygon = stronger overall performance. Uneven shape = specific weakness.")
````

#### LINE-BY-LINE BREAKDOWN

`np.linspace(0, 2 * np.pi, len(categories), endpoint=False)` — divides a full circle (2 pi radians = 360 degrees) into 5 equally-spaced angles. `endpoint=False` excludes 360 degrees (same as 0 degrees, would create a duplicate spoke).

`angles += angles[:1]` — **critical**: appends the first angle to the end. Forces the polygon to close. Without this, the shape is open and incomplete.

`subplot_kw=dict(polar=True)` — switches the axes from rectangular to **polar coordinates** (the circular radar layout).

`values = row.tolist() + row.tolist()[:1]` — same polygon-closing trick for data values. 5 scores + repeat the first = 6 values. Ensures the shape closes cleanly.

`ax.fill(angles, values, color=..., alpha=0.07)` — fills the polygon interior with 7% opacity. Overlapping polygons remain distinguishable.

#### OUTPUT AND INTERPRETATION

5 overlapping coloured pentagons on a circular grid. Each corner = one performance dimension. Larger polygon = better overall performance. One corner extending far out = programme excels specifically at that metric. One corner close to the centre = weakness in that metric. Balanced, even shape = consistent performance across all dimensions.

## Section 8 — Automated Insight Generator

````python
def generate_retail_insights(df):
    print("=" * 65)
    print("   NIGERIA RETAIL SALES 2023 — AUTOMATED INSIGHTS REPORT")
    print("=" * 65)
    total_rev  = df["Revenue_NGN"].sum()
    total_prof = df["Profit_NGN"].sum()
    total_tx   = len(df)
    avg_rating = df["Customer_Rating"].mean()
    print(f"
OVERVIEW")
    print(f"  Total Revenue:       ₦{total_rev:>14,.0f}")
    print(f"  Total Profit:        ₦{total_prof:>14,.0f}")
    print(f"  Profit Margin:        {total_prof/total_rev*100:.1f}%")
    print(f"  Total Transactions:   {total_tx:,}")
    print(f"  Avg Customer Rating:  {avg_rating:.2f} / 5.00")

    top_state = df.groupby("State")["Revenue_NGN"].sum().idxmax()
    top_state_rev = df.groupby("State")["Revenue_NGN"].sum().max()
    top_cat = df.groupby("Category")["Revenue_NGN"].sum().idxmax()
    top_channel = df.groupby("Sales_Channel")["Revenue_NGN"].sum().idxmax()
    top_pct = df.groupby("Sales_Channel")["Revenue_NGN"].sum().max() / total_rev * 100
    print(f"
TOP PERFORMERS")
    print(f"  Highest Revenue State:    {top_state} (₦{top_state_rev/1e6:.1f}M)")
    print(f"  Highest Revenue Category: {top_cat}")
    print(f"  Best Sales Channel:       {top_channel} ({top_pct:.1f}% of revenue)")

    month_rev = df.groupby("Month")["Revenue_NGN"].sum()
    print(f"
SEASONALITY")
    print(f"  Best Month:  {month_rev.idxmax()} (₦{month_rev.max()/1e6:.1f}M)")
    print(f"  Worst Month: {month_rev.idxmin()} (₦{month_rev.min()/1e6:.1f}M)")

    return_rate = df["Item_Returned"].mean() * 100
    avg_disc = df["Discount_Percent"].mean()
    low_cat = df.groupby("Category")["Customer_Rating"].mean().sort_values()
    print(f"
RISK FLAGS")
    print(f"  Return Rate:       {return_rate:.1f}% of transactions")
    print(f"  Avg Discount:      {avg_disc:.1f}%")
    print(f"  Lowest Rated Cat:  {low_cat.index[0]} ({low_cat.iloc[0]:.2f} stars)")
    print("
" + "=" * 65)

generate_retail_insights(df_retail)
````

#### LINE-BY-LINE BREAKDOWN

`def generate_retail_insights(df):` — defines a **reusable function**. Pass any DataFrame with the same column structure and it generates the same report. Call it next year with `generate_retail_insights(df_2024)` — no code changes needed.

`f"  Total Revenue:       ₦{total_rev:>14,.0f}"`:
- `>14` → right-align in a 14-character field (numbers align vertically)
- `,` → thousands separators: 850000000 → 850,000,000
- `.0f` → zero decimal places

`.idxmax()` — returns the **label** of the maximum value. If state revenues are `{"Lagos": 15M, "Abuja": 12M, ...}`, `.idxmax()` returns "Lagos".

`df["Item_Returned"].mean() * 100` — Item_Returned is binary (0 or 1). Mean of binary = fraction of 1s = return rate as decimal. Times 100 = percentage.

`low_cat.index[0]` — after `.sort_values()` ascending, index position `[0]` = the lowest-rated category.

#### OUTPUT AND INTERPRETATION

A formatted text report with 4 sections: Overview, Top Performers, Seasonality, Risk Flags.

Business value: replaces 10+ separate queries with one function call. Schedulable and automatable. Risk Flags section surfaces problems immediately: high return rate = product quality issue, high average discount = margin erosion.


````python
def generate_student_insights(df, prog_summary):
    print("=" * 65)
    print("   PRODIGY TRAINING HUB — STUDENT PERFORMANCE INSIGHTS")
    print("=" * 65)
    total_students = len(df)
    completion_rate = (df["Completion_Status"] == "Completed").mean() * 100
    employment_rate = df["Employed_After"].mean() * 100
    avg_score = df["Total_Score"].mean()
    avg_sat = df["Satisfaction_Rating"].mean()
    print(f"
OVERVIEW")
    print(f"  Total Students Enrolled:  {total_students}")
    print(f"  Completion Rate:          {completion_rate:.1f}%")
    print(f"  Employment Rate:          {employment_rate:.1f}%")
    print(f"  Average Total Score:      {avg_score:.1f} / 100")
    print(f"  Average Satisfaction:     {avg_sat:.2f} / 5")

    best_score  = prog_summary.sort_values("Avg_Total", ascending=False).iloc[0]
    best_employ = prog_summary.sort_values("Employment_Rate", ascending=False).iloc[0]
    print(f"
TOP PROGRAMMES")
    print(f"  Highest Avg Score:    {best_score['Programme']} ({best_score['Avg_Total']:.1f})")
    print(f"  Highest Employment:   {best_employ['Programme']} ({best_employ['Employment_Rate']*100:.1f}%)")

    grade_dist = df["Grade"].value_counts(normalize=True) * 100
    print(f"
GRADE DISTRIBUTION")
    for g in ["A","B","C","D","F"]:
        print(f"  Grade {g}: {grade_dist.get(g, 0):.1f}%")

    dropout_rate = (df["Completion_Status"] == "Dropped Out").mean() * 100
    low_attend = (df["Attendance_Pct"] < 50).mean() * 100
    print(f"
RISK FLAGS")
    print(f"  Dropout Rate:                {dropout_rate:.1f}%")
    print(f"  Students below 50% Attend:   {low_attend:.1f}%")
    print("
" + "=" * 65)

generate_student_insights(students_raw, prog_summary)
````

#### LINE-BY-LINE BREAKDOWN

`(df["Completion_Status"] == "Completed").mean() * 100` — creates a True/False series, takes the mean (fraction of True values), multiplies by 100 to get a percentage.

`grade_dist.get(g, 0)` — `.get(key, default)` — returns the percentage for that grade if it exists; returns 0.0 if somehow absent.

`(df["Attendance_Pct"] < 50).mean() * 100` — counts the fraction of students with attendance below 50%. A key at-risk indicator.

`prog_summary.sort_values("Avg_Total", ascending=False).iloc[0]` — sorts by Avg_Total descending, then `.iloc[0]` takes the first row. `.iloc` = integer-location based indexing (select by position).

#### OUTPUT AND INTERPRETATION

A formatted student performance report covering overview KPIs, top programmes, grade distribution, and risk flags.

Key metrics: dropout rate above 15% → urgently review programme delivery. Students below 50% attendance → highly likely to fail; build an early-warning intervention system. Grade distribution centred at B/C = healthy cohort.

#### LINE-BY-LINE BREAKDOWN

Three `plt.savefig()` calls saving three different formats:

| Format | Best for | Notes |
|--------|----------|-------|
| .png | PowerPoint, Word, email | `dpi=150` = crisp on screen AND when printed |
| .svg | Websites, Figma, Illustrator | Vector — scales to any size without blurring |
| .pdf | Academic journals, formal reports | Vector, small file size |

`dpi` guide: 72 = screen default (blurry when printed). 150 = good balance for business reports. 300 = print quality for academic journals.

`bbox_inches="tight"` — trims all excess whitespace automatically.

**CRITICAL**: `plt.savefig()` MUST be called BEFORE `plt.show()`. After `show()`, matplotlib clears the figure from memory — saving after produces a blank file.

#### OUTPUT AND INTERPRETATION

Three files saved in your notebook's folder. PNG → paste into PowerPoint or email. SVG → upload to website or open in Canva. PDF → include in formal printed report.
