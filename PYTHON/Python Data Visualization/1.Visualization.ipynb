## 2. Line Chart

When to Use

- Time series data

- Trends over time

- Continuous data

Example Use-Cases

- Stock prices over time

- Monthly sales growth

- Temperature changes

```python
# Sample Data
days = [1, 2, 3, 4, 5]
sales = [200, 250, 300, 280, 350]

plt.plot(days, sales)
plt.xlabel('Days')
plt.ylabel('Sales')
plt.title('Daily Sales Trend')
plt.show()
```

### Line-by-Line Explanation

| Line                                | Explanation                                                  |
| ----------------------------------- | ------------------------------------------------------------ |
| `days = [1, 2, 3, 4, 5]`            | Creates the x-axis values (days).                            |
| `sales = [200, 250, 300, 280, 350]` | Creates the y-axis values (sales figures).                   |
| `plt.plot(days, sales)`             | Draws a line chart using days on x-axis and sales on y-axis. |
| `plt.xlabel('Days')`                | Labels the x-axis as "Days".                                 |
| `plt.ylabel('Sales')`               | Labels the y-axis as "Sales".                                |
| `plt.title('Daily Sales Trend')`    | Adds a chart title.                                          |
| `plt.show()`                        | Displays the chart.                                          |

## 6. Box Plot

When to Use

- Identify outliers

- Compare distributions

Example Use-Cases

| Line                         | Explanation                                                 |

| `scores = [...]`             | Dataset.                                                    |

Shows

- Minimum

- Q1

- Median

- Q3

- Maximum

- Outliers

### Box plot with outliers

In data science, `an outlier is a value or observation that lies an abnormal distance from other values in a dataset.` For example, in the dataset below, showing the scores of students `{50,55,65,70,80, 100, 150, 220}`, the number 220 is an outlier because it deviates wildly from the main cluster of scores.

**NOTE: Outliers can be caused by human error (like a typo) or by genuine, extreme anomalies.**

```python
# Box plot with outliers
scores = [50,55,65,70,80, 100, 150, 220]
sns.boxplot(data=[scores])
plt.title('Score Spread') 
plt.show()
```

| `plt.title(...)`                | Adds title.                                              |
| `plt.show()`                    | Displays heatmap.                                        |

`Note : In real projects, heatmaps usually visualize correlation matrices.`

### Sales Age Segmentation Heatmap

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample sales dataset
data = {
    'Age_Group': ['18-25', '18-25', '18-25',
                  '26-35', '26-35', '26-35',
                  '36-45', '36-45', '36-45',
                  '46-55', '46-55', '46-55'],

    'Product': ['Laptop', 'Phone', 'Tablet',
                'Laptop', 'Phone', 'Tablet',
                'Laptop', 'Phone', 'Tablet',
                'Laptop', 'Phone', 'Tablet'],

    'Sales': [12000, 15000, 8000,
              25000, 30000, 18000,
              20000, 22000, 12000,
              10000, 9000, 5000]
}

df = pd.DataFrame(data)

heatmap_data = df.pivot(
    index='Age_Group',
    columns='Product',
    values='Sales'
)

plt.figure(figsize=(8,5))

sns.heatmap(
    heatmap_data,
    annot=True,
    cmap='YlGnBu',
    fmt=','
)

plt.title('Sales by Age Group and Product')
plt.xlabel('Product')
plt.ylabel('Age Group')

plt.show()
```

## 15a. Basic Donut Chart

```python
import matplotlib.pyplot as plt

# Product revenue data
products = ['Product A', 'Product B', 'Product C']
revenue = [5000, 7000, 6000]

# Create figure
plt.figure(figsize=(6, 6))

# Create donut chart
plt.pie(
    revenue,
    labels=products,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops=dict(width=0.4)
)

# Add title
plt.title('Revenue Share by Product')

# Keep chart circular
plt.axis('equal')

# Display chart
plt.show()
```

### LINE BY LINE EXPLANATION

| Line of Code                                         | What It Does                                                                                                                      |
| ---------------------------------------------------- | --------------------------------------------------------------------------------------------------------------------------------- |
| `import matplotlib.pyplot as plt`                    | Imports Matplotlib's plotting module and assigns it the alias `plt`. This library is used for creating charts and visualizations. |
| `products = ['Product A', 'Product B', 'Product C']` | Creates a list containing the product names that will appear as labels on the donut chart.                                        |
| `revenue = [5000, 7000, 6000]`                       | Creates a list containing the revenue values associated with each product.                                                        |
| `plt.figure(figsize=(6, 6))`                         | Creates a new figure that is 6 inches wide and 6 inches high. A square figure helps keep the donut chart circular.                |
| `plt.pie(...)`                                       | Creates a pie chart which will later be transformed into a donut chart.                                                           |
| `revenue`                                            | Supplies the numerical values used to determine the size of each slice.                                                           |
| `labels=products`                                    | Displays product names beside the corresponding slices.                                                                           |
| `autopct='%1.1f%%'`                                  | Displays percentage values on the chart with one decimal place.                                                                   |
| `startangle=90`                                      | Rotates the chart so the first slice begins at the top (90 degrees).                                                              |
| `wedgeprops=dict(width=0.4)`                         | Creates the donut hole by reducing the slice thickness to 40% of the radius.                                                      |
| `plt.title('Revenue Share by Product')`              | Adds a title describing the chart.                                                                                                |
| `plt.axis('equal')`                                  | Ensures the chart maintains a perfect circular shape instead of appearing oval.                                                   |
| `plt.show()`                                         | Displays the completed donut chart.                                                                                               |


## 15b. Professional Donut Chart

```python
import matplotlib.pyplot as plt

# Product revenue data
products = ['Product A', 'Product B', 'Product C']
revenue = [5000, 7000, 6000]

# Create figure
fig, ax = plt.subplots(figsize=(7, 7))

# Create donut chart
ax.pie(
    revenue,
    labels=products,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops=dict(width=0.4)
)

# Add center text
ax.text(
    0, 0,
    'Revenue\n18,000',
    ha='center',
    va='center',
    fontsize=14,
    fontweight='bold'
)

# Add title
plt.title('Revenue Share by Product')

# Display chart
plt.show()
```

| Line of Code                                         | What It Does                                                           |
| ---------------------------------------------------- | ---------------------------------------------------------------------- |
| `import matplotlib.pyplot as plt`                    | Imports Matplotlib's plotting module for creating visualizations.      |
| `products = ['Product A', 'Product B', 'Product C']` | Defines the product categories.                                        |
| `revenue = [5000, 7000, 6000]`                       | Defines revenue values for each product.                               |
| `fig, ax = plt.subplots(figsize=(7,7))`              | Creates a figure and axes object with dimensions of 7 by 7 inches.     |
| `ax.pie(...)`                                        | Creates the donut chart on the axes object.                            |
| `revenue`                                            | Supplies the values that determine slice sizes.                        |
| `labels=products`                                    | Displays product names around the donut chart.                         |
| `autopct='%1.1f%%'`                                  | Displays percentages with one decimal place.                           |
| `startangle=90`                                      | Rotates the chart so the first slice begins from the top.              |
| `wedgeprops=dict(width=0.4)`                         | Creates the hole in the center by adjusting slice thickness.           |
| `ax.text(...)`                                       | Adds custom text to the center of the donut chart.                     |
| `0, 0`                                               | Specifies the center coordinates of the chart.                         |
| `'Revenue\n18,000'`                                  | Displays the text "Revenue" and the total revenue value in the center. |
| `ha='center'`                                        | Horizontally centers the text.                                         |
| `va='center'`                                        | Vertically centers the text.                                           |
| `fontsize=14`                                        | Sets the text size to 14 points.                                       |
| `fontweight='bold'`                                  | Makes the center text bold.                                            |
| `plt.title('Revenue Share by Product')`              | Adds a title to the visualization.                                     |
| `plt.show()`                                         | Displays the completed professional donut chart.                       |


## Key Visualization Rule for Data Science

| Goal                   | Best Chart    |
| ---------------------- | ------------- |
| Trend over time        | Line Chart    |
| Compare categories     | Bar Chart     |
| Distribution           | Histogram     |
| Outliers               | Box Plot      |
| Distribution + Density | Violin Plot   |
| Relationship           | Scatter Plot  |
| Correlation            | Heatmap       |
| Multiple Variables     | Pair Plot     |
| Category Counts        | Count Plot    |
| Part-to-Whole          | Pie Chart     |
| Cumulative Trend       | Area Chart    |
| Interactive Dashboard  | Plotly Charts |

#### NOTE: For `Data Science, Machine Learning, and Analytics`, the charts you'll use most often are `Histogram, Scatter Plot, Box Plot, Heatmap, Pair Plot, and Count Plot` because they are essential for `Exploratory Data Analysis (EDA)`.
