# Load Data

```python
import os
import pandas as pd

all_files = os.listdir()
all_files

# Note: os is a module in Python which stands for Operating System
# It is a built-in Python module used to interact with your computer's operating system

# os.listdir()                          # List files in folder
# os.mkdir("test")                      # Create a folder
# os.remove("a.txt")                    # Delete a file
# os.path.join("folder", "file.csv")    # Safely joins paths
# os.getcwd()                           # Get current working directory
# os.name                               # OS type (Windows, Linux)
# os.chdir                              # Change folder
```

```python
all_data = [] # This line create an empty list to store all monthly datasets

# Get only CSV files
all_files = [file for file in os.listdir()  # os.listdir() => lists all files in the folder
             if file.endswith('.csv')]      # file.endswith('.csv') => keeps only csv files
                                            # [...] => creates list
# Loop through each file
for file in all_files:                      # for file in all_files: => This loops through each CSV file one by one
    df = pd.read_csv(file)                  # pd.read_csv(file) => Reads each CSV file into a DataFrame
                                            # df => temporary dataset for each month
    all_data.append(df)                     # all_data.append(df) => Adds each monthly DataFrame into the list

# Combines all files
all_data = pd.concat(all_data, ignore_index=True)  # pd.concat() => merges all DataFrames into one
                                                   # ignore_index=True => resets index (0,1,2,.....)
```

# Analyze

### What is the most successful month for sales?

In this question, we use three features, namely: 'Quantity Ordered', 'Price Each', and 'Order Date'.
- Order Quantity and Price Each are used to see how many sales.
- Order Date is used to indicate the month in which sales were highest

We extract these three features so that we can answer our questions. To see which month has the most sales, we use a bar chart.

```python
all_data[['Quantity Ordered', 'Price Each', 'Order Date']].dtypes
```

```python
# Extract 'Month' from 'Order date'
all_data['Month'] = all_data['Order Date'].apply(lambda x: x.split('/')[0])
all_data['Month'].head()
```

```python
# Ensure data is in the form of numbers
all_data['Month'].unique()
```

```python
# Remove 'order date' data
filter_month = all_data['Month'] == 'Order Date'

all_data = all_data[~filter_month]
```

```python
# Customize the data type of each feature

all_data['Month'] = all_data['Month'].astype(int)
all_data['Quantity Ordered'] = all_data['Quantity Ordered'].astype(int)
all_data['Price Each'] = all_data['Price Each'].astype(float)
```

```python
all_data[['Quantity Ordered', 'Price Each', 'Order Date']].dtypes
```

```python
# Viewing 'Sales' by creating the 'Sales' feature
all_data['Sales'] = all_data['Quantity Ordered'] * all_data['Price Each']
all_data['Sales'].head()

# Explanation on the code
# all_data['Sales'] = means we are creating a new column inside our full dataset called `all_data`
# all_data['Quantity Ordered'] * all_data['Price Each'] means `Sales = Quantity x Price Each`
# all_data['Sales'].head() shows the first five(5) rows of the Sales Column
```

```python
import plotly.graph_objs as go # We are importing Plotly's graph objects module. This module let's us build interactive charts (not static like Matplotlib)

# Group by month and sum the sales
monthly_sales = all_data.groupby(['Month'])['Sales'].sum()
# groupby(['Month']) => This groups the dataset based on each month (Jan, Feb, e.t.c)
# ['Sales'] => Selects only the Sales Column
# .sum() => Adds up all sales within each month

# Create a line plot trace
trace = go.Scatter(                     # Creating the line chart(Trace)
    x=monthly_sales.index,              # X-axis => Months
    y=monthly_sales.values,             # Y-axis => Total Sales per month 
    mode='lines+markers',               # Draws a line connecting points and markers (dots) on each point
    marker=dict(color='blue'),          # Sets point colour to blue
    name='Monthly Sales'                # Label shown in the chart legend
)

# Create layout - This controls how the chart looks
layout = go.Layout(
    title='Monthly Sales',              # This gives us the Chart title
    xaxis=dict(title='Month'),          # This labels the X-axis
    yaxis=dict(title='Sales'),          # This labels the Y-axis
    hovermode='closest',                # When you hover, it shows the closest data point info
)

# Create figure
fig = go.Figure(data=[trace], layout=layout)  # This combines everything together into one object 
                                              # data=[trace] => Your actual plotted data (the line chart we created earlier)
                                              # layout=layout => Your design settings (title, axis labels, e.t.c
# In conclusion, trace => the content(data), layout => the design(style), Figure => the final complete chart
# Show plot
fig.show()              # This renders (displays) the chart on your screen
                        # Without this line, nothing will show
```

**Highlight**:

In 2019, the highest sales were in December.

### Which city has the highest number of orders?

To see the distribution of the most sales in each city, we use the "'Purchase Address'" feature which contains user purchase addresses. We extract this feature and just take the city. After that we count how many cities there are. To display it, use a pie chart.

```python
all_data['Purchase Address'].head()
```

```python
# Extracting the City
# The code below extracts the city from the address column
# .str.split => Splits each address like "917 1st St", "Dallas", "TX 75001"
# .str.get(1) => Picks the second element (index 1) which is the city
# all_data['City'].head() => Previews the result
all_data['City'] = all_data['Purchase Address'].str.split(',').str.get(1)
all_data['City'].head()
```

```python
# Creating the Pie Chart
# all_data['City'].value_counts() => Counts how many orders came from each city
# .index                          => Gives the city names used as Labels
# .values                         => Gives the counts used as slice sizes
# go.Pie (...)                    => Builds the pie chart
# go.Figure(...)                  => Wraps everything into a complete chart object
fig = go.Figure(data=[go.Pie(labels=all_data['City'].value_counts().index, values=all_data['City'].value_counts().values)])

# Update layout for better appearance
fig.update_layout(
    title='Distribution of Cities', # This adds a title to the chart
                                    # "Distribution of Cities" = shows how orders are shared across cities
)

# Show the plot
fig.show()                          # Renders the interactive pie chart
```

### Which product had the highest sales volume, and what factors contributed to its popularity?

To see which items have the highest sales, we use the "Quantity Ordered" feature, we add up the transactions for each item. Then we display it using a bar chart.

To see what factors cause high sales, we use the "Price Each" feature. We look at the average price for the number of transactions for each item. After that we display it using a scatter plot.

We combine these two charts (bar and scatter) and then analyze the relationship between item sales and their average price.To see which items have the highest sales, we use the "Quantity Ordered" feature, we add up the transactions for each item. Then we display it using a bar chart.

```python
# The goal of the code in this cell is to know which product sold the most, and why?
# So, we are analyzing the `Quantity Ordered => sales volume` and `Price Each => pricing factor`
# .groupby(['Product']) => Groups all rows by each product
# .agg({...}) => This performs multiple aggregations at once
# 'Quantity Ordered':'sum' => Adds up total units sold per product
# 'Price Each':'mean' => Calculates the average price of each product
# .reset_index() => Converts grouped index back into a normal column
order_count = all_data.groupby(['Product']).agg({'Quantity Ordered':'sum', 'Price Each':'mean'}).reset_index()
order_count.head()
```

```python
# Import Subplots => This allows you to create advanced charts and especially useful for multiple axes (like this case)
from plotly.subplots import make_subplots

# Create subplots
fig = make_subplots(specs=[[{"secondary_y": True}]]) # You are creating a figure with secondary Y-axis
                                                     # You are creating a chart with two Y-axes
                                                     # Because you are plotting Quantity Ordered (large numbers) and Price Each (small numbers)
                                                     # Note: If you use one axis, the chart will look distorted
# Add bar plot for Quantity Ordered
# This creates a bar chart which shows how many units each product sold
# secondary_y=False => Uses the left Y-axis
# Left axis => Quantity
fig.add_trace(go.Bar(x=order_count['Product'], y=order_count['Quantity Ordered'], name='Quantity Ordered'), secondary_y=False)

# Add line plot for Price Each
# This creates a line chart which shows price trend across products
# secondary_y=True => Uses the right Y-axis
# Right axis => Price
fig.add_trace(go.Scatter(x=order_count['Product'], y=order_count['Price Each'], mode='lines', name='Price Each'), secondary_y=True)

# Update layout
fig.update_layout(
    title='Quantity Ordered and Price Each for Products',           # Chart title
    xaxis=dict(title='Product'),                                    # X-axis => Product names on horizontal axis
    yaxis=dict(title='Quantity Ordered', rangemode='tozero'),       # Y-axis (Left axis) => Quantity Ordered
    yaxis2=dict(title='Price Each', overlaying='y', side='right'),  # Y-axis(2) (Right axis) => Price Each
                                                                    # overlaying='y' => both charts share same space
                                                                    # side='right' => shows on right side
)

# Show the plot
fig.show()          # Shows your combined chart
```

**Data**

The sales with the most transactions are AAA Batteries, followed by AA Batteries.

**Insights**

From the two items with the most transactions, we can conclude that people have many electronic devices that are powered by batteries, such as: Remote controls, audio devices, toys, etc.

**Action**

The company stocks various types of batteries, not just AAA and AA types. Because electronic devices that use batteries currently use various types of batteries, such as: C and D, 9V batteries, lithium batteries, and rechargeable batteries.

```python
most_sold_products_data.groupby(['Month','Product']).size().unstack()
```

```python
grouped_data = most_sold_products_data.groupby(['Month', 'Product']).size().unstack()

# Create traces for each product
traces = []
for column in grouped_data.columns:
    trace = go.Scatter(
        x=grouped_data.index,
        y=grouped_data[column],
        mode='lines+markers',
        name=column
    )
    traces.append(trace)

# Create layout
layout = go.Layout(
    title='Most Sold Products Over Time',
    xaxis=dict(title='Month'),
    yaxis=dict(title='Number of Sales'),
    hovermode='closest',
)

# Create figure
fig = go.Figure(data=traces, layout=layout)

# Show plot
fig.show()
```

### Which products are frequently sold together?

In analyzing which items are frequently purchased together, we use the "Order ID" and "Products" features.

```python
duplicated_data = all_data[all_data['Order ID'].duplicated(keep=False)]
duplicated_data.head()
```

```python
duplicated_data.shape
```

```python
grouped_products = duplicated_data.groupby(['Order ID'])['Product'].apply(lambda x: ','.join(x)).reset_index().rename(columns={'Product':'Grouped Products'})
grouped_products.head()
```

```python
grouped_products.shape
```

```python
duplicate_products = duplicated_data.merge(grouped_products, how='left', on='Order ID')
duplicate_products.head()
```

```python
duplicate_products.shape
```

```python
drop_duplicate_order_id = duplicate_products.drop_duplicates(subset=['Order ID'])
drop_duplicate_order_id.head()
```

```python
drop_duplicate_order_id.shape
```

```python
# Get the top 5 grouped products
top_products = drop_duplicate_order_id['Grouped Products'].value_counts()[:5]

# Create pie chart trace
trace = go.Pie(
    labels=top_products.index,
    values=top_products.values,
    hole=0.3,
)

# Create layout
layout = go.Layout(
    title='Top 5 Grouped Products Distribution'
)

# Create figure
fig = go.Figure(data=[trace], layout=layout)

# Show plot
fig.show()
```

**Data**

Items that are often purchased together are:

1. iPhone. Often buyers when buying an iPhone also buy a lightning charging cable and wired headphones.
2. Google phone. Often buyers when buying a Google phone also buy a USB-C charging cable and wired headphones.
3. Vareebadd phone. Often buyers when buying a Vareebadd phone also buy a USB-C charging cable.

**Insight**

Of these 3 devices (iPhone, Google phone, vareebadd phone) buyers always buy charging cables and headphones too. This is because this device does not provide these accessories.

**Action**

Based on the data and insight provided, online stores must provide a variety of smartphone accessories so that buyers can complete their smartphone needs.
