## 1. What is `match`?

Python's `match` statement compares a value against one or more patterns.

```python
match value:
    case pattern:
        ...
    case another_pattern:
        ...
    case _:
        ...
```

`match` was introduced in Python 3.10. Unlike a traditional switch statement, it can inspect the **structure** of data.


### Code Example

````python
status = "success"

match status:
    case "success":
        print("Operation completed")
    case "failed":
        print("Operation failed")
    case _:
        print("Unknown status")
````

### Line-by-Line Breakdown

**Line 1**
- **Code:** `status = "success"`
- **Meaning:** Assigns `"success"` to `status`.

**Line 2**
- **Code:** `match status:`
- **Meaning:** Starts structural pattern matching. Python evaluates the value after `match`: `status`.

**Line 3**
- **Code:** `case "success":`
- **Meaning:** Defines a pattern that Python tests against the value being matched: `"success"`.

**Line 4**
- **Code:** `print("Operation completed")`
- **Meaning:** Displays information in the console: `print("Operation completed")`.

**Line 5**
- **Code:** `case "failed":`
- **Meaning:** Defines a pattern that Python tests against the value being matched: `"failed"`.

**Line 6**
- **Code:** `print("Operation failed")`
- **Meaning:** Displays information in the console: `print("Operation failed")`.

**Line 7**
- **Code:** `case _:`
- **Meaning:** Defines the wildcard/catch-all case. It runs if no earlier case matches.

**Line 8**
- **Code:** `print("Unknown status")`
- **Meaning:** Displays information in the console: `print("Unknown status")`.


### Expected Output

```text

```

## 2. `match` versus `if/elif`

For simple conditions, `if/elif` may be clearer:

```python
if day == 1:
    ...
elif day == 2:
    ...
```

The equivalent `match` is:

```python
match day:
    case 1:
        ...
    case 2:
        ...
```

`match` becomes especially powerful when you need to inspect tuples, lists, dictionaries, nested data, or object structure.


### Code Example

````python
day = 2

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case _:
        print("Invalid day")
````

## 7. Capturing values

## 8. Literal matching versus capture patterns

This distinction is critical.

```python
case 10:
```

matches the literal value `10`.

But:

```python
case number:
```

captures whatever value is matched into `number`.

Therefore, a bare name is generally **not** a test against an existing variable.


### Code Example

````python
x = 10

match x:
    case 10:
        print("Exactly 10")
````

## 15. Guards with `if`

A guard adds a condition after the pattern:

```python
case pattern if condition:
```

Guards are excellent for business rules and numeric thresholds.


### Code Example

````python
score = 85

match score:
    case score if score >= 70:
        print("Excellent")
    case score if score >= 50:
        print("Pass")
    case _:
        print("Fail")
````

- **Code:** `score = 85`
- **Meaning:** Assigns `85` to `score`.

**Line 2**
- **Code:** `match score:`
- **Meaning:** Starts structural pattern matching. Python evaluates the value after `match`: `score`.

**Line 3**
- **Code:** `case score if score >= 70:`
- **Meaning:** Defines a pattern plus a guard. The pattern must match and the `if` condition must be true: `score if score >= 70`.

**Line 4**
- **Code:** `print("Excellent")`
- **Meaning:** Displays information in the console: `print("Excellent")`.

**Line 5**
- **Code:** `case score if score >= 50:`
- **Meaning:** Defines a pattern plus a guard. The pattern must match and the `if` condition must be true: `score if score >= 50`.

**Line 6**
- **Code:** `print("Pass")`
- **Meaning:** Displays information in the console: `print("Pass")`.

**Line 7**
- **Code:** `case _:`
- **Meaning:** Defines the wildcard/catch-all case. It runs if no earlier case matches.

# 23. Common Mistakes

### Mistake 1 - Using Python < 3.10
`match` was introduced in Python 3.10.

### Mistake 2 - Forgetting `case`
Incorrect:
```python
match value:
    "A":
        print("A")
```

Correct:
```python
match value:
    case "A":
        print("A")
```

### Mistake 3 - Putting `_` first
A catch-all case placed before specific cases makes later cases unreachable.

### Mistake 4 - Confusing capture patterns with literals
`case name:` captures a value. It does not generally mean “equal to the existing variable `name`”.

### Mistake 5 - Using `match` for everything
Simple Boolean conditions are often clearer with `if/elif`.


# 24. When Should You Use `match`?

## Use `if/elif` when:
- You are mainly testing Boolean conditions.
- You have unrelated conditions.
- You are doing simple range comparisons.
- Pattern structure provides little benefit.

## Use `match` when:
- There are many discrete alternatives.
- Data has a meaningful structure.
- You need tuple/list/dictionary destructuring.
- You are routing commands or events.
- You are handling structured API responses.
- You need nested pattern matching.

### Simple rule

**`if` asks:** “Is this condition true?”

**`match` asks:** “What value or structure does this data have?”

# 26. Interview Questions and Answers

### 1. What is Python `match`?
A structural pattern matching statement introduced in Python 3.10.

### 2. What does `case` do?
Defines a pattern that Python attempts to match against the subject value.

### 3. What does `_` mean?
A wildcard/catch-all pattern.

### 4. What is structural pattern matching?
Matching based on the structure and contents of data, not only equality.

### 5. Can `match` handle dictionaries?
Yes.

### 6. Can it handle lists and tuples?
Yes.

### 7. What is a guard?
An extra `if` condition attached to a case.

Example:
```python
case x if x > 100:
```

### 8. What does `|` mean?
Alternative patterns:
```python
case "yes" | "y":
```

### 9. Which Python version introduced it?
Python 3.10.

### 10. Is `match` always better than `if/elif`?
No. Use the construct that makes the logic clearer. `match` is especially powerful for structured data.


# 27. Cheat Sheet

```python
# Exact value
match value:
    case 10:
        ...

# Multiple alternatives
match value:
    case 1 | 2 | 3:
        ...

# Wildcard
match value:
    case _:
        ...

# Tuple
match point:
    case (x, y):
        ...

# List
match values:
    case [x, y]:
        ...

# First, middle, last
match values:
    case [first, *middle, last]:
        ...

# Dictionary
match customer:
    case {"name": name, "age": age}:
        ...

# Guard
match score:
    case score if score >= 70:
        ...

# Nested dictionary
match data:
    case {"user": {"name": name}}:
        ...

# Class/object
match employee:
    case Employee(name=name, department=department):
        ...
```
