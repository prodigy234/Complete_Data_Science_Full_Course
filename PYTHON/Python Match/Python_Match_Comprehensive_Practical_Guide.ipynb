## 1. What is `match`?

Python's `match` statement compares a value against one or more patterns.

```python
match value:
    case pattern:
        ...
    case another_pattern:
        ...
    case _:
        ...
```

`match` was introduced in Python 3.10. Unlike a traditional switch statement, it can inspect the **structure** of data.


````python
status = "success"

match status:
    case "success":
        print("Operation completed")
    case "failed":
        print("Operation failed")
    case _:
        print("Unknown status")
````

## 2. `match` versus `if/elif`

For simple conditions, `if/elif` may be clearer:

```python
if day == 1:
    ...
elif day == 2:
    ...
```

The equivalent `match` is:

```python
match day:
    case 1:
        ...
    case 2:
        ...
```

`match` becomes especially powerful when you need to inspect tuples, lists, dictionaries, nested data, or object structure.


````python
day = 2

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case _:
        print("Invalid day")
````

## 5. Multiple patterns with `|`

Use `|` when several patterns should execute the same branch.

```python
case "yes" | "y":
```


````python
answer = "y"

match answer:
    case "yes" | "y":
        print("Approved")
    case "no" | "n":
        print("Rejected")
    case _:
        print("Invalid response")
````

## 6. The wildcard `_`

`case _:` is the catch-all/default case. It is normally placed last.

```python
status = "pending"

match status:
    case "approved":
        print("Approved")
    case "rejected":
        print("Rejected")
    case _:
        print("Pending or unknown")
```


````python
status = "pending"

match status:
    case "approved":
        print("Approved")
    case "rejected":
        print("Rejected")
    case _:
        print("Pending or unknown")
````

## 7. Capturing values

Names inside patterns can capture parts of the matched value.

For a tuple `(name, age)`:

```python
user = ("Gbenga", 30)

match user:
    case (name, age):
        print(name, age)
```

The captured values become variables inside the case body.


````python
user = ("Gbenga", 30)

match user:
    case (name, age):
        print(f"Name: {name}")
        print(f"Age: {age}")
````

## 8. Literal matching versus capture patterns

This distinction is critical.

```python
case 10:
```

matches the literal value `10`.

But:

```python
case number:
```

captures whatever value is matched into `number`.

Therefore, a bare name is generally **not** a test against an existing variable.


````python
x = 10

match x:
    case 10:
        print("Exactly 10")
````

## 9. Tuple pattern matching

**Question:** Determine whether a coordinate is the origin, on the X-axis, on the Y-axis, or a general point.

````python
point = (0, 5)

match point:
    case (0, 0):
        print("Origin")
    case (0, y):
        print(f"Y-axis: y={y}")
    case (x, 0):
        print(f"X-axis: x={x}")
    case (x, y):
        print(f"General point: ({x}, {y})")
````

## 10. List pattern matching

Patterns can describe the length/shape of a list.

```python
case []:          # empty
case [x]:         # one item
case [x, y]:      # two items
```


````python
numbers = [1, 2, 3]

match numbers:
    case []:
        print("Empty list")
    case [x]:
        print("One item:", x)
    case [x, y]:
        print("Two items:", x, y)
    case [x, y, z]:
        print("Three items:", x, y, z)
    case _:
        print("Other size")
````

## 11. Capturing many list elements with `*`

`[first, *middle, last]` captures the first item, all middle items, and the last item.

This is useful when the number of middle values is unknown.


````python
numbers = [1, 2, 3, 4, 5]

match numbers:
    case [first, *middle, last]:
        print("First:", first)
        print("Middle:", middle)
        print("Last:", last)
````

## 12. Dictionary pattern matching

Dictionary patterns are useful for API responses, customer records, configuration data, transactions and ML results.

```python
case {"name": name, "age": age}:
```

The keys are matched and the values are captured.


````python
customer = {"name": "Gbenga", "age": 30}

match customer:
    case {"name": name, "age": age}:
        print(f"Name: {name}, Age: {age}")
````

## 13. Dictionary values - transaction classification

**Question:** Handle deposits, withdrawals, and unknown transaction types.

````python
transaction = {
    "type": "deposit",
    "amount": 50000
}

match transaction:
    case {"type": "deposit", "amount": amount}:
        print(f"Deposit received: ₦{amount:,}")
    case {"type": "withdrawal", "amount": amount}:
        print(f"Withdrawal made: ₦{amount:,}")
    case _:
        print("Unknown transaction")
````

## 14. Extra dictionary keys and `**rest`

A dictionary pattern can match selected keys without requiring every key to be listed.

You can capture remaining keys with:

```python
case {"name": name, "age": age, **rest}:
```


````python
customer = {
    "name": "Gbenga",
    "age": 30,
    "city": "Lagos"
}

match customer:
    case {"name": name, "age": age, **rest}:
        print("Name:", name)
        print("Age:", age)
        print("Other data:", rest)
````

## 15. Guards with `if`

A guard adds a condition after the pattern:

```python
case pattern if condition:
```

Guards are excellent for business rules and numeric thresholds.


````python
score = 85

match score:
    case score if score >= 70:
        print("Excellent")
    case score if score >= 50:
        print("Pass")
    case _:
        print("Fail")
````

## 16. Practical grading system

Rules:

- 70+ → A
- 60–69 → B

````python
score = 67

match score:
    case score if score >= 70:
        grade = "A"
    case score if score >= 60:
        grade = "B"
    case score if score >= 50:
        grade = "C"
    case score if score >= 45:
        grade = "D"
    case _:
        grade = "F"

print("Grade:", grade)
````

## 17. Nested dictionary matching

You can match nested structures directly.

This is useful for structured JSON/API data.


````python
order = {
    "customer": {"name": "Gbenga"},
    "payment": {"status": "successful"}
}

match order:
    case {
        "customer": {"name": name},
        "payment": {"status": "successful"}
    }:
        print(f"Order paid successfully by {name}")
    case _:
        print("Payment not successful")
````

## 18. Nested list matching

A pattern can describe several levels of a list.

Example: `[location, [product, amount]]`.


````python
sale = ["Lagos", ["Laptop", 250000]]

match sale:
    case [location, [product, amount]]:
        print("Location:", location)
        print("Product:", product)
        print("Amount:", amount)
````

## 19. Matching by data shape

Pattern matching is useful when the same variable may contain different structures.

Examples:

```python
case ():
case (x,):
case (x, y):
case (x, y, *rest):
```


````python
data = (10, 20)

match data:
    case ():
        print("No values")
    case (x,):
        print("One value:", x)
    case (x, y):
        print("Two values:", x, y)
    case (x, y, *rest):
        print("Three or more values:", x, y, rest)
````

## 20. Matching classes and objects

Pattern matching also supports class patterns.

For named attributes:

```python
case Employee(name=name, department=department):
```


````python
class Employee:
    def __init__(self, name, department):
        self.name = name
        self.department = department

employee = Employee("Gbenga", "Data Science")

match employee:
    case Employee(name=name, department=department):
        print(f"{name} works in {department}.")
````

## 21. Positional class matching with `__match_args__`

A class can define:

```python
__match_args__ = ("name", "department")
```

Then positional matching can be used:

```python
case Employee(name, department):
```

This is an advanced feature useful when designing classes for structural matching.


````python
class Employee:
    __match_args__ = ("name", "department")

    def __init__(self, name, department):
        self.name = name
        self.department = department

employee = Employee("Gbenga", "AI Engineering")

match employee:
    case Employee(name, department):
        print("Name:", name)
        print("Department:", department)
````

# 23. Common Mistakes

### Mistake 1 - Using Python < 3.10
`match` was introduced in Python 3.10.

### Mistake 2 - Forgetting `case`
Incorrect:
```python
match value:
    "A":
        print("A")
```

Correct:
```python
match value:
    case "A":
        print("A")
```

### Mistake 3 - Putting `_` first
A catch-all case placed before specific cases makes later cases unreachable.

### Mistake 4 - Confusing capture patterns with literals
`case name:` captures a value. It does not generally mean “equal to the existing variable `name`”.

### Mistake 5 - Using `match` for everything
Simple Boolean conditions are often clearer with `if/elif`.


# 24. When Should You Use `match`?

## Use `if/elif` when:
- You are mainly testing Boolean conditions.
- You have unrelated conditions.
- You are doing simple range comparisons.
- Pattern structure provides little benefit.

## Use `match` when:
- There are many discrete alternatives.
- Data has a meaningful structure.
- You need tuple/list/dictionary destructuring.
- You are routing commands or events.
- You are handling structured API responses.
- You need nested pattern matching.

### Simple rule

**`if` asks:** “Is this condition true?”

**`match` asks:** “What value or structure does this data have?”

# 26. Interview Questions and Answers

### 1. What is Python `match`?
A structural pattern matching statement introduced in Python 3.10.

### 2. What does `case` do?
Defines a pattern that Python attempts to match against the subject value.

### 3. What does `_` mean?
A wildcard/catch-all pattern.

### 4. What is structural pattern matching?
Matching based on the structure and contents of data, not only equality.

### 5. Can `match` handle dictionaries?
Yes.

### 6. Can it handle lists and tuples?
Yes.

### 7. What is a guard?
An extra `if` condition attached to a case.

Example:
```python
case x if x > 100:
```

### 8. What does `|` mean?
Alternative patterns:
```python
case "yes" | "y":
```

### 9. Which Python version introduced it?
Python 3.10.

### 10. Is `match` always better than `if/elif`?
No. Use the construct that makes the logic clearer. `match` is especially powerful for structured data.
