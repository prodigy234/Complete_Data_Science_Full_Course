# 1. Introduction to Exploratory Data Analysis

## 1.1 What is EDA?

Exploratory Data Analysis, commonly shortened to EDA, is the practice of investigating a dataset before applying any formal modeling or statistical testing. The goal is to understand the shape, structure, quality and story hidden inside the data using summary statistics and visualizations. The term was popularized by the statistician John Tukey, who argued that data should first be explored with an open mind rather than forced immediately into a hypothesis.

In practical Python data science work, EDA is the bridge between raw, messy data and a clean, trustworthy dataset that is ready for machine learning, dashboards or reporting. It answers questions such as: What does each column represent? Are there missing values? Are there outliers? Which variables are related to each other? Is the data balanced or skewed?

## 1.2 Why EDA Matters

- It reveals data quality problems early, such as missing values, duplicates, wrong data types and inconsistent categories, before they silently corrupt a model.
- It builds intuition about relationships between variables, which guides feature engineering and model selection.
- It helps detect outliers and anomalies that could distort statistical summaries or bias a machine learning model.
- It confirms or challenges assumptions, for example whether a variable is normally distributed, which many statistical tests rely on.
- It produces visual evidence that is easy to communicate to non-technical stakeholders.

## 1.3 The Typical EDA Workflow

Although every project is different, a typical EDA workflow in Python follows this general sequence:

1. Import the necessary libraries (pandas, numpy, matplotlib, seaborn and so on).
1. Load the raw data into a DataFrame from a file or database.
1. Perform an initial inspection: shape, column names, data types, first and last rows.
1. Check and handle missing values.
1. Check and handle duplicate rows.
1. Explore each variable individually (univariate analysis).
1. Explore relationships between pairs of variables (bivariate analysis).
1. Explore relationships among three or more variables (multivariate analysis).
1. Detect and treat outliers.
1. Engineer or transform features where useful.
1. Summarize findings, either in a notebook narrative, a report, or a dashboard.

# 6. Understanding Data Types and Structure

## 6.1 Common Pandas Data Types

- int64: whole numbers, such as a count of items or an age in years.
- float64: decimal numbers, such as price or a percentage.
- object: usually text/strings, but also used for mixed-type or unparsed columns.
- bool: True/False values.
- datetime64[ns]: date and time values.
- category: a memory-efficient type for columns with a limited number of repeating text values, such as a country or a product category.

## 6.2 Converting Data Types

| `pd.to_numeric(df['price'], errors='coerce')` | Converts the price column to a numeric type; the errors='coerce' argument tells pandas to turn any value it cannot convert (such as the text 'unknown') into NaN instead of raising an error and stopping the program. |
| `df['region'].astype('category')` | Converts the region column to pandas' category dtype, which stores each unique value once internally and uses far less memory when there are many repeated values. |
| `df['is_returning_customer'].astype(bool)` | Converts the column to True/False boolean values. |

# 7. Handling Missing Data

## 7.1 Detecting Missing Values

```python
df.isnull().sum()
df.isnull().sum() / len(df) * 100
df.isnull().sum().sort_values(ascending=False)
```

| Code / Item | Explanation |
|---|---|
| `df.isnull()` | Returns a DataFrame of the same shape as df, but filled with True where a value is missing (NaN) and False where it is present. |
| `.sum()` | Chained onto isnull(), this adds up the True values (True counts as 1) down each column, giving the total number of missing values per column. |
| `/ len(df) * 100` | Divides the missing count of each column by the total number of rows (len(df)) and multiplies by 100, converting the raw count into a percentage of the dataset that is missing. |
| `.sort_values(ascending=False)` | Sorts the resulting counts from highest to lowest, so the columns with the most missing data appear at the top. |

## 7.2 Visualizing Missing Data

```python
import missingno as msno
msno.matrix(df)
plt.show()

msno.heatmap(df)
plt.show()
```

| Code / Item | Explanation |
|---|---|
| `msno.matrix(df)` | Draws a matrix where each column is a vertical bar and each row is a horizontal position; white gaps in the bar show exactly where values are missing, revealing patterns like whole blocks of missing rows. |
| `msno.heatmap(df)` | Draws a correlation-style heatmap that shows whether the missingness in one column tends to occur alongside missingness in another column, which can hint at a shared root cause. |
| `plt.show()` | Renders the currently built matplotlib figure to the screen (or notebook cell output). |

## 7.3 Strategies for Handling Missing Values

- Deletion: remove rows or columns that are missing, appropriate when the missing amount is small and random.
- Simple imputation: fill with the mean, median or mode of the column.
- Constant/flag imputation: fill with a fixed value like 'Unknown' and optionally add a boolean column flagging that the value was originally missing.
- Model-based imputation: predict the missing value using other columns, for example with K-Nearest Neighbors imputation.
- Forward/backward fill: for time series data, carry the last known value forward or the next known value backward.

### 7.3.1 Dropping Missing Data

```python
df_clean = df.dropna()
df_clean = df.dropna(subset=['price', 'quantity'])
df_clean = df.dropna(axis=1, thresh=int(0.7 * len(df)))
```

| Code / Item | Explanation |
|---|---|
| `df.dropna()` | Removes every row that contains at least one missing value anywhere; can be very aggressive if many columns have scattered missing values. |
| `df.dropna(subset=['price', 'quantity'])` | Only drops a row if it has a missing value specifically in the price or quantity columns, ignoring missing values elsewhere. |
| `df.dropna(axis=1, thresh=int(0.7 * len(df)))` | axis=1 switches the operation to columns instead of rows; thresh sets a minimum number of non-null values a column must have to survive, here calculated as 70 percent of the total row count, effectively dropping any column that is more than 30 percent empty. |

### 7.3.2 Filling (Imputing) Missing Data

```python
df['age'] = df['age'].fillna(df['age'].median())
df['category'] = df['category'].fillna(df['category'].mode()[0])
df['notes'] = df['notes'].fillna('Unknown')
df['temperature'] = df['temperature'].fillna(method='ffill')
```

| Code / Item | Explanation |
|---|---|
| `df['age'].fillna(df['age'].median())` | Replaces every missing value in the age column with the median (middle value) of that column, a robust choice for numeric data that resists distortion from outliers. |
| `df['category'].fillna(df['category'].mode()[0])` | mode() returns the most frequently occurring value(s) as a Series; [0] picks the first one. fillna() then replaces missing entries in the category column with that most common category. |
| `df['notes'].fillna('Unknown')` | Replaces missing text values with the literal string 'Unknown', useful when the absence of a note is itself meaningful information rather than something to estimate. |
| `df['temperature'].fillna(method='ffill')` | 'ffill' stands for forward fill; it carries the last valid observation forward to fill the gap, which is appropriate for ordered time series data such as sensor readings. |

### 7.3.3 Flagging Missingness Before Imputing

```python
df['age_was_missing'] = df['age'].isnull()
df['age'] = df['age'].fillna(df['age'].median())
```

The first line creates a new boolean column that records, for every row, whether the age value was originally missing before it gets overwritten. This preserves information: sometimes the fact that a value is missing is itself predictive (for example, customers who skip an age field might behave differently), and this pattern would be lost if the column were silently filled without a trace.

# 8. Handling Duplicate Records

## 8.1 Detecting Duplicates

```python
df.duplicated().sum()
df[df.duplicated()]
df.duplicated(subset=['customer_id', 'order_date']).sum()
```

| Code / Item | Explanation |
|---|---|
| `df.duplicated()` | Returns a boolean Series, one value per row, marked True if that entire row is an exact repeat of an earlier row in the DataFrame. |
| `.sum()` | Counts how many rows are flagged as duplicates, since True is treated as 1 when summed. |
| `df[df.duplicated()]` | Uses the boolean Series from duplicated() to filter the DataFrame down to only the rows that are duplicates, so they can be inspected directly. |
| `df.duplicated(subset=['customer_id', 'order_date'])` | Restricts the duplicate check to only the listed columns, flagging a row as duplicate if that specific combination of customer_id and order_date has appeared before, even if other columns differ. |

## 8.2 Removing Duplicates

```python
df = df.drop_duplicates()
df = df.drop_duplicates(subset=['customer_id', 'order_date'], keep='first')
```

| Code / Item | Explanation |
|---|---|
| `df.drop_duplicates()` | Removes rows that are exact duplicates of a previous row, keeping only the first occurrence by default. |
| `subset=['customer_id', 'order_date']` | Limits the comparison to only these two columns when deciding what counts as a duplicate. |
| `keep='first'` | Specifies that when duplicates are found, the first occurrence should be kept and later ones removed; alternatives are 'last' (keep the last occurrence) or False (remove all copies, keeping none). |

> **Tip:** Always check duplicates before dropping missing values or aggregating; duplicate rows can silently inflate sums, counts and averages during EDA.

## 9.1 Descriptive Statistics

```python
df['price'].mean()
df['price'].median()
df['price'].std()
df['price'].var()
df['price'].min(), df['price'].max()
df['price'].quantile([0.25, 0.5, 0.75])
df['price'].skew()
df['price'].kurt()
```

| Code / Item | Explanation |
|---|---|
| `.mean()` | Calculates the arithmetic average of all values in the column: the sum of values divided by the count of values. |
| `.median()` | Calculates the middle value when all values are sorted; less sensitive to extreme outliers than the mean. |
| `.std()` | Calculates the standard deviation, a measure of how spread out the values are around the mean. |
| `.var()` | Calculates the variance, which is the standard deviation squared. |
| `.min(), .max()` | Return the smallest and largest values in the column respectively. |
| `.quantile([0.25, 0.5, 0.75])` | Returns the value below which 25 percent, 50 percent (the median) and 75 percent of the data falls; these are the first, second and third quartiles. |
| `.skew()` | Measures the asymmetry of the distribution; a positive value means a longer tail on the right (high-value outliers pull the mean up), a negative value means a longer tail on the left. |
| `.kurt()` | Measures 'tailedness', how much of the data is concentrated in the tails versus the center compared to a normal distribution; high kurtosis means more extreme outliers than a normal curve. |

## 9.3 Box Plots

```python
plt.figure(figsize=(6, 4))
sns.boxplot(x=df['price'], color='lightgreen')
plt.title('Box Plot of Price')
plt.show()
```

A box plot summarizes a numeric column into five key statistics drawn visually: the minimum (excluding outliers), the first quartile (bottom of the box), the median (line inside the box), the third quartile (top of the box), and the maximum (excluding outliers). Points plotted individually beyond the 'whiskers' are flagged as potential outliers. sns.boxplot(x=df['price'], ...) tells seaborn to plot price along the x-axis, producing a horizontal box; using y=df['price'] instead would produce a vertical box.

## 9.4 Violin Plots

```python
sns.violinplot(x=df['price'], color='plum')
plt.title('Violin Plot of Price')
plt.show()
```

A violin plot combines a box plot with a rotated, mirrored kernel density plot, so it shows both the summary statistics and the full shape of the distribution, including whether it is bimodal (has two peaks), which a plain box plot would hide.

# 13. Outlier Detection and Treatment

## 13.1 What Counts as an Outlier

An outlier is a data point that differs substantially from the rest of the observations. Outliers are not automatically errors; some are genuine, important extreme values (a legitimate very large transaction), while others are the result of data entry mistakes or sensor faults. EDA aims to find them so a deliberate decision can be made about each one, rather than letting them silently distort statistics or models.

## 13.2 The IQR (Interquartile Range) Method

```python
Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = df[(df['price'] < lower_bound) | (df['price'] > upper_bound)]
print(f'Number of outliers: {len(outliers)}')
```

| Code / Item | Explanation |
|---|---|
| `Q1 = df['price'].quantile(0.25)` | Calculates the first quartile, the value below which 25 percent of the data falls, and stores it in the variable Q1. |
| `Q3 = df['price'].quantile(0.75)` | Calculates the third quartile, the value below which 75 percent of the data falls, and stores it in Q3. |
| `IQR = Q3 - Q1` | The interquartile range, the width of the middle 50 percent of the data, calculated by subtracting Q1 from Q3. |
| `lower_bound = Q1 - 1.5 * IQR` | Defines the lower fence; any value below this is flagged as a potential outlier. 1.5 times the IQR is the conventional multiplier introduced by Tukey. |
| `upper_bound = Q3 + 1.5 * IQR` | Defines the upper fence in the same way, on the high side of the distribution. |
| `df[(df['price'] < lower_bound) \| (df['price'] > upper_bound)]` | Filters the DataFrame to keep only rows where price falls outside either fence; the pipe symbol \| is the element-wise logical OR operator, and each condition must be wrapped in parentheses. |
| `len(outliers)` | Counts how many rows were flagged, giving a quick sense of how large the outlier problem is. |

## 13.3 The Z-Score Method

```python
from scipy import stats
z_scores = stats.zscore(df['price'].dropna())
outliers_z = df.loc[df['price'].dropna().index[abs(z_scores) > 3]]
```

| Code / Item | Explanation |
|---|---|
| `stats.zscore(df['price'].dropna())` | Calculates the z-score for every value: how many standard deviations that value is from the column mean. dropna() removes missing values first, since zscore cannot handle NaN. |
| `abs(z_scores) > 3` | A common rule of thumb: any value more than 3 standard deviations away from the mean (in either direction, hence the absolute value) is treated as an outlier, since under a normal distribution over 99.7 percent of values fall within 3 standard deviations. |
| `df.loc[df['price'].dropna().index[...]]` | Uses .loc, pandas' label-based indexer, to pull out the original rows corresponding to the flagged outlier positions. |

## 13.4 Visualizing Outliers

```python
sns.boxplot(x=df['price'])
plt.title('Outliers Visible as Points Beyond the Whiskers')
plt.show()
```

In a box plot, the whiskers typically extend to 1.5 times the IQR beyond the box edges, exactly matching the IQR method above; any point plotted individually beyond the whiskers is a flagged outlier under that same rule, which is why box plots and the IQR method are natural companions.

## 13.5 Deciding What To Do With Outliers

- Investigate first: check whether the value is a genuine data entry error (for example, a price of -50 or an age of 300).
- Cap/Winsorize: clip extreme values to a maximum and minimum threshold rather than deleting them entirely.
- Transform: apply a log or square root transform to reduce the influence of extreme values while keeping them in the dataset.
- Remove: drop the row only when there is strong evidence it is an error or irrelevant to the analysis.
- Leave as is: some models (like tree-based models) are naturally robust to outliers, so no treatment may be necessary.

```python
df['price_capped'] = df['price'].clip(lower=lower_bound, upper=upper_bound)
```

The .clip() method replaces any value below lower_bound with lower_bound itself, and any value above upper_bound with upper_bound itself, keeping the row in the dataset while limiting the extremity of the value, a technique known as winsorizing.

# 14. Correlation Analysis in Depth

## 14.1 Pearson vs Spearman vs Kendall

- Pearson correlation: measures the strength of a linear relationship; assumes both variables are roughly normally distributed and sensitive to outliers.
- Spearman correlation: measures the strength of a monotonic relationship using ranked values; does not assume linearity or normality and is robust to outliers.
- Kendall's Tau: another rank-based correlation, generally more robust for small sample sizes and produces slightly different values than Spearman, but tells a similar story.

```python
df[['price', 'quantity', 'discount']].corr(method='pearson')
df[['price', 'quantity', 'discount']].corr(method='spearman')
df[['price', 'quantity', 'discount']].corr(method='kendall')
```

## 14.2 Important Caution: Correlation Is Not Causation

A strong correlation between two variables only shows that they move together statistically, not that one causes the other. Both could be driven by a third, unobserved variable (a confounder), or the relationship could simply be a coincidence in the particular sample collected. EDA should describe relationships carefully and avoid asserting causal claims without further experimental or causal-inference evidence.

## 14.3 Checking for Multicollinearity

Multicollinearity happens when two or more predictor variables are highly correlated with each other, which can destabilize certain models (particularly linear regression). A simple EDA check is to scan the correlation heatmap from section 12.1 for any off-diagonal values close to 1 or -1. A more formal measure is the Variance Inflation Factor (VIF).

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
X = df[['price', 'quantity', 'discount']].dropna()
vif_data = pd.DataFrame()
vif_data['feature'] = X.columns
vif_data['VIF'] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
```

| Code / Item | Explanation |
|---|---|
| `from statsmodels... import variance_inflation_factor` | Imports a function from the statsmodels library that calculates how much a variable's variance is inflated due to its correlation with the other predictor variables. |
| `X = df[[...]].dropna()` | Selects the numeric predictor columns to test and removes any rows with missing values, since VIF cannot be computed with NaN present. |
| `vif_data['feature'] = X.columns` | Creates a new column in an empty results DataFrame listing each predictor's name. |
| `[variance_inflation_factor(X.values, i) for i in range(X.shape[1])]` | A list comprehension that loops over each column index i in X, calculating its VIF score against all the other columns, and collects the results into a list; a VIF above roughly 5 to 10 is often considered a sign of concerning multicollinearity. |

# 17. Statistical Summary Techniques

## 17.1 Testing for Normality

```python
from scipy import stats
stat, p_value = stats.shapiro(df['price'].dropna().sample(min(5000, len(df))))
print(f'Statistic={stat:.4f}, p-value={p_value:.4f}')
```

| Code / Item | Explanation |
|---|---|
| `stats.shapiro(...)` | Runs the Shapiro-Wilk test, a formal statistical test for whether a sample plausibly comes from a normal distribution; it returns a test statistic and a p-value. |
| `.sample(min(5000, len(df)))` | The Shapiro-Wilk test becomes unreliable and can flag trivial deviations as significant on very large samples, so this line caps the sample used for the test at 5000 rows (or the full dataset if it is smaller). |
| `p_value` | By common convention, if the p-value is below 0.05, there is enough evidence to reject the assumption that the data is normally distributed; if it is above 0.05, normality cannot be ruled out. |

## 17.2 Grouped Statistical Comparison (t-test)

```python
group_a = df[df['region'] == 'North']['price'].dropna()
group_b = df[df['region'] == 'South']['price'].dropna()
t_stat, p_val = stats.ttest_ind(group_a, group_b, equal_var=False)
print(f't-statistic={t_stat:.3f}, p-value={p_val:.4f}')
```

| Code / Item | Explanation |
|---|---|
| `df[df['region'] == 'North']['price']` | Filters the DataFrame down to only rows where region equals 'North', then selects the price column from that filtered subset. |
| `stats.ttest_ind(group_a, group_b, equal_var=False)` | Runs an independent two-sample t-test, checking whether the mean price differs significantly between the two regions; equal_var=False runs Welch's t-test variant, which does not assume the two groups have equal variance, a safer default. |

## 17.3 Chi-Square Test of Independence

```python
contingency = pd.crosstab(df['region'], df['product_category'])
chi2, p, dof, expected = stats.chi2_contingency(contingency)
print(f'Chi2={chi2:.2f}, p-value={p:.4f}, degrees of freedom={dof}')
```

| Code / Item | Explanation |
|---|---|
| `pd.crosstab(df['region'], df['product_category'])` | Builds the frequency table of counts for every combination of the two categorical variables, which is the required input for a chi-square test. |
| `stats.chi2_contingency(contingency)` | Tests whether the two categorical variables are statistically independent of each other, returning the chi-square statistic, the p-value, the degrees of freedom, and the table of expected counts under the assumption of independence. |

# 19. EDA Checklist and Best Practices

## 19.1 The Master Checklist

- Load the data and confirm the row and column counts match expectations.
- Check df.info() for data types and unexpected nulls.
- Check df.describe() for implausible minimums, maximums or means.
- Quantify missing values per column, both as a count and a percentage.
- Decide and apply a missing-value strategy per column, not one blanket rule.
- Check and remove exact duplicate rows.
- Check for near-duplicates using a subset of key columns.
- Standardize text columns (trim whitespace, fix casing, unify category spellings).
- Convert columns to their correct data type (dates, numbers, categories).
- Plot the distribution of every important numeric column.
- Plot the frequency of every important categorical column.
- Check skewness and consider a transform for heavily skewed numeric columns.
- Detect outliers with the IQR or z-score method and decide on treatment.
- Explore relationships between the target variable (if any) and each feature.
- Build a correlation heatmap and flag highly correlated feature pairs.
- Engineer any obviously useful new features (date parts, ratios, bins).
- Document findings and decisions in markdown cells or a report as you go.
- Save a cleaned version of the dataset separately from the raw file.

## 19.2 General Best Practices

- Never overwrite the raw data file; always keep an unmodified original.
- Make cleaning steps reproducible in code rather than editing values manually in a spreadsheet.
- Document the reasoning behind every cleaning decision, not just the code, since future readers (including the analyst themselves) need to know why a choice was made.
- Prefer simple, transparent charts over complex ones when communicating with non-technical audiences.
- Revisit EDA after feature engineering; new variables need their own exploration too.
- Treat EDA as iterative, not linear; a finding in step 10 often sends the analyst back to step 3.

# 20. Common Pitfalls and How to Avoid Them

## 20.1 Frequent Mistakes

| Code / Item | Explanation |
|---|---|
| `Dropping rows with any missing value without checking impact` | Can silently delete a large fraction of the dataset. Always check what percentage of rows would be lost, per column, before choosing dropna(). |
| `Filling all missing values with the same statistic` | Different columns need different strategies; blindly using the mean everywhere ignores skewed distributions where the median would be more representative. |
| `Forgetting errors='coerce' when converting types` | pd.to_numeric() or pd.to_datetime() will raise an error and stop execution on the first bad value unless errors='coerce' is set to convert bad values to NaN instead. |
| `Not checking for duplicate rows before aggregating` | Duplicated rows will double-count in sums, means and counts, producing misleading summary statistics. |
| `Ignoring the difference between object columns that are text vs numbers stored as text` | A price column read as object type due to a stray currency symbol will not sort or aggregate numerically until it is explicitly cleaned and converted. |
| `Using pie charts for many categories` | Beyond about five slices, pie charts become very hard to compare accurately; a sorted bar chart communicates the same information more clearly. |
| `Interpreting correlation as causation` | A strong correlation coefficient only shows statistical association; it never proves that one variable causes the other. |
| `Not setting a random seed when sampling for a preview` | df.sample() without a fixed random_state will return a different sample every run, making results hard to reproduce or discuss with a team. |

## 20.2 A Safer Sampling Pattern

```python
df.sample(5, random_state=42)
```

| Code / Item | Explanation |
|---|---|
| `random_state=42` | Fixes the random number generator's starting seed so the same 'random' rows are returned every time the code is run, making the analysis reproducible for collaborators or for re-running later. |
