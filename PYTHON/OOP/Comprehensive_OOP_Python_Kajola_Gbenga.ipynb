# 1. What is OOP?

OOP means **Object-Oriented Programming**.

It is a programming approach where we organize code around **objects**.

An object represents something we can describe using:

- Data or characteristics.
- Actions or behaviors.

## Real-life example - A Bank Account

A bank account has characteristics:

- Account holder name
- Account number
- Balance

A bank account can perform actions:

- Deposit money
- Withdraw money
- Check balance

In OOP:

- The characteristics are usually called **attributes**.
- The actions are usually called **methods**.
- A blueprint for creating bank accounts is called a **class**.
- A real bank account created from the blueprint is called an **object**.

# 7. The `__init__()` Constructor

The `__init__()` method is a special method that runs automatically when an object is created.

It allows us to give each object its initial data.


````python
class Student:

    def __init__(self, name, age, course):
        self.name = name
        self.age = age
        self.course = course


student1 = Student("Ade", 20, "Python")

print(student1.name)
print(student1.age)
print(student1.course)
````

## Complete line-by-line breakdown

```python
class Student:
```

Creates a class called `Student`.

```python
def __init__(self, name, age, course):
```

- `def` is used to define a function.
- `__init__` is a special method called automatically when an object is created.
- `self` represents the current object.
- `name`, `age`, and `course` are values we want to provide when creating an object.

```python
self.name = name
```

- The value passed into `name` is stored inside the object as `self.name`.

```python
self.age = age
```

- Stores the age inside the current object.

```python
self.course = course
```

- Stores the course inside the current object.

```python
student1 = Student("Ade", 20, "Python")
```

Python automatically sends:

- `"Ade"` into `name`
- `20` into `age`
- `"Python"` into `course`

The `__init__()` method then stores those values inside the object.

The object now has its own data.


# 8. Understanding `self`

`self` is one of the most important concepts in Python OOP.

`self` refers to the **current object**.

Consider:


````python
class Employee:

    def __init__(self, name, department):
        self.name = name
        self.department = department


employee1 = Employee("Amina", "Finance")
employee2 = Employee("Tunde", "Technology")

print(employee1.name)
print(employee2.name)
````

## What is happening?

When Python creates:

```python
employee1 = Employee("Amina", "Finance")
```

The current object is `employee1`.

Therefore:

```python
self.name = name
```

stores `"Amina"` inside `employee1`.

When Python creates:

```python
employee2 = Employee("Tunde", "Technology")
```

The current object is now `employee2`.

Therefore, `self.name` stores `"Tunde"` inside `employee2`.

`self` allows each object to keep its own separate data.


# 9. Methods

A method is a function defined inside a class.

Methods describe what an object can do.


````python
class Employee:

    def __init__(self, name, department):
        self.name = name
        self.department = department

    def introduce(self):
        print(f"My name is {self.name}")
        print(f"I work in the {self.department} department")


employee1 = Employee("Amina", "Finance")
employee1.introduce()
````

## Line-by-line breakdown

```python
def introduce(self):
```

- Defines a method named `introduce`.
- `self` gives the method access to the data belonging to the current object.

```python
print(f"My name is {self.name}")
```

- Reads the `name` stored in the current object.
- Displays the value using an f-string.

```python
employee1.introduce()
```

- Calls the `introduce` method for `employee1`.

The method can access the attributes of `employee1` because of `self`.

self.salary = self.salary + amount

# 13. Instance Methods, Class Methods, and Static Methods

Python provides different types of methods.

## Instance Method

Works with data belonging to a specific object.

## Class Method

Works with the class itself.

## Static Method

A utility method logically related to the class but does not need object or class data.


````python
class Employee:

    company = "Kajola Tech Solutions"

    def __init__(self, name):
        self.name = name

    def introduce(self):
        print(f"I am {self.name}")

    @classmethod
    def show_company(cls):
        print(cls.company)

    @staticmethod
    def is_valid_age(age):
        return age >= 18


employee1 = Employee("Amina")

employee1.introduce()
Employee.show_company()
print(Employee.is_valid_age(20))
````

## Line-by-line breakdown

### Instance method

```python
def introduce(self):
```

Uses `self`, so it can access object-specific data.

### Class method

```python
@classmethod
```

This decorator tells Python that the next method is a class method.

```python
def show_company(cls):
```

- `cls` refers to the class.
- It is similar to `self`, but `self` refers to an object while `cls` refers to the class.

### Static method

```python
@staticmethod
```

The method does not need `self` or `cls`.

It is useful for helper functions related to the class.


# 14. Encapsulation

Encapsulation means controlling how data inside an object is accessed and modified.

In business applications, some information should not be freely changed.

Examples:

- Bank account balances
- Employee salaries
- Passwords
- Medical records

Python uses naming conventions to indicate different levels of access.


````python
class BankAccount:

    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.__balance = balance

    def check_balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Deposit successful")
        else:
            print("Deposit amount must be greater than zero")


account1 = BankAccount("John", 50000)

print(account1.check_balance())

account1.deposit(10000)

print(account1.check_balance())
````

## Breakdown of private attributes

```python
self.__balance = balance
```

The double underscore indicates that the attribute should be treated as private.

Instead of directly changing the balance, we use methods such as:

```python
deposit()
check_balance()
```

This gives the program more control over what happens to important data.


# 15. Business Scenario 2 - Banking System

A bank account should support:

- Deposits
- Withdrawals
- Balance checking

The system should also prevent invalid transactions.


````python
class BankAccount:

    def __init__(self, account_holder, account_number, balance=0):
        self.account_holder = account_holder
        self.account_number = account_number
        self.__balance = balance

    def deposit(self, amount):
        if amount <= 0:
            print("Invalid deposit amount")
            return

        self.__balance += amount
        print(f"Deposit successful. New balance: {self.__balance}")

    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid withdrawal amount")

        elif amount > self.__balance:
            print("Insufficient balance")

        else:
            self.__balance -= amount
            print(f"Withdrawal successful. New balance: {self.__balance}")

    def check_balance(self):
        print(f"Current balance: {self.__balance}")


account = BankAccount("Gbenga", "ACC001", 100000)

account.deposit(50000)
account.withdraw(30000)
account.check_balance()
````

## Business scenario explanation

This class keeps banking information and banking actions together.

### `deposit()`

Validates that the amount is greater than zero before adding money.

### `withdraw()`

Checks:

1. Is the amount valid?
2. Does the customer have enough money?

### `check_balance()`

Displays the protected account balance.

This is a realistic example of encapsulation and business rules.


# 16. Inheritance

Inheritance allows one class to inherit features from another class.

The original class is often called:

- Parent class
- Base class
- Superclass

The new class is often called:

- Child class
- Derived class
- Subclass

## Example

A general `Employee` may have:

- Name
- Salary

A `DataScientist` is also an employee, but may have additional information such as programming language.


````python
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display_details(self):
        print(f"Name: {self.name}")
        print(f"Salary: {self.salary}")


class DataScientist(Employee):

    def __init__(self, name, salary, programming_language):
        super().__init__(name, salary)
        self.programming_language = programming_language

    def build_model(self):
        print(f"{self.name} is building a machine learning model")


data_scientist = DataScientist(
    "Amina",
    800000,
    "Python"
)

data_scientist.display_details()
data_scientist.build_model()

print(data_scientist.programming_language)
````

## Line-by-line breakdown

```python
class DataScientist(Employee):
```

This means:

- Create a class called `DataScientist`.
- Inherit from the `Employee` class.

The child class can now reuse features from the parent class.

```python
super().__init__(name, salary)
```

`super()` gives access to the parent class.

This line runs the parent class constructor and initializes:

- `name`
- `salary`

The child class then adds:

```python
self.programming_language = programming_language
```

This is one of the biggest benefits of inheritance: code reuse.


# 17. Method Overriding

A child class can change how an inherited method behaves.

This is called method overriding.


````python
class Employee:

    def work(self):
        print("The employee is working")


class DataScientist(Employee):

    def work(self):
        print("The Data Scientist is analyzing data and building models")


class AIEngineer(Employee):

    def work(self):
        print("The AI Engineer is building and deploying AI systems")


employee1 = DataScientist()
employee2 = AIEngineer()

employee1.work()
employee2.work()
````

Both child classes have a `work()` method, but each performs a different behavior.

This leads us to polymorphism.


# 18. Polymorphism

Polymorphism means **many forms**.

Different objects can respond to the same method name in different ways.


````python
class DataScientist:

    def work(self):
        print("Analyzing data and building predictive models")


class AIEngineer:

    def work(self):
        print("Building, integrating, and deploying AI applications")


class DataAnalyst:

    def work(self):
        print("Cleaning data and creating business insights")


employees = [
    DataScientist(),
    AIEngineer(),
    DataAnalyst()
]

for employee in employees:
    employee.work()
````

## Breakdown

Each object has a method named:

```python
work()
```

However, each object performs a different action.

The loop does not need to know the exact type of employee in advance.

It simply calls:

```python
employee.work()
```

Python allows each object to respond appropriately.

This is polymorphism.


# 19. Abstraction

Abstraction means focusing on what an object should do while hiding unnecessary implementation details.

For example, when using an ATM, a customer presses "Withdraw".

The customer does not need to know:

- Database operations
- Security checks
- Network communication
- Internal transaction processing

The customer only interacts with the important interface.

In Python, abstraction can be implemented using the `abc` module.


````python
from abc import ABC, abstractmethod


class Payment(ABC):

    @abstractmethod
    def process_payment(self, amount):
        pass


class BankTransfer(Payment):

    def process_payment(self, amount):
        print(f"Processing bank transfer of {amount}")


class CardPayment(Payment):

    def process_payment(self, amount):
        print(f"Processing card payment of {amount}")


payment1 = BankTransfer()
payment1.process_payment(50000)
````

## Key idea

The abstract class `Payment` defines a rule:

Every payment type must provide:

```python
process_payment()
```

Different payment systems can implement the rule differently.

This is useful in large business systems.

# 21. Business Scenario 3 - E-commerce System

An e-commerce business sells products.

We want to represent products using OOP.

Each product should have:

- Name
- Price
- Quantity

The product should be able to:

- Display information.
- Calculate total inventory value.


````python
class Product:

    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def display_product(self):
        print(f"Product: {self.name}")
        print(f"Price: {self.price}")
        print(f"Quantity: {self.quantity}")

    def inventory_value(self):
        return self.price * self.quantity


product1 = Product("Laptop", 800000, 10)

product1.display_product()

print("Total Inventory Value:", product1.inventory_value())
````

## Line-by-line business explanation

```python
self.price * self.quantity
```

This calculates:

**Price of one product x Number of units in stock**

For example:

800,000 x 10 = 8,000,000

This is useful for:

- Inventory management

- Financial reporting
- E-commerce dashboards


# 22. Business Scenario 4 - Customer and VIP Customer

A business may have normal customers and VIP customers.

Both have:

- Name
- Email

VIP customers may receive discounts.

This is a good use case for inheritance.


````python
class Customer:

    def __init__(self, name, email):
        self.name = name
        self.email = email

    def customer_details(self):
        print(f"Name: {self.name}")
        print(f"Email: {self.email}")


class VIPCustomer(Customer):

    def __init__(self, name, email, discount_rate):
        super().__init__(name, email)
        self.discount_rate = discount_rate

    def calculate_discount(self, purchase_amount):
        return purchase_amount * self.discount_rate


vip = VIPCustomer(
    "Grace",
    "grace@email.com",
    0.10
)

vip.customer_details()

discount = vip.calculate_discount(200000)

print("Discount:", discount)
````

# 23. Business Scenario 5 - Sales Performance System

A company wants to track sales representatives.

Each representative has:

- Name
- Monthly sales

The system should calculate whether the representative reached a target.


````python
class SalesRepresentative:

    target = 1000000

    def __init__(self, name, monthly_sales):
        self.name = name
        self.monthly_sales = monthly_sales

    def check_target(self):
        if self.monthly_sales >= self.target:
            print(f"{self.name} achieved the sales target")
        else:
            remaining = self.target - self.monthly_sales
            print(f"{self.name} needs {remaining} more to reach the target")


salesperson1 = SalesRepresentative("Daniel", 1200000)
salesperson2 = SalesRepresentative("Joy", 750000)

salesperson1.check_target()
salesperson2.check_target()
````

# 24. Business Scenario 6 - Simple AI Model Class

OOP is very useful in AI Engineering.

We can create a class representing a machine learning model.


````python
class AIModel:

    def __init__(self, model_name, algorithm):
        self.model_name = model_name
        self.algorithm = algorithm
        self.is_trained = False

    def train(self):
        self.is_trained = True
        print(f"{self.model_name} is now trained")

    def predict(self):
        if self.is_trained:
            print(f"{self.model_name} is making predictions")
        else:
            print("Train the model before making predictions")


model = AIModel(
    "Customer Churn Predictor",
    "Random Forest"
)

model.predict()
model.train()
model.predict()
````

## AI Engineering breakdown

The object stores the model's state:

```python
self.is_trained
```

Initially:

```python
self.is_trained = False
```

After calling:

```python
model.train()
```

The state changes to:

```python
True
```

This is an example of an object maintaining information about its current state.

# 26. Practical Project - Mini Banking Application

This project combines several topics learned so far.

The application allows a customer to:

- Create an account.
- Deposit money.
- Withdraw money.
- Check account details.


````python
class BankAccount:

    bank_name = "Kajola Digital Bank"

    def __init__(self, account_holder, account_number, opening_balance=0):
        self.account_holder = account_holder
        self.account_number = account_number
        self.__balance = opening_balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposit of {amount} was successful")
        else:
            print("Invalid deposit amount")

    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid withdrawal amount")
        elif amount > self.__balance:
            print("Insufficient funds")
        else:
            self.__balance -= amount
            print(f"Withdrawal of {amount} was successful")

    def check_balance(self):
        print(f"Current balance: {self.__balance}")

    def account_details(self):
        print(f"Bank: {self.bank_name}")
        print(f"Account Holder: {self.account_holder}")
        print(f"Account Number: {self.account_number}")
        print(f"Balance: {self.__balance}")


customer = BankAccount(
    "Kajola Gbenga",
    "KGB001",
    100000
)

customer.deposit(50000)
customer.withdraw(25000)
customer.check_balance()
customer.account_details()
````
