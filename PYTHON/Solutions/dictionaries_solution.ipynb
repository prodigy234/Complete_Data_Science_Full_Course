# Solution to Task 3

## Using get()

````python
student = {
    "name": "John",
    "age": 22
}
print(student.get("course", "Course Not Found"))
````

### Syntax of get():

```
dictionary.get(key, default_value)

key - the key to search for.

default_value (optional) - the value returned if the key is not found. If omitted, None is returned.
```
