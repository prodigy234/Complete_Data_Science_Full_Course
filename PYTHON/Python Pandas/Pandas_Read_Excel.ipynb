## 1️⃣ Introduction to `read_excel()`

`pandas.read_excel()` is used to read Excel files into a Pandas DataFrame.

### Basic Syntax

```python
pd.read_excel(io, sheet_name=0)
```

### Required Import


````python

import pandas as pd
````


## 2️⃣ Reading a Single Excel Sheet

### Example 1: Read First Sheet (Default)


````python

df = pd.read_excel("sample.xlsx")
df.head()
````


### Example 2: Read Specific Sheet by Name


````python

df = pd.read_excel("sample.xlsx", sheet_name="Sales")
df.head()
````


### Example 3: Read Sheet by Index


````python

df = pd.read_excel("sample.xlsx", sheet_name=1)
df.head()
````


## 3️⃣ Reading Multiple Sheets


## 4️⃣ Important Parameters of `read_excel()`

### 1. `usecols`
Read specific columns only.


````python

df = pd.read_excel("sample.xlsx", usecols="A:C")
df.head()
````


### 2. `skiprows`
Skip unwanted rows.


````python

df = pd.read_excel("sample.xlsx", skiprows=2)
df.head()
````


### 3. `nrows`
Limit number of rows.


````python

df = pd.read_excel("sample.xlsx", nrows=10)
df.head()
````


### 4. `header`
Define header row.


````python

df = pd.read_excel("sample.xlsx", header=2)
df.head()
````


### 5. `index_col`
Set index column.


````python

df = pd.read_excel("sample.xlsx", index_col=0)
df.head()
````


### 6. `dtype`
Force specific data types.


````python

df = pd.read_excel("sample.xlsx", dtype={"ID": str})
df.dtypes
````


### 7. `parse_dates`
Parse date columns automatically.


What parse_dates Does in pd.read_excel()

parse_dates tells Pandas:

"Convert this column from normal text (object) into real datetime format automatically while reading the file."

Without parse_dates

If you run:

````python
df = pd.read_excel("sample.xlsx", sheet_name="Finance")
df.dtypes
````

That means Pandas is treating the date like ordinary text.

And text dates:

- Cannot be used for time filtering properly

- Cannot extract year/month/day easily

- Cannot do time-based analysis

`   It becomes a true datetime object.  `

````python

df = pd.read_excel("sample.xlsx", parse_dates=["Date"])
df.dtypes # The error is happening because:
          # There is no column named "Date" in sample.xlsx.
````

If Reading Sales SheetIf Reading Sales Sheet

````python
df = pd.read_excel("sample.xlsx", sheet_name="Sales", parse_dates=["OrderDate"])
df.dtypes
````

If Reading Finance Sheet

````python
df = pd.read_excel("sample.xlsx", sheet_name="Finance", parse_dates=["TransactionDate"])
df.dtypes
````

If Reading Health Sheet

````python
df = pd.read_excel("sample.xlsx", sheet_name="Health",
                   parse_dates=["AdmissionDate", "DischargeDate"])
df.dtypes
````

`   NOTE: Before using parse_dates, always check the column names:  `

`   This prevents 80% of beginner errors.    `

Why This Is Powerful

Once converted, you can now do:

````python
# Extract Year

df["TransactionDate"].dt.year
````

````python
# Extract Month

df["TransactionDate"].dt.month
````

````python
# Filter by Date

df[df["TransactionDate"] > "2025-06-01"]
````

````python
# Time-Based Grouping

df.groupby(df["TransactionDate"].dt.month)["Amount"].sum()
````

We grouped transactions by month and calculated the total amount spent in each month.

What This Code Means

(1) df["TransactionDate"]

Selects the date column.

(2) .dt.month

Extracts the month number from each date.

Example:

TransactionDate	        dt.month
2025-01-15	            1
2025-03-22	            3
2025-11-01	            11

So now Pandas is grouping by month number (1–12).

(3) df.groupby(...)

This tells Pandas:

“Group rows that belong to the same month together.”

(4) ["Amount"]

Select only the Amount column.

(5) .sum()

Add up all the amounts inside each month group.

The left column (1–12) is called the Group Index.

It represents:

- 1 = January

- 2 = February

- 3 = March

- ...

- 12 = December

So:

Month 4 (April) had total expenses of 122,857

Month 11 had total expenses of 61,291
