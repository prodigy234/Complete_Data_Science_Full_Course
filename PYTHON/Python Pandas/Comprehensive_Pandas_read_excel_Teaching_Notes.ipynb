---
# 1️⃣ Basic Usage of read_excel()

## Syntax:
`pd.read_excel(io, sheet_name=0)`


```python
# Read first sheet (default)
df = pd.read_excel('sample.xlsx')
df.head()
```

---
# 2️⃣ Reading a Specific Sheet

You can specify sheet by:
- Sheet index (0,1,2...)
- Sheet name ("Sales", "Finance")


```python
# By index
df1 = pd.read_excel('sample.xlsx', sheet_name=1)

# By sheet name
df2 = pd.read_excel('sample.xlsx', sheet_name='Sales')

df2.head()
```

---
# 3️⃣ Reading Multiple Sheets at Once

To load ALL sheets:


```python
all_sheets = pd.read_excel('sample.xlsx', sheet_name=None)

# all_sheets becomes a dictionary
print(type(all_sheets))

# Access individual sheet
all_sheets['Sales'].head()
```

### Combine Multiple Sheets


```python
combined_df = pd.concat(all_sheets.values(), ignore_index=True)
combined_df.head()
```

---
# 4️⃣ Working with Headers


```python
# If header is on row 3
df = pd.read_excel('sample.xlsx', header=2)

# No header in file
df = pd.read_excel('sample.xlsx', header=None)

# Rename columns manually
df.columns = ['Col1', 'Col2', 'Col3']
df.head()
```

---
# 5️⃣ Skipping Rows


```python
# Skip first 2 rows
df = pd.read_excel('sample.xlsx', skiprows=2)

# Skip specific rows
df = pd.read_excel('sample.xlsx', skiprows=[0,2,4])
df.head()
```

---
# 6️⃣ Selecting Specific Columns


```python
# Select by column names
df = pd.read_excel('sample.xlsx', usecols=['Name', 'Sales'])

# Select by column letters
df = pd.read_excel('sample.xlsx', usecols='A:C')
df.head()
```

---
# 7️⃣ Handling Missing Values


```python
# Define custom missing values
df = pd.read_excel('sample.xlsx', na_values=['N/A', 'Missing'])
df.isna().sum()
```

---
# 8️⃣ Setting Data Types


```python
df = pd.read_excel('sample.xlsx', dtype={'Sales': float, 'Age': int})
df.dtypes
```

---
# 9️⃣ Working with Dates


```python
df = pd.read_excel('sample.xlsx', parse_dates=['Order_Date'])
df.info()
```

---
# 🔟 Reading Large Excel Files Efficiently
- Load only required columns
- Load specific sheets
- Use nrows to limit rows


```python
df = pd.read_excel('sample.xlsx', nrows=1000)
df.head()
```
