# 4. Python Sets

A set is an unordered collection of unique values.

Sets are especially useful for:
- Removing duplicates
- Membership testing
- Union
- Intersection
- Difference

## Question 1: Remove duplicates

````python
numbers = [10, 20, 10, 30, 20, 40, 30]

unique_numbers = set(numbers)

print(unique_numbers)
````

### Line-by-line breakdown

**Line 1:** Creates a list containing duplicate values.

**Line 3:** `set(numbers)` converts the list into a set and automatically removes duplicates.

**Line 5:** Displays the unique values.

The order of a set should not be relied upon.

## Question 2: Union and intersection

````python
python_students = {"Ada", "John", "Mary"}
data_students = {"Mary", "David", "John"}

all_students = python_students.union(data_students)
both_courses = python_students.intersection(data_students)

print("All students:", all_students)
print("Students in both:", both_courses)
````

### Line-by-line breakdown

**Lines 1-2:** Create two sets.

**Line 4:** `union()` combines all unique values.

**Line 5:** `intersection()` returns values found in both sets.

This can be useful when comparing groups of customers, students or records.

## Question 3: Difference

````python
online_students = {"Ada", "John", "Mary"}
physical_students = {"John", "David"}

online_only = online_students.difference(physical_students)

print(online_only)
````

### Line-by-line breakdown

The `difference()` operation returns values present in the first set but not the second set.

# 6. Conditional Statements

Conditional statements allow a program to make decisions.

Basic structure:

```python
if condition:
    statement
elif another_condition:
    statement
else:
    statement
```

## Question 1: Pass or fail

````python
score = 67

if score >= 50:
    print("Pass")
else:
    print("Fail")
````

### Line-by-line breakdown

**Line 1:** Stores the score.

**Line 3:** Tests whether the score is at least 50.

**Line 4:** Executes when the condition is true.

**Line 5:** Separates the alternative path.

**Line 6:** Executes when the condition is false.

## Question 2: Grade classification

````python
score = 82

if score >= 70:
    grade = "A"
elif score >= 60:
    grade = "B"
elif score >= 50:
    grade = "C"
elif score >= 45:
    grade = "D"
else:
    grade = "F"

print("Grade:", grade)
````

### Line-by-line breakdown

Python evaluates the conditions from top to bottom.

- 70 or above gets A.
- 60 to 69 gets B.
- 50 to 59 gets C.
- 45 to 49 gets D.
- Anything below 45 gets F.

Once Python finds a true condition, it skips the remaining `elif` and `else` branches.

## Question 3: Combine conditions

````python
age = 25
has_id = True

if age >= 18 and has_id:
    print("Access granted")
else:
    print("Access denied")
````

### Line-by-line breakdown

**Line 1:** Stores the person's age.

**Line 2:** Stores a Boolean value.

**Line 4:** `and` requires both conditions to be true.

**Line 5:** Runs when both requirements are satisfied.

**Line 7:** Runs otherwise.

# 7. Functions

A function is a reusable block of code designed to perform a specific task.

Basic syntax:

```python
def function_name(parameters):
    statements
    return result
```

## Question 1: Create a simple function

````python
def greet():
    print("Welcome to Python")

greet()
````

**Line 2:** Contains the function body.

## Question 2: Function with a parameter

## Question 3: Function that returns a value

````python
def calculate_total(price, quantity):
    total = price * quantity
    return total

result = calculate_total(2500, 4)

print("Total:", result)
````

### Line-by-line breakdown

**Line 1:** Defines a function with two parameters.

**Line 2:** Multiplies price by quantity.

**Line 3:** Returns the calculated result.

## Question 4: Function for data classification

````python
def classify_score(score):
    if score >= 70:
        return "Excellent"
    elif score >= 50:
        return "Pass"
    else:
        return "Fail"

print(classify_score(82))
print(classify_score(55))
print(classify_score(30))
````

### Line-by-line breakdown

The function receives a score and returns a category.

The important idea is that the decision-making logic is packaged into a reusable function. This avoids repeating the same code throughout a program.

## Question 5: Function with a default parameter

````python
def calculate_discount(price, discount=10):
    discounted_price = price - (price * discount / 100)
    return discounted_price

print(calculate_discount(10000))
print(calculate_discount(10000, 20))
````

### Line-by-line breakdown

**Line 1:** `discount=10` provides a default value.

**Line 2:** Calculates the discount amount and subtracts it from the original price.

**Line 5:** Uses the default 10 percent discount.

**Line 6:** Overrides the default and uses 20 percent.

## Question 2: User input with a while loop

Keep asking for a score until the user enters 0.

````python
total_sales = 0
sale = float(input("Enter sale amount, or 0 to stop: "))

while sale != 0:
    total_sales = total_sales + sale
    sale = float(input("Enter another sale, or 0 to stop: "))

print("Total sales:", total_sales)
````

### Line-by-line breakdown

**Line 1:** Requests the first score and converts the input from text to an integer.

**Line 3:** Continues while the score is not zero.

**Lines 4-7:** Classify the score.

**Line 9:** Requests another score.

**Line 11:** Executes after the user enters zero.

This pattern is useful for interactive programs.

````python
total_sales = 0

while True:
    sale = input("Enter sale amount, or 0 to stop: ")

    if sale == "":
        print("Please enter a number.")
        continue

    sale = float(sale)

    if sale == 0:
        break

    total_sales = total_sales + sale

print("Total sales:", total_sales)
````

Ask for a sale → validate the input → convert it to a number → check whether it is zero → add it to the total → repeat until zero is entered → display the final total.

### Line-by-line breakdown

**Line 2:** Collects the first sale.

**Line 4:** Continues until the user enters zero.

**Line 5:** Adds each sale to the running total.

**Line 6:** Requests the next sale.

**Line 8:** Prints the final total.

# 10. break, continue and pass

## break

Stops the loop immediately.

## continue

Skips the current iteration and moves to the next iteration.

## pass

Does nothing. It is commonly used as a placeholder.

## Question 1: break

Search for a number and stop when it is found.

````python
numbers = [10, 20, 30, 50, 70, 90]

for number in numbers:
    if number == 50:
        print("Number found")
        break
    print(number)
````

### Line-by-line breakdown

The loop examines values in order.

When `number == 50` becomes true, Python prints the message and immediately executes `break`, ending the loop.

## Question 2: continue

Print only even numbers.

````python
numbers = [1, 2, 3, 4, 5, 6]

for number in numbers:
    if number % 2 != 0:
        continue
    print(number)
````

### Line-by-line breakdown

**Line 4:** `%` calculates the remainder.

If a number has a remainder when divided by 2, it is odd.

**Line 5:** `continue` skips odd numbers.

**Line 6:** Only even numbers reach the print statement.

## Question 3: pass

````python
age = 17

if age >= 18:
    pass
else:
    print("Under 18")
````

### Line-by-line breakdown

`pass` allows Python to have an intentionally empty block. It does not produce output or change the program state.

## Question 1: Student performance analyzer

Create a function that receives student scores, calculates the average and classifies performance.

````python
def analyze_scores(scores):
    total = 0

    for score in scores:
        total = total + score

    average = total / len(scores)

    if average >= 70:
        performance = "Excellent"
    elif average >= 50:
        performance = "Pass"
    else:
        performance = "Fail"

    return total, average, performance


scores = [75, 80, 65, 90, 70]

total, average, performance = analyze_scores(scores)

print("Total:", total)
print("Average:", average)
print("Performance:", performance)
````

### Detailed line-by-line breakdown

**Line 1:** Defines the reusable function.

**Line 2:** Creates an accumulator for the total.

**Line 4:** Loops through every score.

**Line 5:** Adds each score to the total.

**Line 7:** Calculates the average.

**Lines 9-15:** Classify the average using conditional statements.

**Line 17:** Returns three values.

**Line 20:** Creates the score list.

**Line 22:** Calls the function.

**Line 24:** Uses tuple unpacking to receive the three returned values.

**Lines 26-28:** Display the results.

This single example demonstrates functions, lists, loops, arithmetic, conditions and tuple unpacking.

## Question 5: Simple sales report

Use a list of dictionaries to represent sales records and calculate total revenue.

**Line 12:** Displays revenue for the current product.

**Line 14:** Displays total revenue.

This structure is very common when processing JSON-like business data.
