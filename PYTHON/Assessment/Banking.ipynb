## Write a Python program that processes a customer's bank transaction based on the transaction type.

The program should accept:
```

- Account balance

- Transaction type: "deposit", "withdrawal", or "transfer"

- Transaction amount

```

Use if, elif, and else to apply the following rules:

````
- 1. If the transaction is deposit, add the amount to the account balance and display the new balance.

- 2. If the transaction is withdrawal:

- (a) If the amount is greater than the balance, display "Insufficient funds".

- (b) Otherwise, deduct the amount and display the new balance.

- 3. If the transaction is transfer:

- (a) If the amount is greater than the balance, display "Transfer failed: Insufficient funds".

- (b) Otherwise, deduct the amount and display the new balance.

- 4. If the transaction type is anything else, display "Invalid transaction type".

# Solution to Question 1

`````python
# Bank Transaction Program

balance = float(input("Enter account balance: "))
transaction_type = input("Enter transaction type (deposit, withdrawal, transfer): ").lower()
amount = float(input("Enter transaction amount: "))

if transaction_type == "deposit":
    balance = balance + amount
    print("Deposit successful.")
    print("New balance:", balance)

elif transaction_type == "withdrawal":
    if amount > balance:
        print("Insufficient funds")
    else:
        balance = balance - amount
        print("Withdrawal successful.")
        print("New balance:", balance)

elif transaction_type == "transfer":
    if amount > balance:
        print("Transfer failed: Insufficient funds")
    else:
        balance = balance - amount
        print("Transfer successful.")
        print("New balance:", balance)

else:
    print("Invalid transaction type")
`````
