## Section 1: Python Data Types


Every value in Python has a data type. A data type tells Python what kind of value
is being stored and what operations can be performed on it. Python is dynamically
typed, which means you do not need to declare the type of a variable before using
it - Python figures it out automatically based on the value you assign. Python is
also strongly typed, which means it will not silently convert between unrelated
types (for example, adding a string to an integer will raise an error instead of
guessing what you meant).

In this section we cover every core built in data type: integers, floating point
numbers, complex numbers, booleans, strings, and the special value None. We also
cover how to check a type, and how to convert (cast) one type into another.



### 1.1 Checking a Type with type()

Before learning individual types, it helps to know how to inspect the type of any
value. Python gives us the built in type() function for this purpose. Every single
value that exists in Python, including numbers, text, functions, and even classes,
has a type that Python can report back to you.


````python
weight = 70
print(type(weight))

name = "Kajola"
print(type(name))

is_active = True
print(type(is_active))

nothing = None
print(type(nothing))
````

**Expected output:**
```
<class 'int'>
<class 'str'>
<class 'bool'>
<class 'NoneType'>
```


**Line by line breakdown:**

- `weight = 70`
  Creates a variable named weight and binds it to the integer value 70.
- `print(type(weight))`
  Calls the built in type() function on weight, which returns the class of the object (int), then prints it.
- `name = "Kajola"`
  Creates a variable named name and binds it to a string value.
- `print(type(name))`
  Prints the type of name, which will be str (short for string).
- `is_active = True`
  Creates a variable is_active and binds it to the boolean value True.
- `print(type(is_active))`
  Prints the type of is_active, which will be bool.
- `nothing = None`
  Creates a variable nothing and binds it to the special null value None.
- `print(type(nothing))`
  Prints the type of nothing, which will be NoneType.


> **Notes and tips:**
> - type() always returns a class object, shown as <class 'typename'>.
> - Use isinstance(value, SomeType) instead of type(value) == SomeType when you also want to correctly handle subclasses.


### 1.2 Integers (int)

An integer is a whole number, positive, negative, or zero, with no decimal point.
Python integers have unlimited precision, meaning they can grow as large as your
computer's memory allows, unlike many other languweights where integers overflow
after a fixed number of bits.


````python
positive_number = 42
negative_number = -17
zero = 0
big_number = 123_456_789_000

binary_value = 0b1010
octal_value = 0o12
hex_value = 0xA

print(positive_number, negative_number, zero)
print(big_number)
print(binary_value, octal_value, hex_value)
print(type(positive_number))
````

**Expected output:**
```
42 -17 0
123456789000
10 10 10
<class 'int'>
```


**Line by line breakdown:**

- `positive_number = 42`
  Assigns the whole number 42 to positive_number.
- `negative_number = -17`
  Assigns a negative whole number to negative_number using the unary minus sign.
- `zero = 0`
  Assigns zero, which is also a valid integer.
- `big_number = 123_456_789_000`
  The underscore is a visual separator only, it groups digits for readability and is ignored by Python; the value is one hundred twenty three billion plus.
- `binary_value = 0b1010`
  The 0b prefix means the digits that follow are read as binary; 1010 in binary equals 10 in decimal.
- `octal_value = 0o12`
  The 0o prefix means base 8; 12 in octal equals 10 in decimal.
- `hex_value = 0xA`
  The 0x prefix means base 16; A in hexadecimal equals 10 in decimal.
- `print(positive_number, negative_number, zero)`
  Prints the three values separated by a single space, which is the default separator for print().
- `print(big_number)`
  Prints the large integer; Python removes the underscores automatically when displaying the number.
- `print(binary_value, octal_value, hex_value)`
  All three variables hold the integer value 10, because binary 1010, octal 12, and hex A all represent the same decimal number, so 10 10 10 is printed.
- `print(type(positive_number))`
  Confirms the type is int.


> **Notes and tips:**
> - Integer division uses //, while normal division / always returns a float.
> - The ** operator raises a number to a power, for example 2 ** 10 equals 1024.
> - The modulo operator % returns the remainder of a division, for example 10 % 3 equals 1.


### 1.3 Floating Point Numbers (float)

A float represents a real number that includes a decimal point, such as 3.14 or
-0.5. Floats can also be written using scientific notation. Because computers
store floats in binary, some decimal values cannot be represented exactly, which
can lead to tiny rounding differences.


````python
price = 19.99
temperature = -3.5
scientific = 1.5e3

result = 0.1 + 0.2
print(price, temperature, scientific)
print(result)
print(round(result, 2))
print(type(price))
````

**Expected output:**
```
19.99 -3.5 1500.0
0.70000000000000004
0.3
<class 'float'>
```


**Line by line breakdown:**

- `price = 19.99`
  Assigns a floating point value with two decimal places to price.
- `temperature = -3.5`
  Assigns a negative float to temperature.
- `scientific = 1.5e3`
  The e3 means multiply by 10 to the power of 3, so this equals 1500.0.
- `result = 0.1 + 0.2`
  Adds two floats; due to binary floating point representation this produces 0.70000000000000004 instead of exactly 0.3.
- `print(price, temperature, scientific)`
  Prints all three float values.
- `print(result)`
  Prints the imprecise sum, demonstrating the classic floating point rounding quirk.
- `print(round(result, 2))`
  The round() function rounds result to 2 decimal places, producing a clean 0.3 for display purposes.
- `print(type(price))`
  Confirms the type is float.


> **Notes and tips:**
> - Never compare floats with == for equality in critical code; instead check that the difference is smaller than a tiny tolerance (epsilon).
> - Use the decimal module when you need exact decimal arithmetic, for example in financial calculations.


### 1.4 Complex Numbers (complex)

Python has built in support for complex numbers, which are made up of a real part
and an imaginary part. The imaginary part is written using a trailing j.


````python
z = 3 + 4j
print(z)
print(z.real)
print(z.imag)
print(abs(z))
print(type(z))
````

**Expected output:**
```
(3+4j)
3.0
4.0
5.0
<class 'complex'>
```


**Line by line breakdown:**

- `z = 3 + 4j`
  Creates a complex number where 3 is the real part and 4j is the imaginary part.
- `print(z)`
  Prints the complex number in the form (3+4j).
- `print(z.real)`
  Accesses the .real attribute, returning 3.0 as a float.
- `print(z.imag)`
  Accesses the .imag attribute, returning 4.0 as a float.
- `print(abs(z))`
  abs() on a complex number returns its magnitude, calculated as the square root of (real squared plus imag squared), which is 5.0 here.
- `print(type(z))`
  Confirms the type is complex.


> **Notes and tips:**
> - Complex numbers are mostly used in scientific and engineering computing, for example signal processing.


### 1.5 Booleans (bool)

A boolean can only ever be True or False. Booleans are the result of comparisons
and logical operations, and in Python bool is technically a subclass of int, so
True behaves like 1 and False behaves like 0 in arithmetic.


````python
is_valid = True
is_empty = False

print(is_valid, is_empty)
print(True + True)
print(5 > 3)
print(5 == 3)
print(bool(0), bool(1), bool(""), bool("hello"), bool([]))
````

**Expected output:**
```
True False
2
True
False
False True False True False
```


**Line by line breakdown:**

- `is_valid = True`
  Assigns the boolean literal True.
- `is_empty = False`
  Assigns the boolean literal False.
- `print(is_valid, is_empty)`
  Prints True False.
- `print(True + True)`
  Because bool is a subclass of int, True is treated as 1, so True + True equals 2.
- `print(5 > 3)`
  The comparison evaluates to True.
- `print(5 == 3)`
  The equality check evaluates to False since 5 does not equal 3.
- `print(bool(0), bool(1), bool(""), bool("hello"), bool([]))`
  Shows how bool() converts other types: 0 is falsy, 1 is truthy, an empty string is falsy, a non empty string is truthy, and an empty list is falsy.


> **Notes and tips:**
> - Falsy values in Python include: 0, 0.0, None, False, empty string, empty list, empty tuple, empty dict, and empty set.
> - Everything else is truthy by default.


### 1.6 Strings (str)

A string is an ordered, immutable sequence of characters used to represent text.
Strings can be created with single quotes, double quotes, or triple quotes for
multi line text. Because strings are sequences, they support indexing, slicing,
and many built in methods.


````python
first_name = "Kajola"
last_name = 'Gbenga'
full_title = f"{first_name} {last_name} - Data Scientist & AI Engineer"

print(full_title)
print(full_title.upper())
print(full_title.lower())
print(len(full_title))
print(full_title[0])
print(full_title[-1])
print(full_title[0:6])
print(full_title.split(" "))
print("-".join(["a", "b", "c"]))
print(full_title.replace("Engineer", "Practitioner"))
````

**Expected output:**
```
Kajola Gbenga - Data Scientist & AI Engineer
Kajola GBENGA - DATA SCIENTIST & AI ENGINEER
Kajola gbenga - data scientist & ai engineer
44
K
r
Kajola
['Kajola', 'Gbenga', '-', 'Data', 'Scientist', '&', 'AI', 'Engineer']
a-b-c
Kajola Gbenga - Data Scientist & AI Practitioner
```


**Line by line breakdown:**

- `first_name = "Kajola"`
  Creates a string using double quotes.
- `last_name = 'Gbenga'`
  Creates a string using single quotes; both quote styles behave identically in Python.
- `full_title = f"{first_name} {last_name} - Data Scientist & AI Engineer"`
  An f string (formatted string literal) lets you embed variables directly inside {} placeholders; Python replaces them with their values when the string is built.
- `print(full_title)`
  Prints the fully assembled string.
- `print(full_title.upper())`
  The .upper() method returns a new string with every character converted to uppercase; the original string is not changed because strings are immutable.
- `print(full_title.lower())`
  The .lower() method returns a new string in all lowercase.
- `print(len(full_title))`
  len() returns the number of characters in the string, including spaces.
- `print(full_title[0])`
  Indexing with [0] returns the first character of the string; Python indexing starts at 0.
- `print(full_title[-1])`
  Negative indexing counts from the end; [-1] returns the last character.
- `print(full_title[0:6])`
  Slicing with [0:6] returns a substring starting at index 0 up to but not including index 6.
- `print(full_title.split(" "))`
  The .split(" ") method breaks the string into a list of substrings wherever a space occurs.
- `print("-".join(["a", "b", "c"]))`
  The .join() method glues together the items of an iterable using the string it is called on as the separator, producing 'a-b-c'.
- `print(full_title.replace("Engineer", "Practitioner"))`
  The .replace() method returns a new string with every occurrence of the first argument swapped for the second argument.


> **Notes and tips:**
> - Strings are immutable, meaning every string method returns a brand new string rather than modifying the original in place.
> - Use plain hyphens (-) for ranges or compound words in code and text, not en dashes or em dashes, to keep text consistent and copy paste safe.
> - Common string methods worth remembering: strip(), startswith(), endswith(), find(), count(), title(), format().


### 1.7 NoneType

None is a special built in constant that represents the absence of a value. It is
the only value of the type NoneType. It is commonly used as a default placeholder,
or as the return value of a function that does not explicitly return anything.


````python
result = None

def do_nothing():
    pass

print(result)
print(result is None)
print(do_nothing())
print(type(None))
````

**Expected output:**
```
None
True
None
<class 'NoneType'>
```


**Line by line breakdown:**

- `result = None`
  Assigns the special None value to result, signalling that there is currently no meaningful value.
- `def do_nothing(): pass`
  Defines a function whose body is just the pass keyword, meaning it does nothing and implicitly returns None.
- `print(result)`
  Prints None.
- `print(result is None)`
  The is operator checks identity; this is the correct and recommended way to check for None, rather than using ==.
- `print(do_nothing())`
  Calling the function and printing its return value shows None, because a function with no explicit return statement returns None automatically.
- `print(type(None))`
  Shows that the type of None is NoneType.


> **Notes and tips:**
> - Always compare to None using is None or is not None, never == None, because is checks the exact singleton identity.


### 1.8 Type Casting (Conversion)

Type casting is the process of converting a value from one data type into another,
either explicitly using functions like int(), float(), str(), bool(), or list(),
or automatically by Python in certain expressions (implicit casting).


````python
text_number = "25"
converted = int(text_number)
print(converted, type(converted))

number = 25
as_text = str(number)
print(as_text, type(as_text))

value = "3.14"
as_float = float(value)
print(as_float, type(as_float))

mixed_result = 5 + int("10")
print(mixed_result)

whole_number = int(9.99)
print(whole_number)
````

**Expected output:**
```
25 <class 'int'>
25 <class 'str'>
3.14 <class 'float'>
15
9
```


**Line by line breakdown:**

- `text_number = "25"`
  Creates a string that looks like a number but is not one yet.
- `converted = int(text_number)`
  int() parses the string and converts it into the actual integer 25.
- `print(converted, type(converted))`
  Confirms the value is now 25 with type int.
- `number = 25`
  A normal integer.
- `as_text = str(number)`
  str() converts the integer into the string "25".
- `print(as_text, type(as_text))`
  Confirms the value is now the string "25" with type str.
- `value = "3.14"`
  A string that represents a decimal number.
- `as_float = float(value)`
  float() parses the string and converts it into the actual float 3.14.
- `print(as_float, type(as_float))`
  Confirms the conversion worked.
- `mixed_result = 5 + int("10")`
  int("10") converts the string to 10 first, then Python adds it to 5, giving 15.
- `print(mixed_result)`
  Prints 15.
- `whole_number = int(9.99)`
  Converting a float to int truncates (chops off) the decimal part rather than rounding, so this becomes 9, not 10.
- `print(whole_number)`
  Prints 9.


> **Notes and tips:**
> - int() on a non numeric string, such as int("abc"), raises a ValueError.
> - int() truncates floats toward zero; use round() first if you want normal rounding behaviour.

## Section 2: Python Data Structures


Data structures are containers that hold multiple values together in an organized
way. Python ships with four core built in data structures: lists, tuples, sets,
and dictionaries. Choosing the right one depends on whether you need order,
whether the collection should be changeable (mutable), and whether duplicate
values are allowed.



### 2.1 Lists

A list is an ordered, mutable (changeable) collection that can hold items of any
type, including a mix of types. Lists are created with square brackets [] and are
one of the most frequently used data structures in Python.


````python
fruits = ["apple", "banana", "cherry"]
print(fruits)
print(fruits[0])
print(fruits[-1])
print(fruits[0:2])

fruits.append("date")
print(fruits)

fruits.insert(1, "avocado")
print(fruits)

fruits.remove("banana")
print(fruits)

popped = fruits.pop()
print(popped, fruits)

fruits.sort()
print(fruits)

numbers = [4, 2, 9, 1]
squares = [n ** 2 for n in numbers]
print(squares)
````

**Expected output:**
```
['apple', 'banana', 'cherry']
apple
cherry
['apple', 'banana']
['apple', 'banana', 'cherry', 'date']
['apple', 'avocado', 'banana', 'cherry', 'date']
['apple', 'avocado', 'cherry', 'date']
date ['apple', 'avocado', 'cherry']
['apple', 'avocado', 'cherry']
[16, 4, 81, 1]
```


**Line by line breakdown:**

- `fruits = ["apple", "banana", "cherry"]`
  Creates a list of three strings using square brackets, separated by commas.
- `print(fruits)`
  Prints the whole list including brackets and quotes.
- `print(fruits[0])`
  Indexing returns the first element, apple, since indexing starts at 0.
- `print(fruits[-1])`
  Negative indexing returns the last element, cherry.
- `print(fruits[0:2])`
  Slicing returns a new list containing elements at index 0 and 1, stopping before index 2: ['apple', 'banana'].
- `fruits.append("date")`
  The .append() method adds a single new item to the end of the list, modifying it in place.
- `print(fruits)`
  Shows the list now contains four items, ending with date.
- `fruits.insert(1, "avocado")`
  The .insert() method places a new item at a specific index, pushing later items one position to the right.
- `print(fruits)`
  Shows avocado now sitting at index 1.
- `fruits.remove("banana")`
  The .remove() method deletes the first matching value from the list.
- `print(fruits)`
  Confirms banana is gone.
- `popped = fruits.pop()`
  The .pop() method removes and returns the last item of the list by default.
- `print(popped, fruits)`
  Prints the removed item and the list after removal.
- `fruits.sort()`
  The .sort() method reorders the list alphabetically in place, returning nothing.
- `print(fruits)`
  Shows the alphabetically sorted list.
- `numbers = [4, 2, 9, 1]`
  Creates a list of integers in no particular order.
- `squares = [n ** 2 for n in numbers]`
  A list comprehension: for every value n in numbers, compute n squared, and collect all results into a brand new list.
- `print(squares)`
  Prints the new list of squared values in the same order as numbers.


> **Notes and tips:**
> - Lists are mutable: you can change, add, or remove items after creation.
> - Lists allow duplicate values and preserve insertion order.
> - Common methods: append, insert, remove, pop, sort, reverse, extend, count, index, clear.


### 2.2 Tuples

A tuple is an ordered, immutable collection, created with parentheses () or simply
by separating values with commas. Once created, a tuple's contents cannot be
changed, which makes tuples useful for fixed collections of values, and for
returning multiple values from a function safely.


````python
coordinates = (10, 20)
print(coordinates)
print(coordinates[0])

single_item_tuple = (5,)
print(type(single_item_tuple))

x, y = coordinates
print(x, y)

person = ("Kajola", "Gbenga", "Data Scientist & AI Engineer")
name_first, name_last, job_title = person
print(name_first, name_last, job_title)

try:
    coordinates[0] = 99
except TypeError as error:
    print("Error:", error)
````

**Expected output:**
```
(10, 20)
10
<class 'tuple'>
10 20
Kajola Gbenga Data Scientist & AI Engineer
Error: 'tuple' object does not support item assignment
```


**Line by line breakdown:**

- `coordinates = (10, 20)`
  Creates a tuple with two integer values using parentheses.
- `print(coordinates)`
  Prints the tuple, shown with parentheses: (10, 20).
- `print(coordinates[0])`
  Indexing works the same way as lists; this returns 10.
- `single_item_tuple = (5,)`
  The trailing comma is required to make Python treat this as a tuple containing one item; without the comma, (5) would just be the integer 5 in parentheses.
- `print(type(single_item_tuple))`
  Confirms the type is tuple.
- `x, y = coordinates`
  Tuple unpacking: Python assigns the first value of coordinates to x and the second to y in a single line.
- `print(x, y)`
  Prints 10 20.
- `person = ("Kajola", "Gbenga", "Data Scientist & AI Engineer")`
  Creates a tuple holding three related pieces of information about a person.
- `name_first, name_last, job_title = person`
  Unpacks all three values into three separate variables in one statement.
- `print(name_first, name_last, job_title)`
  Prints the three unpacked values.
- `try: coordinates[0] = 99`
  Attempts to change the first item of the tuple, which is not allowed because tuples are immutable.
- `except TypeError as error: print("Error:", error)`
  Python raises a TypeError when you try to modify a tuple; this catches it and prints a friendly messweight instead of crashing the program.


> **Notes and tips:**
> - Tuples are faster and use less memory than lists because Python knows they cannot change.
> - Use a tuple when the collection represents a fixed record, such as a coordinate pair or a database row.


### 2.3 Sets

A set is an unordered collection of unique values, created with curly braces {}
or the set() function. Sets automatically remove duplicates and are optimized for
fast membership testing (checking whether a value exists) and mathematical set
operations such as union and intersection.


````python
skills = {"python", "sql", "python", "machine learning"}
print(skills)

skills.add("deep learning")
print(skills)

skills.discard("sql")
print(skills)

set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

print(set_a.union(set_b))
print(set_a.intersection(set_b))
print(set_a.difference(set_b))
print(set_a.symmetric_difference(set_b))
print(3 in set_a)

unique_lengths = {len(word) for word in ["hi", "hello", "hey", "yo"]}
print(unique_lengths)
````

**Expected output:**
```
{'python', 'sql', 'machine learning'}
{'python', 'sql', 'machine learning', 'deep learning'}
{'python', 'machine learning', 'deep learning'}
{1, 2, 3, 4, 5, 6}
{3, 4}
{1, 2}
{1, 2, 5, 6}
True
{2, 3, 5}
```


**Line by line breakdown:**

- `skills = {"python", "sql", "python", "machine learning"}`
  Creates a set literal; even though python appears twice, sets only keep unique values, so the duplicate is automatically dropped.
- `print(skills)`
  Prints the set with only 3 unique items; the display order is not guaranteed.
- `skills.add("deep learning")`
  The .add() method inserts a single new item into the set.
- `print(skills)`
  Shows the set now containing 4 unique items.
- `skills.discard("sql")`
  The .discard() method removes a value if it exists, and does nothing (no error) if it does not exist.
- `print(skills)`
  Shows sql has been removed.
- `set_a = {1, 2, 3, 4}`
  Creates a set of four integers.
- `set_b = {3, 4, 5, 6}`
  Creates another set of four integers, sharing 3 and 4 with set_a.
- `print(set_a.union(set_b))`
  union() combines all unique values from both sets: {1, 2, 3, 4, 5, 6}.
- `print(set_a.intersection(set_b))`
  intersection() returns only the values present in both sets: {3, 4}.
- `print(set_a.difference(set_b))`
  difference() returns values in set_a that are not in set_b: {1, 2}.
- `print(set_a.symmetric_difference(set_b))`
  symmetric_difference() returns values that are in exactly one of the two sets, but not both: {1, 2, 5, 6}.
- `print(3 in set_a)`
  The in operator checks membership; sets are extremely fast at this check regardless of size, evaluating to True.
- `unique_lengths = {len(word) for word in ["hi", "hello", "hey", "yo"]}`
  A set comprehension: computes the length of each word and keeps only the unique lengths, here {2, 5, 3}.
- `print(unique_lengths)`
  Prints the resulting set of unique word lengths.


> **Notes and tips:**
> - Sets are unordered, so you cannot index into them with [0].
> - Sets cannot contain mutable items such as lists, because every item must be hashable.
> - Use frozenset() for an immutable version of a set.


### 2.4 Dictionaries

A dictionary stores data as key to value pairs, created with curly braces {} using
a colon to separate each key from its value. Dictionaries are ordered (since
Python 3.7, insertion order is preserved) and mutable, and keys must be unique and
hashable (typically strings, numbers, or tuples).


````python
profile = {
    "name": "Kajola Gbenga",
    "title": "Data Scientist & AI Engineer",
    "skills": ["Python", "Machine Learning", "Data Analysis"]
}

print(profile["name"])
print(profile.get("title"))
print(profile.get("company", "Not specified"))

profile["years_experience"] = 5
print(profile)

profile["title"] = "Senior Data Scientist & AI Engineer"
print(profile["title"])

for key, value in profile.items():
    print(key, "->", value)

print(list(profile.keys()))
print(list(profile.values()))

squares_map = {n: n ** 2 for n in range(1, 6)}
print(squares_map)
````

**Expected output:**
```
Kajola Gbenga
Data Scientist & AI Engineer
Not specified
{'name': 'Kajola Gbenga', 'title': 'Data Scientist & AI Engineer', 'skills': ['Python', 'Machine Learning', 'Data Analysis'], 'years_experience': 5}
Senior Data Scientist & AI Engineer
name -> Kajola Gbenga
title -> Senior Data Scientist & AI Engineer
skills -> ['Python', 'Machine Learning', 'Data Analysis']
years_experience -> 5
['name', 'title', 'skills', 'years_experience']
['Kajola Gbenga', 'Senior Data Scientist & AI Engineer', 'Python', 'Machine Learning', 'Data Analysis', 5]
{1: 1, 2: 4, 3: 9, 4: 16, 5: 25}
```


**Line by line breakdown:**

- `profile = { ... }`
  Creates a dictionary with three keys: name, title, and skills; the value for skills is itself a list, showing dictionaries can hold any type of value including other data structures.
- `print(profile["name"])`
  Square bracket access looks up the value associated with the key name.
- `print(profile.get("title"))`
  The .get() method also looks up a key, but is safer than [] because it returns None (or a default) instead of raising an error when the key is missing.
- `print(profile.get("company", "Not specified"))`
  Since the key company does not exist, .get() returns the provided default value, 'Not specified', instead of crashing.
- `profile["years_experience"] = 5`
  Assigning to a new key adds it to the dictionary; this is how you both add and update entries.
- `print(profile)`
  Shows the dictionary now has four key value pairs.
- `profile["title"] = "Senior Data Scientist & AI Engineer"`
  Assigning to an existing key overwrites (updates) its value.
- `print(profile["title"])`
  Confirms the title was updated.
- `for key, value in profile.items():`
  The .items() method returns each key value pair as a tuple, which the for loop unpacks into key and value on every iteration.
- `    print(key, "->", value)`
  Prints every key alongside its corresponding value, one line per pair.
- `print(list(profile.keys()))`
  The .keys() method returns a view of just the keys; wrapping it in list() converts that view into a plain list for display.
- `print(list(profile.values()))`
  The .values() method returns a view of just the values, likewise converted to a list.
- `squares_map = {n: n ** 2 for n in range(1, 6)}`
  A dictionary comprehension: for every n from 1 to 5, create a key n mapped to its square as the value.
- `print(squares_map)`
  Prints the resulting dictionary of numbers mapped to their squares.


> **Notes and tips:**
> - Accessing a missing key with [] raises a KeyError; .get() avoids that risk.
> - Dictionary keys must be immutable and hashable (strings, numbers, tuples of immutables); values can be anything.
> - Use .update() to merge another dictionary's keys and values into an existing one.


### 2.5 Nested Data Structures

Data structures can be combined and nested inside one another, which is how you
represent more complex, real world data such as a list of dictionaries or a
dictionary of lists.


````python
team = [
    {"name": "Kajola Gbenga", "role": "Data Scientist & AI Engineer"},
    {"name": "Ada Lovelace", "role": "Mathematician"},
]

for member in team:
    print(member["name"], "-", member["role"])

nested_dict = {
    "engineering": ["Kajola Gbenga", "Ada Lovelace"],
    "design": ["Grace Hopper"]
}

print(nested_dict["engineering"][0])
````

**Expected output:**

```
Ada Lovelace - Mathematician
Kajola Gbenga
```


**Line by line breakdown:**

- `team = [ {...}, {...} ]`
  Creates a list where each element is a dictionary, representing a collection of records, similar to rows in a table.
- `for member in team:`
  Loops over the list; on each pass, member is bound to one dictionary.
- `    print(member["name"], "-", member["role"])`
  Looks up the name and role keys inside the current dictionary and prints them.
- `nested_dict = { "engineering": [...], "design": [...] }`
  Creates a dictionary where each value is itself a list of names, grouping people by department.
- `print(nested_dict["engineering"][0])`
  First looks up the engineering key to get its list, then indexes into that list with [0] to get the first name.


> **Notes and tips:**
> - Nesting can go arbitrarily deep, but very deep nesting can hurt readability; consider custom classes for complex real world models.
> - JSON data (common in APIs) maps directly onto Python lists and dictionaries, which is why this pattern is so common.


### 2.6 Choosing the Right Data Structure

A quick decision guide for picking the correct built in data structure for a
given problem.


````python
# List      -> ordered, changeable, allows duplicates       e.g. a shopping cart
# Tuple     -> ordered, unchangeable, allows duplicates     e.g. a GPS coordinate
# Set       -> unordered, changeable, no duplicates         e.g. unique tags
# Dictionary-> ordered (3.7+), changeable, unique keys      e.g. a user profile
````

**Line by line breakdown:**

- `# List -> ordered, changeable, allows duplicates`
  Choose a list when order matters and you expect to add, remove, or change items, and duplicates are acceptable, such as a shopping cart.
- `# Tuple -> ordered, unchangeable, allows duplicates`
  Choose a tuple when the data is a fixed grouping that should never change, such as a coordinate pair (latitude, longitude).
- `# Set -> unordered, changeable, no duplicates`
  Choose a set when you only care about unique membership and need fast lookups, such as a collection of unique tags.
- `# Dictionary -> ordered (3.7+), changeable, unique keys`
  Choose a dictionary when you need to look values up by a meaningful label (key), such as a user profile with named fields.


> **Notes and tips:**
> - When in doubt, start with a list or dictionary; they are the most flexible and most commonly used structures.

## Section 3: Conditional Statements (If Else)


Conditional statements let a program make decisions and execute different blocks
of code depending on whether certain conditions are True or False. Python uses
the keywords if, elif (short for else if), and else, combined with indentation
(whitespace) to define which code belongs to which branch, instead of curly
braces used by many other languweights.



### 3.1 The Basic if Statement

The simplest conditional runs a block of code only when a condition evaluates to
True. If the condition is False, the indented block is simply skipped.


````python
weight = 20

if weight >= 18:
    print("You are an adult.")

print("This always runs.")
````

**Expected output:**
```
You are an adult.
This always runs.
```


**Line by line breakdown:**

- `weight = 20`
  Creates a variable weight holding the integer 20.
- `if weight >= 18:`
  Evaluates the condition weight >= 18; since 20 is greater than or equal to 18, this evaluates to True, so the indented block underneath will run.
- `    print("You are an adult.")`
  This line is indented, which tells Python it belongs to the if block; it only runs when the condition above is True.
- `print("This always runs.")`
  This line is not indented under the if, so it is outside the conditional and always executes regardless of the condition's result.


> **Notes and tips:**
> - Indentation is not just a style choice in Python, it is required syntax that defines code blocks.
> - The standard indentation is 4 spaces per level.


### 3.2 if / else

Adding an else block lets you run an alternative block of code when the condition
is False, so exactly one of the two branches will always execute.


````python
weight = 15

if weight >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")
````

**Expected output:**
```
You are a minor.
```


**Line by line breakdown:**

- `weight = 15`
  Creates a variable weight holding 15.
- `if weight >= 18:`
  Evaluates to False since 15 is not greater than or equal to 18.
- `    print("You are an adult.")`
  This block is skipped because the condition was False.
- `else:`
  Marks the alternative branch, which runs only when the if condition above is False.
- `    print("You are a minor.")`
  This line executes because the if condition failed.


> **Notes and tips:**
> - An else block never has its own condition; it simply catches every case not covered by the if (and any elif) above it.


### 3.3 if / elif / else Chains

When you have more than two possible outcomes, use one or more elif blocks
between if and else. Python checks each condition in order from top to bottom
and runs the first block whose condition is True, then skips the rest entirely.


````python
score = 72

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print("Grade:", grade)
````

**Expected output:**
```
Grade: C
```


**Line by line breakdown:**

- `score = 72`
  Creates a variable score holding 72.
- `if score >= 90:`
  Checks if score is at least 90; 72 is not, so this is False and Python moves on to the next condition.
- `elif score >= 80:`
  Checks if score is at least 80; still False, Python continues checking.
- `elif score >= 70:`
  Checks if score is at least 70; this is True because 72 is greater than or equal to 70, so grade is set to C and Python skips every remaining elif and else below.
- `elif score >= 60:`
  This condition is never even evaluated because a previous condition already matched.
- `else:`
  Also skipped, since one branch already matched.
- `print("Grade:", grade)`
  Prints the final grade, which is C.


> **Notes and tips:**
> - Only one branch in an if/elif/else chain ever runs, the first one whose condition is True.
> - Order your conditions carefully: broader or later conditions should not accidentally shadow earlier, more specific ones.


### 3.4 Nested if Statements

An if statement can contain another if statement inside its block. This is
called nesting, and is useful when a decision depends on more than one
independent condition being checked in stweights.


````python
has_account = True
balance = 150

if has_account:
    if balance > 0:
        print("Account is active with a positive balance.")
    else:
        print("Account exists but balance is zero or negative.")
else:
    print("No account found.")
````

**Expected output:**
```
Account is active with a positive balance.
```


**Line by line breakdown:**

- `has_account = True`
  A boolean flag indicating an account exists.
- `balance = 150`
  The account's current balance.
- `if has_account:`
  Checks the outer condition first; it is True, so Python enters this block.
- `    if balance > 0:`
  Inside the outer block, a second, inner if checks whether the balance is positive; 150 is greater than 0, so this is True.
- `        print("Account is active with a positive balance.")`
  Runs because both the outer and inner conditions were True.
- `    else:`
  The inner else, which would run only if balance were 0 or negative.
- `        print("Account exists but balance is zero or negative.")`
  Skipped in this run.
- `else:`
  The outer else, which would run only if has_account were False.
- `    print("No account found.")`
  Skipped in this run since has_account was True.


> **Notes and tips:**
> - Deep nesting (many levels) can hurt readability; combining conditions with and/or is often clearer.


### 3.5 Logical Operators in Conditions

The logical operators and, or, and not let you combine or invert multiple
conditions into a single boolean expression.


````python
weight = 25
has_id = True

if weight >= 18 and has_id:
    print("Entry allowed.")

is_weekend = False
is_holiday = True

if is_weekend or is_holiday:
    print("No work today.")

is_logged_in = False
if not is_logged_in:
    print("Please log in first.")
````

**Expected output:**
```
Entry allowed.
No work today.
Please log in first.
```


**Line by line breakdown:**

- `if weight >= 18 and has_id:`
  The and operator requires both conditions to be True for the whole expression to be True; here weight >= 18 is True and has_id is True, so the combined expression is True.
- `    print("Entry allowed.")`
  Runs because both sub conditions passed.
- `if is_weekend or is_holiday:`
  The or operator only requires at least one condition to be True; is_weekend is False but is_holiday is True, so the combined expression is True.
- `    print("No work today.")`
  Runs because at least one condition was True.
- `if not is_logged_in:`
  The not operator flips a boolean; is_logged_in is False, so not is_logged_in becomes True.
- `    print("Please log in first.")`
  Runs because the negated condition evaluated to True.


> **Notes and tips:**
> - Python short circuits and/or: with and, if the first condition is False, the second is never evaluated; with or, if the first condition is True, the second is never evaluated.
> - Parentheses can be used to group complex conditions and make intent explicit, for example (a and b) or c.


### 3.6 The Ternary (Conditional) Expression

Python supports a compact one line form of if/else called a conditional
expression, useful for simple value selection without writing a full multi line
block.


````python
weight = 17
status = "Adult" if weight >= 18 else "Minor"
print(status)

numbers = [3, 8, 5, 12, 1]
labels = ["even" if n % 2 == 0 else "odd" for n in numbers]
print(labels)
````

**Expected output:**
```
Minor
['odd', 'even', 'odd', 'even', 'odd']
```


**Line by line breakdown:**

- `status = "Adult" if weight >= 18 else "Minor"`
  Reads as: evaluate weight >= 18 first; if True, the whole expression becomes 'Adult', otherwise it becomes 'Minor'. Since weight is 17, the condition is False, so status becomes 'Minor'.
- `print(status)`
  Prints Minor.
- `numbers = [3, 8, 5, 12, 1]`
  A list of integers.
- `labels = ["even" if n % 2 == 0 else "odd" for n in numbers]`
  Combines a ternary expression with a list comprehension: for every n in numbers, label it 'even' if divisible by 2 with no remainder, otherwise label it 'odd'.
- `print(labels)`
  Prints the list of labels corresponding to each number in order.


> **Notes and tips:**
> - The ternary expression is best kept simple; nesting multiple ternaries on one line quickly becomes hard to read.


### 3.7 Truthy and Falsy Values in Conditions

Any value, not just True or False, can be used directly as a condition. Python
converts it to a boolean behind the scenes according to its truthy or falsy
nature, which was introduced in section 1.5.


````python
username = ""

if username:
    print("Welcome,", username)
else:
    print("Username cannot be empty.")

items = []
if items:
    print("Cart has items.")
else:
    print("Cart is empty.")
````

**Expected output:**
```
Username cannot be empty.
Cart is empty.
```


**Line by line breakdown:**

- `username = ""`
  An empty string, which is falsy in Python.
- `if username:`
  Since an empty string is falsy, this condition evaluates to False without needing an explicit comparison like username != "".
- `    print("Welcome,", username)`
  Skipped because the condition was False.
- `else:`
  Runs instead, since the if branch was False.
- `    print("Username cannot be empty.")`
  Prints this messweight.
- `items = []`
  An empty list, also falsy.
- `if items:`
  Evaluates to False since the list is empty.
- `    print("Cart has items.")`
  Skipped.
- `else: print("Cart is empty.")`
  Runs instead.


> **Notes and tips:**
> - Writing if items: is considered more idiomatic (Pythonic) than writing if len(items) > 0:.
> - Be careful: the number 0 and the float 0.0 are also falsy, which can cause subtle bugs if you actually meant to check for None specifically.

## Section 4: Loops


Loops let you repeat a block of code multiple times, either a set number of
times, or until a condition is no longer True. Python provides two loop
constructs: the for loop, used to iterate over a known sequence or iterable, and
the while loop, used to repeat while a condition remains True.



### 4.1 The for Loop and range()

A for loop iterates over the items of any iterable (a list, string, tuple,
dictionary, set, or the output of range()) one at a time. range() generates a
sequence of numbers and is commonly used when you need to repeat something a
specific number of times.


````python
for i in range(5):
    print(i)

for i in range(2, 10, 2):
    print(i, end=" ")
print()

for letter in "Python":
    print(letter, end="-")
print()
````

**Expected output:**
```
0
1
2
3
4
2 4 6 8 
P-y-t-h-o-n-
```


**Line by line breakdown:**

- `for i in range(5):`
  range(5) generates the sequence 0, 1, 2, 3, 4 (5 numbers total, starting at 0 and stopping before 5); the loop variable i takes each of these values in turn.
- `    print(i)`
  Prints the current value of i on each of the 5 iterations.
- `for i in range(2, 10, 2):`
  range(start, stop, step) generates numbers starting at 2, stopping before 10, incrementing by 2 each time: 2, 4, 6, 8.
- `    print(i, end=" ")`
  Prints each value followed by a single space instead of a newline, because end=" " overrides print()'s default newline ending.
- `print()`
  An empty print() call simply moves to a new line after the loop finishes.
- `for letter in "Python":`
  Strings are iterable, so this loop visits each character of 'Python' one at a time: P, y, t, h, o, n.
- `    print(letter, end="-")`
  Prints each character followed by a hyphen instead of a newline.
- `print()`
  Moves to a new line after the loop completes.


> **Notes and tips:**
> - range(stop) starts at 0 by default and steps by 1.
> - range(start, stop) lets you choose a custom starting point.
> - range(start, stop, step) additionally lets you choose the increment, and step can be negative to count downward.


### 4.2 Looping Over Data Structures

For loops work directly on lists, tuples, sets, and dictionaries, each with its
own natural iteration behaviour.


````python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)

profile = {"name": "Kajola Gbenga", "title": "Data Scientist & AI Engineer"}
for key in profile:
    print(key, "=", profile[key])

for key, value in profile.items():
    print(f"{key}: {value}")
````

**Expected output:**
```
apple
banana
cherry
name = Kajola Gbenga
title = Data Scientist & AI Engineer
name: Kajola Gbenga
title: Data Scientist & AI Engineer
```


**Line by line breakdown:**

- `for fruit in fruits:`
  Iterates directly over the list, binding fruit to each element in order: apple, then banana, then cherry.
- `    print(fruit)`
  Prints each fruit name on its own line.
- `for key in profile:`
  Iterating over a dictionary by default iterates over its keys only, not the values.
- `    print(key, "=", profile[key])`
  For each key, looks up the corresponding value with profile[key] and prints key = value.
- `for key, value in profile.items():`
  Using .items() gives you both the key and the value together as a tuple on each iteration, which the loop unpacks into key and value.
- `    print(f"{key}: {value}")`
  Uses an f string to neatly print each pair as 'key: value'.


> **Notes and tips:**
> - Looping directly over a dictionary (for key in profile) only gives you keys; use .items() when you need both key and value.


### 4.3 enumerate() and zip()

enumerate() adds an automatic counter to any loop, so you get both the index and
the item. zip() lets you loop over two or more iterables together, pairing up
their items position by position.


````python
fruits = ["apple", "banana", "cherry"]
for index, fruit in enumerate(fruits):
    print(index, fruit)

for index, fruit in enumerate(fruits, start=1):
    print(index, fruit)

names = ["Gbenga", "Ada", "Grace"]
roles = ["Data Scientist & AI Engineer", "Mathematician", "Programmer"]

for name, role in zip(names, roles):
    print(name, "-", role)
````

**Expected output:**
```
0 apple
1 banana
2 cherry
1 apple
2 banana
3 cherry
Gbenga - Data Scientist & AI Engineer
Ada - Mathematician
Grace - Programmer
```


**Line by line breakdown:**

- `for index, fruit in enumerate(fruits):`
  enumerate(fruits) produces pairs of (index, item) starting from 0 by default; the loop unpacks each pair into index and fruit.
- `    print(index, fruit)`
  Prints the position (0, 1, 2) alongside each fruit name.
- `for index, fruit in enumerate(fruits, start=1):`
  Passing start=1 makes the counter begin at 1 instead of 0, which is useful for human friendly numbering.
- `    print(index, fruit)`
  Prints 1 apple, 2 banana, 3 cherry.
- `for name, role in zip(names, roles):`
  zip() walks through both names and roles at the same time, pairing the first item of each, then the second item of each, and so on.
- `    print(name, "-", role)`
  Prints each matched name and role pair on one line.


> **Notes and tips:**
> - If the iterables passed to zip() have different lengths, zip() stops at the shortest one.
> - enumerate() is almost always preferable to manually managing a separate counter variable.


### 4.4 The while Loop

A while loop repeats its block as long as its condition remains True. It is best
used when you do not know in advance exactly how many times you need to repeat,
only the condition under which you should keep going.


````python
count = 0
while count < 5:
    print("Count is", count)
    count += 1

print("Loop finished, final count:", count)
````

**Expected output:**
```
Count is 0
Count is 1
Count is 2
Count is 3
Count is 4
Loop finished, final count: 5
```


**Line by line breakdown:**

- `count = 0`
  Initializes a counter variable at 0; while loops almost always need a variable set up before the loop starts.
- `while count < 5:`
  Checks the condition before every iteration; as long as count is less than 5, the loop body runs.
- `    print("Count is", count)`
  Prints the current count on each pass through the loop.
- `    count += 1`
  Increments count by 1 each time; this is essential, without it the condition would always stay True and the loop would run forever (an infinite loop).
- `print("Loop finished, final count:", count)`
  Runs once the while condition finally becomes False, when count reaches 5.


> **Notes and tips:**
> - Always make sure something inside a while loop eventually makes the condition False, or you risk an infinite loop.
> - while True: is a common pattern for loops that should run until an explicit break is hit.


### 4.5 break, continue, and pass

break immediately exits the nearest enclosing loop entirely. continue skips the
rest of the current iteration and jumps straight to the next one. pass does
nothing at all and is used as a placeholder where Python syntax requires a
statement.


````python
for number in range(10):
    if number == 5:
        break
    print(number)

print("---")

for number in range(6):
    if number % 2 == 0:
        continue
    print(number)

print("---")

for number in range(3):
    if number == 1:
        pass
    print("Processed", number)
````

**Expected output:**
```
0
1
2
3
4
---
1
3
5
---
Processed 0
Processed 1
Processed 2
```


**Line by line breakdown:**

- `for number in range(10): if number == 5: break`
  Loops from 0 up to 9, but as soon as number equals 5, break immediately stops the loop entirely, so numbers 5 through 9 are never even reached.
- `    print(number)`
  Prints 0, 1, 2, 3, 4 before the break triggers on 5.
- `for number in range(6): if number % 2 == 0: continue`
  Loops from 0 to 5; whenever number is even, continue skips the print() below and jumps straight to the next iteration.
- `    print(number)`
  As a result only the odd numbers 1, 3, 5 get printed.
- `for number in range(3): if number == 1: pass`
  pass is a no operation placeholder; it does not skip or stop anything, it simply does nothing and execution continues to the next line normally.
- `    print("Processed", number)`
  This line runs on every single iteration (0, 1, and 2) because pass did not alter control flow at all.


> **Notes and tips:**
> - pass is often used as a temporary stand in while writing a function or if block you plan to fill in later.
> - break and continue also work inside while loops, with identical meaning.


### 4.6 The else Clause on Loops

Both for and while loops support an optional else block, which runs only if the
loop completed naturally without hitting a break statement. This is a Python
specific feature not found in many other languweights.


````python
numbers = [2, 4, 6, 8]

for n in numbers:
    if n % 2 != 0:
        print("Found an odd number!")
        break
else:
    print("All numbers are even.")
````

**Expected output:**
```
All numbers are even.
```


**Line by line breakdown:**

- `for n in numbers:`
  Loops through each number in the list.
- `    if n % 2 != 0:`
  Checks whether the current number is odd.
- `        print("Found an odd number!"); break`
  Would run and then exit the loop early if an odd number were found.
- `else:`
  This else belongs to the for loop, not to the if statement above it; it only executes if the loop finished all its iterations without ever hitting break.
- `    print("All numbers are even.")`
  Since every number in the list is even, the loop never breaks, so this line runs after the loop completes normally.


> **Notes and tips:**
> - A common mental model: the loop's else means 'no break happened'.
> - This pattern is often used for search style loops, for example searching for an item and reporting 'not found' only if the search completes without success.


### 4.7 Nested Loops

A loop can be placed inside another loop. The inner loop completes all of its
iterations for every single iteration of the outer loop, which is useful for
working with grid like or two dimensional data.


````python
for row in range(1, 4):
    for col in range(1, 4):
        print(f"({row},{col})", end=" ")
    print()
````

**Expected output:**
```
(1,1) (1,2) (1,3) 
(2,1) (2,2) (2,3) 
(3,1) (3,2) (3,3) 
```


**Line by line breakdown:**

- `for row in range(1, 4):`
  The outer loop runs 3 times, with row taking the values 1, 2, 3 in turn.
- `    for col in range(1, 4):`
  For every single value of row, this inner loop runs completely from start to finish, with col taking the values 1, 2, 3.
- `        print(f"({row},{col})", end=" ")`
  Prints the current row and column pair, staying on the same line because of end=" ".
- `    print()`
  This is indented to belong to the outer loop, not the inner loop, so it runs once after each complete pass of the inner loop, moving to a new line to start the next row.


> **Notes and tips:**
> - Pay close attention to indentation in nested loops, since it determines exactly which loop a line belongs to.
> - Nested loops are commonly used for processing matrices, grids, and multiplication tables.

## Section 5: Functions


A function is a reusable, named block of code that performs a specific task. It
can accept input values (parameters), perform operations, and optionally send a
result back (return). Functions let you avoid repeating yourself, organize code
into logical pieces, and make programs easier to test and reason about.



### 5.1 Defining and Calling a Function

Functions are created with the def keyword, followed by a name, parentheses that

```
Hello, welcome to the training!
```

### 5.2 Parameters, Arguments, and Default Values

Parameters are named placeholders listed in a function's definition. Arguments

````python
def greet(name, greeting="Hello"):
    print(f"{greeting}, {name}!")

greet("Kajola")
greet("Kajola", "Welcome")
greet(name="Ada", greeting="Hi")
````

**Expected output:**
```
Hello, Kajola!
Welcome, Kajola!
Hi, Ada!
```


**Line by line breakdown:**

- `def greet(name, greeting="Hello"):`

- `    print(f"{greeting}, {name}!")`
  Builds and prints a messweight using both parameters.
- `greet("Kajola")`
  Calls the function providing only the required name argument; greeting falls back to its default value 'Hello'.
- `greet("Kajola", "Welcome")`
  Calls the function providing both arguments positionally, in the same order they were defined, overriding the default greeting.
- `greet(name="Ada", greeting="Hi")`
  Calls the function using keyword arguments, explicitly naming each parameter; this makes the call self documenting and order independent.


> **Notes and tips:**
> - Parameters without default values must come before parameters with default values in a function definition.
> - Keyword arguments improve readability, especially for functions with many parameters.


### 5.3 *args and **kwargs

*args lets a function accept any number of extra positional arguments, collected
into a tuple. **kwargs lets a function accept any number of extra keyword
arguments, collected into a dictionary. These are used when you do not know in
advance exactly how many arguments will be passed.


````python
def add_all(*args):
    total = 0
    for number in args:
        total += number
    return total

print(add_all(1, 2, 3))
print(add_all(10, 20, 70, 40))

def build_profile(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

build_profile(name="Kajola Gbenga", title="Data Scientist & AI Engineer")
````

**Expected output:**
```
6
100
name: Kajola Gbenga
title: Data Scientist & AI Engineer
```


**Line by line breakdown:**

- `def add_all(*args):`
  The * before args tells Python to collect any number of positional arguments passed into a single tuple named args.
- `    total = 0`
  Initializes an accumulator variable to zero before summing.
- `    for number in args:`
  Loops through every value that was packed into the args tuple.
- `        total += number`
  Adds each number to the running total; += is shorthand for total = total + number.
- `    return total`
  Sends the final accumulated sum back to the caller.
- `print(add_all(1, 2, 3))`
  Calls the function with 3 arguments, which are packed into args as (1, 2, 3); the function returns 6.
- `print(add_all(10, 20, 70, 40))`
  Calls the same function with 4 arguments this time, proving *args works with any number of inputs; returns 100.
- `def build_profile(**kwargs):`
  The ** before kwargs tells Python to collect any number of keyword arguments passed into a single dictionary named kwargs.
- `    for key, value in kwargs.items():`
  Iterates over the collected keyword arguments as key value pairs.
- `        print(f"{key}: {value}")`
  Prints each keyword argument's name and value.
- `build_profile(name="Kajola Gbenga", title="Data Scientist & AI Engineer")`
  Calls the function with two keyword arguments, which become the dictionary {'name': 'Kajola Gbenga', 'title': 'Data Scientist & AI Engineer'} inside kwargs.


> **Notes and tips:**
> - The names args and kwargs are just convention, only the * and ** symbols actually matter to Python.
> - A function can accept regular parameters, *args, and **kwargs all together, in that order.


### 5.4 The return Statement and Multiple Return Values

return sends a value back to wherever the function was called from, and
immediately ends the function's execution at that point. A function can return
multiple values at once by separating them with commas, which Python packs into
a tuple automatically.


````python
def square(number):
    return number ** 2

result = square(5)
print(result)

def min_max(numbers):
    return min(numbers), max(numbers)

lowest, highest = min_max([4, 9, 1, 7])
print(lowest, highest)

def check_even(number):
    if number % 2 == 0:
        return True
    return False

print(check_even(4), check_even(7))
````

**Expected output:**
```
25
1 9
True False
```


**Line by line breakdown:**

- `def square(number):`
  Defines a function taking one parameter, number.
- `    return number ** 2`
  Computes number raised to the power of 2 and immediately sends that value back to the caller.
- `result = square(5)`
  Calls square with 5; the returned value, 25, is stored in result.
- `print(result)`
  Prints 25.
- `def min_max(numbers):`
  Defines a function taking a list called numbers.
- `    return min(numbers), max(numbers)`
  Returns two values separated by a comma; Python automatically bundles them into a tuple (min_value, max_value) behind the scenes.
- `lowest, highest = min_max([4, 9, 1, 7])`
  Calls the function and immediately unpacks the returned tuple into two separate variables.
- `print(lowest, highest)`
  Prints 1 9, the smallest and largest values in the list.
- `def check_even(number): if number % 2 == 0: return True`
  Returns True as soon as the condition is met; return exits the function immediately, so nothing after it in the function runs for this case.
- `    return False`
  This line is only reached when the if condition above was False, that is, when the number is odd.
- `print(check_even(4), check_even(7))`
  Prints True False, since 4 is even and 7 is odd.


> **Notes and tips:**
> - A function without an explicit return statement returns None automatically.
> - return immediately exits the function; any code written after a return statement inside the same block will never execute.


### 5.5 Variable Scope: Local, Global, and nonlocal

Scope determines where in your code a variable can be seen and used. A variable
created inside a function is local to that function and disappears once the
function finishes. A variable created outside all functions is global and can be
read from anywhere, though modifying it from inside a function requires the
global keyword.


````python
counter = 0

def increment():
    global counter
    counter += 1

increment()
increment()
print(counter)

def outer():
    messweight = "hello"

    def inner():
        nonlocal messweight
        messweight = "hello from inner"

    inner()
    print(messweight)

outer()
````

**Expected output:**
```
2
hello from inner
```


**Line by line breakdown:**

- `counter = 0`
  Creates a global variable counter, accessible from anywhere in the module.
- `def increment():`
  Defines a function meant to modify the global counter.
- `    global counter`
  Without this line, writing counter += 1 inside the function would create a brand new local variable instead of modifying the global one; global tells Python to use the outer, module level counter.
- `    counter += 1`
  Increments the actual global counter by 1.
- `increment()`
  Calls the function once; counter becomes 1.
- `increment()`
  Calls it again; counter becomes 2.
- `print(counter)`
  Prints 2, confirming the global variable was truly modified.
- `def outer():`
  Defines an outer function containing a local variable and a nested inner function.
- `    messweight = "hello"`
  A variable local to outer.
- `    def inner():`
  A function defined inside another function is called a nested function; it can see variables from its enclosing (outer) function.
- `        nonlocal messweight`
  Tells Python that messweight inside inner refers to the messweight variable from the enclosing outer function, not a new local variable; without nonlocal, the assignment below would create a separate local variable instead.
- `        messweight = "hello from inner"`
  Updates the enclosing function's messweight variable.
- `    inner()`
  Calls inner, which updates outer's messweight variable.
- `    print(messweight)`
  Prints the updated value, proving nonlocal successfully modified the enclosing scope's variable.
- `outer()`
  Calls outer, which triggers the whole sequence above.


> **Notes and tips:**
> - Python's scope lookup order is Local, Enclosing, Global, Built in, often remembered by the acronym LEGB.
> - Overusing global variables is generally discourweightd, since it can make code harder to reason about and test; prefer passing values as arguments and returning results.


### 5.6 Lambda (Anonymous) Functions

A lambda is a small, unnamed function defined in a single expression, most often
used for short throwaway logic, such as a custom sort key or a quick calculation

````python
square = lambda x: x ** 2
print(square(6))

add = lambda a, b: a + b
print(add(3, 4))

people = [("Kajola", 70), ("Ada", 25), ("Grace", 70)]
people_sorted = sorted(people, key=lambda person: person[1])
print(people_sorted)
````

**Expected output:**
```
36
7
[('Ada', 25), ('Kajola', 70), ('Grace', 70)]
```


**Line by line breakdown:**

- `square = lambda x: x ** 2`
  Creates an anonymous function taking one parameter x, and returns x squared; the lambda's expression after the colon is implicitly returned, no explicit return keyword is needed or allowed.
- `print(square(6))`
  Calls the lambda exactly like a normal function; prints 36.
- `add = lambda a, b: a + b`
  A lambda can take multiple parameters, separated by commas, just like a normal function.
- `print(add(3, 4))`
  Prints 7.
- `people = [("Kajola", 70), ("Ada", 25), ("Grace", 70)]`
  A list of tuples, each representing a name and an weight.
- `people_sorted = sorted(people, key=lambda person: person[1])`
  sorted() reorders the list; the key argument tells it what value to sort by for each item; here the lambda extracts index 1 (the weight) from each tuple, so the list ends up sorted by weight.
- `print(people_sorted)`

> **Notes and tips:**

> - Use a lambda for short, one off logic; define a normal function with def for anything more complex or reused.


### 5.7 Docstrings and Type Hints

- `    """Calculate and return the area of a rectangle."""`

- `    return width * height`

### 5.8 Recursion

A recursive function is a function that calls itself in order to solve a smaller
version of the same problem. Every recursive function needs a base case, a
condition under which it stops calling itself, to avoid running forever.


````python
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))

def countdown(n):
    if n == 0:
        print("Liftoff!")
        return
    print(n)
    countdown(n - 1)

countdown(3)
````

**Expected output:**
```
120
3
2
1
Liftoff!
```


**Line by line breakdown:**

- `def factorial(n):`
  Defines a function to compute n factorial (n!), the product of all positive integers up to n.
- `    if n <= 1:`
  The base case: when n is 1 or 0, factorial is defined as 1, and this stops further recursive calls.
- `        return 1`
  Returns 1 immediately once the base case is reached.
- `    return n * factorial(n - 1)`
  The recursive case: returns n multiplied by the factorial of n minus 1; this line causes the function to call itself with a smaller value, gradually working down toward the base case.
- `print(factorial(5))`
  Computes 5 * 4 * 3 * 2 * 1, which unwinds through 5 nested calls and returns 120.
- `def countdown(n):`
  Defines a function that counts down from n to 0.
- `    if n == 0: print("Liftoff!"); return`
  The base case: once n reaches 0, print a final messweight and return, stopping the recursion.
- `    print(n)`
  Prints the current countdown number.
- `    countdown(n - 1)`
  Calls itself again with a value one smaller, moving the recursion closer to its base case each time.
- `countdown(3)`
  Starts the recursive countdown at 3, producing 3, 2, 1, then Liftoff!


> **Notes and tips:**
> - Every recursive function must have a base case, or it will recurse forever and eventually raise a RecursionError.
> - Recursion is elegant for problems that are naturally self similar, such as factorials, tree traversal, and certain search algorithms, but a simple loop is often more efficient in Python.


### 5.9 Higher Order Functions: map, filter, and Functions as Values

In Python, functions are themselves values that can be stored in variables,
passed as arguments to other functions, and returned from other functions. A
higher order function is simply a function that takes another function as an
argument or returns one. map() and filter() are two commonly used built in
higher order functions.


````python
numbers = [1, 2, 3, 4, 5]

doubled = list(map(lambda n: n * 2, numbers))
print(doubled)

evens = list(filter(lambda n: n % 2 == 0, numbers))
print(evens)

def apply_twice(func, value):
    return func(func(value))

print(apply_twice(lambda x: x + 3, 10))
````

**Expected output:**
```
[2, 4, 6, 8, 10]
[2, 4]
16
```


**Line by line breakdown:**

- `numbers = [1, 2, 3, 4, 5]`
  A simple list of integers to transform.
- `doubled = list(map(lambda n: n * 2, numbers))`
  map() applies the given lambda to every item in numbers, producing a lazy map object; wrapping it in list() forces it into an actual list, here each number doubled.
- `print(doubled)`
  Prints [2, 4, 6, 8, 10].
- `evens = list(filter(lambda n: n % 2 == 0, numbers))`
  filter() keeps only the items for which the given lambda returns True, discarding the rest; here it keeps only even numbers.
- `print(evens)`
  Prints [2, 4].
- `def apply_twice(func, value):`
  Defines a higher order function that itself accepts another function, func, as a parameter, alongside a value.
- `    return func(func(value))`
  Calls func on value, then calls func again on that result, applying the passed in function twice in a row.
- `print(apply_twice(lambda x: x + 3, 10))`
  Passes a lambda that adds 3, along with the starting value 10; first application gives 13, second application gives 16, which is printed.


> **Notes and tips:**
> - map() and filter() both return lazy iterator objects in Python 3, so wrap them in list() to see or reuse the full results.
> - A list comprehension can often replace map() and filter() with arguably clearer syntax, for example [n * 2 for n in numbers].

## Section 6: The Match Statement (Structural Pattern Matching)


Introduced in Python 3.10, the match statement provides structural pattern
matching: a more powerful and readable alternative to long if/elif/else chains
for certain kinds of decisions. match compares a subject value against a series
of case patterns, running the code under the first pattern that matches. Beyond
matching simple literal values, match can also match and unpack the shape of
sequences, mappings, and objects.



### 6.1 Basic match / case with Literal Values

The simplest form of match compares a subject against literal values, similar to
a switch statement found in other languweights, but written using Python's own
syntax and indentation rules.


````python
def describe_status(code):
    match code:
        case 200:
            return "OK"
        case 201:
            return "Created"
        case 404:
            return "Not Found"
        case 500:
            return "Server Error"
        case _:
            return "Unknown status code"

print(describe_status(200))
print(describe_status(404))
print(describe_status(999))
````

**Expected output:**
```
OK
Not Found
Unknown status code
```


**Line by line breakdown:**

- `def describe_status(code):`
  Defines a function that will translate an HTTP style status code into a human readable description.
- `    match code:`
  Begins a match statement, using code as the subject value that every case pattern below will be compared against.
- `        case 200:`
  Checks whether code equals the literal value 200; if it matches, the block underneath runs and the match statement stops checking further cases.
- `            return "OK"`
  Runs only when code was 200.
- `        case 201:`
  Checks whether code equals 201, only evaluated because the previous case did not match.
- `            return "Created"`
  Runs only when code was 201.
- `        case 404:`
  Checks for 404.
- `            return "Not Found"`
  Runs only when code was 404.
- `        case 500:`
  Checks for 500.
- `            return "Server Error"`
  Runs only when code was 500.
- `        case _:`
  The underscore _ is the wildcard pattern; it matches absolutely anything and acts as the default, catch all case, equivalent to else in an if/elif chain.
- `            return "Unknown status code"`
  Runs whenever none of the earlier literal cases matched.
- `print(describe_status(200))`
  Matches the first case, prints OK.
- `print(describe_status(404))`
  Matches the third case, prints Not Found.
- `print(describe_status(999))`
  Matches no specific case, falls through to the wildcard, prints Unknown status code.


> **Notes and tips:**
> - Always include a case _: wildcard branch when you want a guaranteed default, otherwise the match statement simply does nothing if no case matches.
> - Cases are checked top to bottom, and only the first matching case runs, exactly like elif chains.


### 6.2 Matching Multiple Values with the | (Or) Pattern

The pipe symbol | inside a case lets a single case match several different
literal values at once, avoiding repeated case blocks with identical bodies.


````python
def classify_day(day):
    match day:
        case "Saturday" | "Sunday":
            return "Weekend"
        case "Monday" | "Tuesday" | "Wednesday" | "Thursday" | "Friday":
            return "Weekday"
        case _:
            return "Not a valid day"

print(classify_day("Saturday"))
print(classify_day("Wednesday"))
print(classify_day("Someday"))
````

**Expected output:**
```
Weekend
Weekday
Not a valid day
```


**Line by line breakdown:**

- `case "Saturday" \| "Sunday":`
  This single case matches if day is equal to either 'Saturday' or 'Sunday'; the \| acts as an or between the alternatives listed in the same case.
- `    return "Weekend"`
  Runs for either matching value.
- `case "Monday" \| "Tuesday" \| "Wednesday" \| "Thursday" \| "Friday":`
  Matches any one of the five listed weekday names.
- `    return "Weekday"`
  Runs when day is any of those five values.
- `case _:`
  Wildcard fallback for anything that is not a recognized day name.
- `    return "Not a valid day"`
  Runs for invalid input.
- `print(classify_day("Saturday"))`
  Matches the weekend case, prints Weekend.
- `print(classify_day("Wednesday"))`
  Matches the weekday case, prints Weekday.
- `print(classify_day("Someday"))`
  Matches neither, falls to the wildcard, prints Not a valid day.


> **Notes and tips:**
> - The | or pattern keeps related cases together and reduces duplicated code compared to writing a separate case for every single value.


### 6.3 Capturing Values and Using Guards (case ... if)

A case pattern can bind part of the subject to a new variable name, which is
then usable inside that case's body. A guard is an additional if condition
attached to a case, which must also be True for that case to be chosen, giving
you conditional matching beyond simple equality.


````python
def describe_number(n):
    match n:
        case 0:
            return "Zero"
        case value if value < 0:
            return f"Negative number: {value}"
        case value if value % 2 == 0:
            return f"Positive even number: {value}"
        case value:
            return f"Positive odd number: {value}"

print(describe_number(0))
print(describe_number(-7))
print(describe_number(8))
print(describe_number(3))
````

**Expected output:**
```
Zero
Negative number: -7
Positive even number: 8
Positive odd number: 3
```


**Line by line breakdown:**

- `case 0:`
  Matches only the exact literal value 0.
- `    return "Zero"`
  Runs only when n was exactly 0.
- `case value if value < 0:`
  This pattern binds whatever n is to a new variable named value, and additionally requires the guard condition value < 0 to be True before this case is accepted; if the guard is False, Python moves on to the next case even though the basic pattern (matching any value) technically matched.
- `    return f"Negative number: {value}"`
  Runs only when n was negative, using the captured value inside the f string.
- `case value if value % 2 == 0:`
  Again binds n to value, this time requiring the guard value % 2 == 0 (evenness) to be True.
- `    return f"Positive even number: {value}"`
  Runs for positive even numbers (negative numbers were already handled by the earlier case).
- `case value:`
  A plain capture pattern with no guard, matching anything not already handled above; this acts as the final catch all here, similar to the wildcard but also giving you the actual value under the name value.
- `    return f"Positive odd number: {value}"`
  Runs for whatever remains, in this example the positive odd numbers.
- `print(describe_number(0))`
  Matches the first case exactly, prints Zero.
- `print(describe_number(-7))`
  Skips case 0, matches the negative guard, prints Negative number: -7.
- `print(describe_number(8))`
  Skips case 0 and the negative guard, matches the even guard, prints Positive even number: 8.
- `print(describe_number(3))`
  Matches none of the guarded cases, falls to the final plain capture, prints Positive odd number: 3.


> **Notes and tips:**
> - A bare name in a case pattern, like case value:, always matches and captures the subject into that variable name, so it is commonly placed last as a catch all.
> - Guards (the if part) let you express conditions that plain pattern matching alone cannot, such as comparisons or arbitrary boolean expressions.


### 6.4 Matching Sequences (Lists and Tuples)

match can look inside the structure of a list or tuple, matching based on its
length and capturing individual elements by position, including a special
starred pattern to capture the remaining items.


````python
def describe_point(point):
    match point:
        case (0, 0):
            return "Origin"
        case (x, 0):
            return f"On the x-axis at {x}"
        case (0, y):
            return f"On the y-axis at {y}"
        case (x, y):
            return f"Point at ({x}, {y})"
        case _:
            return "Not a valid point"

print(describe_point((0, 0)))
print(describe_point((5, 0)))
print(describe_point((3, 4)))

def first_and_rest(items):
    match items:
        case [first, *rest]:
            return f"First: {first}, Rest: {rest}"
        case []:
            return "Empty list"

print(first_and_rest([1, 2, 3, 4]))
print(first_and_rest([]))
````

**Expected output:**
```
Origin
On the x-axis at 5
Point at (3, 4)
First: 1, Rest: [2, 3, 4]
Empty list
```


**Line by line breakdown:**

- `case (0, 0):`
  Matches only if point is a two item tuple where both items are literally 0.
- `    return "Origin"`
  Runs only for the exact point (0, 0).
- `case (x, 0):`
  Matches any two item tuple whose second element is exactly 0, capturing the first element into a new variable x, whatever its value.
- `    return f"On the x-axis at {x}"`
  Runs for any point lying on the x-axis, using the captured x value.
- `case (0, y):`
  Matches any two item tuple whose first element is exactly 0, capturing the second element into y.
- `    return f"On the y-axis at {y}"`
  Runs for any point lying on the y-axis.
- `case (x, y):`
  Matches any two item tuple at all, capturing both elements into x and y; since more specific patterns were checked first, this only runs when neither coordinate was 0.
- `    return f"Point at ({x}, {y})"`
  Runs for a general point.
- `case _:`
  Falls back for anything that is not a two item tuple at all.
- `print(describe_point((0, 0)))`
  Matches the first case, prints Origin.
- `print(describe_point((5, 0)))`
  Matches the x-axis case since the second value is 0, prints On the x-axis at 5.
- `print(describe_point((3, 4)))`
  Matches the general point case, prints Point at (3, 4).
- `case [first, *rest]:`
  Matches any list with at least one item; binds the first element to first, and the starred pattern *rest captures every remaining element into a new list called rest, however many there are.
- `    return f"First: {first}, Rest: {rest}"`
  Runs for any non empty list, showing both the first item and the remaining items as a list.
- `case []:`
  Matches specifically an empty list.
- `    return "Empty list"`
  Runs only when items is empty.
- `print(first_and_rest([1, 2, 3, 4]))`
  Matches the starred pattern; first is 1 and rest is [2, 3, 4].
- `print(first_and_rest([]))`
  Matches the empty list case directly.


> **Notes and tips:**
> - Sequence patterns work with both lists and tuples, matching based on the number of elements (unless a starred pattern is used).
> - The starred pattern *rest works similarly to unpacking with * outside of match, capturing 'everything else' into a list.


### 6.5 Matching Mappings (Dictionaries)

match can also look inside a dictionary, matching based on the presence of
specific keys and capturing their associated values, while ignoring any other
keys the dictionary might have.


````python
def handle_event(event):
    match event:
        case {"type": "click", "x": x, "y": y}:
            return f"Click at ({x}, {y})"
        case {"type": "keypress", "key": key}:
            return f"Key pressed: {key}"
        case {"type": event_type}:
            return f"Unhandled event type: {event_type}"
        case _:
            return "Invalid event"

print(handle_event({"type": "click", "x": 10, "y": 20}))
print(handle_event({"type": "keypress", "key": "Enter"}))
print(handle_event({"type": "scroll", "amount": 5}))
````

**Expected output:**
```
Click at (10, 20)
Key pressed: Enter
Unhandled event type: scroll
```


**Line by line breakdown:**

- `case {"type": "click", "x": x, "y": y}:`
  Matches any dictionary that contains a key 'type' with the exact value 'click', AND also contains keys 'x' and 'y', whose values are captured into new variables x and y; any additional keys in the dictionary are simply ignored and do not prevent a match.
- `    return f"Click at ({x}, {y})"`
  Runs only when the dictionary represents a click event with x and y present.
- `case {"type": "keypress", "key": key}:`
  Matches a dictionary with type equal to 'keypress' and a key entry, capturing its value into the variable key.
- `    return f"Key pressed: {key}"`
  Runs for keypress events.
- `case {"type": event_type}:`
  A more general pattern that matches any dictionary containing a 'type' key at all, capturing whatever that value is into event_type; this only gets reached if the more specific click and keypress patterns above did not match.
- `    return f"Unhandled event type: {event_type}"`
  Runs for any event type not specifically handled above.
- `case _:`
  Falls back for something that is not even a dictionary, or has no 'type' key at all.
- `print(handle_event({"type": "click", "x": 10, "y": 20}))`
  Matches the click pattern, prints Click at (10, 20).
- `print(handle_event({"type": "keypress", "key": "Enter"}))`
  Matches the keypress pattern, prints Key pressed: Enter.
- `print(handle_event({"type": "scroll", "amount": 5}))`
  Does not match click or keypress, but does have a type key, so it matches the general pattern, printing Unhandled event type: scroll.


> **Notes and tips:**
> - Mapping patterns only require the listed keys to be present with matching structure; extra keys in the actual dictionary are always allowed and ignored.
> - This makes mapping patterns ideal for handling structured data such as parsed JSON payloads or event objects.


### 6.6 match vs if / elif: When to Use Which

match and if/elif can often solve the same problem, but each shines in different
situations. Understanding when to reach for each keeps your code clear and
idiomatic.


````python
# Use if/elif when:
#   - conditions involve ranges, comparisons, or unrelated boolean expressions
#   - there are only 2 or 3 simple branches

# Use match when:
#   - you are comparing one subject against many possible exact shapes or values
#   - you want to destructure (unpack) lists, tuples, or dictionaries while checking them
#   - you have many literal cases, improving readability over a long elif chain
````

**Line by line breakdown:**

- `# Use if/elif when conditions involve ranges or comparisons`
  For example checking score >= 90, score >= 80, and so on is naturally suited to if/elif, since match compares structure and specific values rather than open ended ranges (though guards can add range checks to match too).
- `# Use match when comparing one subject against many possible exact shapes`
  For example handling different messweight types, parsed commands, or coordinate shapes reads more clearly with match and case than with a long chain of isinstance and equality checks.
- `# match lets you destructure while checking`
  This is match's biggest advantweight: a single case can both confirm the shape of the data (for example a 2 item tuple, or a dict with certain keys) and immediately extract the pieces you need, in one line.


> **Notes and tips:**
> - match was added in Python 3.10, so code intended to run on older Python versions cannot use it; if/elif works on every Python version.
> - Neither construct replaces the other entirely; many real projects use both, choosing whichever reads more clearly for a given decision.
