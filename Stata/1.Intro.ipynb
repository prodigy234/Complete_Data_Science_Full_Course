# STATA

## INTRODUCTION TO STATA FOR DATA ANALYTICS

### 1. WHAT IS STATA?

Stata is a statistical software package used for:

- Data analysis

- Data management

- Statistical modeling

- Data visualization

- Econometrics and research

It is widely used in:

- Academia (Economics, Social Sciences, Public Health)

- Government institutions

- Financial and policy research organizations

- NGOs and development agencies

Unlike Python and R:

- Stata is menu-driven + command-driven

- Designed specifically for statistical analysis

- Less coding-heavy but highly structured

### 2. KEY FEATURES OF STATA

1. User-Friendly Interface
Point-and-click (GUI)
Command-line interface for automation

2. Powerful Statistical Tools
Regression (linear, logistic, panel)
Time series analysis
Hypothesis testing

3. Data Management Capabilities
Cleaning
Transformation
Merging datasets

4. Reproducibility
Uses Do-files (.do) similar to scripts in Python/R

5. Built-in Documentation
Excellent help system (help command)

### 3. STATA INTERFACE OVERVIEW

When you open Stata, you see:

Main Components

Command Window
- Where you type commands
- Equivalent to Python terminal or R console

Results Window
- Displays output

Variables Window
- Shows all variables in dataset

Data Editor
- Spreadsheet-like view

Review Window
- History of commands

Do-file Editor
- Script editor for writing reusable code

### 4. STATA WORKFLOW

Typical Workflow:

- Load data

- Explore data

- Clean data

- Analyze data

- Visualize data

- Save results

### 5. DATA TYPES IN STATA

Variable Types

| Type    | Description | Example      |
| ------- | ----------- | ------------ |
| Numeric | Numbers     | age, income  |
| String  | Text        | name, gender |
| Date    | Time-based  | 2024-01-01   |

### 6. BASIC STATA COMMANDS

🔹 General Syntax

command variable, options

🔹 1. Display Data

list
list age income

🔹 2. Describe Dataset

describe

🔹 3. Summarize Data

summarize
summarize age income

🔹 4. Detailed Summary

summarize age, detail

🔹 5. Tabulation

tabulate gender
tabulate gender region

### 7. IMPORTING DATA INTO STATA

Supported Formats:

- Excel (.xlsx)

- CSV (.csv)

- Stata (.dta)

🔹 Import Excel

import excel "file.xlsx", firstrow clear

🔹 Import CSV

import delimited "file.csv", clear

🔹 Load Stata File

use "file.dta", clear

### 8. DATA MANAGEMENT IN STATA

🔹 1. Creating Variables
generate income_usd = income * 0.001

🔹 2. Modifying Variables
replace income = income * 1000

🔹 3. Renaming Variables
rename oldname newname

🔹 4. Dropping Variables
drop age

🔹 5. Keeping Variables
keep age income gender

🔹 6. Filtering Data
keep if age > 18
drop if income < 1000

### 9. SORTING AND ORDERING

sort age

gsort -income   // descending

### 10. HANDLING MISSING VALUES

In Stata, missing values are represented as:

🔹 Check Missing

- misstable summarize

🔹 Drop Missing

- drop if missing(age)

### 11. DATA VISUALIZATION

**Histogram**

- histogram income

**Scatter Plot**

- scatter income age

**Line Graph**

- line sales year

**Bar Chart**

- graph bar income, over(gender)

### 12. STATISTICAL ANALYSIS

🔹 1. Correlation

correlate age income

🔹 2. Linear Regression

regress income age education

🔹 3. Logistic Regression

logit employed age education

🔹 4. T-Test

ttest income, by(gender)

### 13. DO-FILES (VERY IMPORTANT)

Equivalent to:

- Python scripts (.py)

- R scripts (.R)

🔹 Why Use Do-files?
- Reproducibility

- Automation

- Clean workflow

🔹 Example Do-file

clear

import excel "data.xlsx", firstrow

summarize

regress income age education

### 14. SAVING WORK

🔹 Save Dataset

save "data_clean.dta", replace

🔹 Export to Excel

export excel "output.xlsx", replace

### 15. HELP SYSTEM (VERY POWERFUL)

### 17. WHEN TO USE STATA

Use Stata when:

- Doing econometric analysis

- Working with structured datasets

- Performing policy research

- Running quick statistical tests

### 18. LIMITATIONS OF STATA

Not ideal for:

- Machine Learning

- Deep Learning

- Big Data processing

- Paid software (unlike Python/R)

### 19. BEST PRACTICES

✔ Always use Do-files

✔ Comment your code

✔ Keep raw data untouched

✔ Save versions of cleaned data

✔ Use meaningful variable names
